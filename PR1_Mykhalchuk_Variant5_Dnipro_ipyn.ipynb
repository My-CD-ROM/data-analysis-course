```python
# Завдання 1. Створення масивів

# Варіант 5 — Дніпро
temps = np.array([-4, -3, 3, 11, 17, 21, 23, 22, 16, 9, 3, -2])

print("Температури Дніпра:")
print(temps)

print("Тип:", temps.dtype)
print("Форма:", temps.shape)
print("Кількість елементів:", temps.size)
print("Кількість вимірів:", temps.ndim)
```

```python
# Завдання 2. Векторизовані обчислення

days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

temps_f = temps * 9 / 5 + 32

warm_days = days_in_month[temps > 15].sum()

print("Температури у °F:")
print(temps_f)

print("Кількість днів з температурою вище 15°C:", warm_days)

print("Кількість днів у році:", days_in_month.sum())
print("Перевірка 365 днів:", days_in_month.sum() == 365)
```

```python
# Завдання 3. Індексація і фільтрація

max_index = temps.argmax()
min_index = temps.argmin()

print("Найтепліший місяць:")
print("Температура:", temps[max_index], "°C")
print("Номер місяця:", month_numbers[max_index])
print("Індекс:", max_index)

print("\nНайхолодніший місяць:")
print("Температура:", temps[min_index], "°C")
print("Номер місяця:", month_numbers[min_index])
print("Індекс:", min_index)

print("\nМісяці з температурою нижче 0°C:")
print(month_numbers[temps < 0])

print("\nТемператури за зростанням:")
print(np.sort(temps))

print("\nОригінальний масив:")
print(temps)
```

```python
# Завдання 5. reshape і broadcasting

rows = np.array([temps, days_in_month])
correction = np.array([1, 0])

print("Форма rows:", rows.shape)
print("Форма correction:", correction.shape)

# Спроба додавання без reshape
try:
    result_error = rows + correction
except ValueError as e:
    print("\nПомилка при rows + correction:")
    print(e)

# Змінюємо форму correction з (2,) на (2, 1)
correction_reshaped = correction.reshape(2, 1)

print("\nНова форма correction:", correction_reshaped.shape)

result = rows + correction_reshaped

print("\nРезультат після broadcasting:")
print(result)
```

## Відповіді на питання

**1. Яка різниця між shape == (12,) і shape == (1, 12)?**  
(12,) — це одновимірний масив із 12 елементів.  
(1, 12) — це двовимірний масив з одним рядком і 12 стовпцями.

**2. Чому temps[temps > 15] має іншу довжину, ніж temps?**  
Тому що булева індексація залишає тільки ті елементи масиву, які задовольняють умову temps > 15.

**3. Що станеться при додаванні масиву форми (3,) до масиву форми (3, 12)?**  
Виникне помилка broadcasting, оскільки останні виміри 3 і 12 несумісні.

**4. Чому rows + correction дає помилку, а rows + correction.reshape(2, 1) — ні?**  
rows має форму (2, 12), а correction — (2,), тому виміри 12 і 2 несумісні. Після reshape correction має форму (2, 1). Розмір 1 може розширитися до 12, тому broadcasting працює і кожна поправка додається до відповідного рядка.
