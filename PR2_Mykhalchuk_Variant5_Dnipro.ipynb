```python
import pandas as pd

print("Практична робота №2")
print("Варіант 5 — Дніпро")
print("Михальчук Юліан, ІТ-42")

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

avg_temp = [-4, -3, 3, 11, 17, 21, 23, 22, 16, 9, 3, -2]

precip = [36, 34, 30, 32, 42, 50, 54, 44, 32, 32, 38, 40]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip
})

print("\nDataFrame:")
print(df)

print("\nФорма:")
print(df.shape)

print("\nТипи даних:")
print(df.dtypes)
```

```python
# Завдання 1. Індекс

# 1. Індекс за замовчуванням
print("DataFrame з індексом за замовчуванням:")
print(df)

print("\nІндекс:")
print(df.index)

# 2. Робимо month індексом
df_month = df.set_index("month")

print("\nDataFrame з індексом month:")
print(df_month)

print("\nРядок для місяця Лип:")
print(df_month.loc["Лип"])

# 3. Повертаємо звичайну нумерацію
df = df_month.reset_index()

print("\nПісля повернення звичайного індексу:")
print(df)
print("\nІндекс:", df.index)
```

```python
# Завдання 2. Відбір рядків і стовпців

# Числовий індекс від 1 до 12
df.index = range(1, 13)

# 1. avg_temp для місяців 1–6 через .loc
print("Через .loc:")
print(df.loc[1:6, "avg_temp"])

# 2. Те саме через .iloc за позиціями 0–5
print("\nЧерез .iloc:")
print(df.iloc[0:6, df.columns.get_loc("avg_temp")])

# .loc працює за підписами індексу,
# а .iloc — за позиціями рядків.

# 3. Лише month і precip
print("\nСтовпці month і precip:")
print(df[["month", "precip"]])
```

```python
# Завдання 3. Фільтрація за умовою

# 1. Місяці, де середня температура > 15°C
print("Місяці, де avg_temp > 15:")
print(df[df["avg_temp"] > 15])

# 2. Одночасно avg_temp > 10 і precip > 50
print("\nМісяці, де avg_temp > 10 і precip > 50:")
print(df[(df["avg_temp"] > 10) & (df["precip"] > 50)])

# 3. avg_temp < 0 або precip > 80
print("\nМісяці, де avg_temp < 0 або precip > 80:")
print(df[(df["avg_temp"] < 0) | (df["precip"] > 80)])
```

```python
# Завдання 4. Обчислювані стовпці

# 1. Температура у Фаренгейтах
df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32

# 2. True, якщо опади більші за середнє значення
df["is_wet"] = df["precip"] > df["precip"].mean()

# 3. Визначення пори року
def get_season(month_num):
    if month_num in [12, 1, 2]:
        return "зима"
    elif month_num in [3, 4, 5]:
        return "весна"
    elif month_num in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"

df["season"] = df.index.to_series().apply(get_season)

print(df)
```

## Відповіді на питання

**1. Чому `.loc[1:6]` і `.iloc[0:6]` можуть повернути різну кількість рядків?**  
`.loc` працює за підписами індексу і включає кінцевий підпис 6. `.iloc` працює за позиціями, де права межа не включається. Тому результат залежить від того, який індекс встановлений у таблиці.

**2. Чому `df["avg_temp"] > 10 and df["precip"] > 50` викликає помилку, а варіант з `&` — ні?**  
`and` призначений для окремих логічних значень, а стовпці pandas містять багато значень. Оператор `&` виконує логічну операцію поелементно для двох умов.

**3. У чому різниця між обчислюваним стовпцем через формулу і через `.apply()`?**  
Формула виконує операцію одразу над усім стовпцем і зручна для простих векторизованих обчислень. `.apply()` застосовує власну функцію до кожного значення та зручний для складнішої логіки, наприклад визначення пори року.
