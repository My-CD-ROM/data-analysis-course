```python
import pandas as pd

print("Практична робота №3")
print("Варіант 5 — Дніпро")
print("Михальчук Юліан, ІТ-42")

wide = pd.DataFrame({
    "місто": ["Дніпро"],
    "Січ": [74],
    "Лют": [70],
    "Бер": [63],
    "Кві": [53],
    "Тра": [45],
    "Чер": [39],
    "Лип": [35],
    "Сер": [37],
    "Вер": [47],
    "Жов": [60],
    "Лис": [72],
    "Гру": [78]
})

print("\nWide-таблиця:")
print(wide)

print("\nФорма таблиці:")
print(wide.shape)
```

```python
# Завдання 2. Перетворення wide → tidy

tidy = wide.melt(
    id_vars=["місто"],
    var_name="місяць",
    value_name="хмарність"
)

print("Tidy-таблиця:")
print(tidy)

print("\nФорма таблиці:")
print(tidy.shape)
```

## Завдання 3. Класифікація змінних

- **місто — номінальна шкала**, оскільки назва міста є категорією без природного порядку.
- **місяць — порядкова шкала**, оскільки місяці мають природну календарну послідовність.
- **хмарність — шкала відношень**, оскільки 0% означає відсутність хмарності та значення можна змістовно порівнювати за відношенням.

```python
# Завдання 4. Зворотне перетворення й перевірка

back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

print("Таблиця після pivot():")
print(back)

# Порівнюємо значення по місяцях
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

original_values = wide[months].iloc[0].to_numpy()
back_values = back.loc["Дніпро", months].to_numpy()

print("\nПочаткові значення:")
print(original_values)

print("\nЗначення після melt() і pivot():")
print(back_values)

print("\nДані збігаються:", (original_values == back_values).all())
```

```python
# Завдання 5. Читання й запис CSV та JSON

# 1. Зберігаємо tidy-таблицю
tidy.to_csv("variant.csv", index=False)
tidy.to_json(
    "variant.json",
    orient="records",
    force_ascii=False
)

# 2. Читаємо файли назад
csv_df = pd.read_csv("variant.csv")
json_df = pd.read_json("variant.json")

print("Дані з CSV:")
print(csv_df)

print("\nДані з JSON:")
print(json_df)

# Перевіряємо збіг
print("\nCSV збігається з оригіналом:", csv_df.equals(tidy))
print("JSON збігається з оригіналом:", json_df.equals(tidy))

# 3. Переглядаємо структуру файлів
print("\n--- variant.csv ---")
print(open("variant.csv", encoding="utf-8").read())

print("\n--- variant.json ---")
print(open("variant.json", encoding="utf-8").read())
```

## Відповіді на питання

**1. Що означає принцип tidy data «одна змінна — один стовпець»?**  
Кожна окрема характеристика повинна зберігатися у своєму стовпці. У нашій таблиці це місто, місяць і хмарність.

**2. Чому wide-формат зручний для перегляду, але не завжди зручний для аналізу?**  
Wide-формат компактно показує значення за місяцями, але назви місяців стають окремими стовпцями. У tidy-форматі кожне спостереження займає окремий рядок, тому дані простіше фільтрувати, групувати та аналізувати.

**3. У чому різниця між CSV та JSON?**  
CSV зберігає дані у вигляді рядків і стовпців, розділених символом-роздільником. JSON зберігає структуровані дані як об'єкти та пари «ключ — значення». У нашому випадку `orient="records"` представляє кожен рядок таблиці як окремий JSON-об'єкт.

### Різниця структури створених файлів
`variant.csv` має табличну структуру з назвами стовпців у першому рядку.  
`variant.json` містить список об'єктів, де кожен об'єкт відповідає одному рядку tidy-таблиці.
