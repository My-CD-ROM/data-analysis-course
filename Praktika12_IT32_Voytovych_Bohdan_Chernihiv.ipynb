Варіант 6 — Чернігів (липень = 20°C)

```python
import numpy as np

np.random.seed(6)   # номер варіанта
july_temp = 20      # липнева середня темп. Чернігів з таблиці
daily_temps = np.random.normal(loc=july_temp, scale=2.5, size=30)

print("Щоденні температури липня:", daily_temps)
print(f"Середнє: {daily_temps.mean():.3f}")
print(f"Стандартне відхилення (сип.): {daily_temps.std(ddof=1):.3f}")
```

```python
from scipy import stats

n = len(daily_temps)
mean = daily_temps.mean()
s = daily_temps.std(ddof=1)
se = s / np.sqrt(n)

# 95% довірчий інтервал за t-розподілом
ci_mean = stats.t.interval(0.95, df=n-1, loc=mean, scale=se)
print(f"95% довірчий інтервал для середньої: {ci_mean}")
```

```python
p_hat = (daily_temps > july_temp).mean()
n_p = n * p_hat
n_1p = n * (1 - p_hat)

print(f"Оцінка частки днів вище {july_temp}°C: {p_hat:.3f}")
print(f"n * p_hat = {n_p:.1f}, n * (1 - p_hat) = {n_1p:.1f}")

se_p = np.sqrt(p_hat * (1 - p_hat) / n)
ci_prop = stats.norm.interval(0.95, loc=p_hat, scale=se_p)
print(f"95% довірчий інтервал для частки: {ci_prop}")
```

```python
confidence_levels = [0.90, 0.95, 0.99]
intervals = []
widths = []

for cl in confidence_levels:
    ci = stats.t.interval(cl, df=n-1, loc=mean, scale=se)
    intervals.append(ci)
    widths.append(ci[1] - ci[0])
    print(f"{int(cl*100)}% інтервал: {ci}, ширина: {widths[-1]:.3f}")
```

Завдання 5. Інтерпретація довірчого інтервалу словами
95% довірчий інтервал означає, що якби ми повторили збір вибірок такого ж розміру багато разів (наприклад, 100 разів) і для кожної побудували довірчий інтервал тим самим методом, то близько 95 із 100 таких інтервалів містили б справжнє середнє значення генеральної сукупності. Але для конкретного інтервалу ймовірність того, що він містить справжнє значення, або ні — не визначена.

Контрольні питання
Різниця між точковою оцінкою і довірчим інтервалом
Точкова оцінка — це одне число (наприклад, середнє вибірки), але воно має похибку. Довірчий інтервал дає діапазон, де з визначеним рівнем довіри, побудованим на теорії ймовірностей, має перебувати справжнє значення параметра. Один чисельний показник не показує цю невизначеність.

Чому при невідомому σ і не дуже великому n застосовують t-розподіл?
t-розподіл враховує додаткову невизначеність через оцінювання σ з вибірки, на відміну від нормального розподілу, де σ відоме. При малих n т-розподіл має більші "хвости", що робить інтервал ширшим і реалістичнішим.

Що означає "95% довіри"?
Це частотна інтерпретація: якби повторювати процедуру багато разів, близько 95% побудованих інтервалів містили би справжнє значення параметра. Помилка — вважати, що для конкретного інтервалу ймовірність покриття дорівнює 95%.

Які фактори визначають ширину довірчого інтервалу і як вони впливають?

Рівень довіри (чем вищий, тим ширший інтервал).
Стандартне відхилення вибірки (чим більше, тим ширше).
Розмір вибірки (чим більше n, тим вужче, бо SE зменшується)
