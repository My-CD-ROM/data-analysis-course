Параметр генерації липня: 20°C, стара норма μ₀ = 19.4°C

```python
import numpy as np

np.random.seed(6)
daily_temps = np.random.normal(loc=20, scale=2.5, size=30)

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)

print(f"Вибіркове середнє x̄ = {x_bar:.3f}")
print(f"Вибіркове стандартне відхилення s = {s:.3f}")
```

```python
from scipy import stats

n = 30
mu0 = 19.4
se = s / np.sqrt(n)
z = (x_bar - mu0) / se

p_value = 2 * (1 - stats.norm.cdf(abs(z)))  # двобічний тест

print(f"Стандартна похибка se = {se:.4f}")
print(f"z-статистика = {z:.3f}")
print(f"p-value = {p_value:.4f}")
```

```python
alpha = 0.05

if p_value < alpha:
    decision = "Відхиляємо H0"
else:
    decision = "Не відхиляємо H0"

print(f"Рішення при α = 0.05: {decision}")
print(f"Висновок: Зібрані дані {'дають' if p_value < alpha else 'не дають'} статистично значущих підстав вважати, що середня липнева температура міста Чернігів відрізняється від старої норми {mu0}°C (p = {p_value:.4f}, двобічний тест).")
```

```python
ci_low, ci_high = x_bar - 1.96 * se, x_bar + 1.96 * se

print(f"95% довірчий інтервал для середньої: ({ci_low:.3f}, {ci_high:.3f})")
print(f"Чи містить інтервал стару норму {mu0}? {'Так' if ci_low <= mu0 <= ci_high else 'Ні'}")
```

Контрольні питання
Що означає p-value?
Це ймовірність отримати спостережуване або більш екстремальне значення статистики (z) за умови, що H₀ істинна. Воно НЕ є ймовірністю істинності H₀.

Чому тип тесту слід обирати до отримання результату?
Щоб уникнути статистичної помилки підлаштування під дані (постхоковий вибір), який викривлює результати.

Що означають помилки I і II роду саме в цьому контексті?
I рід — помилкове визнання зміни клімату без підстав; II рід — пропуск фактичної зміни. Ці дві помилки мають різну роль і наслідки в науковому та практичному контексті.

Чи збігається результат тесту з висновком довірчого інтервалу?
Так, при тому ж α, якщо μ0 лежить всередині 95% довірчого інтервалу — H₀ не відхиляємо; якщо поза — відхиляємо. Це закономірність теорії.
