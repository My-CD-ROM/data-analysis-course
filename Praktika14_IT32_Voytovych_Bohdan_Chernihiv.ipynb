```python
import numpy as np
from scipy import stats

# Ранкові значення вологості (%) вашого міста — Чернігів
morning = np.array([85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86])
normal = 75  # норма за рік, %

# Вечірні амплітуди для корекції (по місяцях)
amplitude_correction = np.array([3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3])

# Вечірні значення (ранкові - амплітуда)
evening = morning - amplitude_correction
print("Вечірні значення:", evening)
```

```python
t_stat, p_val = stats.ttest_1samp(morning, popmean=normal)
print(f"t_stat = {t_stat:.3f}, p_value = {p_val:.4f}")

if p_val < 0.05:
    print("Відхиляємо H0: середня вологість рік відрізняється від норми.")
else:
    print("Не відхиляємо H0: підстав вважати відмінність від норми немає.")
```

```python
n = len(morning)
mean = morning.mean()
std = morning.std(ddof=1)
t_manual = (mean - normal) / (std / np.sqrt(n))
print(f"Ручна t-статистика: {t_manual:.3f}")
```

```python
lviv = np.array([86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87])

# Перевірка рівності дисперсій (Levene)
stat_lev, p_lev = stats.levene(morning, lviv)
print(f"Levene stat: {stat_lev:.3f}, p_value: {p_lev:.4f}")

equal_var = p_lev >= 0.05
t_stat_ind, p_val_ind = stats.ttest_ind(morning, lviv, equal_var=equal_var)
print(f"t-statistic: {t_stat_ind:.3f}, p-value: {p_val_ind:.4f}")
if p_val_ind < 0.05:
    print("Статистично значуща різниця середніх вологості між Черніговом і Львовом.")
else:
    print("Підстав для вважати середні різними немає.")
```

```python
poltava = np.array([84, 81, 76, 69, 63, 61, 59, 61, 67, 74, 80, 85])

stat_lev2, p_lev2 = stats.levene(morning, poltava)
print(f"Levene stat: {stat_lev2:.3f}, p_value: {p_lev2:.4f}")

equal_var2 = p_lev2 >= 0.05
t_stat_ind2, p_val_ind2 = stats.ttest_ind(morning, poltava, equal_var=equal_var2)
print(f"t-statistic: {t_stat_ind2:.3f}, p-value: {p_val_ind2:.4f}")

print("Обґрунтування вибору equal_var:")
if equal_var2:
    print("Дисперсії рівні, обираємо standard t-test (equal_var=True).")
else:
    print("Дисперсії різні, обираємо Welch t-test (equal_var=False).")
```

```python
  t_rel, p_rel = stats.ttest_rel(morning, evening)
print(f"Парний t-тест: t = {t_rel:.3f}, p = {p_rel:.4f}")

if p_rel < 0.05:
    print("Існують статистично значущі добові відмінності вологості.")
else:
    print("Добові відмінності вологості статистично незначущі.")

# Пояснення: тут парний критерій правильний, бо ранкові та вечірні значення — залежні пари,
# і обчислення різниці усуває міжособистісну варіативність.
```

Короткі письмові відповіді (приклади)
Чому t-, а не z-розподіл?
σ генеральної сукупності невідомо, застосовуємо t-розподіл, який враховує це через вибіркове std.

Чому важливо враховувати equal_var?
Неправильний вибір може сильно викривити p-value через різні припущення про розподіл дисперсій у вибірках.

Чому парний критерій потужніший?
Він "усуває" вплив міжпарної варіабельності, розглядаючи лише різницю в кожній парі, що підвищує чутливість тесту.

Чи збігається висновок Levene з оглядом std?
Зазвичай так: помітна різниця у std підтверджується низьким p-value в Levene.
