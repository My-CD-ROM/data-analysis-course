```python
from scipy.optimize import linprog

# Оскільки linprog мінімізує, беремо від’ємні коефіцієнти цільової функції
c = [-55, -40]

# Матриця обмежень A_ub x <= b_ub
A_ub = [
    [2, 3],
    [6, 2],
    [2, 1]
]

b_ub = [210, 250, 125]

bounds = [(0, None), (0, None)]  # x1, x2 >= 0

res = linprog(c, A_ub=A_ub, b_ub=b_ub, bounds=bounds, method='highs')

print("Оптимальний план виробництва (виробів А і Б):", res.x)
print("Максимальний прибуток:", -res.fun)
print("Статус розв’язку:", res.status, "-", res.message)
```

```python
print("Запаси ресурсів по обмеженнях (застосовано рішення):")
used_resources = [2*res.x[0] + 3*res.x[1],
                  6*res.x[0] + 2*res.x[1],
                  2*res.x[0] + 1*res.x[1]]
limits = b_ub

for i, (used, limit) in enumerate(zip(used_resources, limits), 1):
    slack = limit - used
    print(f"Ресурс {i}: Використано {used:.2f} з {limit}. Запас: {slack:.2f} - {'активне' if slack < 1e-6 else 'неактивне'} обмеження")
```

```python
# Візьмемо активний ресурс (наприклад, перше)
new_limits_active = b_ub.copy()
new_limits_active[0] *= 1.15  # збільшили на 15%

res_new_active = linprog(c, A_ub=A_ub, b_ub=new_limits_active, bounds=bounds, method='highs')
print("Збільшення активного ресурсу на 15%:")
print("Новий план:", res_new_active.x)
print("Новий прибуток:", -res_new_active.fun)

# Візьмемо неактивний ресурс (наприклад, третій)
new_limits_inactive = b_ub.copy()
new_limits_inactive[2] *= 1.5  # збільшили на 50%

res_new_inactive = linprog(c, A_ub=A_ub, b_ub=new_limits_inactive, bounds=bounds, method='highs')
print("\nЗбільшення неактивного ресурсу на 50%:")
print("Новий план:", res_new_inactive.x)
print("Новий прибуток:", -res_new_inactive.fun)
```
