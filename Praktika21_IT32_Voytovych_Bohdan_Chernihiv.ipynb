```python
import numpy as np

# Варіант 6
np.random.seed(6)

def monte_carlo_integral(N):
    x = np.random.uniform(0,1,N)
    fx = np.exp(-x)
    return fx.mean()

N = 1000
estimate = monte_carlo_integral(N)
analytic = 1 - 1/np.e
error = abs(estimate - analytic)

print(f"Оцінка інтеграла для N={N}: {estimate:.5f}")
print(f"Точне значення: {analytic:.5f}")
print(f"Абсолютна похибка: {error:.5f} ({error/analytic*100:.2f}%)")
```

```python
import matplotlib.pyplot as plt

Ns = [100, 1000, 10000, 100000]
estimates = []
errors = []

for N in Ns:
    np.random.seed(6)
    est = monte_carlo_integral(N)
    err = abs(est - analytic)
    estimates.append(est)
    errors.append(err)
    print(f"N={N:6d} | Оцінка: {est:.6f} | Похибка: {err:.6f}")

plt.figure(figsize=(8,5))
plt.plot(Ns, errors, marker='o')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Кількість вибірок N (логарифмічна шкала)')
plt.ylabel('Абсолютна похибка (логарифмічна шкала)')
plt.title('Залежність похибки Monte Carlo від кількості вибірок')
plt.grid(True, which='both', ls='--')
plt.show()
```

## Завдання 5. Теоретичні відповіді

1. Оцінка методом Монте-Карло є вибірковим середнім незалежних випадкових величин, що гарантує закон великих чисел: при зростанні числа \( N \) вона збігається до точного математичного очікування (інтеграла).

2. Центральна гранична теорема пояснює, що помилка оцінки наближається до нормального розподілу з дисперсією, яка зменшується пропорційно \( 1/\sqrt{N} \). Тому збільшення \( N \) в 4 рази зменшує похибку приблизно в 2 рази.

3. У випадку інтеграла ми генеруємо рівномірні точки по інтервалу і беремо середнє значення функції, помножене на довжину відрізка, що дає наближення до інтеграла.

4. Формули похибки і збіжності не є унікальними для методу Монте-Карло, а випливають із теорії вибіркових середніх.
