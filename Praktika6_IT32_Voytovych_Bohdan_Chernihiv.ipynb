```python
import pandas as pd
import numpy as np

# Дані відносної вологості для 10 міст за 12 місяців
data = {
    "Місто": ["Київ", "Львів", "Одеса", "Харків", "Дніпро", "Чернігів", "Івано-Франківськ", "Полтава", "Суми", "Ужгород"],
    "Регіон": ["Центр", "Захід", "Південь", "Схід", "Схід", "Центр", "Захід", "Центр", "Схід", "Захід"],
    "Січ":    [84, 85, 82, 83, 82, 85, 86, 84, 85, 84],
    "Лют":    [82, 83, 80, 80, 79, 83, 84, 81, 82, 82],
    "Бер":    [78, 80, 78, 75, 74, 79, 81, 76, 77, 79],
    "Кві":    [70, 73, 74, 68, 67, 71, 75, 69, 70, 73],
    "Тра":    [65, 68, 71, 62, 61, 66, 70, 63, 64, 68],
    "Чер":    [63, 65, 69, 60, 59, 63, 67, 61, 62, 66],
    "Лип":    [61, 63, 67, 58, 57, 61, 65, 59, 60, 64],
    "Сер":    [63, 65, 68, 60, 59, 63, 67, 61, 62, 66],
    "Вер":    [68, 70, 72, 66, 65, 69, 72, 67, 68, 71],
    "Жов":    [75, 77, 76, 73, 72, 76, 78, 74, 75, 77],
    "Лис":    [81, 82, 80, 79, 78, 82, 83, 80, 81, 82],
    "Гру":    [85, 86, 83, 84, 83, 86, 87, 85, 86, 85],
}

df = pd.DataFrame(data)

# Завдання 1: описова статистика для власного міста
# Припустимо, варіант 6 — Чернігів
humidity = df[df["Місто"] == "Чернігів"].iloc[:, 2:].T.squeeze()  # беремо всі місяці, повертаємо у Series

print("Завдання 1 — describe():\n", humidity.describe(), "\n")

# Пояснення:
print("""
Пояснення для describe():
- count: кількість спостережень
- mean: середнє значення (міра положення)
- std: стандартне відхилення (міра розсіювання)
- min: мінімальне значення (для розмаху)
- 25%, 50%, 75%: квантілі (25% і 75% — для обчислення інтерквартильного розмаху)
- max: максимальне значення (для розмаху)
""")

# Завдання 2: ручні обчислення
rozmakh = humidity.max() - humidity.min()
disp_ddof1 = humidity.var(ddof=1)  # дисперсія з корекцією (за вибіркою)
disp_ddof0 = humidity.var(ddof=0)  # дисперсія без корекції (за генеральною сукупністю)
std_ddof1 = humidity.std(ddof=1)
std_ddof0 = humidity.std(ddof=0)
q1, q3 = humidity.quantile([0.25, 0.75])
iqr = q3 - q1
cv = std_ddof1 / humidity.mean() * 100

print(f"Завдання 2:")
print(f"Розмах = {rozmakh}")
print(f"Дисперсія (ddof=1) = {disp_ddof1}")
print(f"Дисперсія (ddof=0) = {disp_ddof0}")
print(f"Стандартне відхилення (ddof=1) = {std_ddof1}")
print(f"Стандартне відхилення (ddof=0) = {std_ddof0}")
print(f"IQR = {iqr}")
print(f"Коефіцієнт варіації = {cv:.2f}%\n")

# Перевірка: 25% і 75% співпадають з describe()

# Завдання 3: порівняння регіонів
df["Середня Вологість"] = df.iloc[:, 2:].mean(axis=1)

# Вибір підмножини для регіону "Центр" (регіон власного міста)
center_humidity = df[df["Регіон"] == "Центр"]["Середня Вологість"]
print("Описова статистика середньої вологості для регіону Центр:\n", center_humidity.describe(), "\n")

# Вибір іншого регіону, наприклад "Захід"
west_humidity = df[df["Регіон"] == "Захід"]["Середня Вологість"]
print("Описова статистика середньої вологості для регіону Захід:\n", west_humidity.describe(), "\n")

# Порівняння розкиду та коефіцієнтів варіації
std_center = center_humidity.std()
cv_center = 100 * std_center / center_humidity.mean()

std_west = west_humidity.std()
cv_west = 100 * std_west / west_humidity.mean()

print(f"Порівняння регіонів:")
print(f"Середнє STD Центр: {std_center:.4f}, CV: {cv_center:.2f}%")
print(f"Середнє STD Захід: {std_west:.4f}, CV: {cv_west:.2f}%")

print("\nЗазвичай CV точніше порівнює варіативність між групами з різними середніми.")

# Завдання 4: аналіз форми розподілу для власного міста
mean_humidity = humidity.mean()
median_humidity = humidity.median()

print(f"\nЗавдання 4:")
print(f"Середнє: {mean_humidity:.2f}")
print(f"Медіана: {median_humidity:.2f}")

if abs(mean_humidity - median_humidity) < 1e-2:
    print("Розподіл близький до симетричного. Середнє і медіана майже співпадають.")
    rep_mesure = "Середнє є репрезентативнішою мірою положення."
else:
    print("Розподіл скошений. Значна різниця між середнім і медіаною.")
    rep_mesure = "Медіана є репрезентативнішою мірою положення."

print(f"Рекомендація: {rep_mesure}")

# Завдання 5: порівняння з іншим містом з іншого регіону
# Для прикладу оберемо місто з регіону "Південь" - Одеса

odesa_humidity = df[df["Місто"] == "Одеса"].iloc[:, 2:14].T.squeeze()

std_odesa = odesa_humidity.std(ddof=1)
cv_odesa = 100 * std_odesa / odesa_humidity.mean()

print(f"\nЗавдання 5:")
print(f"Коефіцієнт варіації вологості для Одеси: {cv_odesa:.2f}%")
print(f"Коефіцієнт варіації вологості для Чернігова: {cv:.2f}%")

print("""
Пояснення:
Коефіцієнт варіації — це відносна міра розсіювання, що враховує масштаб (середнє) вибірки.
Навіть при близьких середніх значеннях порівняння абсолютних стандартних відхилень може бути хибним.
CV дозволяє більш коректно порівнювати дисперсію в різних групах чи містах.
""")
```
