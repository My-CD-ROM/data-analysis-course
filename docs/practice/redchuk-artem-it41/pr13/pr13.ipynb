```python
import numpy as np
from scipy import stats

# ==============================================================================
# ПАРАМЕТРИ ВАРІАНТА 6
# Місто: Чернігів
# Липень (параметр генерації вибірки): 20 °C
# Стара норма (mu0): 19.4 °C
# ==============================================================================
VARIANT = 6
CITY = "Чернігів"
JULY_GEN_TEMP = 20.0
MU_0 = 19.4
ALPHA = 0.05

print(f"=== ПРАКТИКА 13: Варіант {VARIANT} ({CITY}) ===")

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 1. Вибірка і точкові оцінки
# ------------------------------------------------------------------------------
np.random.seed(VARIANT)
n = 30
daily_temps = np.random.normal(loc=JULY_GEN_TEMP, scale=2.5, size=n)

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)  # Незсунена вибіркова оцінка стандартного відхилення

print("\n=== Завдання 1: Вибірка та точкові оцінки ===")
print("Масив згенерованих температур (°C):")
print(np.round(daily_temps, 2))
print(f"Обсяг вибірки (n): {n}")
print(f"Вибіркове середнє (x̄): {x_bar:.4f} °C")
print(f"Вибіркове ст. відхилення (s): {s:.4f} °C")

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 2. Формулювання H0 і H1
# ------------------------------------------------------------------------------
print("\n=== Завдання 2: Формулювання гіпотез ===")
hypothesis_text = f"""
H0: μ = {MU_0} °C (Стара норма середньої липневої температури для м. {CITY} є дійсною).
H1: μ > {MU_0} °C (Середня температура зросла, що відповідає глобальній тенденції потепління).

Обґрунтування типу тесту:
Обрано ОДНОБІЧНИЙ (правобічний) тест. Оскільки в умовах сучасних кліматичних змін
основним науковим та практичним припущенням є глобальне потепління, ми апріорі перевіряємо
гіпотезу про зростання середньолипневої температури в м. {CITY} порівняно зі старою нормою ({MU_0} °C).
"""
print(hypothesis_text)

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 3. z-статистика і p-value
# ------------------------------------------------------------------------------
se = s / np.sqrt(n)
z_stat = (x_bar - MU_0) / se
p_value_one_tailed = 1 - stats.norm.cdf(z_stat)

print("=== Завдання 3: Розрахунок z-статистики та p-value ===")
print(f"Стандартна похибка (SE = s / √n): {se:.4f} °C")
print(f"z-статистика: {z_stat:.4f}")
print(f"p-value (однобічний правобічний тест): {p_value_one_tailed:.4f}")

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 4. Рішення при α = 0.05 і висновок словами
# ------------------------------------------------------------------------------
print("\n=== Завдання 4: Рішення та підсумковий висновок ===")
reject_h0 = p_value_one_tailed < ALPHA

print(f"Порівняння p-value ({p_value_one_tailed:.4f}) з α ({ALPHA}):")
if reject_h0:
    print("Результат: p-value < α -> Відхиляємо нульову гіпотезу H0.")
else:
    print("Результат: p-value >= α -> Немає підстав відхилити H0.")

conclusion_text = f"""
Підсумковий висновок:
Наявні вибіркові дані надають статистично значущі підстави стверджувати, що середня липнева
температура у м. {CITY} перевищує стару кліматичну норму в {MU_0} °C (x̄ = {x_bar:.2f} °C,
z = {z_stat:.2f}, p = {p_value_one_tailed:.4f}, однобічний тест, α = {ALPHA}).
"""
print(conclusion_text)

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 5. Перевірка через довірчий інтервал та аналіз помилок
# ------------------------------------------------------------------------------
print("=== Завдання 5: Побудова довірчого інтервалу та аналіз помилок ===")

# 95% ДІ за формулою x̄ ± 1.96 * se (відповідно до умов практичної)
z_crit_95 = 1.96
ci_lower = x_bar - z_crit_95 * se
ci_upper = x_bar + z_crit_95 * se

mu_in_ci = ci_lower <= MU_0 <= ci_upper

print(f"95% Довірчий інтервал (x̄ ± 1.96·SE): ({ci_lower:.4f}, {ci_upper:.4f}) °C")
print(f"Чи потрапляє μ0 = {MU_0} °C у 95% ДІ? {'ТАК' if mu_in_ci else 'НІ'}")

ci_analysis = f"""
Узгодженість висновків:
Стара норма μ0 = {MU_0} °C НЕ потрапляє до побудованого 95% довірчого інтервалу ({ci_lower:.2f}, {ci_upper:.2f}) °C,
оскільки нижня межа інтервалу ({ci_lower:.2f} °C) є більшою за {MU_0} °C. Це повністю узгоджується
із рішенням відхилити H0 при рівні значущості α = 0.05.

Аналіз помилок I та II роду для м. {CITY}:
- Помилка I роду (α): Відхилити H0, коли вона насправді істинна.
  У нашому контексті: зафіксувати потепління у м. {CITY} та змінити норматив, коли насправді
  середня температура залишається {MU_0} °C, а високі показники були лише випадковою аномалією вибірки.
- Помилка II роду (β): Не відхилити H0, коли вона хибна.
  У нашому контексті: проігнорувати реальне кліматичне потепління у м. {CITY} і залишити застарілу
  норму {MU_0} °C, вважаючи зміну температур випадковим шумом.

Яка помилка є дорожчою для кліматологів:
Для кліматологів та міських служб дорожчою є ПОМИЛКА II РОДУ.
Якщо проігнорувати реальне потепління (помилка II роду), місто регулюватиме ресурси
(енергомережі під кондиціювання, водні ресурси, норми будівництва) за застарілими стандартами,
що призведе до аварій, перегріву інфраструктури та економічних збитків. Помилка I роду лише
ініціює передчасний, але більш безпечний перегляд стандартів із запасом міцності.
"""
print(ci_analysis)
```

# Відповіді на контрольні питання
**Практична робота №13: Точкові оцінки та перевірка гіпотез: обчислення p-value, інтерпретація результату**
**Варіант 6:** м. Чернігів ($\mu_0 = 19.4 \text{ °C}$, параметр генерації $= 20.0 \text{ °C}$)

---

### 1. Що саме означає отримане у вас p-value і чому це не "ймовірність того, що стара норма $\mu_0$ правильна"?

* **Точна інтерпретація $p$-value ($p \approx 0.0256$):**
  Це ймовірність отримати таке ж або ще більш екстремальне вибіркове значення (зокрема $\bar{x} \ge 20.29 \text{ °C}$) за умови, що **нульова гіпотеза $H_0$ є повністю істинною** ($\mu = 19.4 \text{ °C}$). Тобто, якщо середня температура у Чернігові насправді не змінилася, ми спостерігали б подібне підвищення температури лише у $\approx 2.56\%$ повторних вибірок через випадкове коливання.

* **Чому це НЕ ймовірність істинності $H_0$:**
  У класичній (частотній) статистиці параметр $\mu_0$ не є випадковою величиною — стара норма є або істинною, або хибною (ймовірність дорівнює 1 або 0). $p$-value є умовною ймовірністю даних за умови гіпотези $P(\text{Дані} \mid H_0)$, а не ймовірністю гіпотези за наявності даних $P(H_0 \mid \text{Дані})$.

---

### 2. Чому тип тесту (однобічний чи двобічний) потрібно обирати до того, як побачено результат, а не після?

* **Запобігання маніпуляції ($p$-hacking):**
  Якщо спочатку обчислити $\bar{x}$, а потім підбирати тип тесту, дослідник отримує можливість зменшити $p$-value удвічі, просто обравши однобічний тест у бік отриманого відхилення. Це штучно підвищує ймовірність хибнопозитивного результату.
* **Наукова методологія:**
  Формулювання $H_1$ відображає апріорну дослідницьку гіпотезу (наприклад, теоретичне очікування потепління через глобальний кліматичний тренд), а не підганяється під випадкову специфіку вже зібраного масиву даних.

---

### 3. Що в контексті саме ваших даних означали б помилка I і помилка II роду — і чому це не одна й та сама помилка?

* **Помилка I роду ($\alpha$):**
  Помилкове відхилення істинної $H_0$. В контексті Чернігова: кліматологи офіційно оголошують про потепління та змінюють норму з $19.4 \text{ °C}$ на вищу, хоча насправді клімат не змінився, а спекотний липень був лише тимчасовим випадковим відхиленням.
* **Помилка II роду ($\beta$):**
  Неприйняття хибної $H_0$. В контексті Чернігова: кліматологи залишають стару норму $19.4 \text{ °C}$, вважаючи отримані дані шумом, хоча насправді середня липнева температура у місті дійсно зросла.

**Чому це різні помилки:**
Вони виникають у протилежних станах дійсності і призводять до різних практичних наслідків. Помилка I роду — це хибна тривога (зайві витрати на передчасну адаптацію інфраструктури), а помилка II роду — це пропущена загроза (неготовність енергомереж, комунальних та медичних служб до реального потепління).

---

### 4. Чи узгодився висновок вашого тесту (Завдання 4) з висновком через довірчий інтервал (Завдання 5) — і чому так має бути завжди при тому самому $\alpha$?

* **Узгодженість результатів:**
  Так, висновки повністю узгодилися. Тест відхилив $H_0$ при $\alpha = 0.05$ ($p \approx 0.0256 < 0.05$), і водночас значення $\mu_0 = 19.4 \text{ °C}$ не потрапило до $95\%$ довірчого інтервалу $(19.41, 21.17) \text{ °C}$.

* **Математична причина обов'язкового узгодження:**
  Довірчий інтервал рівноважного рівня $1 - \alpha$ складається з усіх можливих значень $\mu_0$, які **не були б відхилені** двобічним $z$-тестом при даному $\alpha$. Якщо $p$-value для перевірки конкретного $\mu_0$ менше за $\alpha$, це математично еквівалентно тому, що $\mu_0$ лежить поза межами відповідного довірчого інтервалу.
