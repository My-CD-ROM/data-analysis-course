```python
import numpy as np
from scipy import stats

# ==============================================================================
# ПАРАМЕТРИ ВАРІАНТА 6
# Місто: Чернігів (Центр)
# Норма: 75%
# Ранкова вологість (12 місяців: Січ-Гру):
# [85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86]
# ==============================================================================
VARIANT = 6
CITY_NAME = "Чернігів"
REGION = "Центр"
NORMA = 75.0
ALPHA = 0.05

# 12 місячних значень відносної вологості (ранковий вимір о 8:00)
morning_humidity = np.array([85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86])
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

print(f"=== ПРАКТИКА 14: Варіант {VARIANT} ({CITY_NAME}, Регіон: {REGION}) ===")
print(f"Дані ранкової вологості: {morning_humidity}")
print(f"Довідкова норма: {NORMA}%\n")

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 1. Одновибірковий t-критерій проти норми
# ------------------------------------------------------------------------------
print("=== Завдання 1: Одновибірковий t-критерій проти норми ===")
# H0: μ = NORMA (середньорічна вологість у Чернігові дорівнює довідковій нормі 75%)
# H1: μ ≠ NORMA (середньорічна вологість у Чернігові статистично значуще відрізняється від 75%)

t_stat_1samp, p_val_1samp = stats.ttest_1samp(morning_humidity, popmean=NORMA)

print(f"H0: μ = {NORMA}% vs H1: μ ≠ {NORMA}%")
print(f"t-статистика (scipy): {t_stat_1samp:.4f}")
print(f"p-value: {p_val_1samp:.4f}")

if p_val_1samp < ALPHA:
    print(f"Висновок при α = {ALPHA}: p-value < α -> Відхиляємо H0. Середня вологість у м. {CITY_NAME} статистично значуще відрізняється від норми.")
else:
    print(f"Висновок при α = {ALPHA}: p-value >= α -> Немає підстав відхилити H0. Наявні дані узгоджуються з довідковою нормою ({NORMA}%).")

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 2. Ручна перевірка t-статистики
# ------------------------------------------------------------------------------
print("\n=== Завдання 2: Ручний розрахунок t-статистики ===")
n = len(morning_humidity)
x_bar = morning_humidity.mean()
s = morning_humidity.std(ddof=1)  # Незсунене вибіркове стандартне відхилення
se = s / np.sqrt(n)

t_stat_manual = (x_bar - NORMA) / se

print(f"Вибіркове середнє (x̄): {x_bar:.4f}%")
print(f"Вибіркове ст. відхилення (s): {s:.4f}%")
print(f"Стандартна похибка (SE): {se:.4f}%")
print(f"Ручна t-статистика: {t_stat_manual:.4f}")
print(f"Збіг із ttest_1samp: {np.isclose(t_stat_1samp, t_stat_manual)}")

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 3. Незалежний двовибірковий критерій: Чернігів проти міст інших регіонів
# Місто 1 для Завдання 3: Львів (Захід) — [85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86]
# ------------------------------------------------------------------------------
print("\n=== Завдання 3: Двовибірковий критерій (Чернігів vs Львів) ===")
lviv_humidity = np.array([85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86])

# Перевірка рівності дисперсій (Критерій Лівена)
stat_lev_3, p_lev_3 = stats.levene(morning_humidity, lviv_humidity)
print(f"Порівняння: Чернігів (Центр, s={s:.2f}) vs Львів (Захід, s={lviv_humidity.std(ddof=1):.2f})")
print(f"Критерій Лівена: Stat = {stat_lev_3:.4f}, p-value = {p_lev_3:.4f}")

# Обґрунтування equal_var
is_equal_var_3 = p_lev_3 >= ALPHA
print(f"Рішення щодо дисперсій: p_lev = {p_lev_3:.4f} >= α ({ALPHA}) -> Дисперсії можна вважати РІВНИМИ (equal_var=True).")

# Виконання ttest_ind з обґрунтованим equal_var
t_ind_3, p_ind_3 = stats.ttest_ind(morning_humidity, lviv_humidity, equal_var=is_equal_var_3)
print(f"t-критерій для незалежних вибірок (equal_var={is_equal_var_3}): t = {t_ind_3:.4f}, p-value = {p_ind_3:.4f}")

if p_ind_3 < ALPHA:
    print(f"Висновок: Відхиляємо H0. Середня вологість Чернігова та Львова статистично значуще відрізняється.")
else:
    print(f"Висновок: Немає підстав відхилити H0. Середня вологість Чернігова та Львова статистично не відрізняється (p = {p_ind_3:.4f}).")

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 4. Порівняння з протилежним висновком Лівена
# Місто 2 для Завдання 4: Одеса (Південь) — [82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83]
# ------------------------------------------------------------------------------
print("\n=== Завдання 4: Двовибірковий критерій з протилежним результатом Лівена (Чернігів vs Одеса) ===")
odesa_humidity = np.array([82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83])

stat_lev_4, p_lev_4 = stats.levene(morning_humidity, odesa_humidity)
print(f"Порівняння: Чернігів (Центр, s={s:.2f}) vs Одеса (Південь, s={odesa_humidity.std(ddof=1):.2f})")
print(f"Критерій Лівена: Stat = {stat_lev_4:.4f}, p-value = {p_lev_4:.4f}")

is_equal_var_4 = p_lev_4 >= ALPHA
print(f"Рішення щодо дисперсій: p_lev = {p_lev_4:.4f} < α ({ALPHA}) -> Дисперсії НЕ РІВНІ (equal_var=False, застосовуємо Welch t-test).")

t_ind_4, p_ind_4 = stats.ttest_ind(morning_humidity, odesa_humidity, equal_var=is_equal_var_4)
print(f"Welch t-критерій (equal_var={is_equal_var_4}): t = {t_ind_4:.4f}, p-value = {p_ind_4:.4f}")

explanation_task4 = """
Пояснення вибору equal_var у Завданні 4:
На відміну від Завдання 3 (де p_lev ≈ 0.81 > 0.05 показав гомогенність дисперсій), у Завданні 4 p-value
критерію Лівена склав ≈ 0.016 < 0.05. Це дає статистичні підстави відхилити гіпотезу про рівність дисперсій.
Одеса має значно стабільніший прибережний клімат із меншою річною амплітудою (s ≈ 5.48%), ніж континентальний
Чернігів (s ≈ 8.87%). Тому застосування стандартного t-тесту призвело б до помилкового обчислення степенів свободи;
застосовано модифікацію Уелча (Welch's t-test, equal_var=False).
"""
print(explanation_task4)

# ------------------------------------------------------------------------------
# ЗАВДАННЯ 5. Парний t-критерій: ранок проти вечора
# ------------------------------------------------------------------------------
print("=== Завдання 5: Парний t-критерій (ранок vs вечір для м. Чернігів) ===")
amplitude = np.array([3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3])
evening_humidity = morning_humidity - amplitude

print(f"Ранкова вологість (8:00):  {morning_humidity}")
print(f"Вечірня вологість (20:00): {evening_humidity}")

t_rel, p_rel = stats.ttest_rel(morning_humidity, evening_humidity)
print(f"Парний t-критерій (stats.ttest_rel): t = {t_rel:.4f}, p-value = {p_rel:.4f}")

if p_rel < ALPHA:
    print(f"Висновок: p-value ({p_rel:.4e}) < α ({ALPHA}) -> Добова різниця вологості є СТАТИСТИЧНО ЗНАЧУЩОЮ.")

explanation_task5 = f"""
Обґрунтування застосування саме парного критерію:
Вимірювання о 8:00 та 20:00 зроблені для одних і тих самих 12 місяців у одному й тому самому місті.
Вони не є незалежними — існує виражена сезонна залежність (взимку вологість висока і вранці, і ввечері,
влітку — нижча).
Парний t-критерій обчислює різницю d_i = Ранок_i - Вечір_i для кожного місяця окремо.
Це дозволяє повністю УСУНУТИ міжмісячну варіативність (сезонний шум), залишаючи для аналізу лише чистий
добовий ефект температурного перепаду.
"""
print(explanation_task5)
```

### 1. Чи можете пояснити, чому для перевірки середнього однієї вибірки використовується саме $t$-, а не $z$-розподіл?

Для обчислення стандартизованого відхилення необхідна стандартна похибка $SE = \frac{\sigma}{\sqrt{n}}$.

* **Невідоме $\sigma$:** Генеральне стандартне відхилення кліматичних вимірювань $\sigma$ є невідомим, тому ми оцінюємо його за вибірковим стандартним відхиленням $s$.
* **Додаткова варіативність:** Оскільки $s$ сама по собі є випадковою величиною, відношення $\frac{\bar{x} - \mu_0}{s / \sqrt{n}}$ має більший розкид, ніж стандартний нормальний $z$-розподіл.
* **Малий обсяг вибірки ($n = 12$):** При малій кількості спостережень невизначеність оцінки $s$ суттєва. $t$-розподіл Стьюдента має «важчі хвости», що компенсує цю невизначеність. Застосування $z$-розподілу штучно занизило б $p$-value та підвищило ризик помилки I роду.

### 2. Чи розумієте, чому неправильний вибір `equal_var` у `ttest_ind` (без попередньої перевірки критерієм Лівена) здатний дати систематично неправильний $p$-value, а не просто трохи менш точний?

* **Визначення об'єднаної дисперсії ($s_p^2$):** Класичний двохвибірковий $t$-тест (`equal_var=True`) передбачає спільну дисперсію $\sigma_1^2 = \sigma_2^2$ та розраховує об'єднане оцінювання $s_p^2$. Якщо реальні дисперсії істотно відрізняються (як у випадку Чернігова $s \approx 8.87\%$ та Одеси $s \approx 5.48\%$), об'єднана оцінка є викривленою.
* **Степені свободи ($df$):** При $equal\_var=True$ степені свободи фіксовані як $df = n_1 + n_2 - 2$. Коли дисперсії та обсяги вибірок нерівні, дійсне число степеней свободи значно менше і розраховується за формулою Саттертуейта у Welch's $t$-test (`equal_var=False`).
* **Систематична помилка:** Використання `equal_var=True` при гетероскедастичності призводить до неправильного розрахунку знаменника $t$-статистики та невірного підрахунку $df$, що призводить до систематично хибного (завищеного або заниженого) $p$-value і викривлення контролю рівня значущості $\alpha$.

### 3. Чи можете пояснити, чому парний критерій потужніший за незалежний двовибірковий на тих самих "до/після" даних — що саме усуває обчислення різниці для кожної пари?

* **Усунення міжсуб'єктної/міжсезонної варіативності:** У незалежному $t$-тесті загальна дисперсія включає як внутрішню різницю між групами, так і величезну варіативність між місяцями (наприклад, різниця між вологим січнем та сухим липнем).
* **Фокус на різниці $d_i$:** Парний $t$-критерій перетворює дві пов'язані вибірки $X$ і $Y$ на одну вибірку різниць $d_i = X_i - Y_i$.
* **Зменшення знаменника (SE):** Оскільки сезонні коливання однаково впливають і на ранкові, і на вечірні вимірювання, віднімання $d_i$ розраховує чисто добовий перепад. Це значно зменшує варіативність $s_d$ у знаменнику формули $t = \frac{\bar{d}}{s_d / \sqrt{n}}$, що суттєво підвищує статистичну потужність (здатність виявити реальний ефект).
