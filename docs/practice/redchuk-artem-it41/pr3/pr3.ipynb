Варіант 6 Місто Чернігів

```python
import pandas as pd

print("-"*6, "Завданння 1", "-"*6)

wide = pd.DataFrame({
    "Місто" : ["Чернігів"],
    "Січ" : 80,
    "Лют" : 76,
    "Бер" : 70,
    "Кві" : 60,
    "Тра" : 52,
    "Чер" : 46,
    "Лип" : 42,
    "Сер" : 44,
    "Вер" : 54,
    "Жов" : 68,
    "Лис" : 78,
    "Гру" : 84
})

print(wide)

print("-"*6, "Завданння 2", "-"*6)

tidy = wide.melt(id_vars="Місто", var_name="Місяць", value_name="Хмарність")
print(tidy)
print(tidy.shape)
# Згідно з принципами Tidy Data: Кожна змінна має утворювати окремий стовпець. Кожне спостереження має утворювати окремий рядок. У вихідному wide-форматі один рядок містив одразу 12 різних спостережень (рівень хмарності для 12 окремих місяців). Під час розгортання операцією melt() кожне з цих 12 спостережень отримало власний окремий рядок (1 рядок * 12 місяців = 12 рядків).

# місто — Номінальна (якісна) шкала, оскільки значення є назвами категорій без природного порядку чи можливості виконання математичних операцій.
# місяць — Порядкова (рангова) шкала, тому що місяці мають чітку хронологічну послідовність (січень, лютий тощо), але різниця між ними не вимірюється однаковими математичними інтервалами.
# хмарність — Відносна шкала (якщо вимірюється у % від 0 до 100) або інтервальна шкала (якщо у балах), оскільки значення є числовими, їх можна порівнювати між собою математично, а 0% хмарності означає абсолютну відсутність хмар (природний нуль).

print("-"*6, "Завданння 4", "-"*6)

back = tidy.pivot(index="Місто", columns="Місяць", values="Хмарність")
print(back)

print("-"*6, "Завданння 5", "-"*6)

tidy.to_csv("variant.csv", index=False)
tidy.to_json("variant.json", orient="records", force_ascii=False)

readed_csv = pd.read_csv("variant.csv")
readed_json = pd.read_json("variant.json")

print(readed_csv)
print(readed_json)
# Опис відмінностей у структурі файлів
# Формат запису даних та плоскість проти вкладеності:
# CSV (Comma-Separated Values) має плоску текстову та табличну структуру, де кожний рядок описує запис, а значення відокремлюються комами. Заголовки стовпців записуються лише один раз у першому рядку файлу.
# JSON (JavaScript Object Notation) у режимі orient="records" являє собою масив об'єктів (список словників). Назви стовпців (ключі "місто", "місяць", "хмарність") повторюються для кожного окремого рядка-об'єкта.
# Типізація та синтаксичні елементи:
# CSV зберігає всі дані як суцільний текст без явних синтаксичних позначень типів даних (числа та текстові рядки записані однаково).
# JSON має чітко виражену синтаксичну типізацію: текстові значення обов'язково обрамлені подвійними лапками ("Київ"), числові значення записуються без лапок (8.1), а вся структура загорнута в квадратні [...] та фігурні {...} дужки.
```

### 1. Шкали вимірювання стовпців

* **`місто` — Номінальна шкала:** Назви міст є текстовими категоріями, які слугують лише як мітки. Для них не існує природного порядку (наприклад, Київ не є «більшим» чи «меншим» за Львів), і над ними неможливо виконувати математичні операції.
* **`хмарність` — Шкала відношень (відносна):** Вимірюється чисельно (від 0% до 100% або від 0 до 10 балів) та має **абсолютний (природний) нуль**, де 0 означає повну відсутність хмарності (безхмарне небо). Це дозволяє не лише порівнювати різницю, а й казати, що в один місяць хмарність у $N$ разів більша, ніж в інший.

---

### 2. Причини трансформації у Tidy Data (`melt()`)

* **Чому wide-формат не є tidy:** Згідно з концепцією Tidy Data, **кожна змінна має бути в окремому стовпці**, а **кожне спостереження — в окремому рядку**. У wide-форматі назви стовпців («Січень», «Лютий»...) є не назвами змінних, а **значеннями** однієї змінної («місяць»). Крім того, один рядок містив одразу 12 окремих вимірювань.
* **Що змінює `melt()`:** Операція розгортає таблицю по вертикалі — створює 2 нові стовпці (`місяць` і `хмарність`) та перетворює кожен місячний показник на окремий рядок. У результаті розмірність змінюється з `(1, 13)` на `(12, 3)`, де кожен рядок відповідає єдиній атомарній події: *хмарність у конкретному місці в конкретний місяць*.

### 3. Принципова різниця між CSV та JSON

| Характеристика | CSV (`variant.csv`) | JSON (`variant.json`, `orient="records"`) |
| :--- | :--- | :--- |
| **Структура** | Плоска двовимірна таблиця, де кожний рядок — це запис, а значення розділені комами. | Масив вкладених об'єктів (список словників `[{...}, {...}]`). |
| **Дублювання ключі** | Назви стовпців вказуються **лише один раз** у першому рядку-заголовку. | Ключі (назви стовпців) **повторюються для кожного рядка** всередині JSON-об'єкта. |
| **Типи даних** | Зберігає все як **чистий текст** без синтаксичного розрізнення типів (число `8.1` та текст `Київ` записуються однаково). | Має **вбудовану синтаксичну типізацію** (числа записуються без лапок `8.1`, а рядки — в лапках `"Київ"`). |
