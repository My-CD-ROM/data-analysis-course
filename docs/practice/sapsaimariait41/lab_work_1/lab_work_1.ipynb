**1. Чи можете ви пояснити різницю між `shape == (12,)` і `shape == (1, 12)`?**

`shape == (12,)` — це одновимірний масив із 12 елементів.
`shape == (1, 12)` — це двовимірний масив із 1 рядком і 12 стовпцями. Це різні форми масивів, хоча кількість елементів однакова.

**2. Чи розумієте, чому `temps[temps > 15]` повертає масив іншої довжини, ніж `temps`?**

`temps > 15` створює булевий масив, у якому для кожної температури вказано `True` або `False`. При індексації `temps[temps > 15]` вибираються тільки ті температури, для яких умова має значення `True`. Тому результат містить лише температури, що більші за 15°C, і може мати меншу довжину, ніж `temps`.

**3. Чи можете пояснити, що станеться, якщо додати масив форми `(3,)` до масиву форми `(3, 12)`?**

Виникне помилка broadcasting. NumPy порівнює форми справа наліво: `12` і `3` не є сумісними розмірами. Тому масив форми `(3,)` не можна безпосередньо додати до масиву форми `(3, 12)`.

**4. Чи можете пояснити, чому `rows + correction` спочатку дає помилку, а `rows + correction.reshape(2, 1)` — ні?**

`rows` має форму `(2, 12)`, а `correction` — `(2,)`. При додаванні NumPy порівнює розміри справа наліво, тому `12` і `2` несумісні. Після `correction.reshape(2, 1)` форма стає `(2, 1)`. Тепер розміри `12` і `1` сумісні, тому NumPy повторює значення поправок по стовпцях і виконує додавання.

```python
import numpy as np

temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])

days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])

month_numbers = np.arange(1, 13)

print("temps:", temps)
print("shape:", temps.shape)
print("dtype:", temps.dtype)

print("days_in_month:", days_in_month)
print("shape:", days_in_month.shape)
print("dtype:", days_in_month.dtype)

print("month_numbers:", month_numbers)
print("shape:", month_numbers.shape)
print("dtype:", month_numbers.dtype)
```

```python
temps_f = temps * 9 / 5 + 32

print("Температура у °F:")
print(temps_f)

warm_days = days_in_month[temps > 15]

print("Дні у місяцях з температурою вище 15°C:")
print(warm_days)

print("Всього днів тепліше 15°C:", warm_days.sum())

print("Сума днів у році:", days_in_month.sum())
print("Чи дорівнює 365:", days_in_month.sum() == 365)
```

```python
warmest_index = temps.argmax()
coldest_index = temps.argmin()

print("Найтепліший місяць:")
print("Температура:", temps[warmest_index], "°C")
print("Індекс:", warmest_index)
print("Номер місяця:", month_numbers[warmest_index])

print("\nНайхолодніший місяць:")
print("Температура:", temps[coldest_index], "°C")
print("Індекс:", coldest_index)
print("Номер місяця:", month_numbers[coldest_index])

cold_months = month_numbers[temps < 0]

print("\nМісяці з температурою нижче 0°C:")
print(cold_months)

sorted_temps = np.sort(temps)

print("\nТемператури за зростанням:")
print(sorted_temps)

print("\nОригінальний масив temps:")
print(temps)
```

```python
sun_hours = np.array([50, 70, 110, 150, 190, 230, 250, 240, 180, 130, 70, 45])

data = np.array([temps, days_in_month, sun_hours])

print("Двовимірний масив:")
print(data)

print("\nФорма масиву:")
print(data.shape)

print("\nСума кожного рядка:")
print(data.sum(axis=1))

print("\nСереднє кожного рядка:")
print(data.mean(axis=1))

print("\nСума кожного стовпця:")
print(data.sum(axis=0))

print("\nСума для січня:", data[:, 0].sum())
```

Для січня отримане число означає суму трьох значень: температури, кількості днів у місяці та умовної кількості сонячних годин: -4 + 31 + 50 = 77. Сума для січня дорівнює 77.
