```python
import numpy as np
from scipy import stats

# Практика 12. Довірчі інтервали для середнього і частки
# Варіант 1 — Київ

np.random.seed(1)

july_temp = 21

daily_temps = np.random.normal(
    loc=july_temp,
    scale=2.5,
    size=30
)

print("Варіант: 1 — Київ")
print("Вибірка денних температур:")
print(daily_temps)

mean = daily_temps.mean()
s = daily_temps.std(ddof=1)
n = len(daily_temps)

print("Середня температура:", mean)
print("Вибіркове стандартне відхилення:", s)
print("Кількість спостережень:", n)
```

```python
# Завдання 2. Довірчий інтервал для середньої температури

se = s / np.sqrt(n)

ci_mean = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean,
    scale=se
)

print("Стандартна похибка:", se)
print("95% довірчий інтервал для середньої температури:", ci_mean)

print("""
Чому використовуємо t-розподіл:
Справжнє стандартне відхилення генеральної сукупності σ невідоме.
Ми оцінюємо його за вибірковим стандартним відхиленням s,
тому за невеликої вибірки використовуємо t-розподіл із n - 1
ступенями свободи, а не z-критичне значення.
""")
```

```python
# Завдання 3. Довірчий інтервал для частки

p_hat = (daily_temps > july_temp).mean()

n_p = n * p_hat
n_1p = n * (1 - p_hat)

se_p = np.sqrt(p_hat * (1 - p_hat) / n)

ci_prop = stats.norm.interval(
    0.95,
    loc=p_hat,
    scale=se_p
)

print("Частка днів із температурою вище 21°C:", p_hat)
print("n * p_hat:", n_p)
print("n * (1 - p_hat):", n_1p)
print("95% довірчий інтервал для частки:", ci_prop)
print("Чи потрапляє 0.5 в інтервал:", ci_prop[0] <= 0.5 <= ci_prop[1])

if n_p >= 5 and n_1p >= 5:
    print("Обидва значення не менші за 5, нормальне наближення прийнятне.")
else:
    print("Принаймні одне значення менше за 5, нормальне наближення ненадійне.")

if ci_prop[0] <= 0.5 <= ci_prop[1]:
    print("Інтервал містить 0.5 — результат узгоджується з теоретичною часткою.")
else:
    print("""
Інтервал не містить 0.5. Через випадковість вибірки з 30 спостережень
окремий 95% довірчий інтервал може не накрити теоретичну частку.
За повторення процедури приблизно 5% таких інтервалів не міститимуть
справжнього значення.
""")
```

```python
# Завдання 4. Порівняння ширини довірчих інтервалів

confidence_levels = [0.90, 0.95, 0.99]
intervals = []

for level in confidence_levels:
    interval = stats.t.interval(
        level,
        df=n - 1,
        loc=mean,
        scale=se
    )

    width = interval[1] - interval[0]

    intervals.append({
        "Рівень довіри": f"{level * 100:.0f}%",
        "Нижня межа": interval[0],
        "Верхня межа": interval[1],
        "Ширина інтервалу": width
    })

for item in intervals:
    print(item)

print("""
Пояснення:
Чим вищий рівень довіри, тим ширший довірчий інтервал,
оскільки для більшої надійності потрібно охопити ширший діапазон значень.
Щоб звузити інтервал при рівні довіри 99%, можна збільшити обсяг вибірки:
зі збільшенням n стандартна похибка зменшується.
""")
```

```python
# Завдання 5. Інтерпретація результату

print("Інтерпретація 95% довірчого інтервалу:")
print(
    f"Для нашої вибірки 95% довірчий інтервал для середньої "
    f"температури становить від {ci_mean[0]:.2f}°C до {ci_mean[1]:.2f}°C."
)
print(
    "Якби ми багато разів генерували нові вибірки такого самого розміру "
    "і щоразу будували інтервал тим самим методом, приблизно 95% "
    "отриманих інтервалів містили б справжнє середнє значення, "
    "а приблизно 5% — ні. Це не означає, що ймовірність потрапляння "
    "справжнього середнього в уже обчислений інтервал дорівнює 95%."
)

print("\nКонтрольні запитання")

print("""
1. Точкова оцінка — це одне число, наприклад вибіркове середнє.
Довірчий інтервал показує діапазон правдоподібних значень параметра
з урахуванням випадкової мінливості вибірки.

2. Якщо справжнє стандартне відхилення σ невідоме і вибірка невелика,
використовують t-розподіл, оскільки σ оцінюють за вибірковим
стандартним відхиленням s. t-розподіл враховує додаткову невизначеність.

3. 95% довіри означає, що при багаторазовому повторенні процедури
приблизно 95% побудованих таким способом інтервалів міститимуть
справжній параметр. Для одного вже обчисленого інтервалу не можна
стверджувати, що ймовірність містити параметр становить 95%:
параметр фіксований, а інтервал уже визначений вибіркою.

4. Ширина інтервалу залежить від рівня довіри, мінливості даних
і обсягу вибірки. Вищий рівень довіри збільшує ширину; більша
мінливість збільшує ширину; більша вибірка зменшує ширину.
""")
```
