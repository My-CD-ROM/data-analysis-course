```python
def get_season(month_num):
    if month_num in [12, 1, 2]:
        return "зима"
    elif month_num in [3, 4, 5]:
        return "весна"
    elif month_num in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"
```

**1. Чому `df.loc[1:6]` і `df.iloc[0:6]` можуть повертати різну кількість рядків залежно від індексу?**

`loc` працює з мітками (значеннями) індексу, а `iloc` — з позиціями рядків. У `loc[1:6]` обидві межі включаються, тому будуть вибрані рядки з індексами 1, 2, 3, 4, 5 і 6. У `iloc[0:6]` права межа 6 не включається, тому вибираються позиції від 0 до 5. Якщо індекс має інші значення або не збігається з позиціями рядків, результати можуть відрізнятися.

**2. Чому `df["avg_temp"] > 10 and df["precip"] > 50` дає помилку, а `(df["avg_temp"] > 10) & (df["precip"] > 50)` працює?**

У `pandas` кожна умова повертає цілу серію значень `True` або `False`. Оператор `and` призначений для роботи з одиночними логічними значеннями і не може визначити одне значення істинності для всієї Series. Оператор `&` виконує поелементне логічне «І», тому він дозволяє перевірити дві умови для кожного рядка окремо. Кожну умову потрібно брати в дужки.

**3. Чим відрізняється створення обчислюваного стовпця через формулу від використання `.apply()`?**

При створенні стовпця через формулу операція виконується одразу над усім стовпцем або масивом значень. Наприклад, `df["avg_temp"] * 9/5 + 32` векторизовано перетворює всі значення температури. `.apply()` дозволяє застосувати власну функцію до окремих значень або рядків і використовувати складнішу логіку, наприклад визначення пори року за номером місяця.
