```python
import pandas as pd
```

```python
wide = pd.DataFrame({
    "місто": ["Івано-Франківськ"],
    "Січ": [84],
    "Лют": [80],
    "Бер": [74],
    "Кві": [64],
    "Тра": [57],
    "Чер": [50],
    "Лип": [46],
    "Сер": [48],
    "Вер": [58],
    "Жов": [72],
    "Лис": [82],
    "Гру": [86],
})

wide
```

```python
tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність"
)

tidy
```

```python
tidy.shape
```

```python
back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

back
```

```python
back = back.reindex(columns=wide.columns[1:])
back
```

```python
back_reset = back.reset_index()
back_reset.equals(wide)
```

```python
tidy.to_csv("variant.csv", index=False)
tidy.to_json("variant.json", orient="records", force_ascii=False)
```

**1. Чому «місто» — номінальна шкала, а «хмарність» — шкала відношень?**

«Місто» — номінальна шкала, тому що це назва категорії без числового значення та природного порядку. «Хмарність» — шкала відношень, оскільки має змістовний нуль: 0% означає відсутність хмарності, а співвідношення між значеннями можна інтерпретувати.

**2. Чому wide-таблиця не є tidy і що змінюється після `melt()`?**

У wide-таблиці місяці представлені окремими стовпцями, тому одна змінна «місяць» розподілена між назвами стовпців. Після `melt()` назви місяців перетворюються на значення одного стовпця «місяць», а значення хмарності — на стовпець «хмарність». У результаті кожен рядок описує одне спостереження.

**3. Чим CSV відрізняється від JSON?**

CSV зберігає дані у вигляді рядків і стовпців, розділених роздільником. JSON з `orient="records"` зберігає таблицю як масив об'єктів, де кожен об'єкт містить пари «назва поля: значення». JSON має більш структурований формат і може зберігати складніші вкладені дані, тоді як CSV простіший для табличних даних.
