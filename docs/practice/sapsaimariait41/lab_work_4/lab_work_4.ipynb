```python
import numpy as np
import pandas as pd
```

```python
orders = pd.DataFrame([
    {"дата": "2026-09-01", "місто": "Івано-Франківськ", "товар": "Товар 7", "ціна": 760, "кількість": 3},
    {"дата": "2026-09-02", "місто": " Івано-Франківськ ", "товар": "Товар 7", "ціна": "760 грн", "кількість": 4},
    {"дата": "2026-09-03", "місто": "ІВАНО-ФРАНКІВСЬК", "товар": "Товар 7", "ціна": 1520, "кількість": np.nan},
    {"дата": "2026-09-04", "місто": "Івано-Франківськ", "товар": "Товар 7", "ціна": "760 грн", "кількість": 2},
    {"дата": "2026-09-05", "місто": "Івано-Франківськ", "товар": "Товар 7", "ціна": 7600, "кількість": 5},
    {"дата": "2026-09-06", "місто": "Івано-Франківськ", "товар": "Товар 7", "ціна": 700, "кількість": np.nan},
    {"дата": "2026-09-07", "місто": "Івано-Франківськ", "товар": "Товар 7", "ціна": 780, "кількість": 3},
    {"дата": "2026-09-07", "місто": "Івано-Франківськ", "товар": "Товар 7", "ціна": 780, "кількість": 3},
])

orders
```

```python
orders.isna().sum()
```

```python
orders["кількість"].median()
```

```python
orders["кількість"] = orders["кількість"].fillna(orders["кількість"].median())

orders
```

```python
orders.duplicated().sum()
```

```python
orders.duplicated(subset=["дата", "товар"]).sum()
```

```python
orders = orders.drop_duplicates(subset=["дата", "товар"])

orders
```

```python
orders["ціна"] = (
    orders["ціна"]
    .astype(str)
    .str.replace(" грн", "", regex=False)
    .astype(float)
)

orders
```

```python
orders["місто"].unique()
```

```python
orders["місто"] = (
    orders["місто"]
    .str.strip()
    .str.lower()
    .replace({"івано-франківськ": "Івано-Франківськ"})
)

orders["місто"].unique()
```

```python
Q1 = orders["ціна"].quantile(0.25)
Q3 = orders["ціна"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Нижня межа =", lower_bound)
print("Верхня межа =", upper_bound)
```

```python
outliers = orders[
    (orders["ціна"] < lower_bound) |
    (orders["ціна"] > upper_bound)
]

outliers
```

```python
orders.loc[orders["ціна"] == 7600, "ціна"] = 760

orders
```

### 1. Чому для заповнення пропусків у `кількість` було використано медіану, а не середнє?

У наборі даних лише 8 початкових спостережень, тому окремі великі або малі значення можуть суттєво впливати на середнє арифметичне. Медіана є більш стійкою до таких значень, тому для цього невеликого набору даних вона краще підходить для заповнення пропусків. Медіана кількості дорівнює 3, тому пропущені значення були замінені на 3.

### 2. Чим відрізняється `duplicated()` без `subset` від `duplicated()` із `subset`?

`duplicated()` без параметра `subset` перевіряє повний збіг усіх значень у рядках. Якщо всі стовпці однакові, рядок вважається дублікатом.

`duplicated(subset=["дата", "товар"])` перевіряє дублікати лише за вибраними стовпцями. У цій роботі дата та товар були використані як поля, що визначають унікальне замовлення.

### 3. Чому не можна автоматично видаляти всі значення, які виходять за межі IQR?

Значення за межами IQR не обов'язково є помилками. Вони можуть бути реальними та важливими спостереженнями. Тому перед видаленням викиду потрібно перевірити контекст даних.

У цьому наборі значення `7600 грн` було визнано помилкою введення, оскільки базова ціна для варіанта становить `760 грн`, значення `7600` рівно у 10 разів більше, а також перевищує верхню межу IQR `1735 грн`. Ймовірно, під час введення було додано зайвий нуль, тому значення було виправлено на `760 грн`.

### 4. Результат очищення даних

Після очищення у наборі залишилося 7 унікальних рядків. Пропущені значення кількості замінено медіаною, дубль видалено, ціни перетворено на числовий тип, назви міста нормалізовано, а аномальну ціну виправлено.
