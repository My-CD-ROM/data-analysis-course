```python
import numpy as np
import pandas as pd

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
```

```python
humidity = pd.Series(
    [84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85],
    name="Вологість, %"
)

humidity
```

```python
humidity.describe()
```

```python
rozmakh = humidity.max() - humidity.min()
variance_sample = humidity.var(ddof=1)
variance_population = humidity.var(ddof=0)
std = humidity.std()
q1 = humidity.quantile(0.25)
q3 = humidity.quantile(0.75)
iqr = q3 - q1
cv = std / humidity.mean() * 100

print("Розмах:", rozmakh)
print("Дисперсія (ddof=1):", variance_sample)
print("Дисперсія (ddof=0):", variance_population)
print("Стандартне відхилення:", std)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("CV (%):", cv)
```



count — кількість спостережень.

mean — середнє арифметичне, міра положення.

std — стандартне відхилення, міра розсіювання.

min — мінімальне значення.

25% — перший квартиль (Q1), міра положення.

50% — медіана (Q2), міра положення.

75% — третій квартиль (Q3), міра положення.

max — максимальне значення.

```python
print("Q1 з describe():", humidity.describe()["25%"])
print("Q1 вручну:", q1)
print("Q3 з describe():", humidity.describe()["75%"])
print("Q3 вручну:", q3)
```

```python
humidity = pd.Series(
    [86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87],
    name="Вологість"
)

mean = humidity.mean()
median = humidity.median()

print(f"Середнє: {mean:.2f}")
print(f"Медіана: {median:.2f}")
print(f"Різниця між середнім і медіаною: {abs(mean - median):.2f}")
```

```python
humidity = pd.Series(
    [86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87],
    name="Вологість"
)

mean = humidity.mean()
median = humidity.median()

print(f"Середнє: {mean:.2f}")
print(f"Медіана: {median:.2f}")
print(f"Різниця між середнім і медіаною: {abs(mean - median):.2f}")
```

Оскільки середнє та медіана мають дуже близькі значення, розподіл 12 місячних значень вологості Івано-Франківська можна вважати близьким до симетричного, без вираженої скошеності.

Для цих даних середнє значення є репрезентативною мірою положення, оскільки значення не мають сильної скошеності та середнє практично не відрізняється від медіани. Медіана також добре характеризує центр розподілу, але суттєвої переваги над середнім у цьому випадку немає.

```python
kharkiv = pd.Series(
    [83, 80, 75, 68, 62, 60, 58, 60, 66, 73, 79, 84],
    name="Вологість Харків"
)

kharkiv_mean = kharkiv.mean()
kharkiv_std = kharkiv.std()
kharkiv_cv = kharkiv_std / kharkiv_mean * 100

print(f"Середня вологість Харкова: {kharkiv_mean:.2f}%")
print(f"Стандартне відхилення Харкова: {kharkiv_std:.2f}")
print(f"Коефіцієнт варіації Харкова: {kharkiv_cv:.2f}%")
```

```python
ivano_mean = humidity.mean()
ivano_std = humidity.std()
ivano_cv = ivano_std / ivano_mean * 100

print(f"Середня вологість Івано-Франківська: {ivano_mean:.2f}%")
print(f"Стандартне відхилення Івано-Франківська: {ivano_std:.2f}")
print(f"Коефіцієнт варіації Івано-Франківська: {ivano_cv:.2f}%")

print(f"\nКоефіцієнт варіації Харкова: {kharkiv_cv:.2f}%")
```

Для порівняння було обрано Харків, який належить до Східного регіону, тоді як Івано-Франківськ належить до Західного.

Коефіцієнт варіації Івано-Франківська становить приблизно 10,45%, а Харкова — приблизно 13,51%. Отже, відносний розкид значень вологості у Харкова більший.

Коефіцієнт варіації є коректнішим інструментом для порівняння варіативності, оскільки він враховує величину середнього значення і показує стандартне відхилення у відсотках від середнього. Сирий std показує абсолютний розкид і не враховує масштаб даних. Навіть якщо середні значення вологості двох міст близькі, CV дає стандартизовану міру варіативності, тому його використання робить порівняння зрозумілішим.



**1. Чи можете пояснити, яка міра розсіювання з лекції стоїть за кожним із рядків `std`, `25%`/`75%` і `min`/`max` у `.describe()`?**

`std` — це стандартне відхилення, яке показує середній рівень розсіювання значень відносно їхнього середнього. `25%` і `75%` — це перший та третій квартилі (Q1 і Q3), які використовуються для визначення міжквартильного розмаху IQR = Q3 − Q1. `min` і `max` — мінімальне та максимальне значення, за якими можна визначити розмах: max − min.

**2. Чому заповнення пропуску середнім штучно зменшує `std`?**

Якщо пропущене значення замінити середнім, воно потрапляє точно в центр розподілу і не збільшує відхилення від середнього. Через це розкид даних стає меншим, а стандартне відхилення може штучно зменшитися. Тому таке заповнення може створювати хибне враження про меншу варіативність даних.

**3. Чому порівнювати абсолютні значення `std` двох груп без коефіцієнта варіації може привести до хибного висновку?**

Стандартне відхилення показує абсолютний розкид і залежить від масштабу значень. Тому дві групи з різними середніми можуть мати різні значення `std`, які не можна коректно порівнювати лише за абсолютною величиною. Коефіцієнт варіації (`std / mean × 100%`) показує розкид відносно середнього і тому зручніший для порівняння варіативності.
