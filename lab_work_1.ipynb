### Відповіді на контрольні питання:

1. **Чи можете ви пояснити різницю між shape == (12,) і shape == (1, 12)?**
   - `shape == (12,)` — це одновимірний масив (вектор) із 12 елементів. 
   - `shape == (1, 12)` — це двовимірний масив (матриця), що складається з 1 рядка і 12 стовпців.

2. **Чи розумієте, чому temps[temps > 15] повертає масив іншої довжини, ніж temps?**
   - Тому що булева індексація вибирає лише ті елементи масиву, які відповідають умові (`True`). Оскільки не всі температури вищі за 15°C, результуючий масив міститиме менше елементів.

3. **Чи можете пояснити, що станеться, якщо додати масив форми (3,) до масиву форми (3, 12)?**
   - Виникне помилка `ValueError: operands could not be broadcast together`, оскільки розмірності з кінця не збігаються (12 і 3), і NumPy не зможе автоматично узгодити їх за правилами розширення (broadcasting).

4. **Чи можете пояснити чому rows + correction спочатку дає помилку, а rows + correction.reshape(2, 1) — ні?**
   - Масив `correction` мав форму `(2,)`, що несумісно з `(2, 12)`. Після застосування `.reshape(2, 1)` його форма стала сумісною (`(2, 1)`) для поелементного додавання до кожного рядка матриці `rows`.

```python
import numpy as np

temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])

days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])

month_numbers = np.arange(1, 13)

print("--- Temps ---")
print("Shape:", temps.shape, "| Dtype:", temps.dtype)

print("\n--- Days in Month ---")
print("Shape:", days_in_month.shape, "| Dtype:", days_in_month.dtype)

print("\n--- Month Numbers ---")
print("Shape:", month_numbers.shape, "| Dtype:", month_numbers.dtype)
```

```python
rows = np.array([temps, days_in_month])
print("Rows shape:", rows.shape)

correction = np.array([1, 0])

print(
    "Пояснення помилки: форма rows — (2, 12), а correction — (2,). NumPy намагається"
    " вирівняти їх з кінця, порівнюючи 12 і 2, що викликає несумісність."
)

correction_reshaped = correction.reshape(2, 1)
print("Reshaped correction shape:", correction_reshaped.shape)

rows_corrected = rows + correction_reshaped
print("Successfully added after reshape:\n", rows_corrected)
```

```python
sun_hours = np.array([50, 70, 120, 180, 250, 280, 300, 270, 200, 120, 60, 40])

data_2d = np.array([temps, days_in_month, sun_hours])
print("Shape of data_2d:", data_2d.shape)

row_sums = data_2d.sum(axis=1)
row_means = data_2d.mean(axis=1)
print("Row sums (axis=1):", row_sums)
print("Row means (axis=1):", row_means)

col_sums = data_2d.sum(axis=0)
print("Column sums (axis=0):", col_sums)
print(
    "Explanation for January (first column): Сума елементів першого стовпця"
    f" ({-4} + {31} + {50} = {col_sums[0]}) означає загальну суму значень"
    " для січня."
)
```

```python
max_idx = temps.argmax()
min_idx = temps.argmin()
print(
    f"Warmest month: index {max_idx}, value {temps[max_idx]}°C (Month"
    f" {month_numbers[max_idx]})"
)
print(
    f"Coldest month: index {min_idx}, value {temps[min_idx]}°C (Month"
    f" {month_numbers[min_idx]})"
)

sub_zero_months = month_numbers[temps < 0]
print("Months with sub-zero temperatures:", sub_zero_months)

sorted_temps = np.sort(temps)
print("Sorted temps:", sorted_temps)
print("Original temps (unchanged):", temps)
```

```python
temps_f = temps * 9 / 5 + 32
print("Temperatures in °F:", temps_f)

warm_days_mask = temps > 15
total_warm_days = np.sum(days_in_month[warm_days_mask])
print("Total days warmer than 15°C:", total_warm_days)

total_days = days_in_month.sum()
print("Total days in year:", total_days, "-> Is 365:", total_days == 365)
```
