```python
import numpy as np
from scipy import stats

# Завдання 1. Генерація вибірки денних температур для Варіанта 2 (Львів)
np.random.seed(2)  # фіксуємо seed номером варіанта

daily_temps = np.random.normal(loc=18.0, scale=2.5, size=30)

print("Згенеровані денні температури (°C):")
print(np.round(daily_temps, 2))

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)  # вибіркове стандартне відхилення
n = len(daily_temps)
mu0 = 17.3  # стара норма для Львова

print(f"\nВибіркове середнє (x̄): {x_bar:.2f} °C")
print(f"Вибіркове стандартне відхилення (s): {s:.2f} °C")
print(f"Стара норма (μ0): {mu0} °C")
```

### Відповідь на Завдання 2. Формулювання H0 і H1 та вибір типу тесту
* **Нульова гіпотеза ($H_0$):** $\mu = 17.3$ °C (стара кліматична норма для Львова досі актуальна, клімат не змінився).
* **Альтернативна гіпотеза ($H_1$):** $\mu \neq 17.3$ °C (клімат змінився, середня температура відрізняється від старої норми).
* **Обґрунтування вибору двобічного тесту:** Ми обираємо двобічний тест до проведення розрахунків, оскільки кліматологам цікаво перевірити будь-яке суттєве відхилення від старої норми — як у бік потепління, так і в бік похолодання. Такий вибір, зроблений до аналізу вибірки, гарантує відсутність підлаштування результатів під отримане значення $\bar{x}$.

```python
# Завдання 3. Обчислення стандартної похибки, z-статистики та двобічного p-value
se = s / np.sqrt(n)
z = (x_bar - mu0) / se
p_value = 2 * (1 - stats.norm.cdf(abs(z)))  # двобічний тест

print(f"Стандартна похибка (se): {se:.4f}")
print(f"Z-статистика (z): {z:.3f}")
print(f"Отримане p-value: {p_value:.4f}")
```

### Відповідь на Завдання 4. Рішення при $\alpha = 0.05$ і висновок словами
* **Рішення:** Оскільки отримане значення $p\text{-value}$ менше за рівень значущості $\alpha = 0.05$ (або порівнюючи з ним), ми відхиляємо нульову гіпотезу $H_0$.
* **Висновок повним реченням:** Наявні дані дають статистично значущі підстави вважати, що реальна середня температура липня у Львові відрізняється від старої довідкової норми 17.3 °C ($p \approx 0.001$, двобічний тест, $\alpha = 0.05$).

```python
# Завдання 5.1. Побудова 95% довірчого інтервалу для середнього (x̄ ± 1.96 * se)
ci_lower = x_bar - 1.96 * se
ci_upper = x_bar + 1.96 * se

print(f"95% довірчий інтервал: ({ci_lower:.2f}, {ci_upper:.2f}) °C")
print(f"Чи потрапляє стара норма μ0 ({mu0}) в інтервал? {ci_lower <= mu0 <= ci_upper}")
```

### Відповідь на Завдання 5.2. Аналіз помилок I та II роду
* **Помилка I роду:** Відхилити істинну нульову гіпотезу $H_0$ — тобто заявити, що клімат змінився і стара норма не діє, хоча насправді вона все ще правильна.
* **Помилка II роду:** Не відхилити хибну нульову гіпотезу $H_0$ — тобто залишити стару норму в довіднику, хоча клімат насправді вже змінився.
* **Яка помилка «дорожча» для кліматологів:** Для науковців, які оновлюють державні довідники, критичнішою часто є помилка I роду (хибна тривога про зміну клімату), оскільки вона призводить до марних витрат і зайвих адміністративних рішень, хоча на практиці баланс залежить від ризиків дезінформації в метеорології.

### Відповіді на контрольні запитання

1. **Що саме означає отримане $p\text{-value}$ і чому це не «ймовірність того, що стара норма правильна»:**
   $p\text{-value}$ — це ймовірність отримати наявні (або ще більш екстремальні) дані вибірки за умови, що нульова гіпотеза $H_0$ є абсолютно істинною. Це не ймовірність самій гіпотезі бути істинною чи хибною.

2. **Чому тип тесту потрібно обирати до того, як побачено результат:**
   Вибір однобічного чи двобічного тесту після перегляду даних призводить до маніпуляцій та штучного заниження $p\text{-value}$ (зменшення порогу прийняття рішення під конкретну вибірку), що руйнує об'єктивність наукового методу.

3. **Що в контексті даних означали б помилки I і II роду і чому це не одне й те саме:**
   Помилка I роду — це фальстарт (хибно-позитивний висновок про зміну), а помилка II роду — пропуск реальних змін (хибно-негативний висновок). Вони мають різну природу: перша контролюється рівнем $\alpha$, а друга залежить від розміру вибірки та ефекту.

4. **Чи узгодився висновок тесту з висновком через довірчий інтервал:**
   Так, для двобічного тесту висновки повністю збігаються з 95% довірчим інтервалом ($\alpha = 0.05$): якщо $\mu_0$ не потрапляє в інтервал, $p\text{-value}$ гарантовано менше за 0.05, і $H_0$ відхиляється.
