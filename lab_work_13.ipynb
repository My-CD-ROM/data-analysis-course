```python
import numpy as np
import pandas as pd
from scipy import stats

VARIANT = 1
np.random.seed(VARIANT)

loc_param = 21.0    # параметр генерації вибірки
mu0 = 20.0          # стара довідникова норма
n = 30
alpha = 0.05
```

```python
daily_temps = np.random.normal(loc=loc_param, scale=2.5, size=n)

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)  # несміщена вибіркова оцінка s

print(f"Згенерована вибірка (перші 10 значень): {daily_temps[:10].round(2)}")
print(f"Вибіркове середнє (x̄): {x_bar:.4f} °C")
print(f"Вибіркове стандартне відхилення (s): {s:.4f} °C")
```

```python
se = s / np.sqrt(n)

z_stat = (x_bar - mu0) / se

p_value = 1 - stats.norm.cdf(z_stat)

print(f"Стандартна похибка (se): {se:.4f} °C")
print(f"z-статистика: {z_stat:.4f}")
print(f"p-value (правобічний тест): {p_value:.6f}")
```

```python
print(f"p-value ({p_value:.6f}) vs alpha ({alpha})")
if p_value < alpha:
    print("Рішення: Відхиляємо H0 на користь H1.")
else:
    print("Рішення: Немає підстав відхилити H0.")
```

```python
z_crit = stats.norm.ppf(0.975)  # 1.95996
ci_lower = x_bar - z_crit * se
ci_upper = x_bar + z_crit * se

print(f"95% довірчий інтервал для середнього: ({ci_lower:.4f} °C, {ci_upper:.4f} °C)")
print(f"Чи потрапляє μ0 = {mu0} у довірчий інтервал? {ci_lower <= mu0 <= ci_upper}")
```
