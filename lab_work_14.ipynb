```python
import numpy as np
import pandas as pd
from scipy import stats

norma_kyiv = 74.0
morning_kyiv = np.array([84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85])
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

morning_odesa = np.array([82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83])
morning_lviv = np.array([85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86])
```

```python
# Виконання одновибіркового t-критерію
t_stat_1samp, p_val_1samp = stats.ttest_1samp(morning_kyiv, popmean=norma_kyiv)

print(f"Вибіркове середнє (x̄): {morning_kyiv.mean():.4f}%")
print(f"t-статистика: {t_stat_1samp:.4f}")
print(f"p-value: {p_val_1samp:.4f}")
```

```python
x_bar = morning_kyiv.mean()
s = morning_kyiv.std(ddof=1)  # несміщене вибіркове стандартне відхилення
n = len(morning_kyiv)

t_manual = (x_bar - norma_kyiv) / (s / np.sqrt(n))

print(f"Ручна t-статистика:     {t_manual:.6f}")
print(f"t_stat з ttest_1samp:   {t_stat_1samp:.6f}")
print(f"Різниця:                {abs(t_manual - t_stat_1samp):.10e}")
```

```python
stat_lev_1, p_lev_1 = stats.levene(morning_kyiv, morning_odesa)
print(f"Критерій Лівена (Київ - Одеса): stat = {stat_lev_1:.4f}, p-value = {p_lev_1:.4f}")

equal_var_flag_1 = p_lev_1 >= 0.05
t_ind_1, p_ind_1 = stats.ttest_ind(morning_kyiv, morning_odesa, equal_var=equal_var_flag_1)

print(f"Обрано equal_var = {equal_var_flag_1} (бо p_lev = {p_lev_1:.4f} >= 0.05)")
print(f"ttest_ind: t = {t_ind_1:.4f}, p-value = {p_ind_1:.4f}")
```

```python
stat_lev_2, p_lev_2 = stats.levene(morning_kyiv, morning_lviv)
print(f"Критерій Лівена (Київ - Львів): stat = {stat_lev_2:.4f}, p-value = {p_lev_2:.4f}")

equal_var_flag_2 = p_lev_2 >= 0.05
t_ind_2, p_ind_2 = stats.ttest_ind(morning_kyiv, morning_lviv, equal_var=equal_var_flag_2)

print(f"Обрано equal_var = {equal_var_flag_2}")
print(f"ttest_ind: t = {t_ind_2:.4f}, p-value = {p_ind_2:.4f}")
```

```python
deductions = np.array([3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3])
evening_kyiv = morning_kyiv - deductions

t_rel, p_rel = stats.ttest_rel(morning_kyiv, evening_kyiv)

print(f"Ранкова середня вологість:  {morning_kyiv.mean():.2f}%")
print(f"Вечірня середня вологість:  {evening_kyiv.mean():.2f}%")
print(f"ttest_rel: t = {t_rel:.4f}, p-value = {p_rel:.10e}")
```
