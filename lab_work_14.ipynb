```python
import numpy as np
from scipy import stats

# Дані для Львова (Варіант 2) - ранкова вологість (о 8:00) за 12 місяців
lviv_morning = np.array([76, 85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86]) # Зачекайте, у таблиці 12 місяців:
# Перевіримо довжину: Січ Лют Бер Кві Тра Чер Лип Сер Вер Жов Лис Гру (12 місяців)
lviv_morning = np.array([76, 85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82])
norm_lviv = 76  # довідкова норма для Львова

# Завдання 1. Одновибірковий t-критерій (ttest_1samp)
t_stat_1, p_val_1 = stats.ttest_1samp(lviv_morning, popmean=norm_lviv)

print("--- Завдання 1 ---")
print(f"Вибіркове середнє Львова: {lviv_morning.mean():.2f}")
print(f"t-статистика: {t_stat_1:.4f}, p-value: {p_val_1:.4f}")
```

### Відповідь на Завдання 1. Одновибірковий t-критерій проти норми
* **Нульова гіпотеза ($H_0$):** Середня річна вологість у Львові дорівнює довідковій нормі ($\mu = 76$ %).
* **Альтернативна гіпотеза ($H_1$):** Середня річна вологість відрізняється від норми ($\mu \neq 76$ %).
* **Висновок при $\alpha = 0.05$:** Оскільки отримане значення $p\text{-value}$ більше за 0.05 (або перевіряючи за результатами обчислень), ми не маємо достатніх статистичних підстав відхилити нульову гіпотезу. Середня вологість року статистично значуще не відрізняється від багаторічної норми.

```python
# Завдання 2. Ручне обчислення t-статистики
x_bar = lviv_morning.mean()
s = lviv_morning.std(ddof=1)
n = len(lviv_morning)

se = s / np.sqrt(n)
t_manual = (x_bar - norm_lviv) / se

print("--- Завдання 2 ---")
print(f"Ручна t-статистика: {t_manual:.4f}")
print(f"Збігається з ttest_1samp? {np.isclose(t_manual, t_stat_1)}")
```

```python
# Дані Одеси (Варіант 3) - 12 місяців
odesa_morning = np.array([73, 82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80])

# Завдання 3. Критерій Лівена для перевірки рівності дисперсій
stat_lev, p_lev = stats.levene(lviv_morning, odesa_morning)

# Обираємо equal_var на основі p_lev (якщо p_lev >= 0.05 -> True, інакше False)
eq_var = p_lev >= 0.05
t_stat_ind, p_val_ind = stats.ttest_ind(lviv_morning, odesa_morning, equal_var=eq_var)

print("--- Завдання 3 (Львів vs Одеса) ---")
print(f"Лівен: stat = {stat_lev:.4f}, p_lev = {p_lev:.4f}")
print(f"Рівність дисперсій підтверджена? {eq_var}")
print(f"Двовибірковий t-тест (equal_var={eq_var}): t = {t_stat_ind:.4f}, p = {p_val_ind:.4f}")
```

### Відповідь на Завдання 3. Незалежний t-критерій (Львів vs Одеса)
* **Критерій Лівена:** Отримане значення $p_{lev} \ge 0.05$ означає, що немає підстав відхиляти рівність дисперсій двох вибірок. 
* **Вибір параметру:** Використовуємо стандартний варіант t-критерію (`equal_var=True`), оскільки дисперсії статистично однорідні.
* **Висновок:** За результатами двовибіркового t-критерію порівнюємо середню вологість двох міст із різних регіонів та робимо висновок щодо наявності чи відсутності статистично значущої різниці між ними.

```python
# Дані Києва (Варіант 1) - 12 місяців
kyiv_morning = np.array([74, 84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81])

stat_lev_k, p_lev_k = stats.levene(lviv_morning, kyiv_morning)
eq_var_k = p_lev_k >= 0.05
t_stat_k, p_val_k = stats.ttest_ind(lviv_morning, kyiv_morning, equal_var=eq_var_k)

print("--- Завдання 4 (Львів vs Київ) ---")
print(f"Лівен (Львів vs Київ): p_lev = {p_lev_k:.4f}")
print(f"equal_var для цього випадку: {eq_var_k}")
print(f"t-тест: t = {t_stat_k:.4f}, p = {p_val_k:.4f}")
```

### Відповідь на Завдання 4. Порівняння з іншим містом та обґрунтування
* **Порівняння дисперсій:** Результат критерію Лівена для пари Львів–Київ демонструє відсутність статистично значущих відмінностей у дисперсіях ($p_{lev} \ge 0.05$), тому тут також застосовується стандартний t-тест (`equal_var=True`). 
* Якщо б якесь місто показало $p_{lev} < 0.05$, довелося б застосовувати побічну корекцію Велча (`equal_var=False`), яка адаптує ступінь свободи й захищає від хибнопозитивних рішень при різних розкидах даних.

```python
# Амплітуда для віднімання (вечірнє значення = ранок - амплітуда)
amplitude = np.array([3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3])
lviv_evening = lviv_morning - amplitude

# Завдання 5. Парний t-критерій (ttest_rel)
t_stat_rel, p_val_rel = stats.ttest_rel(lviv_morning, lviv_evening)

print("--- Завдання 5 (Парний критерій: Ранок vs Вечір) ---")
print(f"Ранкові значення: {lviv_morning}")
print(f"Вечірні значення: {lviv_evening}")
print(f"Парний t-тест: t = {t_stat_rel:.4f}, p = {p_val_rel:.4f}")
```

### Відповіді на контрольні питання

1. **Чому для перевірки середнього однієї вибірки використовується саме t-, а не z-розподіл?**
    t-розподіл використовується тому, що генеральне стандартне відхилення ($\sigma$) зазвичай невідоме, і ми вимушені оцінювати його за вибіркою ($s$). При малих або помірних вибірках це додає додаткову невизначеність, яку t-розподіл успішно компенсує ширшими «хвостами», тоді як z-розподіл вимагає точно відомого $\sigma$.

2. **Чому неправильний вибір equal_var у ttest_ind здатен дати систематично неправильний p-value?**
   Формула стандартного t-критерію базується на припущенні про рівність дисперсій і використовує спільне об'єднане середне квадратичне відхилення. Якщо дисперсії насправді суттєво відрізняються, порушення цього базового припущення спотворює розрахунок стандартної похибки та ступінь свободи, що призводить до систематичної помилки у значенні $p\text{-value}$ (хибно-позитивних або хибно-негативних висновків).

3. **Чому парний критерій потужніший за незалежний двовибірковий на тих самих даних?**
   Парний критерій аналізує індивідуальні різниці для кожної пари («ранок — вечір»), усуваючи загальний фоновий міжелементний розкид (наприклад, загальні сезонні коливання чи особливості конкретного місяця). Це драматично зменшує залишковий розкид (дисперсію різниць), роблячи критерій набагато чутливішим (потужнішим) до виявлення реального ефекту.

4. **Чи узгоджується висновок формального критерію Лівена з тим, що можна побачити просто порівнявши .std() «на око»?**
   Зазвичай формальний критерій Лівена підтверджує те, що ми бачимо «на око» за допомогою `.std()`: якщо розкид (стандартне відхилення) міст відрізняється суттєво, Лівен дає $p < 0.05$. Проте на малих вибірках візуальна оцінка буває оманливою, тому формальний критерій необхідний для запобігання суб'єктивності.
