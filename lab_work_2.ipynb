```python
import pandas as pd

months = [
    "Січ",
    "Лют",
    "Бер",
    "Кві",
    "Тра",
    "Чер",
    "Лип",
    "Сер",
    "Вер",
    "Жов",
    "Лис",
    "Гру",
]
avg_temp = [-4, -3, 1, 7, 12, 15, 17, 16, 12, 7, 2, -2]
precip = [44, 42, 46, 58, 90, 105, 100, 82, 62, 52, 56, 50]

df = pd.DataFrame({"month": months, "avg_temp": avg_temp, "precip": precip})

print(" Перевірка DataFrame ")
print(df.head())
print("Форма (shape):", df.shape)
print("\nТипи даних (dtypes):\n", df.dtypes)
```

```python
print("Завдання 1")
print("Вихідний індекс:", df.index)

df_by_month = df.set_index("month")
print("\nРядок для місяця 'Чер':")
print(df_by_month.loc["Чер"])

df_reset = df_by_month.reset_index()
print("\nІндекс після reset_index():", df_reset.index)
```

```python
print("Завдання 2")
df_1_12 = df.copy()
df_1_12.index = range(1, 13)


print("Вибір через .loc[1:6, 'avg_temp']:")
print(df_1_12.loc[1:6, "avg_temp"])


print("\nВибір через .iloc[0:5, 1] (дев'ятий стовпець avg_temp):")
print(df_1_12.iloc[0:5, 1])


print("\nСтовпці month та precip:")
print(df[["month", "precip"]])
```

```python
print(" Завдання 3 ")
print("1. Місяці з avg_temp > 15:")
print(df[df["avg_temp"] > 15])

print("\n2. Місяці з avg_temp > 10 та precip > 50:")
print(df[(df["avg_temp"] > 10) & (df["precip"] > 50)])

print("\n3. Місяці з avg_temp < 0 або precip > 80:")
print(df[(df["avg_temp"] < 0) | (df["precip"] > 80)])
```

```python
print("Завдання 4")
df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32

df["is_wet"] = df["precip"] > df["precip"].mean()

def get_season(month_num):
    if month_num in [12, 1, 2]:
        return "зима"
    elif month_num in [3, 4, 5]:
        return "весна"
    elif month_num in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"


df["season"] = (df.index + 1).map(get_season)

print("Підсумкова таблиця зі створеними стовпцями:")
print(df)
```
