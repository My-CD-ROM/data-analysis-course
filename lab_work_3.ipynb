```python
import pandas as pd

wide_df = pd.DataFrame(
    {
        "Місто": ["Івано-Франківськ"],
        "Січ": [84],
        "Лют": [80],
        "Бер": [74],
        "Кві": [64],
        "Тра": [57],
        "Чер": [50],
        "Лип": [46],
        "Сер": [48],
        "Вер": [58],
        "Жов": [72],
        "Лис": [82],
        "Гру": [86],
    }
)

print("Завдання 1: Вихідна Wide-таблиця")
print(wide_df)
print("Форма (shape):", wide_df.shape)
```

```python
tidy_df = wide_df.melt(
    id_vars="Місто", var_name="Місяць", value_name="Хмарність"
)

print("\n Завдання 2: Tidy (Long) таблиця ")
print(tidy_df)
print("Нова форма (shape):", tidy_df.shape)
```

```python
# Завдання 3. Класифікація змінних
# 1. Місто — Номінальна шкала (текстова категорія без природного порядку чи ранжування).
# 2. Місяць — Порядкова шкала (категорії мають чітку часову послідовність, але кроки/тривалість місяців трохи різняться).
# 3. Хмарність — Шкала відношень (кількісна змінна у відсотках із природним нулем, де 0% означає повну відсутність хмарності).
```

```python
pivot_df = tidy_df.pivot(
    index="Місто", columns="Місяць", values="Хмарність"
).reset_index()

month_order = [
    "Місто",
    "Січ",
    "Лют",
    "Бер",
    "Кві",
    "Тра",
    "Чер",
    "Лип",
    "Сер",
    "Вер",
    "Жов",
    "Лис",
    "Гру",
]
pivot_df = pivot_df[month_order]

print("\n Завдання 4: Зворотне перетворення (Pivot) ")
print(pivot_df)
print("Дані збігаються з оригіналом:", pivot_df.equals(wide_df))
```

```python
tidy_df.to_csv("variant.csv", index=False)
tidy_df.to_json("variant.json", orient="records", force_ascii=False)

reloaded_csv = pd.read_csv("variant.csv")
reloaded_json = pd.read_json("variant.json")

print("\n Завдання 5: Перевірка зчитування ")
print("Зчитано з CSV:\n", reloaded_csv.head(3))
print("Зчитано з JSON:\n", reloaded_json.head(3))

print("\n Вміст CSV (перші 3 рядки)")
with open("variant.csv", "r", encoding="utf-8") as f:
    print("".join(f.readlines()[:4]))

print("\n Вміст JSON (перші 2 об'єкти) ")
with open("variant.json", "r", encoding="utf-8") as f:
    print(f.read()[:150] + "...")
```
