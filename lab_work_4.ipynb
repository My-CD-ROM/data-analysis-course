```python
import numpy as np
import pandas as pd


dirty_data = [
    {
        "замовлення_id": 101,
        "місто": "Івано-Франківськ",
        "ціна": 760.0,
        "кількість": 3.0,
    },
    {
        "замовлення_id": 101,
        "місто": "Івано-Франківськ",
        "ціна": 760.0,
        "кількість": 3.0,
    },  
    {
        "замовлення_id": 102,
        "місто": " Івано-Франківськ ",
        "ціна": "760 грн",
        "кількість": np.nan,
    },  
    {
        "замовлення_id": 103,
        "місто": "ІВАНО-ФРАНКІВСЬК",
        "ціна": "760 грн",
        "кількість": 3.0,
    },  
    {
        "замовлення_id": 104,
        "місто": "Івано-Франківськ",
        "ціна": 760.0,
        "кількість": np.nan,
    }, 
    {
        "замовлення_id": 105,
        "місто": "Івано-Франківськ",
        "ціна": 7600.0,
        "кількість": 3.0,
    },  
    {
        "замовлення_id": 106,
        "місто": "Івано-Франківськ",
        "ціна": 760.0,
        "кількість": 2.0,
    },
    {
        "замовлення_id": 107,
        "місто": "Івано-Франківськ",
        "ціна": 760.0,
        "кількість": 4.0,
    },
]

df = pd.DataFrame(dirty_data)
print("Завдання 1: Вихідний брудний DataFrame")
print(df)
```

```python
print("\n Завдання 2: Кількість пропусків у стовпцях ")
print(df.isna().sum())

median_qty = df["кількість"].median()
df["кількість"] = df["кількість"].fillna(median_qty)

print(f"\nМедіана для 'кількість': {median_qty}")
print("Пропуски після заповнення:")
print(df.isna().sum())
```

```python
print("\n Завдання 3: Пошук та видалення дублікатів ")
print("Повних дублікатів (без subset):", df.duplicated().sum())
print(
    "Дублікатів за ID (subset=['замовлення_id']):",
    df.duplicated(subset=["замовлення_id"]).sum(),
)

df = df.drop_duplicates(subset=["замовлення_id"]).reset_index(drop=True)
print(f"Кількість рядків після видалення дублікатів: {len(df)}")
```

```python
df["ціна"] = (
    df["ціна"]
    .astype(str)
    .str.replace(" грн", "", regex=False)
    .str.strip()
    .astype(float)
)

df["місто"] = df["місто"].str.strip().str.capitalize()

print("\n Завдання 4: Перевірка типів та унікальних міст ")
print(df.dtypes)
print("Унікальні значення міста:", df["місто"].unique())
print("\nОчищений DataFrame (без викидів):")
print(df)
```

```python
q1 = df["ціна"].quantile(0.25)
q3 = df["ціна"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df["ціна"] < lower_bound) | (df["ціна"] > upper_bound)]

print("\n Завдання 5: Аналіз викидів за IQR ")
print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
print(f"Межі для нормальних значень: [{lower_bound}, {upper_bound}]")
print("Знайдені викиди:")
print(outliers)
```
