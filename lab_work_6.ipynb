```python
import numpy as np
import pandas as pd

# Дані відносної вологості (%) для Варіанта 2 (Львів) по місяцях (з січня по грудень)
humidity_data = [85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86]
hum_series = pd.Series(humidity_data, name="вологість_львів")

print("--- ЗАВДАННЯ 1. Виклик .describe() ---")
desc = hum_series.describe()
display(desc)

print("\n--- ЗАВДАННЯ 2. Ручні обчислення ---")
# 1. Розмах (max - min)
rozmakh = hum_series.max() - hum_series.min()

# 2. Дисперсія (ddof=1 — вибіркова, ddof=0 — генеральна)
var_sample = hum_series.var(ddof=1)
var_pop = hum_series.var(ddof=0)

# 3. Стандартне відхилення (ddof=1)
std_val = hum_series.std(ddof=1)

# 4. Квантилі та IQR
q1 = hum_series.quantile(0.25)
q3 = hum_series.quantile(0.75)
iqr = q3 - q1

# 5. Коефіцієнт варіації (%)
mean_val = hum_series.mean()
cv = (std_val / mean_val) * 100

print(f"Розмах: {rozmakh}")
print(f"Дисперсія (вибіркова, ddof=1): {var_sample:.4f}")
print(f"Дисперсія (генеральна, ddof=0): {var_pop:.4f}")
print(f"Стандартне відхилення (ddof=1): {std_val:.4f}")
print(f"Q1 (25%): {q1} | Q3 (75%): {q3} | IQR: {iqr}")
print(f"Коефіцієнт варіації (CV): {cv:.2f}%")
```

### Відповідь на Завдання 1. Відповідність рядків .describe() мірам з лекції 7:
* **count** — обсяг вибірки (кількість місяців, $n = 12$).
* **mean** — середнє арифметичне (основна міра положення).
* **std** — вибіркове стандартне відхилення (міра розсіювання, квадратний корінь з вибіркової дисперсії).
* **min / max** — мінімальне та максимальне значення (використовуються для знаходження розмаху вибірки).
* **25% (Q1) та 75% (Q3)** — перший та третій квартилі (використовуються для обчислення міжквартильного розмаху $IQR = Q3 - Q1$).
* **50%** — медіана (другий квартиль, серединне значення варіаційного ряду).

### Завдання 4. Форма розподілу власних даних (Львів)
1. **Порівняння середнього та медіани:** 
   Для річних даних вологості Львова середнє арифметичне ($\text{mean} \approx 74.33$) та медіана (50% квартиль $\approx 74.0$) дуже близькі між собою. Це свідчить про те, що розподіл вологості протягом року є **симетричним** (без різких перекосів чи довгих хвостів у бік мінімумів/максимумів).
2. **Вибір репрезентативнішої міри положення:**
   Оскільки розподіл близький до симетричного і в наборі немає екстремальних викидів, і **середнє арифметичне**, і **медіана** є однаково репрезентативними мірами положення. Проте середнє частіше використовується в подальшому аналізі, оскільки враховує точне значення кожного місяця.

### Завдання 5. Порівняння варіативності з іншим містом (наприклад, Харків зі Сходу)
1. Для порівняння візьмемо місто **Харків** (регіон Схід). Його середньорічна вологість і стандартне відхилення відрізняються від львівських.
2. Обчислюючи коефіцієнт варіації ($CV = \frac{std}{mean} \cdot 100$), ми позбавляємося впливу абсолютного масштабу значень. 
3. **Чому саме CV, а не сирий std?** Навіть якщо вологість двох міст відрізняється в середньому на кілька відсотків, абсолютне стандартне відхилення само по собі не показує відносної стабільності процесу. Коефіцієнт варіації дозволяє коректно порівняти відносний розкид річних коливань клімату між містами з різними базовими показниками.

### Відповіді на контрольні запитання

1. **Яка міра розсіювання стоїть за кожним із рядків std, 25%/75% і min/max у виводі .describe()?**
   * За рядок `std` відповідає вибіркове стандартне відхилення (квадратний корінь із вибіркової дисперсії з мірою відхилення від середнього).
   * За рядки `25%` (Q1) та `75%` (Q3) відповідають квартилі, на основі яких розраховується міжквартильний розмах ($IQR = Q3 - Q1$), що є стійкою до викидів мірою розсіювання.
   * За рядки `min` і `max` відповідають мінімальне та максимальне значення, різниця між якими дає розмах вибірки ($\text{Range} = \text{max} - \text{min}$).

2. **Чому заповнення пропуску середнім штучно зменшує std, і яка механіка цього на числах?**
   Якщо у вибірці є пропуск і ми заповнюємо його значенням середнього арифметичного, ми додаємо до набору число, яке ідеально збігається із центром розподілу. Це створює штучне скупчення даних навколо середнього і зменшує суму квадратів відхилень від нього ($\sum (x_i - \bar{x})^2$), що при діленні на $n-1$ неминуче знижує значення дисперсії та стандартного відхилення (`std`).

3. **Чому порівнювати абсолютні значення std двох груп різного масштабу без коефіцієнта варіації може привести до хибного висновку?**
   Стандартне відхилення (`std`) вимірюється в тих самих одиницях, що й самі дані, і залежить від їхнього загального рівня (масштабу). Група з більшими середніми значеннями природно має більший абсолютний розкид. Коефіцієнт варіації ($CV$) нормує `std` до середнього значення, дозволяючи порівнювати відносну мінливість об'єктів різного масштабу.

4. **Чи симетричний власний розподіл і яка міра положення репрезентативніша?**
   Розподіл вологості Львова протягом року є близьким до симетричного (оскільки середнє значення та медіана практично збігаються). У такому випадку і середнє арифметичне, і медіана є однаково репрезентативними, але середнє є кращим вибором для симетричних вибірок без аномальних викидів.
