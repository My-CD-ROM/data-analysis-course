```python
import numpy as np
import pandas as pd

humidity_if = pd.Series(
    [86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87], name="Вологість"
)

desc_if = humidity_if.describe()
print(" Завдання 1: Результат describe() для Івано-Франківська ")
print(desc_if)
```

```python
rozmakh = humidity_if.max() - humidity_if.min()


var_sample = humidity_if.var(ddof=1)  # Вибіркова (ddof=1)
var_pop = humidity_if.var(ddof=0)  # Генеральна (ddof=0)

std_if = humidity_if.std(ddof=1)

q1_manual = humidity_if.quantile(0.25)
q3_manual = humidity_if.quantile(0.75)
iqr_manual = q3_manual - q1_manual


mean_if = humidity_if.mean()
cv_if = (std_if / mean_if) * 100

print(" Завдання 2: Ручне обчислення статистик ")
print(
    f"Мінімум: {humidity_if.min()}, Максимум: {humidity_if.max()}, Розмах: {rozmakh}"
)
print(f"Дисперсія вибіркова (ddof=1): {var_sample:.4f}")
print(f"Дисперсія генеральна (ddof=0): {var_pop:.4f}")
print(f"Різниця дисперсій: {var_sample - var_pop:.4f}")
print(f"Стандартне відхилення (std, ddof=1): {std_if:.4f}")
print(f"Q1 (25%): {q1_manual:.2f}, Q3 (75%): {q3_manual:.2f}, IQR: {iqr_manual:.2f}")
print(f"Коефіцієнт варіації (CV): {cv_if:.2f}%")

assert q1_manual == desc_if["25%"], "Q1 не збігається!"
assert q3_manual == desc_if["75%"], "Q3 не збігається!"
print("\n[Перевірка] 25% та 75% повністю збігаються з .describe()!")
```
