```python
import numpy as np
import pandas as pd

VARIANT = 1
np.random.seed(VARIANT)

base_price = 850  # Варіант 1 
base_qty = 3
```

```python
ages = [22, 25, 28, 35, 42, 45, 51, 58, 63, 67]
incomes = [18000, 22000, 31000, 45000, 52000, 38000, 60000, 29000, 75000, 34000]
quantities = [2, 3, 3, 4, 2, 5, 3, 1, 4, 3]

cities = ["Київ", "Львів", "Одеса", "Харків", "Дніпро", 
          "Київ", "Львів", "Одеса", "Київ", "Харків"]

clients = pd.DataFrame({
    "вік": ages,
    "дохід": incomes,
    "кількість": quantities,
    "місто_доставки": cities
})

print("Створений DataFrame клієнтів (Варіант 1 — Київ):")
print(clients)
```

```python
clients["сума_замовлення"] = clients["кількість"] * base_price
clients["частка_доходу"] = clients["сума_замовлення"] / clients["дохід"]

print("\nТаблиця з похідними змінними:")
print(clients[["дохід", "кількість", "сума_замовлення", "частка_доходу"]].round(4))
```

```python
clients["вік_cut"] = pd.cut(clients["вік"], bins=3)
cut_counts = clients["вік_cut"].value_counts().sort_index()

clients["вік_qcut"] = pd.qcut(clients["вік"], q=3)
qcut_counts = clients["вік_qcut"].value_counts().sort_index()

print(" Розподіл pd.cut() (однакова ширина) ")
print(cut_counts)

print("\n Розподіл pd.qcut() (однакова кількість) ")
print(qcut_counts)
```

```python
clients["вік_norm"] = (clients["вік"] - clients["вік"].min()) / (clients["вік"].max() - clients["вік"].min())
clients["дохід_norm"] = (clients["дохід"] - clients["дохід"].min()) / (clients["дохід"].max() - clients["дохід"].min())

clients["вік_z"] = (clients["вік"] - clients["вік"].mean()) / clients["вік"].std()
clients["дохід_z"] = (clients["дохід"] - clients["дохід"].mean()) / clients["дохід"].std()

print("Порівняння масштабованих колонок (перші 5 рядків):")
scaled_cols = ["вік", "дохід", "вік_norm", "дохід_norm", "вік_z", "дохід_z"]
print(clients[scaled_cols].round(3).head())
```

```python
city_dummies = pd.get_dummies(clients["місто_доставки"], prefix="місто", dtype=int)

clients_encoded = pd.concat([clients, city_dummies], axis=1)

print(f"Кількість унікальних міст: {clients['місто_доставки'].nunique()}")
print(f"Кількість створених dummy-стовпців: {city_dummies.shape[1]}")
print("\nПерші 5 рядків закодованого DataFrame:")
encoded_display_cols = ["місто_доставки"] + list(city_dummies.columns)
print(clients_encoded[encoded_display_cols].head())
```
