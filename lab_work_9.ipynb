```python
import numpy as np
import pandas as pd

# Фіксуємо seed номером варіанта (2 для Львова)
np.random.seed(2)

# Базова ціна для Варіанта 2 (Львів)
base_price = 720
base_quantity = 4

# Генеруємо 10 замовлень клієнтів
clients_data = [
    {"вік": 24, "дохід": 22000, "кількість": 4, "місто_доставки": "Львів"},
    {"вік": 35, "дохід": 45000, "кількість": 2, "місто_доставки": "Київ"},
    {"вік": 42, "дохід": 31000, "кількість": 6, "місто_доставки": "Одеса"},
    {"вік": 29, "дохід": 18500, "кількість": 3, "місто_доставки": "Харків"},
    {"вік": 51, "дохід": 65000, "кількість": 5, "місто_доставки": "Львів"},
    {"вік": 38, "дохід": 39000, "кількість": 4, "місто_доставки": "Дніпро"},
    {"вік": 22, "дохід": 15000, "кількість": 1, "місто_доставки": "Київ"},
    {"вік": 47, "дохід": 52000, "кількість": 5, "місто_доставки": "Одеса"},
    {"вік": 31, "дохід": 28000, "кількість": 3, "місто_доставки": "Харків"},
    {"вік": 58, "дохід": 78000, "кількість": 8, "місто_доставки": "Львів"}
]

df_clients = pd.DataFrame(clients_data)

print("Набір даних клієнтів (10 замовлень):")
display(df_clients)
```

```python
# Завдання 2. Обчислення похідних змінних
df_clients["сума_замовлення"] = base_price * df_clients["кількість"]
df_clients["частка_доходу"] = df_clients["сума_замовлення"] / df_clients["дохід"]

print("Таблиця з новими похідними змінними:")
display(df_clients[["дохід", "сума_замовлення", "частка_доходу"]])
```

### Відповідь на Завдання 2. Чому `частка_доходу` корисніша?
Змінна `частка_доходу` відіграє роль відносного показника фінансового навантаження замовлення на конкретного клієнта. Сама по себе `сума_замовлення` чи `дохід` окремо дають лише абсолютні величини: наприклад, замовлення на 2880 грн для клієнта з доходом 15 000 грн є суттєвою фінансовою операцією (~19.2% доходу), тоді як для клієнта з доходом 78 000 грн ця ж сума становить лише ~3.7%. Саме відносна частка дозволяє коректно порівнювати платоспроможність і поведінку різних покупців.

```python
# Завдання 3.1. Дискретизація віку на 3 рівні інтервали за довжиною (pd.cut)
df_clients["вік_група_cut"] = pd.cut(df_clients["вік"], bins=3)
print("Розподіл за pd.cut (рівні інтервали):")
print(df_clients["вік_група_cut"].value_counts())

# Завдання 3.2. Дискретизація віку на 3 групи за рівною кількістю спостережень (pd.qcut)
df_clients["вік_група_qcut"] = pd.qcut(df_clients["вік"], q=3)
print("\nРозподіл за pd.qcut (рівна кількість у групах):")
print(df_clients["вік_група_qcut"].value_counts())
```

### Відповідь на Завдання 3. Порівняння cut і qcut
* **`pd.cut`** ділить діапазон значень змінної (від мінімального до максимального віку) на рівні за довжиною інтервали. Через це кількість клієнтів у кожній групі може суттєво відрізнятися, оскільки спостереження у вибірці розподілені нерівномірно.
* **`pd.qcut`** орієнтується на квантилі та гарантує розбиття на групи з приблизно рівною кількістю спостережень у кожній, але межі самих інтервалів при цьому виходять нерівними.

```python
# Завдання 4.1 & 4.2. Масштабування віку та доходу
age = df_clients["вік"]
income = df_clients["дохід"]

# Min-Max нормалізація в діапазон [0, 1]
df_clients["вік_minmax"] = (age - age.min()) / (age.max() - age.min())
df_clients["дохід_minmax"] = (income - income.min()) / (income.max() - income.min())

# Z-стандартизація (середнє = 0, стандартне відхилення = 1)
df_clients["вік_z"] = (age - age.mean()) / age.std()
df_clients["дохід_z"] = (income - income.mean()) / income.std()

print("Результати масштабування (перші 5 рядків):")
display(df_clients[["вік", "дохід", "вік_minmax", "дохід_minmax", "вік_z", "дохід_z"]].head())
```

### Відповідь на Завдання 4. Чому необхідне масштабування?
Дохід вимірюється десятками тисяч гривень, а вік — десятками років. Через таку різницю масштабів алгоритми, чутливі до відстаней (наприклад, метод k-середніх чи KNN) або до числових амплітуд, надаватимуть перевагу доходу, сприймаючи його варіацію як набагато важливішу. Масштабування (Min-Max або Z-оцінка) приводить змінні до спільного діапазону чи стандартного відхилення, забезпечуючи їх рівноправний внесок у моделі.

```python
# Завдання 5.1 & 5.2. One-hot кодування для стовпця "місто_доставки"
city_dummies = pd.get_dummies(df_clients["місто_доставки"], prefix="місто")

# Долучаємо нові стовпці до загального DataFrame
df_clients = pd.concat([df_clients, city_dummies], axis=1)

print(f"Кількість унікальних міст у даних: {df_clients['місто_доставки'].nunique()}")
print(f"Кількість новостворених індикаторних стовпців: {city_dummies.shape[1]}")

print("Фінальна таблиця з індикаторними стовпцями:")
# Виправляємо об'єднання списків для display та прибираємо зайвий бекслеш у тексті
cols_to_show = ["місто_доставки"] + list(city_dummies.columns)
display(df_clients[cols_to_show].head())
```

### Відповідь на Завдання 5. Чому One-hot, а не порядкове кодування для міст?
Назва міста є номінальною категорією (наприклад, Львів, Київ, Одеса, Харків не мають природного кількісного чи порядкового впорядкування). Якби ми застосували просте числове кодування (наприклад, Київ=1, Львів=2, Одеса=3), алгоритми машинного навчання хибно сприйняли б ці цифри як математичні величини (де Одеса «більша» за Київ). One-hot кодування створює окремі бінарні стовпці для кожного міста, уникаючи штучного порядку.

### Відповіді на контрольні запитання

1. **Різниця між `pd.cut()` і `pd.qcut()`:**
   `pd.cut()` використовується для створення інтервалів однакової довжини (на основі числових діапазонів значень), тоді як `pd.qcut()` гарантує розбиття на групи з однаковою кількістю спостережень (на основі квантилів вибірки).

2. **Чому для номінальних змінних застосовують one-hot кодування, а не порядкове чи числове?**
   Оскільки номінальні змінні не мають природного ієрархічного чи кількісного ранжування, числове кодування (1, 2, 3) створило б хибні математичні зв'язки між категоріями. One-hot кодування перетворює кожну категорію на незалежну бінарну ознаку.

3. **Навіщо масштабувати змінні перед дистанційними чи регуляризованими методами?**
   Різні одиниці вимірювання та велика різниця в масштабах (наприклад, тисячі гривень проти десятків років) призводять до того, що алгоритми починають домінуюче враховувати змінні з більшими числовими значеннями. Масштабування вирівнює їхній вплив.

4. **Чому похідна змінна часто несе більше сенсу для аналізу, ніж вихідні складники окремо?**
   Похідні змінні (як-от відносні частки, інтенсивність чи показники ефективності) поєднують кілька факторів у єдину логічну метрику, яка безпосередньо відображає суть явища або поведінковий паттерн, недоступний при аналізі «сирих» показників ізольовано.
