```python
import numpy as np
import pandas as pd

np.random.seed(42)

base_temp = 10.0
amplitude = 14
city = "Дніпро"

rows = []

for year in [2021, 2022, 2023, 2024]:
    for month in range(1, 13):
        seasonal = amplitude * np.cos((month - 7) / 12 * 2 * np.pi)
        noise = np.random.normal(0, 1.0)

        rows.append({
            "місто": city,
            "рік": year,
            "місяць": month,
            "температура": round(base_temp + seasonal + noise, 1)
        })

climate = pd.DataFrame(rows)

print("Варіант 5 — Дніпро")
print(climate)
```

```python
year_stats = climate.groupby("рік")["температура"].agg(
    ["mean", "min", "max"]
)

print(year_stats)
```

Середня температура:
2021 — 10.28 °C
2022 — 9.42 °C
2023 — 9.81 °C
2024 — 9.66 °C.

Чіткого тренду до потепління або похолодання за ці чотири роки
не спостерігається. Температура коливається від року до року,
а зміни виглядають випадковими.

```python
month_stats = climate.groupby("місяць")["температура"].agg(
    ["mean", "std"]
)

print(month_stats)

max_std_month = month_stats["std"].idxmax()
max_std = month_stats["std"].max()

print("\nНайбільший розкид:")
print("Місяць:", max_std_month)
print("std:", max_std)
```

Найбільший розкид температури між роками спостерігається
у серпні (місяць 8).

Стандартне відхилення становить приблизно 1.42 °C.

Це означає, що температура серпня в нашому наборі найбільше
відрізнялася між 2021–2024 роками. Оскільки дані синтетичні
та містять випадковий шум, така нестабільність може бути
результатом випадкових коливань.

```python
pivot_temp = climate.pivot_table(
    index="місяць",
    columns="рік",
    values="температура",
    aggfunc="mean"
)

print(pivot_temp)
```

Зведена таблиця pivot_table зручніша для читання людиною,
оскільки можна легко порівнювати температуру одного місяця
між різними роками.

Довгий (tidy) формат climate зручніший для аналізу даних,
групування, фільтрації та побудови графіків.

Отже, зведена форма зручніша для візуального порівняння,
а довгий формат — для подальшої обробки даних.

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "зима"
    elif month in [3, 4, 5]:
        return "весна"
    elif month in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"

climate["сезон"] = climate["місяць"].apply(get_season)

climate["тепліше_за_середнє"] = (
    climate["температура"] > base_temp
)

print(climate.head(15))
```

```python
season_table = pd.crosstab(
    climate["сезон"],
    climate["тепліше_за_середнє"]
)

print(season_table)
```

Результат crosstab відповідає очікуванням.

Усі 12 літніх спостережень мають температуру вище
середньорічної температури 10 °C.

Усі 12 зимових спостережень, навпаки, мають температуру
нижче або рівну цьому рівню.

Навесні та восени спостерігаються обидві категорії,
оскільки це перехідні сезони.

```python
pivot_simple = climate.pivot(
    index="місяць",
    columns="рік",
    values="температура"
)

print(pivot_simple)
```

Метод pivot() успішно працює для набору climate.

Причина полягає в тому, що для кожної комбінації
"місяць + рік" у наборі існує рівно один рядок.
Тому pivot() однозначно знає, яке значення температури
потрібно помістити в кожну клітинку.

Якби для одного міста було декілька метеостанцій і кожна
станція мала окреме вимірювання температури за той самий
місяць і рік, виникли б дублікати комбінацій "місяць + рік".

У такому випадку pivot() завершився б помилкою, оскільки
не знав би, яке з декількох значень вибрати.

Для таких даних можна використати pivot_table() з функцією
агрегації, наприклад aggfunc="mean".
