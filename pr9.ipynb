```python
import pandas as pd

base_price = 810

clients = pd.DataFrame({
    "вік": [19, 23, 27, 31, 36, 42, 48, 53, 61, 68],
    "дохід": [18000, 24000, 32000, 28000, 45000,
              52000, 39000, 61000, 47000, 70000],
    "кількість": [2, 3, 4, 3, 5, 2, 3, 4, 2, 3],
    "місто_доставки": [
        "Дніпро", "Київ", "Львів", "Дніпро", "Одеса",
        "Київ", "Дніпро", "Львів", "Одеса", "Дніпро"
    ]
})

print("Варіант 5 — Дніпро")
clients
```

```python
clients["сума_замовлення"] = (
    base_price * clients["кількість"]
)

clients["частка_доходу"] = (
    clients["сума_замовлення"] / clients["дохід"]
)

clients
```

```python
clients[
    ["дохід", "кількість", "сума_замовлення", "частка_доходу"]
]
```

Частка доходу є кориснішою для аналізу поведінки клієнта,
оскільки вона показує, яку частину свого місячного доходу
клієнт витрачає на замовлення.

Однакова сума замовлення може мати різне значення для
клієнтів із різним доходом. Наприклад, витрата 3000 грн
буде більш суттєвою для клієнта з доходом 18000 грн,
ніж для клієнта з доходом 70000 грн.

Тому частка доходу дозволяє краще порівнювати клієнтів
із різним рівнем доходу.

```python
clients["вік_cut"] = pd.cut(
    clients["вік"],
    bins=3
)

print("Групи через cut:")
print(clients["вік_cut"].value_counts().sort_index())
```

```python
clients["вік_qcut"] = pd.qcut(
    clients["вік"],
    q=3
)

print("Групи через qcut:")
print(clients["вік_qcut"].value_counts().sort_index())
```

```python
print("CUT:")
print(clients["вік_cut"].value_counts().sort_index())

print("\nQCUT:")
print(clients["вік_qcut"].value_counts().sort_index())
```

pd.cut() та pd.qcut() по-різному формують групи.

pd.cut() розділяє весь діапазон значень віку на інтервали
приблизно однакової ширини. Через це кількість клієнтів
у кожному інтервалі може бути різною.

pd.qcut() формує межі груп на основі квантилів, щоб
кількість спостережень у групах була приблизно однаковою.
Тому ширина отриманих інтервалів може відрізнятися.

Отже, cut орієнтується на ширину інтервалів, а qcut —
на кількість спостережень у них.

```python
clients["вік_minmax"] = (
    (clients["вік"] - clients["вік"].min()) /
    (clients["вік"].max() - clients["вік"].min())
)

clients["дохід_minmax"] = (
    (clients["дохід"] - clients["дохід"].min()) /
    (clients["дохід"].max() - clients["дохід"].min())
)
```

```python
clients["вік_z"] = (
    (clients["вік"] - clients["вік"].mean()) /
    clients["вік"].std()
)

clients["дохід_z"] = (
    (clients["дохід"] - clients["дохід"].mean()) /
    clients["дохід"].std()
)
```

```python
clients[[
    "вік",
    "дохід",
    "вік_minmax",
    "дохід_minmax",
    "вік_z",
    "дохід_z"
]].head()
```

До масштабування вік і дохід мають дуже різні масштаби.

Вік вимірюється десятками років, а дохід — десятками
тисяч гривень. Через це в методах, чутливих до масштабу,
дохід міг би мати значно більший вплив лише через свої
більші числові значення.

Min-max нормалізація приводить значення до діапазону
від 0 до 1.

Z-стандартизація показує, на скільки стандартних
відхилень значення знаходиться від середнього.

Після масштабування змінні можна коректніше використовувати
разом у методах, чутливих до масштабу.

```python
city_dummies = pd.get_dummies(
    clients["місто_доставки"],
    prefix="місто"
)

city_dummies
```

```python
clients = pd.concat(
    [clients, city_dummies],
    axis=1
)

clients
```

```python
unique_cities = clients["місто_доставки"].nunique()
dummy_columns = city_dummies.shape[1]

print("Кількість унікальних міст:", unique_cities)
print("Кількість нових one-hot стовпців:", dummy_columns)

print(
    "Кількість збігається:",
    unique_cities == dummy_columns
)
```

Для змінної "місто_доставки" доречно використовувати
one-hot кодування, оскільки міста є номінальними категоріями
і не мають природного порядку.

Якби міста були закодовані числами 1, 2, 3, 4, модель
могла б помилково сприймати ці числа як порядок або
відстань між категоріями.

One-hot кодування створює окремий індикаторний стовпець
для кожного міста і не створює штучного порядку.

Контрольні питання

1. Яка різниця між pd.cut() і pd.qcut()?

pd.cut() ділить діапазон числової змінної на інтервали
однакової або заданої ширини. Кількість спостережень
у таких групах може бути різною.

pd.qcut() використовує квантилі та намагається розподілити
спостереження між групами приблизно порівну. При цьому
ширина інтервалів може бути різною.


2. Чому для міста використовують one-hot кодування?

Місто є номінальною змінною і його категорії не мають
природного порядку.

Якщо закодувати міста числами 1, 2, 3, модель може
помилково вважати, що між ними існує порядок.

One-hot кодування створює незалежний індикатор для кожного
міста і не вносить хибного порядку.


3. Навіщо масштабувати змінні?

Змінні можуть мати різний масштаб. Наприклад, вік має
значення приблизно від 18 до 70, а дохід — десятки тисяч
гривень.

У методах, чутливих до масштабу, змінна з більшими
числовими значеннями може непропорційно впливати
на результат.

Масштабування приводить змінні до порівнянного масштабу.


4. Чому похідна змінна може бути кориснішою?

Похідна змінна може поєднати інформацію з декількох
початкових змінних та надати їй додатковий зміст.

Наприклад, частка доходу показує не просто суму покупки
або дохід клієнта, а те, наскільки великою є покупка
відносно його доходу.

Тому така змінна дозволяє краще порівнювати поведінку
різних клієнтів.
