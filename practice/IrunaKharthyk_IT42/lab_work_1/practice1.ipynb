Письмові відповіді

1. Чи можете ви пояснити різницю між shape == (12,) і shape == (1, 12)?

`shape == (12,)` означає одновимірний масив із 12 елементів. У нього немає окремого виміру для рядка та стовпця.

`shape == (1, 12)` означає двовимірний масив, який має 1 рядок і 12 стовпців.

Тобто `(12,)` і `(1, 12)` — це різні форми масивів, хоча в них однакова кількість елементів.

2. Чи розумієте, чому temps[temps > 15] повертає масив іншої довжини, ніж temps?

`temps > 15` створює булевий масив такої самої довжини, як `temps`. Для кожної температури отримуємо `True` або `False`.

Під час `temps[temps > 15]` NumPy залишає тільки ті елементи, для яких умова має значення `True`. Тому результат містить лише температури понад 15°C і може мати меншу довжину, ніж початковий масив `temps`.

Для мого варіанта це температури 16, 20, 23, 22 і 17°C, тому результат має 5 елементів.

3. Чи можете пояснити, що станеться, якщо додати масив форми (3,) до масиву форми (3, 12)?

Масив форми `(3,)` не зможе напряму додатися до масиву форми `(3, 12)` через broadcasting.

NumPy порівнює форми справа наліво. Отримаємо порівняння `12` і `3`. Вони несумісні, оскільки не однакові і жодне з них не дорівнює 1.

Тому така операція призведе до помилки broadcasting.

Щоб додавання працювало для кожного з трьох рядків, масив поправок можна перетворити на форму `(3, 1)`.

 4. Чи можете пояснити, чому rows + correction спочатку дає помилку, а rows + correction.reshape(2, 1) — ні?

`rows` має форму `(2, 12)`, а `correction` має форму `(2,)`.

Під час broadcasting NumPy порівнює розміри справа наліво: `12` і `2`. Вони несумісні, тому виникає помилка.

Після `correction.reshape(2, 1)` форма поправки стає `(2, 1)`. Тепер форми `(2, 12)` і `(2, 1)` сумісні: перші розміри однакові, а розмір `1` може бути розширений до `12`.

Тому поправка `[1, 0]` застосовується до рядків: до всіх значень першого рядка додається 1, а до другого додається 0.


```python
correction_2d = correction.reshape(2, 1)

print(correction_2d)
print("shape:", correction_2d.shape)
result = rows + correction_2d

print(result)
print("shape:", result.shape)

print("rows:")
print(rows)

print()

print("correction:")
print(correction_2d)

print()

print("rows + correction:")
print(result)

```
