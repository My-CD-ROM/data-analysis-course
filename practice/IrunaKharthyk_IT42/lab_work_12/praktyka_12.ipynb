Контрольне питання №1
Яка різниця між точковою оцінкою і довірчим інтервалом — чому саме число (наприклад, x̄) недостатньо для повної відповіді?

Точкова оцінка — це одне число, яке використовується для оцінювання невідомого параметра генеральної сукупності. Наприклад, вибіркове середнє x̄ є точковою оцінкою генерального середнього. Проте одне число не показує, наскільки оцінка може відрізнятися від справжнього значення через випадковість вибірки. Довірчий інтервал доповнює точкову оцінку діапазоном можливих значень і показує невизначеність оцінювання.

Контрольне питання №2
Чому для довірчого інтервалу середнього при невідомому σ і не дуже великому n використовують t-розподіл, а не нормальний?

Якщо σ генеральної сукупності невідоме, його замінюють вибірковим стандартним відхиленням s. Через це виникає додаткова невизначеність в оцінюванні середнього. t-розподіл враховує цю невизначеність і має товстіші хвости, ніж нормальний розподіл. Тому при невідомому σ та відносно невеликій вибірці для інтервалу середнього використовують t-розподіл.

Контрольне питання №3
Що насправді означає "95% довіри"?

95% довіри означає властивість процедури побудови довірчих інтервалів. Якщо багато разів брати випадкові вибірки з тієї самої генеральної сукупності та кожного разу будувати інтервал тим самим методом, приблизно 95% отриманих інтервалів накриватимуть справжнє значення параметра. Формулювання про "95% ймовірності того, що параметр знаходиться в цьому конкретному інтервалі" є некоректним у класичній частотній інтерпретації, оскільки параметр вважається фіксованим, а випадковим є сам інтервал.

 Контрольне питання №4 — три фактори ширини

1. Рівень довіри
Більший рівень довіри → ширший інтервал.

90% → вузькіший
95% → ширший
99% → ще ширший

2. Розмір вибірки
Більше n → менша стандартна похибка → вужчий інтервал.

3. Варіативність даних
Більше s → більша стандартна похибка → ширший інтервал.


На ширину довірчого інтервалу впливають рівень довіри, розмір вибірки та варіативність даних. Збільшення рівня довіри розширює інтервал. Збільшення розміру вибірки звужує інтервал, оскільки стандартна похибка зменшується як 1/√n. Збільшення стандартного відхилення s збільшує стандартну похибку і тому робить інтервал ширшим.
