Контрольне питання №1 — p-value
Питання:

Що саме означає отримане у вас p-value і чому це не "ймовірність того, що стара норма μ0 правильна"?

p-value — це ймовірність отримати за умови істинності H0 результат, який є не менш екстремальним щодо H0, ніж спостережений. У нашому випадку це показник того, наскільки незвичайним є отримане вибіркове середнє порівняно зі старою нормою 20.0°C, якщо припустити, що справжня середня температура дорівнює 20.0°C. p-value не є ймовірністю того, що H0 правильна, оскільки воно обчислюється за припущення H0, а не дає ймовірність істинності самої гіпотези.

Контрольне питання №2 — чому тест обираємо заздалегідь?
Тип тесту потрібно визначити до перегляду результату, оскільки вибір однобічного або двобічного тесту впливає на p-value та критичну область. Якщо спочатку подивитися на вибіркове середнє, а потім вибрати напрямок тесту під отриманий результат, можна штучно збільшити ймовірність отримання статистично значущого результату. У нашому випадку обрано двобічний тест, оскільки нас цікавить будь-яке відхилення від старої норми 20.0°C.

Контрольне питання №3 — помилки I та II роду
Для наших даних помилка I роду означала б, що ми відхилили H0 і зробили висновок про зміну середньої липневої температури Києва, хоча насправді справжнє середнє все ще дорівнює 20.0°C. Помилка II роду означала б, що ми не відхилили H0, хоча насправді справжня середня температура вже відрізняється від 20.0°C. Ці помилки різні: у першому випадку ми помилково виявляємо зміну, а в другому — не виявляємо реальну зміну.

Контрольне питання №4 — тест та інтервал
Висновок тесту узгоджується з висновком через 95% довірчий інтервал. Якщо стара норма μ0 потрапляє до 95% довірчого інтервалу, то при α = 0.05 немає підстав відхиляти H0. Якщо μ0 не потрапляє до інтервалу, H0 відхиляємо. Для двобічного тесту та відповідного 95% інтервалу це два еквівалентні способи прийняття статистичного рішення.



Для кліматологів, які готують оновлений офіційний довідник, помилка II роду може бути дорожчою, якщо важливо не пропустити реальну зміну середньої температури. У такому випадку помилкове збереження старої норми може призвести до використання застарілої інформації в подальших кліматичних розрахунках. Водночас помилка I роду також має наслідки, оскільки передчасне оновлення норми може створити хибне уявлення про зміну клімату. Тому вибір того, яка помилка дорожча, залежить від практичної мети аналізу.

При α = 0.05 нульову гіпотезу H0 відхиляємо, оскільки p-value < 0.05. Отримані дані дають статистично значущі підстави вважати, що середня липнева температура Києва відрізняється від старої норми 20.0°C. Спостережуване вибіркове середнє становить x̄ = ...°C, z = ..., p = 0.05

```python
if p_value < alpha:
    print(
        f"При α = {alpha} нульову гіпотезу відхиляємо. "
        f"Є статистично значущі підстави вважати, що "
        f"середня липнева температура Києва відрізняється "
        f"від старої норми {mu0}°C."
    )
else:
    print(
        f"При α = {alpha} немає достатніх статистичних підстав "
        f"відхиляти нульову гіпотезу. "
        f"Дані не дають статистично значущих доказів того, "
        f"що середня липнева температура Києва відрізняється "
        f"від старої норми {mu0}°C."
    )
```

```python
alpha = 0.05

if p_value < alpha:
    print("p-value < 0.05")
    print("Відхиляємо H0.")
else:
    print("p-value >= 0.05")
    print("Не відхиляємо H0.")
```
