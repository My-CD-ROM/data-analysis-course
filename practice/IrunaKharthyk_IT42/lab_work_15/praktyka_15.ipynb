Це узгоджується з побудовою набору, оскільки сезонне коливання температури безпосередньо визначає відхилення від середньорічної температури. Завдання 2:

При α = 0.05, якщо p-value < 0.05, H0 про незалежність відхиляється, тобто є статистичні підстави вважати сезон і відхилення від норми пов'язаними. Якщо p-value ≥ 0.05, H0 не відхиляється, тобто недостатньо статистичних підстав стверджувати про зв'язок.

Завдання 3:

Очікувані частоти перевірено на відповідність умові ≥ 5. Якщо всі вони не менші за 5, умова застосовності критерію χ² виконується, тому об'єднання категорій не потрібне.

Завдання 4:

Завдання 5:

H0: розподіл категорій «холодніше», «звичайно», «тепліше» становить відповідно 20%, 60% і 20%. Така пропорція є обґрунтованим припущенням для перевірки, оскільки категорія «звичайно» визначена широким центральним інтервалом від −3°C до +3°C, а дві інші категорії відповідають відхиленню більш ніж на 3°C. Рішення щодо H0 приймається за отриманим p-value при α = 0.05.

```python
rows = []

for year in [2021, 2022, 2023, 2024]:
    for month in range(1, 13):
        seasonal = amplitude * np.cos((month - 7) / 12 * 2 * np.pi)
        noise = np.random.normal(0, 1.0)
        temp = round(base_temp + seasonal + noise, 1)

        diff = temp - base_temp

        if diff < -3:
            norm_cat = "холодніше"
        elif diff > 3:
            norm_cat = "тепліше"
        else:
            norm_cat = "звичайно"

        rows.append({
            "місто": city,
            "рік": year,
            "місяць": month,
            "температура": temp,
            "сезон": season(month),
            "відхилення_від_норми": norm_cat
        })

climate = pd.DataFrame(rows)

climate
```

```python
base_temp = 9.5
amplitude = 14
city = "Київ"

def season(month):
    if month in (12, 1, 2):
        return "зима"
    if month in (3, 4, 5):
        return "весна"
    if month in (6, 7, 8):
        return "літо"
    return "осінь"
```
