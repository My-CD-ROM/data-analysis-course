1. Чому .loc[1:6] і .iloc[0:6] можуть повернути різну кількість рядків?

.loc вибирає рядки за значеннями індексу, тобто за їхніми підписами. .iloc вибирає рядки за їхньою позицією в таблиці. Наприклад, .loc[1:6] вибирає рядки з індексами від 1 до 6 включно, а .iloc[0:6] вибирає перші 6 рядків із позиціями від 0 до 5. Якщо індекс таблиці не збігається з позиціями рядків, результати можуть бути різними.

2. Чому and викликає помилку, а & — ні?

У pandas умови df["avg_temp"] > 10 та df["precip"] > 50 повертають набір значень True або False для кожного рядка. Оператор and призначений для звичайних одиничних логічних значень, тому він не може правильно обробити два стовпці pandas і викликає помилку. Оператор & виконує поелементне логічне «І», тому для об'єднання умов у pandas потрібно використовувати саме &. Кожну умову також потрібно брати в дужки.

3. У чому різниця між обчислюваним стовпцем через формулу та через .apply()?

Стовпець, створений формулою, обчислюється за допомогою векторизованих операцій одразу для всіх значень стовпця. Наприклад, температуру у Фаренгейтах можна отримати за формулою df["avg_temp"] * 9 / 5 + 32. Метод .apply() дозволяє застосувати власну функцію до кожного значення. Він зручний для складнішої логіки, наприклад для визначення пори року за номером місяця.

```python
def get_season(month_number):
    if month_number in [12, 1, 2]:
        return "зима"
    elif month_number in [3, 4, 5]:
        return "весна"
    elif month_number in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"
```
