Питання 1
Чи можете пояснити, чому стовпець "місто" — номінальна шкала, а стовпець "хмарність" — шкала відношень?


Місто належить до номінальної шкали, оскільки це назви категорій, між якими немає природного порядку. Хмарність належить до шкали відношень, оскільки вимірюється у відсотках, має природний нуль — 0% хмарності, а співвідношення між значеннями має зміст.

Питання 2
Чи розумієте, чому вихідна wide-таблиця з місяцями в стовпцях не вважається tidy, і що саме змінюється в її структурі після melt()?


Вихідна wide-таблиця не є tidy, оскільки змінна «місяць» представлена не окремим стовпцем, а назвами 12 різних стовпців. Після застосування melt() усі місяці перетворюються на значення одного стовпця «місяць», а відповідні значення хмарності — на стовпець «хмарність». Тому одна спостережувана величина для кожного місяця займає окремий рядок.

Можна навіть додати:

Було:

місто | Січ | Лют | Бер | ... | Гру

Стало:

місто | місяць | хмарність
Одеса | Січ    | 70
Одеса | Лют    | 66


Питання 3
Чи можете описати, чим формат CSV принципово відрізняється від JSON при зберіганні однієї й тієї самої таблиці?

CSV представляє дані як рядки та стовпці, розділені спеціальним символом, наприклад комою, тому він добре підходить для табличних даних. JSON представляє дані у структурованому вигляді як об'єкти та масиви, де кожне поле має ключ і значення. Тому JSON є більш гнучким для представлення складних структур даних, а CSV є простішим для табличного зберігання.
