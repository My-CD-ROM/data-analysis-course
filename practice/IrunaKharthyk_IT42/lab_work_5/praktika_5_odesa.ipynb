### 4. Що означає накопичена частота на межі третього інтервалу?

Накопичена частота на межі третього інтервалу дорівнює **10**.

\[

Отже, накопичена частота показує, що приблизно **83,3% місяців року** мають кількість опадів до 39 мм включно.


### 3. Чому для кількості опадів потрібна гістограма з інтервалами, а не стовпчикова діаграма за окремими значеннями?

Кількість опадів є кількісною змінною, а значення можуть бути різними або повторюватися лише кілька разів. Якщо побудувати стовпчик для кожного окремого значення, графік для невеликої кількості спостережень буде малокорисним: він показуватиме окремі числа, але гірше демонструватиме загальну форму розподілу.

Гістограма об'єднує близькі значення в інтервали та показує, скільки спостережень потрапляє в кожен інтервал. Тому вона краще підходить для аналізу розподілу кількості опадів.


### 2. Чому «номер місяця» є неоднозначним прикладом щодо шкали вимірювання?

Номер місяця має природний порядок: січень передує лютому, лютий — березню тощо. Тому його можна розглядати як порядкову змінну.

Водночас номери місяців мають рівні кроки між сусідніми місяцями, що може бути аргументом на користь інтервального трактування.

Проте немає природного нульового місяця, який означав би повну відсутність місяця. Крім того, календар є циклічним: грудень переходить у січень. Тому однозначно трактувати номер місяця як звичайну інтервальну шкалу проблематично.

У цій роботі номер місяця трактуємо як **порядкову змінну**, але визнаємо, що можливе й інше обґрунтоване трактування.

### 1. Чому середнє арифметичне некоректне для номінальної змінної «місто», навіть якщо міста можна закодувати числами?

Середнє арифметичне некоректне, тому що числові коди міст не мають кількісного змісту. Наприклад, якщо Одеса має код 1, Київ — 2, а Львів — 3, то число 3 не означає, що Львів у три рази «більший» за Одесу. Так само різниця між кодами 1 і 2 не є змістовною кількісною відстанню.

Якщо змінити кодування, наприклад призначити Одесі 10, Києву 20, а Львову 30, середнє зміниться, хоча самі міста залишаться тими самими. Отже, середнє залежало б від довільного способу кодування, а не від властивостей номінальної змінної.


# Завдання 5. Зв'язок зі шкалою вимірювання

Кількість опадів належить до **відносної шкали**. Вона має природний нуль: 0 мм означає повну відсутність опадів. Крім того, можна змістовно порівнювати як різниці, так і відношення між значеннями.

Саме тому для кількості опадів коректно використовувати всі три міри положення:

- **середнє** — показує середню кількість опадів за місяць;
- **медіана** — показує центральне значення впорядкованого ряду;
- **мода** — показує значення, яке зустрічається найчастіше.

Для змінної «місто», яка має номінальну шкалу, ситуація принципово інша. Міста є категоріями без числового порядку та кількісної відстані між ними. Тому для міста можна визначати лише **моду**.

Якщо, наприклад, закодувати міста числами 1, 2, 3, ці числа будуть лише умовними кодами. Середнє цих кодів не матиме змістовного значення, оскільки зміна коду не означає кількісної зміни самого міста.

Середнє дещо більше за медіану:

\[
33.33 > 31.5

# Завдання 4. Міри положення й квантилі

Для кількості опадів, яка належить до відносної шкали, можна коректно обчислити середнє арифметичне, медіану та моду.


Правила вибору кількості інтервалів

Для \(n=12\):

- за правилом \(\sqrt{n}\): \(\sqrt{12}\approx3.46\), тобто приблизно **3–4 bins**;
- за правилом Стерджеса: \(1+\log_2(12)\approx4.58\), тобто приблизно **5 bins**.

Отже, обидва правила вказують на невелику кількість інтервалів. Це логічно, оскільки маємо лише 12 спостережень. Варіант із 10 bins є надто подрібненим.


```python
sturges = 1 + np.log2(n)

print("Правило Стерджеса:")
print("k =", sturges)
print("Орієнтовна кількість bins:", round(sturges))
```

```python
n = len(rainfall)

sqrt_rule = np.sqrt(n)

print("n =", n)
print("√n =", sqrt_rule)
print("Орієнтовна кількість bins:", round(sqrt_rule))
```

Завдання 3. Гістограма

Гістограма використовується для кількісної змінної «кількість опадів». На відміну від стовпчикової діаграми для окремих категорій, гістограма групує числові значення в інтервали.


 Інтерпретація накопиченої частоти

На межі третього інтервалу накопичена частота дорівнює **10**.

Це означає, що **10 із 12 місяців**, тобто:

\[
\frac{10}{12}\times100\% \approx 83.3\%
\]

місяців року мають кількість опадів **не більшу за 39 мм**.

Отже, приблизно **83,3% місяців** мають до 39 мм опадів включно.


```python

bins = [25, 30, 35, 40, 46]

intervals = pd.cut(
    rainfall,
    bins=bins,
    right=False,
    labels=["25–29", "30–34", "35–39", "40–45"]
)

freq = intervals.value_counts().sort_index()

rel_freq = freq / len(rainfall) * 100

cum_freq = freq.cumsum()

frequency_table = pd.DataFrame({
    "Інтервал, мм": freq.index,
    "Абсолютна частота": freq.values,
    "Відносна частота, %": rel_freq.round(1).values,
    "Накопичена частота": cum_freq.values
})

frequency_table
```

Змінна «номер місяця» є неоднозначним випадком.

З одного боку, місяці мають природний порядок: січень, лютий, березень і так далі. Тому номер місяця можна розглядати як **порядкову шкалу**.

З іншого боку, номери місяців мають рівні календарні кроки: різниця між 2-м і 3-м місяцем така сама за порядковою відстанню, як між 7-м і 8-м. Проте не існує природного «нульового місяця», який означав би повну відсутність місяця. Через це трактування номера місяця як повноцінної інтервальної шкали є дискусійним.

У цій роботі доцільно вважати номер місяця **порядковою змінною**, оскільки основний зміст номера — позиція місяця в календарному порядку.

Для порядкової шкали коректними є **медіана і мода**. Середнє арифметичне для порядкової шкали загалом не є обов'язковою коректною мірою.
