Тому значної асиметрії в розподілі місячної вологості не спостерігається. Розподіл можна вважати близьким до симетричного.

Оскільки середнє і медіана збігаються, середнє є репрезентативною мірою положення для цих даних. Медіана також коректна, але в цьому наборі вона не дає іншої оцінки центрального значення.


### 3. Чому порівнювати абсолютні std двох груп різного масштабу без CV може бути неправильно?

Стандартне відхилення показує абсолютний розкид у тих самих одиницях, що й сама змінна. Тому група з більшим середнім часто може мати і більше абсолютне стандартне відхилення, хоча відносно свого середнього вона може бути навіть стабільнішою.

Коефіцієнт варіації враховує масштаб середнього:

\[
CV=\frac{std}{mean}\times100\%
\]

Тому CV дозволяє порівнювати відносну варіативність різних груп.

Навіть якщо середні значення двох груп дуже близькі, використання CV залишається корисним, оскільки він безпосередньо показує величину стандартного відхилення як частку від середнього.


### 2. Чому заповнення пропуску середнім штучно зменшує std?

Якщо у наборі є пропуск, а замість нього записати середнє значення, нове спостереження буде точно дорівнювати середньому.

Наприклад, маємо значення:

\[
10,\ 20,\ 30
\]

Середнє дорівнює 20. Якщо замість пропуску додати значення 20, то нове спостереження не відхиляється від середнього взагалі.

Тобто воно додає ще одне значення прямо в центр розподілу і не збільшує розкид. У результаті стандартне відхилення стає штучно меншим.

Тому заповнення пропусків середнім може створювати ілюзію меншої варіативності даних.

### 1. Яка міра розсіювання стоїть за кожним із рядків `std`, `25%/75%` і `min/max` у `.describe()`?

Рядок **std** показує стандартне відхилення — міру абсолютного розсіювання значень навколо середнього.

Рядки **25% і 75%** показують перший і третій квартилі. На їх основі можна обчислити міжквартильний розмах:

\[
IQR=Q3-Q1
\]

Рядки **min і max** показують мінімальне та максимальне значення. На їх основі можна обчислити загальний розмах:

\[
R=max-min
\]

Отже, `.describe()` дозволяє одразу побачити основні показники положення та розсіювання.

Коефіцієнт варіації нормує стандартне відхилення на середнє:

\[
CV=\frac{std}{mean}\times100\%
\]

Тому він показує розкид у відносних, а не абсолютних величинах.

У цьому випадку середні значення вологості Одеси та Львова досить близькі, тому різниця між сирими стандартними відхиленнями вже не настільки проблематична. Проте CV все одно є кращим інструментом для коректного порівняння відносної варіативності.
