```python

city_dummies = pd.get_dummies(df["місто_доставки"], prefix="місто")

df = pd.concat([df, city_dummies], axis=1)

print(f"Кількість унікальних міст у даних: {df['місто_доставки'].nunique()}")
print(f"Кількість створених індикаторних стовпців: {city_dummies.shape[1]}")

display(df[["місто_доставки"] + list(city_dummies.columns)].head())
```

```python

df["вік_minmax"] = (df["вік"] - df["вік"].min()) / (df["вік"].max() - df["вік"].min())
df["дохід_minmax"] = (df["дохід"] - df["дохід"].min()) / (df["дохід"].max() - df["дохід"].min())

df["вік_z"] = (df["вік"] - df["вік"].mean()) / df["вік"].std()
df["дохід_z"] = (df["дохід"] - df["дохід"].mean()) / df["дохід"].std()

display(df[["вік", "вік_minmax", "вік_z", "дохід", "дохід_minmax", "дохід_z"]].head())
```

```python
df["вік_cut"] = pd.cut(df["вік"], bins=3)
print("Розподіл через pd.cut (однакова довжина інтервалів):")
print(df["вік_cut"].value_counts())
print("-" * 40)

df["вік_qcut"] = pd.qcut(df["вік"], q=3)
print("Розподіл через pd.qcut (рівна кількість клієнтів):")
print(df["вік_qcut"].value_counts())
```
