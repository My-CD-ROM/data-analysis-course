## Приклад із лекції: вологість і температура за 8 днів

```python
humidity = np.array([42, 55, 61, 38, 47, 66, 59, 44])
print("humidity:", humidity)
print("shape:", humidity.shape)
print("dtype:", humidity.dtype)

# Векторизована операція: відсотки у частки
humidity_frac = humidity / 100
print("Частки вологості:", humidity_frac)

# Булева індексація: дні з вологістю вище 50%
humid_days = humidity[humidity > 50]
print("Вологість вище 50%:", humid_days)

# Індекси найвологішого та найсухішого днів
print("Індекс найвологішого дня:", humidity.argmax())
print("Індекс найсухішого дня:", humidity.argmin())

# Сортування не змінює початковий масив
print("Відсортована вологість:", np.sort(humidity))
print("Початкова вологість:", humidity)

temperature_demo = np.array([18, 20, 19, 22, 21, 17, 23, 20])
data_demo = np.array([humidity, temperature_demo])
print("Середнє по кожному рядку:", data_demo.mean(axis=1))
print("Сума по кожному дню:", data_demo.sum(axis=0))
```

## Вибір власного варіанта

Для цієї роботи обрано **варіант 2 — Львів**. У словнику нижче збережені набори даних усіх варіантів із завдання; для виконання використовується варіант 2.

```python
VARIANT = 2  # Варіант 2 — Львів

variants = {
    1: {"city": "Київ", "temps": [-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2]},
    2: {"city": "Львів", "temps": [-3, -2, 2, 8, 13, 16, 18, 17, 13, 8, 3, -1]},
    3: {"city": "Одеса", "temps": [-1, 0, 4, 10, 16, 20, 23, 22, 17, 11, 5, 1]},
    4: {"city": "Харків", "temps": [-6, -5, 1, 10, 17, 20, 22, 21, 15, 8, 1, -4]},
    5: {"city": "Дніпро", "temps": [-4, -3, 3, 11, 17, 21, 23, 22, 16, 9, 3, -2]},
    6: {"city": "Чернігів", "temps": [-5, -4, 1, 9, 15, 18, 20, 19, 13, 7, 1, -3]},
    7: {"city": "Івано-Франківськ", "temps": [-4, -3, 1, 7, 12, 15, 17, 16, 12, 7, 2, -2]},
    8: {"city": "Полтава", "temps": [-5, -4, 1, 10, 16, 19, 21, 20, 14, 8, 1, -3]},
    9: {"city": "Суми", "temps": [-6, -5, 0, 9, 15, 18, 20, 19, 13, 7, 1, -4]},
    10: {"city": "Ужгород", "temps": [-1, 0, 4, 10, 15, 18, 20, 19, 14, 9, 4, 0]},
}

if VARIANT not in variants:
    raise ValueError("VARIANT має бути цілим числом від 1 до 10")

city = variants[VARIANT]["city"]
print(f"Обраний варіант: {VARIANT} — {city}")
```

## Завдання 1. Створення масивів

```python
months = np.array(["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
                    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"])

temps = np.array(variants[VARIANT]["temps"])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

print(f"Місто: {city}")
print("temps:", temps)
print("temps shape:", temps.shape, "dtype:", temps.dtype)
print("days_in_month:", days_in_month)
print("days_in_month shape:", days_in_month.shape, "dtype:", days_in_month.dtype)
print("month_numbers:", month_numbers)
print("month_numbers shape:", month_numbers.shape, "dtype:", month_numbers.dtype)
```

## Завдання 2. Векторизовані обчислення

```python
# Переведення всіх температур з °C у °F однією векторизованою операцією
temps_f = temps * 9 / 5 + 32
print("Температури у °F:", temps_f)

# Маска вибирає місяці, середня температура яких вища за 15 °C.
# За допомогою тієї самої маски вибираємо кількість днів у цих місяцях.
warm_mask = temps > 15
warm_months = months[warm_mask]
warm_days = days_in_month[warm_mask]
total_warm_days = warm_days.sum()

print("Місяці з температурою вище 15 °C:", warm_months)
print("Кількість днів у цих місяцях:", warm_days)
print("Усього днів у місяцях, де температура вище 15 °C:", total_warm_days)

print("Кількість днів у році:", days_in_month.sum())
print("Перевірка: сума дорівнює 365:", days_in_month.sum() == 365)
```

## Завдання 3. Індексація і фільтрація

```python
warmest_index = temps.argmax()
coldest_index = temps.argmin()

print("Найтепліший місяць:", months[warmest_index])
print("Його номер:", month_numbers[warmest_index])
print("Температура найтеплішого місяця:", temps[warmest_index], "°C")

print("\nНайхолодніший місяць:", months[coldest_index])
print("Його номер:", month_numbers[coldest_index])
print("Температура найхолоднішого місяця:", temps[coldest_index], "°C")

negative_month_numbers = month_numbers[temps < 0]
print("Номери місяців із температурою нижче 0 °C:", negative_month_numbers)

sorted_temps = np.sort(temps)
print("Температури за зростанням:", sorted_temps)
print("Початковий масив temps не змінився:", temps)
```

## Завдання 4. Двовимірний масив і параметр axis

```python
# Умовна кількість годин сонячного сяйва за місяць (навчальні дані)
sunny_hours = np.array([50, 75, 120, 165, 210, 230,
                        245, 220, 165, 110, 60, 45])

weather_data = np.array([temps, days_in_month, sunny_hours])
print("Двовимірний масив:\n", weather_data)
print("Форма масиву:", weather_data.shape)

row_sums = weather_data.sum(axis=1)
row_means = weather_data.mean(axis=1)
print("Сума кожного рядка (axis=1):", row_sums)
print("Середнє кожного рядка (axis=1):", row_means)

column_sums = weather_data.sum(axis=0)
print("Сума кожного стовпця (axis=0):", column_sums)
print("Сума першого стовпця (січень):", column_sums[0])
```

**Пояснення до січня:** перше число в масиві `column_sums` дорівнює сумі трьох значень за січень: температури в °C, кількості днів і умовних годин сонячного сяйва. Це лише арифметична сума значень першого стовпця; вона не є окремою метеорологічною величиною, оскільки додаються різні одиниці вимірювання.

## Завдання 5. reshape і broadcasting

```python
rows = np.array([temps, days_in_month])
correction = np.array([1, 0])

print("Форма rows:", rows.shape)
print("Форма correction:", correction.shape)

# Демонструємо очікувану помилку broadcasting і виводимо її текст.
try:
    invalid_result = rows + correction
except ValueError as error:
    print("Очікувана помилка під час rows + correction:")
    print(type(error).__name__ + ":", error)
else:
    print("Додавання несподівано виконалося:", invalid_result)
```

```python
# Змінюємо форму correction з (2,) на (2, 1)
correction_reshaped = correction.reshape(2, 1)
print("Форма correction після reshape:", correction_reshaped.shape)

corrected_rows = rows + correction_reshaped
print("Результат rows + correction.reshape(2, 1):\n", corrected_rows)
print("Форма результату:", corrected_rows.shape)
```

**Пояснення broadcasting:** у `rows` форма `(2, 12)`, а у `correction` — `(2,)`. NumPy порівнює розміри справа наліво: `12` і `2` не збігаються, а жоден із них не дорівнює `1`, тому форми несумісні. Після `reshape(2, 1)` форма поправок стає `(2, 1)`: розміри `12` і `1` сумісні, а `2` і `2` збігаються. Тому кожна поправка застосовується до всіх 12 елементів відповідного рядка: до температур додається 1, а до кількості днів — 0.

## Письмові відповіді на контрольні питання

**1. Чим відрізняються `shape == (12,)` і `shape == (1, 12)`?**

`(12,)` — це одновимірний масив із 12 елементів. `(1, 12)` — двовимірний масив із одним рядком і 12 стовпцями. Кількість елементів однакова, але кількість вимірів різна: `ndim` дорівнює відповідно 1 і 2. Це може впливати на індексацію та broadcasting.

**2. Чому `temps[temps > 15]` повертає масив іншої довжини, ніж `temps`?**

Вираз `temps > 15` створює булеву маску тієї самої довжини, що й `temps`. Під час індексації NumPy залишає лише елементи, для яких маска дорівнює `True`. Тому результат містить тільки температури вище 15 °C і може бути коротшим за початковий масив.

**3. Що станеться, якщо додати масив форми `(3,)` до масиву форми `(3, 12)`?**

Виникне `ValueError` через несумісні форми. Під час broadcasting форма `(3,)` вирівнюється справа як `(1, 3)`. Останні розміри `12` і `3` не рівні та не дорівнюють 1. Щоб додати по одному значенню до кожного з трьох рядків, вектор потрібно перетворити на форму `(3, 1)`.

**4. Чому `rows + correction` спочатку дає помилку, а `rows + correction.reshape(2, 1)` — ні?**

`rows` має форму `(2, 12)`, а `correction` — `(2,)`. При вирівнюванні справа порівнюються `12` і `2`, тому broadcasting неможливий. Після `reshape(2, 1)` форми стають `(2, 12)` і `(2, 1)`: розміри `2` збігаються, а `12` і `1` сумісні. NumPy повторює кожну поправку вздовж 12 стовпців відповідного рядка.
