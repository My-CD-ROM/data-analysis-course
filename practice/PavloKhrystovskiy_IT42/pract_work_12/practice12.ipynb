## Довірчі інтервали для середнього і частки
**Виконав:** Хрустовський Павло  
**Група:** ІТ-42  
**Варіант:** 6 (Чернігів, липень = 20°C)  
**№ у журналі:** 26


```python
import numpy as np

np.random.seed(6)
ref_temp = 20
july_data = np.random.normal(loc=ref_temp, scale=2.5, size=30)

print("Вибірка температур липня:")
print(np.round(july_data, 2))
print(f"Вибіркове середнє: {july_data.mean():.3f} °C")
print(f"Виправлене стандартне відхилення (s): {july_data.std(ddof=1):.3f} °C")
```

```python
from scipy import stats

n_obs = len(july_data)
mean_val = july_data.mean()
s_val = july_data.std(ddof=1)
se_mean = s_val / np.sqrt(n_obs)

ci_mean_95 = stats.t.interval(0.95, df=n_obs-1, loc=mean_val, scale=se_mean)
print(f"95% довірчий інтервал для середньої температури: ({ci_mean_95[0]:.3f}, {ci_mean_95[1]:.3f})")
```

```python
p_est = (july_data > ref_temp).mean()
cond1 = n_obs * p_est
cond2 = n_obs * (1 - p_est)

print(f"Частка днів із температурою понад {ref_temp}°C: {p_est:.3f}")
print(f"Перевірка умов нормальної апроксимації: n*p={cond1:.1f}, n*(1-p)={cond2:.1f}")

se_prop = np.sqrt(p_est * (1 - p_est) / n_obs)
ci_prop_95 = stats.norm.interval(0.95, loc=p_est, scale=se_prop)
print(f"95% довірчий інтервал для частки: ({ci_prop_95[0]:.3f}, {ci_prop_95[1]:.3f})")
```

```python
conf_levels = [0.90, 0.95, 0.99]
print("Рівень довіри |         Довірчий інтервал         | Ширина інтервалу")
print("-" * 65)
for cl in conf_levels:
    ci_curr = stats.t.interval(cl, df=n_obs-1, loc=mean_val, scale=se_mean)
    w = ci_curr[1] - ci_curr[0]
    print(f"    {int(cl*100)}%      | ({ci_curr[0]:.3f}, {ci_curr[1]:.3f}) |      {w:.3f}")
```

### Інтерпретація отриманого довірчого інтервалу
95% довірчий інтервал для середнього означає, що при багаторазовому незалежному повторенні процедури формування вибірки розміром $n=30$ та розрахунку інтервалів за цим алгоритмом у 95 випадках зі 100 побудовані інтервали накриватимуть істинне значення генерального середнього $\mu$. Для одного конкретного інтервалу ми маємо 95% статистичну впевненість у тому, що параметр потрапив у цей діапазон.


### Відповіді на контрольні запитання

1. **Чим відрізняється точкова оцінка від інтервальної?**  
   Точкова оцінка — це одне фіксоване число (наприклад, $\bar{x}$), яке є найкращим наближенням параметра, але не дає інформації про можливу похибку вимірювання. Довірчий інтервал задає цілий діапазон значень разом із рівнем імовірності, враховуючи випадкову невизначеність вибірки.

2. **Чому при невідомому $\sigma$ та помірному $n$ використовують розподіл Стьюдента?**  
   Оскільки дисперсія генеральної сукупності невідома, її доводиться оцінювати за вибірковим відхиленням $s$. Це вносить додаткову випадковість у знаменник дробу для статистики. Розподіл Стьюдента має "важчі хвости" порівняно з нормальним, що компенсує цю невизначеність і робить межі інтервалу більш надійними.

3. **Що означає поняття "95% рівень довіри"?**  
   Це характеристика надійності самого статистичного методу: процедура побудови інтервалу гарантує покриття істинного параметра у 95% усіх можливих реалізацій вибірок заданого обсягу.

4. **Які чинники впливають на ширину довірчого інтервалу?**  
   - **Рівень довіри:** чим вища довіра (наприклад, 99% проти 90%), тим ширший інтервал.
   - **Розкид даних ($s$):** чим більша варіативність ознаки, тим ширший інтервал.
   - **Обсяг вибірки ($n$):** чим більше спостережень, тим менша стандартна похибка і вужчий інтервал.
