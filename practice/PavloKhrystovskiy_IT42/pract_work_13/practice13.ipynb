**Варіант:** 6 (Чернігів, середня температура липня 20°C, стара норма μ₀ = 19.4°C)  
**№ у журналі:** 26


```python
import numpy as np

np.random.seed(6)
t_data = np.random.normal(loc=20, scale=2.5, size=30)

mean_obs = t_data.mean()
s_obs = t_data.std(ddof=1)

print(f"Вибіркове середнє x̄ = {mean_obs:.3f} °C")
print(f"Стандартне відхилення s = {s_obs:.3f} °C")
```

```python
from scipy import stats

n_sample = 30
mu_null = 19.4
se_val = s_obs / np.sqrt(n_sample)
z_stat = (mean_obs - mu_null) / se_val

p_val = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print(f"Стандартна похибка SE = {se_val:.4f}")
print(f"Значення z-статистики = {z_stat:.3f}")
print(f"Розраховане p-value = {p_val:.4f}")
```

```python
alpha_level = 0.05
reject_h0 = p_val < alpha_level

print(f"Рівень значущості alpha = {alpha_level}")
print(f"Статистичне рішення: {'Відхиляємо H0' if reject_h0 else 'Немає підстав відхилити H0'}")
print(f"Висновок: Дані {'свідчать' if reject_h0 else 'не дають достатніх підстав вважати'}, що середньомісячна температура липня у Чернігові статистично відрізняється від кліматичної норми {mu_null}°C (p = {p_val:.4f}).")
```

```python
ci_l, ci_u = mean_obs - 1.96 * se_val, mean_obs + 1.96 * se_val
contains_null = ci_l <= mu_null <= ci_u

print(f"95% довірчий інтервал для середнього: ({ci_l:.3f}, {ci_u:.3f})")
print(f"Чи накриває інтервал нульове значення {mu_null}°C? {'Так' if contains_null else 'Ні'}")
```

### Відповіді на контрольні запитання

1. **Що являє собою показник p-value?**  
   $p$-value — це ймовірність отримати таке ж або ще більш екстремальне значення тестової статистики порівняно зі спостережуваним за умови, що нульова гіпотеза $H_0$ є повністю правильною. Воно не вимірює ймовірність того, що гіпотеза $H_0$ або $H_1$ правильна.

2. **Чому вибір спрямованості тесту (однобічний чи двобічний) має бути зроблений до аналізу даних?**  
   Вибір альтернативи після огляду вибіркових даних створює систематичне зміщення (p-hacking / data dredging). Дослідник може підігнати тип критерію під отриманий знак відхилення, що штучно вдвічі знижує номінальний рівень $p$-value і спотворює реальну ймовірність помилки I роду.

3. **Що означають помилки I та II роду в контексті цієї задачі?**  
   - **Помилка I роду ($lpha$):** зробити висновок про те, що кліматична норма змінилася, коли насправді зміни немає (хибна тривога).
   - **Помилка II роду ($eta$):** не помітити реальної зміни кліматичної норми через випадкові коливання вибірки.

4. **Чи завжди висновок перевірки гіпотези за $p$-value узгоджується з аналізом довірчого інтервалу?**  
   Так, для двобічного критерію на рівні $lpha=0.05$ і 95% довірчого інтервалу висновки строго дуальні: якщо $p \ge 0.05$, значення $\mu_0$ обов'язково потрапляє всередину довірчого інтервалу; якщо $p < 0.05$, значення $\mu_0$ залишається поза межами інтервалу.
