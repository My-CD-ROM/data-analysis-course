```python
import numpy as np
from scipy import stats

morning_humidity = np.array([85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86])
norm_annual = 75

amp_correction = np.array([3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3])
evening_humidity = morning_humidity - amp_correction

print("Ранкові спостереження вологості:", morning_humidity)
print("Вечірні розрахункові значення:", evening_humidity)
```

```python
t_stat, p_val = stats.ttest_1samp(morning_humidity, popmean=norm_annual)
print(f"Одновибірковий t-критерій: t = {t_stat:.3f}, p-value = {p_val:.4f}")

if p_val < 0.05:
    print("Висновок: Відхиляємо H0 — середня вологість статистично відрізняється від норми.")
else:
    print("Висновок: Не відхиляємо H0 — підстав стверджувати про відхилення від норми немає.")
```

```python
n_obs = len(morning_humidity)
mean_m = morning_humidity.mean()
std_m = morning_humidity.std(ddof=1)

t_manual = (mean_m - norm_annual) / (std_m / np.sqrt(n_obs))
print(f"Аналітичний розрахунок t-статистики: {t_manual:.3f}")
```

```python
lviv_humidity = np.array([86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87])

stat_lev, p_lev = stats.levene(morning_humidity, lviv_humidity)
print(f"Тест Левена (Чернігів vs Львів): W = {stat_lev:.3f}, p = {p_lev:.4f}")

equal_var = p_lev >= 0.05
t_ind, p_ind = stats.ttest_ind(morning_humidity, lviv_humidity, equal_var=equal_var)
print(f"Двовибірковий t-критерій: t = {t_ind:.3f}, p = {p_ind:.4f}")

if p_ind < 0.05:
    print("Висновок: Виявлено статистично значущу різницю середньої вологості між Черніговом і Львовом.")
else:
    print("Висновок: Відсутні статистичні підстави вважати середні значення різними.")
```

```python
poltava_humidity = np.array([84, 81, 76, 69, 63, 61, 59, 61, 67, 74, 80, 85])

stat_lev2, p_lev2 = stats.levene(morning_humidity, poltava_humidity)
print(f"Тест Левена (Чернігів vs Полтава): W = {stat_lev2:.3f}, p = {p_lev2:.4f}")

equal_var2 = p_lev2 >= 0.05
t_ind2, p_ind2 = stats.ttest_ind(morning_humidity, poltava_humidity, equal_var=equal_var2)
print(f"Двовибірковий t-критерій: t = {t_ind2:.3f}, p = {p_ind2:.4f}")

if equal_var2:
    print("Обґрунтування: Дисперсії однорідні, застосовано стандартний t-критерій Стьюдента.")
else:
    print("Обґрунтування: Дисперсії гетерогенні, застосовано критерій Велча.")
```

```python
t_rel, p_rel = stats.ttest_rel(morning_humidity, evening_humidity)
print(f"Парний t-критерій (ранок vs вечір): t = {t_rel:.3f}, p = {p_rel:.4f}")

if p_rel < 0.05:
    print("Висновок: Добові коливання вологості є статистично значущими.")
else:
    print("Висновок: Відсутні докази добових змін вологості.")
```

### Відповіді на контрольні запитання

1. **Чому використовується саме t-розподіл Стьюдента, а не нормальний z-розподіл?**  
   Оскільки точне значення генерального стандартного відхилення $\sigma$ невідоме, замість нього застосовується вибіркова оцінка $s$. Це збільшує невизначеність розрахунків, яку моделює $t$-розподіл зі збереженням точного контролю за рівнем похибки першого роду.

2. **Чому критично враховувати рівність дисперсій (параметр equal_var)?**  
   Класичний $t$-тест Стьюдента спирається на припущення рівності дисперсій. У разі істотної різниці дисперсій вибірок імовірність хибного відхилення $H_0$ суттєво перевищує номінальний рівень $\alpha$. Використання поправки Велча (`equal_var=False`) коригує ефективну кількість ступенів вільності і зберігає коректність висновку.

3. **У чому полягає перевага парного критерію (paired t-test)?**  
   Парний критерій аналізує індивідуальні різниці $d_i = x_{1i} - x_{2i}$ для кожного спостереження, повністю виключаючи міжособистісну або міжсезонну варіацію базового рівня. Це суттєво знижує дисперсію похибки і помітно підвищує статистичну потужність критерію.

4. **Чи узгоджується висновок тесту Левена з оцінкою вибіркових std?**  
   Так, висновок цілком корелює: якщо вибіркові стандартні відхилення майже ідентичні, тест Левена дає високе $p$-value ($p > 0.05$), свідчачи про гомогенність дисперсій.
