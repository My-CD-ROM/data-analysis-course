```python
import pandas as pd

# Варіант 8 — Полтава
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

avg_temp = [38, 36, 34, 36, 46, 56, 60, 48, 36, 36, 44, 44]

# Навчальні дані про середньомісячні опади, мм
precip = [40, 35, 38, 45, 55, 70, 65, 55, 45, 40, 48, 42]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip
})

print("Варіант 8 — Полтава")
display(df)

print("Перші рядки:")
display(df.head())

print("Розмір DataFrame:", df.shape)

print("Типи даних:")
print(df.dtypes)
```

```python
print("DataFrame з індексом за замовчуванням:")
display(df)

print("Індекс:")
print(df.index)
```

```python
df_by_month = df.set_index("month")

print("DataFrame з індексом month:")
display(df_by_month)

print("Рядок для місяця Чер:")
display(df_by_month.loc["Чер"])
```

```python
df = df_by_month.reset_index()

print("Повернули звичайний індекс:")
display(df)
```

```python
df.index = range(1, 13)

print("DataFrame з індексом від 1 до 12:")
display(df)
```

```python
print("avg_temp для місяців 1–6 через .loc:")
display(df.loc[1:6, "avg_temp"])
```

```python
print("avg_temp для позицій 0–5 через .iloc:")
display(df.iloc[0:6, 1])
```

.loc працює з підписами індексу.
.iloc працює з позиціями рядків.

У нашому випадку індекс 1–12, тому:
.loc[1:6] вибирає рядки з підписами 1, 2, 3, 4, 5, 6.
.iloc[0:6] вибирає перші шість рядків — позиції 0, 1, 2, 3, 4, 5.

Тому результати можуть відрізнятися, якщо індекс не збігається
з позицією рядка.

```python
print("Стовпці month і precip:")
display(df[["month", "precip"]])
```

```python
print("Місяці, де avg_temp > 15:")
display(df.loc[df["avg_temp"] > 15, ["month", "avg_temp"]])
```

```python
print("Місяці, де avg_temp > 10 і precip > 50:")

result = df[
    (df["avg_temp"] > 10) &
    (df["precip"] > 50)
]

display(result[["month", "avg_temp", "precip"]])
```

```python
print("Місяці, де avg_temp < 0 або precip > 80:")

result = df[
    (df["avg_temp"] < 0) |
    (df["precip"] > 80)
]

display(result[["month", "avg_temp", "precip"]])
```

```python
df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32

print("Температура у Фаренгейтах:")
display(df)
```

```python
mean_precip = df["precip"].mean()

print("Середнє значення precip:", mean_precip)
```

```python
df["is_wet"] = df["precip"] > mean_precip

print("Стовпець is_wet:")
display(df[["month", "precip", "is_wet"]])
```

```python
def get_season(month_number):
    if month_number in [12, 1, 2]:
        return "зима"
    elif month_number in [3, 4, 5]:
        return "весна"
    elif month_number in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"


df["season"] = df.index.map(get_season)

print("Сезони:")
display(df[["month", "season"]])
```

## Письмові відповіді

### 1. Чим відрізняються `.loc[1:6]` і `.iloc[0:6]`?

`.loc` вибирає рядки за підписами індексу, а `.iloc` — за їхніми позиціями. 
Тому `.loc[1:6]` залежить від значень індексу, а `.iloc[0:6]` вибирає перші шість позицій. 
Якщо індекс не збігається з позицією рядка, ці команди можуть повернути різні рядки або різну кількість рядків.

### 2. Чому `and` викликає помилку?

У pandas порівняння `df["avg_temp"] > 10` повертає цілий Series значень True/False. 
Оператор `and` призначений для роботи з одиничними логічними значеннями, тому він не може правильно обробити два Series. 
Для поелементного логічного І в pandas використовується `&`.

Правильно:

`(df["avg_temp"] > 10) & (df["precip"] > 50)`

Для логічного АБО використовується `|`.

### 3. Чим відрізняється формула від `.apply()`?

Обчислюваний стовпець через формулу виконує однакову математичну операцію над усіма значеннями стовпця. 
Наприклад, температура у Фаренгейтах обчислюється векторизованою формулою.

`.apply()` дозволяє застосувати власну функцію до кожного значення. 
Це зручно для складнішої логіки, наприклад визначення сезону за номером місяця.
