**Задача:** обчислити методом Монте-Карло

\[
\int_0^4 \sqrt{x}\,dx
\]

**Точне значення:** \(16/3 \approx 5.3333\).

Мета роботи — оцінити інтеграл методом Монте-Карло, порівняти результат з точним значенням, дослідити збіжність оцінки та пояснити зв'язок із законом великих чисел і ЦГТ.

## Завдання 1. Оцінка методом Монте-Карло для власного варіанта

Для інтеграла генеруємо `N = 1000` значень `x` рівномірно на `[0, 4]`. Оцінка інтеграла дорівнює довжині відрізка, помноженій на середнє значення `sqrt(x)`.

```python
import numpy as np

np.random.seed(8)

N = 1000
x = np.random.uniform(0, 4, size=N)
estimate = 4 * np.mean(np.sqrt(x))

print("N =", N)
print("Оцінка інтеграла =", estimate)
```

## Завдання 2. Порівняння з точним значенням

Точне аналітичне значення: `16/3`. Абсолютна похибка обчислюється як `|оцінка - точне значення|`.

```python
exact = 16 / 3
absolute_error = abs(estimate - exact)
relative_error_percent = absolute_error / exact * 100

print("Точне значення =", exact)
print("Оцінка =", estimate)
print("Абсолютна похибка =", absolute_error)
print("Відносна похибка, % =", relative_error_percent)
```

**Пояснення:** при `N = 1000` отримана оцінка близька до точного значення `5.3333`. Абсолютна похибка становить приблизно `0.0219`, або близько `0.41%` від точного значення. Для відносно невеликої кількості випробувань це невелика похибка, хоча через випадковість вона не повинна обов'язково монотонно зменшуватися при кожному збільшенні `N`.

## Завдання 3. Збіжність за кількістю випробувань

Перед кожним запуском знову фіксуємо `np.random.seed(8)`, як вимагає умова практичної роботи.

```python
Ns = [100, 1000, 10000, 100000]
results = []

for N in Ns:
    np.random.seed(8)
    x = np.random.uniform(0, 4, size=N)
    estimate_N = 4 * np.mean(np.sqrt(x))
    error_N = abs(estimate_N - exact)
    results.append((N, estimate_N, error_N))

print(f"{'N':>10} {'Оцінка':>15} {'Абсолютна похибка':>22}")
for N, estimate_N, error_N in results:
    print(f"{N:>10} {estimate_N:>15.6f} {error_N:>22.6f}")
```

### Результат

| N | Оцінка | Абсолютна похибка |
|---:|---:|---:|
| 100 | 5.378969 | 0.045636 |
| 1 000 | 5.355249 | 0.021916 |
| 10 000 | 5.309036 | 0.024298 |
| 100 000 | 5.325201 | 0.008132 |

При переході від `N = 100` до `N = 10 000` кількість випробувань зросла у 100 разів, а похибка зменшилася приблизно у 1.88 раза. У конкретній симуляції вона не зобов'язана зменшуватися монотонно: для `N = 10 000` похибка трохи більша, ніж для `N = 1 000`. Теоретично характерний масштаб випадкової похибки змінюється як `1/√N`, тому збільшення `N` у 100 разів дає очікуване зменшення характерної похибки приблизно у 10 разів, а не у 100 разів. Окремі запуски можуть відхилятися від цього співвідношення через випадковість.

## Завдання 4. Графік похибки залежно від N

```python
import matplotlib.pyplot as plt

N_values = [row[0] for row in results]
errors = [row[2] for row in results]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(N_values, errors, marker="o")
ax.set_xscale("log")
ax.set_xlabel("Кількість випробувань N")
ax.set_ylabel("Абсолютна похибка")
ax.set_title("Залежність абсолютної похибки від N")
ax.grid(True)
plt.show()
```

**Пояснення графіка:** загалом зі збільшенням `N` похибка має тенденцію до зменшення, але через випадковість симуляції графік не є строго монотонним. Форма спадання відповідає очікуваній поведінці порядку `1/√N`: для дуже великих `N` характерна похибка стає меншою.

## Завдання 5. Зв'язок із законом великих чисел і ЦГТ

**1. Який факт гарантує наближення до точного значення?**

Основою є **закон великих чисел**. Оцінка інтеграла є вибірковим середнім значень `sqrt(x)`, помноженим на довжину відрізка. За збільшення кількості незалежних випробувань вибіркове середнє наближається до математичного сподівання, тому оцінка інтеграла наближається до точного значення.

**2. Чому стандартна похибка має вигляд `σ/√n`?**

Центральна гранична теорема описує масштаб випадкових коливань вибіркового середнього: стандартна похибка пропорційна `1/√n`. Тому при збільшенні кількості випробувань у 100 разів характерна похибка зменшується приблизно у `√100 = 10` разів. Це не окрема властивість Монте-Карло, а пряме застосування властивостей вибіркового середнього.

## Контрольні питання

**Чому оцінка методом Монте-Карло — це вибіркове середнє?**  
Для інтеграла випадково генерується багато значень `x`, обчислюється `sqrt(x)` у кожній точці, а потім береться середнє. Множення цього середнього на довжину відрізка `[0, 4]` дає оцінку інтеграла.

**Чому вчетверо більша кількість випробувань дає приблизно вдвічі меншу похибку?**  
Тому що у формулі стандартної похибки стоїть корінь: `σ/√n`. Якщо `n` збільшити у 4 рази, то `√n` збільшиться у 2 рази, тому характерна похибка зменшиться приблизно удвічі.

**Висновок:**  
У практичній роботі для варіанта 8 методом Монте-Карло оцінено інтеграл `∫₀⁴ √x dx`. При `N = 1000` отримано оцінку приблизно `5.35525` при точному значенні `5.33333`. Дослідження для різних `N` показало загальну тенденцію зменшення похибки зі зростанням кількості випробувань та узгоджується з теоретичним порядком `1/√N`.
