```python
import pandas as pd

# Практика 3. Tidy Data
# Варіант 8 — Полтава

wide = pd.DataFrame({
    "місто": ["Полтава"],
    "Січ": [77],
    "Лют": [73],
    "Бер": [66],
    "Кві": [56],
    "Тра": [47],
    "Чер": [41],
    "Лип": [37],
    "Сер": [39],
    "Вер": [49],
    "Жов": [63],
    "Лис": [75],
    "Гру": [81]
})

print("Wide-таблиця, варіант 8 — Полтава:")
display(wide)

print("Розмір таблиці:", wide.shape)
```

```python
months = [
    "Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"
]

tidy = wide.melt(
    id_vars="місто",
    value_vars=months,
    var_name="місяць",
    value_name="хмарність"
)

print("Tidy-таблиця:")
display(tidy)

print("Розмір tidy-таблиці:", tidy.shape)
```

## Завдання 3. Класифікація змінних

### Місто — номінальна шкала

Стовпець "місто" є номінальним, тому що назви міст є категоріями без природного порядку.

### Місяць — номінальна шкала

У цій таблиці місяць використовується як назва категорії, а не як числова величина, тому для цього набору даних його можна розглядати як номінальну змінну.

### Хмарність — шкала відношень

Стовпець "хмарність" є шкалою відношень, оскільки значення задані у відсотках, мають природний нуль, а співвідношення між значеннями має зміст: 80% хмарності вдвічі більше за 40%.

```python
back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

print("Wide-таблиця після pivot:")
display(back)
```

```python
# Переставляємо стовпці в початковому порядку
back = back[months]

# Порівнюємо значення
original_values = wide[months].reset_index(drop=True)
back_values = back.reset_index(drop=True)

print("Чи збігаються значення?")
print(original_values.equals(back_values))
```

```python
tidy.to_csv(
    "variant.csv",
    index=False
)

tidy.to_json(
    "variant.json",
    orient="records",
    force_ascii=False
)

print("Файли variant.csv та variant.json створено.")
```

```python
reloaded_csv = pd.read_csv("variant.csv")

print("Дані з CSV:")
display(reloaded_csv)
```

```python
reloaded_json = pd.read_json("variant.json")

print("Дані з JSON:")
display(reloaded_json)
```

```python
print(
    "CSV збігається:",
    tidy.equals(reloaded_csv)
)

print(
    "JSON збігається:",
    tidy.equals(reloaded_json)
)
```

```python
# Письмові відповіді

## 1. Чому "місто" — номінальна шкала, а "хмарність" — шкала відношень?

"Місто" є номінальною шкалою, тому що це назва категорії, яка не має числового порядку або значення.

"Хмарність" є шкалою відношень, оскільки вона вимірюється у відсотках і має природний нуль. Тому можна порівнювати співвідношення значень, наприклад 80% — це вдвічі більше за 40%.

## 2. Чому wide-таблиця з місяцями у стовпцях не є tidy?

Wide-таблиця не відповідає принципу tidy data, тому що змінна "місяць" представлена назвами окремих стовпців. Після застосування `melt()` всі місяці стають значеннями одного стовпця "місяць", а значення хмарності — стовпцем "хмарність". У результаті кожен рядок описує одне спостереження.

## 3. Чим CSV відрізняється від JSON?

CSV зберігає таблицю у вигляді рядків і стовпців, де значення зазвичай розділені комами. Це простий табличний формат.

JSON зберігає дані як структуровані об'єкти та масиви. Він краще підходить для передачі структурованих даних між програмами та може зберігати складнішу структуру.
```
