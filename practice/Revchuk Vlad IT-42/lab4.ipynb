```python
import numpy as np
import pandas as pd

# Практика 4. Чищення набору даних
# Варіант 8 — Полтава
# Базова ціна: 700 грн
# Базова кількість: 4

dirty = pd.DataFrame([
    {"замовлення": 1, "місто": "Полтава",    "товар": "Товар 8", "ціна": 700,        "кількість": 4},
    {"замовлення": 2, "місто": " Полтава ",  "товар": "Товар 8", "ціна": "700 грн",   "кількість": np.nan},
    {"замовлення": 3, "місто": "ПОЛТАВА",    "товар": "Товар 8", "ціна": 700,        "кількість": 5},
    {"замовлення": 4, "місто": "Полтава",    "товар": "Товар 8", "ціна": "700 грн",   "кількість": 3},
    {"замовлення": 5, "місто": " Полтава ",  "товар": "Товар 8", "ціна": 7000,       "кількість": 6},
    {"замовлення": 6, "місто": "ПОЛТАВА",    "товар": "Товар 8", "ціна": 700,        "кількість": np.nan},
    {"замовлення": 7, "місто": "Полтава",    "товар": "Товар 8", "ціна": 700,        "кількість": 2},
    {"замовлення": 8, "місто": "Полтава",    "товар": "Товар 8", "ціна": 700,        "кількість": 4},
])

# Додаємо повний дублікат рядка 8
dirty = pd.concat([dirty, dirty.iloc[[7]]], ignore_index=True)

print("Брудний набір даних:")
display(dirty)

print("Розмір:", dirty.shape)
```

```python
print("Кількість пропусків у кожному стовпці:")
print(dirty.isna().sum())
```

```python
median_quantity = dirty["кількість"].median()

print("Медіана кількості:", median_quantity)

dirty["кількість"] = dirty["кількість"].fillna(median_quantity)

print("Після заповнення пропусків:")
display(dirty)
```

```python
print("Кількість пропусків після очищення:")
print(dirty.isna().sum())
```

```python
print("Повні дублікати:")
display(dirty[dirty.duplicated()])
```

```python
print("Повні дублікати:")
display(dirty[dirty.duplicated()])
```

```python
print("Дублікати за основними полями:")

duplicates_subset = dirty[
    dirty.duplicated(
        subset=["місто", "товар", "ціна", "кількість"],
        keep=False
    )
]

display(duplicates_subset)
```

```python
dirty = dirty.drop_duplicates()

print("Після видалення дубліката:")
display(dirty)

print("Кількість рядків:", len(dirty))
```

```python
print(dirty.dtypes)
```

```python
dirty["ціна"] = (
    dirty["ціна"]
    .astype(str)
    .str.replace(" грн", "", regex=False)
    .astype(float)
)

print("Тип ціни після очищення:")
print(dirty["ціна"].dtype)
```

```python
dirty["місто"] = (
    dirty["місто"]
    .str.strip()
    .str.lower()
    .replace({
        "полтава": "Полтава"
    })
)

print("Унікальні значення міста:")
print(dirty["місто"].unique())
```

```python
q1 = dirty["ціна"].quantile(0.25)
q3 = dirty["ціна"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Нижня межа =", lower)
print("Верхня межа =", upper)
```

```python
outliers = dirty[
    (dirty["ціна"] < lower) |
    (dirty["ціна"] > upper)
]

print("Значення поза межами IQR:")
display(outliers)
```

```python
dirty_clean = dirty[
    (dirty["ціна"] >= lower) &
    (dirty["ціна"] <= upper)
].copy()

print("Очищений набір:")
display(dirty_clean)
```

```python
print("Кількість рядків після очищення:", len(dirty_clean))
print("Унікальні міста:", dirty_clean["місто"].unique())
print("Пропуски:")
print(dirty_clean.isna().sum())
print("Тип ціни:", dirty_clean["ціна"].dtype)
```

```python
print("=== ФІНАЛЬНИЙ ОЧИЩЕНИЙ НАБІР ===")

display(dirty_clean)

print("Форма:", dirty_clean.shape)
print("Типи даних:")
print(dirty_clean.dtypes)
```
