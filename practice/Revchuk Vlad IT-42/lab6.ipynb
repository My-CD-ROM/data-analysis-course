```python
import pandas as pd
import numpy as np

# Варіант 8 — Полтава
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

poltava = pd.Series(
    [84, 81, 76, 69, 63, 61, 59, 61, 67, 74, 80, 85],
    index=months,
    name="Вологість"
)

poltava
```

```python
description = poltava.describe()
description
```

```python
s = poltava

# Середнє
mean_value = s.mean()

# Мінімум і максимум
min_value = s.min()
max_value = s.max()

# Розмах
rozmakh = max_value - min_value

# Дисперсія
variance_ddof1 = s.var(ddof=1)
variance_ddof0 = s.var(ddof=0)

# Стандартне відхилення
std_value = s.std(ddof=1)

# Квартилі
q1 = s.quantile(0.25)
q3 = s.quantile(0.75)

# IQR
iqr = q3 - q1

# Коефіцієнт варіації
cv = std_value / mean_value * 100

print("Середнє:", mean_value)
print("Мінімум:", min_value)
print("Максимум:", max_value)
print("Розмах:", rozmakh)
print("Дисперсія ddof=1:", variance_ddof1)
print("Дисперсія ddof=0:", variance_ddof0)
print("Стандартне відхилення:", std_value)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Коефіцієнт варіації:", cv, "%")
```

```python
print("Q1 з describe():", description["25%"])
print("Q1 вручну:", q1)

print("Q3 з describe():", description["75%"])
print("Q3 вручну:", q3)

print("Q1 збігається:", description["25%"] == q1)
print("Q3 збігається:", description["75%"] == q3)
```

Додаткові 4 питання, які треба здати
1. Що означають std, 25%/75%, min/max?

std — стандартне відхилення, яке характеризує розсіювання значень відносно середнього. 25% — перший квартиль Q1, а 75% — третій квартиль Q3. Значення min і max показують відповідно найменше та найбільше спостереження. Різниця між Q3 і Q1 утворює міжквартильний розмах IQR.

2. Чому заповнення пропуску середнім зменшує std?

Якщо пропущене значення замінити середнім, нове значення знаходиться точно в центрі розподілу і не збільшує відхилення від середнього. Тому після такого заповнення стандартне відхилення може штучно зменшитися. Наприклад, для значень 10, 20, 30 середнє дорівнює 20. Якщо замість пропуску додати 20, нове значення не створює додаткового розкиду.

3. Чому для порівняння краще використовувати CV?

Стандартне відхилення показує абсолютний розкид і залежить від масштабу даних. Тому дві групи з різними середніми можуть мати різні std навіть при подібній відносній варіативності. Коефіцієнт варіації нормує стандартне відхилення на середнє і виражає його у відсотках, тому він краще підходить для порівняння варіативності.

4. Чи симетричний розподіл Полтави?

Розподіл вологості Полтави не є ідеально симетричним. Середнє значення приблизно 71,67%, а медіана — 68%, тому між ними є помітна різниця. Це свідчить про невелику асиметрію. Для цих даних середнє залишається репрезентативним, оскільки немає різко виражених викидів, але медіана також добре характеризує центральне значення.
