```python
# Практика 13. Точкові оцінки та перевірка гіпотез
# Коць Артем, ІТ-42
# Варіант 9 — Суми

import numpy as np
import pandas as pd
from scipy import stats

print("Місто: Суми")
print("Варіант: 9")
print("Липнева температура для генерації: 20 °C")
print("Стара норма μ0: 19.1 °C")
print("Розмір вибірки: 30")
print("Рівень значущості α: 0.05")
```

```python
np.random.seed(9)

daily_temps = np.random.normal(
    loc=20,
    scale=2.5,
    size=30
)

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)

print("Вибірка daily_temps:")
print(daily_temps)

print("\nКількість спостережень:", len(daily_temps))
print("Вибіркове середнє x̄ =", x_bar)
print("Вибіркове стандартне відхилення s =", s)
```

### Завдання 1. Вибірка і точкові оцінки

Для варіанта 9 було згенеровано вибірку з 30 денних температур
міста Суми навколо липневого значення 20 °C.

Вибіркове середнє x̄ є точковою оцінкою теоретичного середнього
температури, а вибіркове стандартне відхилення s — точковою оцінкою
невідомого стандартного відхилення генеральної сукупності.

### Завдання 2. Формулювання H0 і H1

Стара норма для міста Суми становить:

μ₀ = 19.1 °C.

Нульова гіпотеза:

H₀: μ = 19.1 °C.

Альтернативна гіпотеза:

H₁: μ ≠ 19.1 °C.

Обираємо двобічний тест, оскільки нас цікавить будь-яке відхилення
середньої температури від старої норми — як підвищення, так і
зниження.

Тип тесту визначено до обчислення p-value, тому рішення не залежить
від того, у який бік випадково відхилилося вибіркове середнє.

```python
mu0 = 19.1
alpha = 0.05

se = s / np.sqrt(30)

z = (x_bar - mu0) / se

p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("μ0 =", mu0)
print("x̄ =", x_bar)
print("s =", s)
print("SE =", se)
print("z =", z)
print("p-value =", p_value)
```

```python
if p_value < alpha:
    decision = "Відхиляємо H₀"
else:
    decision = "Не відхиляємо H₀"

print("α =", alpha)
print("p-value =", p_value)
print("Рішення:", decision)
```

### Завдання 4. Рішення при α = 0.05 і висновок

При рівні значущості α = 0.05 порівнюємо отримане p-value
з α.

Якщо p-value < 0.05, нульову гіпотезу H₀ відхиляємо. Це означає,
що наявні статистичні підстави вважати, що середня липнева
температура в Сумах відрізняється від старої норми 19.1 °C.

Якщо p-value ≥ 0.05, нульову гіпотезу H₀ не відхиляємо. У такому
випадку наявних даних недостатньо, щоб стверджувати, що середня
липнева температура відрізняється від старої норми 19.1 °C.

Отриманий результат потрібно інтерпретувати саме через p-value:
це не ймовірність того, що H₀ правильна, а міра того, наскільки
сумісними з H₀ є отримані дані.

```python
ci_lower = x_bar - 1.96 * se
ci_upper = x_bar + 1.96 * se

print("95% довірчий інтервал для середнього:")
print(f"({ci_lower:.4f}, {ci_upper:.4f})")

print("Стара норма μ0 =", mu0)
print("Чи потрапляє μ0 в інтервал?", ci_lower <= mu0 <= ci_upper)
```

### Завдання 5. Довірчий інтервал і помилки I та II роду

95% довірчий інтервал для середнього побудовано за формулою:

x̄ ± 1.96 · SE.

Стара норма μ₀ = 19.1 °C перевіряється на належність до цього
довірчого інтервалу.

Для двобічного тесту при α = 0.05 результат тесту та перевірка
довірчого інтервалу повинні узгоджуватися: якщо μ₀ не входить
до 95% інтервалу, H₀ відхиляється; якщо μ₀ входить до інтервалу,
H₀ не відхиляється.

Помилка I роду для Сум означає відхилити H₀ та зробити висновок,
що середня липнева температура відрізняється від 19.1 °C,
хоча насправді стара норма є правильною.

Помилка II роду означає не відхилити H₀, хоча справжня середня
липнева температура вже відрізняється від 19.1 °C.

Для кліматологів помилка II роду може бути суттєвою, якщо важливо
вчасно виявити реальну зміну клімату та не використовувати
застарілу норму. Водночас помилка I роду може призвести
до передчасного оновлення норми.

## Контрольні питання

### 1. Що означає отримане p-value?

p-value — це ймовірність, за умови істинності H₀, отримати результат
із таким самим або більшим відхиленням від μ₀, ніж спостерігається
у вибірці, відповідно до обраного двобічного тесту.

p-value не є ймовірністю того, що H₀ правильна. Воно характеризує,
наскільки спостережені дані узгоджуються з нульовою гіпотезою.

### 2. Чому тип тесту потрібно обирати до перегляду результату?

Тип тесту визначає область екстремальних результатів і спосіб
обчислення p-value. Якщо обирати однобічний або двобічний тест
після перегляду результату, можна штучно підібрати вигідніший тест
і збільшити ризик помилкового відхилення H₀.

У цій роботі обрано двобічний тест, оскільки перевіряється будь-яке
відхилення середньої температури від старої норми 19.1 °C.

### 3. Що означають помилки I та II роду для Сум?

Помилка I роду — відхилити H₀ і вирішити, що середня липнева
температура Сум відрізняється від 19.1 °C, хоча насправді стара
норма правильна.

Помилка II роду — не відхилити H₀, хоча справжня середня липнева
температура вже відрізняється від 19.1 °C.

Отже, помилка I роду означає хибне виявлення зміни, а помилка II
роду — пропуск реально існуючої зміни.

### 4. Чи узгоджується тест із довірчим інтервалом?

Так. Для двобічного тесту при α = 0.05 перевірка H₀: μ = μ₀
узгоджується з перевіркою, чи входить μ₀ до відповідного
95% двобічного довірчого інтервалу.

Якщо μ₀ не входить в інтервал, H₀ відхиляється.

Якщо μ₀ входить в інтервал, H₀ не відхиляється.
