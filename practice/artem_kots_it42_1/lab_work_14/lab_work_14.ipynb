```python
# Практика 14. Параметричні критерії: t-критерій
# Коць Артем, ІТ-42
# Варіант 9 — Суми

import numpy as np
import pandas as pd
from scipy import stats

city = "Суми"
region = "Схід"
variant = 9
norm = 73

morning = np.array([
    85, 82, 77, 70, 64, 62,
    60, 62, 68, 75, 81, 86
])

months = [
    "Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"
]

print("Місто:", city)
print("Регіон:", region)
print("Варіант:", variant)
print("Норма:", norm, "%")
print("Кількість місяців:", len(morning))
```

```python
# Завдання 1. Одновибірковий t-критерій

t_stat, p_value = stats.ttest_1samp(
    morning,
    popmean=norm
)

print("Вибірка ранкової вологості:", morning)
print("Норма μ0 =", norm, "%")
print("Вибіркове середнє =", morning.mean(), "%")
print("Вибіркове стандартне відхилення =", morning.std(ddof=1), "%")
print("t-статистика =", t_stat)
print("p-value =", p_value)
```

### Завдання 1. Одновибірковий t-критерій

Перевіряємо, чи відрізняється середня вологість у Сумах
від довідкової норми 73%.

Нульова гіпотеза:

H₀: μ = 73%.

Альтернативна гіпотеза:

H₁: μ ≠ 73%.

Оскільки нас цікавить будь-яке відхилення від норми,
використовується двобічний t-критерій.

При α = 0.05:
- якщо p-value < 0.05, H₀ відхиляємо;
- якщо p-value ≥ 0.05, H₀ не відхиляємо.

Використання t-критерію є доцільним, оскільки стандартне
відхилення генеральної сукупності невідоме, а оцінюється
за вибірковим стандартним відхиленням.

```python
# Завдання 2. Ручна перевірка t-статистики

n = len(morning)
x_bar = morning.mean()
s = morning.std(ddof=1)

se = s / np.sqrt(n)

t_manual = (x_bar - norm) / se

print("n =", n)
print("x̄ =", x_bar)
print("s =", s)
print("SE =", se)
print("Ручна t-статистика =", t_manual)
print("t-статистика scipy =", t_stat)
print("Збігаються:", np.isclose(t_manual, t_stat))
```

### Завдання 2. Ручна перевірка t-статистики

t-статистика була обчислена вручну за формулою:

t = (x̄ - μ₀) / (s / √n).

Отримане вручну значення порівняно з результатом
stats.ttest_1samp().

Значення збігаються з урахуванням числової точності, тому
можна підтвердити правильність розрахунку t-статистики.

```python
# Завдання 3. Суми проти Львова

lviv = np.array([
    85, 83, 80, 73, 68, 65,
    63, 65, 70, 77, 82, 86
])

print("Суми:", morning)
print("Львів:", lviv)

print("\nСереднє Суми =", morning.mean())
print("Середнє Львів =", lviv.mean())

print("\nStd Суми =", morning.std(ddof=1))
print("Std Львів =", lviv.std(ddof=1))
```

```python
# Перевірка рівності дисперсій критерієм Лівена

stat_lev, p_lev = stats.levene(
    morning,
    lviv
)

print("Статистика Лівена =", stat_lev)
print("p-value Лівена =", p_lev)

if p_lev < 0.05:
    print("Дисперсії статистично відрізняються.")
    print("Для t-критерію використовуємо Welch: equal_var=False.")
else:
    print("Немає статистично значущих підстав вважати дисперсії різними.")
    print("Для t-критерію використовуємо стандартний варіант: equal_var=True.")
```

```python
# Двовибірковий t-критерій

equal_var_choice = p_lev >= 0.05

t_ind, p_ind = stats.ttest_ind(
    morning,
    lviv,
    equal_var=equal_var_choice
)

print("equal_var =", equal_var_choice)
print("t-статистика =", t_ind)
print("p-value =", p_ind)

if p_ind < 0.05:
    print("Відхиляємо H₀: середні вологості статистично значуще відрізняються.")
else:
    print("Не відхиляємо H₀: статистично значущої різниці середніх не виявлено.")
```

### Завдання 3. Незалежний двовибірковий критерій

Для порівняння обрано Суми (регіон Схід) та Львів
(регіон Захід). Це незалежні вибірки з 12 місячних значень.

Спочатку застосовано критерій Лівена для перевірки рівності
дисперсій.

Рішення щодо параметра equal_var прийнято виключно за результатом
критерію Лівена:

- при p-value Лівена < 0.05 використовуємо Welch-варіант
  (equal_var=False);
- при p-value Лівена ≥ 0.05 використовуємо стандартний
  двовибірковий t-критерій (equal_var=True).

Після цього за допомогою stats.ttest_ind() перевірено,
чи відрізняються середні значення вологості Сум і Львова.

```python
# Завдання 4. Пошук міста з протилежним результатом Лівена

odesa = np.array([
    82, 80, 78, 74, 71, 69,
    67, 68, 72, 76, 80, 83
])

stat_lev_odesa, p_lev_odesa = stats.levene(
    morning,
    odesa
)

print("Місто 1: Суми")
print("Місто 2: Одеса")

print("\nСтатистика Лівена =", stat_lev_odesa)
print("p-value Лівена =", p_lev_odesa)

if p_lev_odesa < 0.05:
    print("Дисперсії статистично відрізняються.")
    print("Використовуємо Welch: equal_var=False.")
else:
    print("Немає статистично значущих підстав вважати дисперсії різними.")
    print("Використовуємо стандартний варіант: equal_var=True.")
```

```python
# t-критерій для Сум та Одеси

equal_var_odesa = p_lev_odesa >= 0.05

t_ind_odesa, p_ind_odesa = stats.ttest_ind(
    morning,
    odesa,
    equal_var=equal_var_odesa
)

print("equal_var =", equal_var_odesa)
print("t-статистика =", t_ind_odesa)
print("p-value =", p_ind_odesa)

if p_ind_odesa < 0.05:
    print("Відхиляємо H₀: середні вологості статистично значуще відрізняються.")
else:
    print("Не відхиляємо H₀: статистично значущої різниці середніх не виявлено.")
```

### Завдання 4. Порівняння з протилежним висновком Лівена

Для порівняння обрано Одесу, яка належить до Південного регіону,
тобто до іншого регіону, ніж Суми.

Спочатку було виконано критерій Лівена для перевірки рівності
дисперсій.

На відміну від Завдання 3, отримано протилежний висновок щодо
рівності дисперсій. Тому параметр equal_var для stats.ttest_ind()
обирається відповідно до результату критерію Лівена, а не
на основі візуального припущення.

Це демонструє, що вибір між стандартним і Welch-варіантом
двовибіркового t-критерію потрібно робити після формальної
перевірки дисперсій.

```python
# Завдання 5. Ранок проти вечора

amplitude = np.array([
    3, 3, 4, 5, 6, 7,
    7, 7, 6, 5, 4, 3
])

evening = morning - amplitude

print("Місяць | Ранок | Відняти | Вечір")
print("-" * 35)

for month, m, a, e in zip(months, morning, amplitude, evening):
    print(f"{month:5} | {m:5} | {a:7} | {e:5}")

print("\nРанкова середня вологість:", morning.mean())
print("Вечірня середня вологість:", evening.mean())
```

```python
# Парний t-критерій

t_rel, p_rel = stats.ttest_rel(
    morning,
    evening
)

print("t-статистика =", t_rel)
print("p-value =", p_rel)

if p_rel < 0.05:
    print("Відхиляємо H₀: добова різниця вологості статистично значуща.")
else:
    print("Не відхиляємо H₀: статистично значущої добової різниці не виявлено.")
```

### Завдання 5. Парний t-критерій: ранок проти вечора

Для кожного місяця було сформовано пару значень:
ранкова вологість та відповідна вечірня вологість.

Вечірнє значення отримано шляхом віднімання заданої добової
амплітуди від ранкового значення.

Для перевірки використано stats.ttest_rel(), оскільки ранкове
та вечірнє значення належать до одного й того самого місяця.
Отже, спостереження всередині кожної пари є пов'язаними.

Парний критерій працює з різницями між значеннями кожної пари.
При цьому усувається вплив спільних для пари факторів, зокрема
особливостей конкретного місяця.

Саме тому для порівняння пов'язаних вимірювань коректним є
парний t-критерій, а не незалежний двовибірковий критерій.

## Підсумкові питання

### 1. Чому для перевірки середнього однієї вибірки використовується t-, а не z-розподіл?

У нашій задачі стандартне відхилення генеральної сукупності невідоме.
Замість нього використовується вибіркове стандартне відхилення s.
Тому статистика має t-розподіл Стьюдента з n − 1 ступенями свободи.

Саме тому для перевірки середнього однієї вибірки використовується
stats.ttest_1samp(), а не z-критерій.

### 2. Чому неправильний вибір equal_var може дати неправильний p-value?

Параметр equal_var визначає, яке припущення про дисперсії двох
незалежних вибірок використовується під час розрахунку
двовибіркового t-критерію.

Якщо дисперсії насправді різні, але використати equal_var=True,
може бути неправильно оцінена стандартна помилка різниці середніх.
Через це змінюється t-статистика та p-value.

Тому спочатку потрібно перевірити рівність дисперсій критерієм
Лівена, а вже потім вибирати стандартний або Welch-варіант.

### 3. Чому парний критерій може бути потужнішим за незалежний?

У парному критерії кожне ранкове значення безпосередньо порівнюється
з відповідним вечірнім значенням того самого місяця.

При обчисленні різниці для кожної пари усувається частина варіації,
пов'язаної зі спільними особливостями конкретного місяця.

Тому парний критерій може мати меншу похибку оцінки різниці та більшу
статистичну потужність порівняно з незалежним тестом на тих самих
даних "до/після".

### 4. Чи узгоджується критерій Лівена з порівнянням std() на око?

Критерій Лівена дає формальну статистичну перевірку рівності
дисперсій, тоді як порівняння std() показує лише величину
вибіркового розкиду.

Тому два значення std() можуть виглядати різними, але це ще не
означає, що різниця дисперсій статистично значуща.

У цій роботі вибір equal_var для ttest_ind() робився саме за
результатом критерію Лівена, а не лише за порівнянням std().
