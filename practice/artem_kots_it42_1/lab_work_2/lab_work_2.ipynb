```python
def get_season(month_number):
    if month_number in [12, 1, 2]:
        return "зима"
    elif month_number in [3, 4, 5]:
        return "весна"
    elif month_number in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"
```

```python
df["season"] = df.index.to_series().apply(get_season)
df
```

## Письмові відповіді

### 1. Різниця між `.loc[1:6]` та `.iloc[0:6]`

`.loc` виконує вибір рядків за мітками індексу, а `.iloc` — за їхніми позиціями. Тому `.loc[1:6]` шукає рядки з мітками від 1 до 6 включно, а `.iloc[0:6]` вибирає перші шість рядків за позиціями 0–5. Кількість результатів може відрізнятися, якщо індекс не збігається з позиціями.

### 2. Чому `and` викликає помилку?

Вираз `df["avg_temp"] > 10 and df["precip"] > 50` викликає помилку, тому що `and` призначений для одиничних булевих значень, а pandas Series містить багато значень. Для поелементного логічного порівняння потрібно використовувати `&` та брати кожну умову в дужки:

`(df["avg_temp"] > 10) & (df["precip"] > 50)`.

### 3. Різниця між формулою та `.apply()`

Обчислюваний стовпець через формулу виконує однакову математичну операцію над усіма значеннями стовпця. `.apply()` дозволяє застосувати власну функцію до кожного значення та реалізувати складнішу логіку, наприклад визначення пори року.
