```python
import pandas as pd

wide = pd.DataFrame({
    "місто": ["Суми"],
    "Січ": [79],
    "Лют": [75],
    "Бер": [69],
    "Кві": [59],
    "Тра": [51],
    "Чер": [45],
    "Лип": [41],
    "Сер": [43],
    "Вер": [53],
    "Жов": [67],
    "Лис": [77],
    "Гру": [83],
})

wide
```

```python
tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність"
)

tidy
```

```python
tidy.shape
```

### Шкали вимірювання

- **місто** — номінальна шкала, оскільки назви міст є категоріями без числового порядку.
- **місяць** — номінальна шкала, оскільки назви місяців використовуються як категоріальні позначення.
- **хмарність** — шкала відношень, оскільки вимірюється у відсотках і має природний нуль: 0% означає відсутність хмарності.

```python
back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

back
```

```python
back_sorted = back.reindex(columns=wide.columns[1:])
back_sorted
```

```python
back_sorted.equals(wide.set_index("місто"))
```

```python
tidy.to_csv("variant.csv", index=False)

reloaded_csv = pd.read_csv("variant.csv")

reloaded_csv
```

```python
reloaded_csv.equals(tidy)
```

```python
tidy.to_json(
    "variant.json",
    orient="records",
    force_ascii=False
)

reloaded_json = pd.read_json("variant.json")

reloaded_json
```

```python
reloaded_json.equals(tidy)
```

## Письмові відповіді

### 1. Чому "місто" — номінальна шкала, а "хмарність" — шкала відношень?

"Місто" є номінальною шкалою, тому що це назва категорії, між якими немає числового порядку. "Хмарність" є шкалою відношень, тому що вона вимірюється у відсотках і має природний нуль — 0% означає відсутність хмарності.

### 2. Чому wide-таблиця не є tidy?

У wide-таблиці місяці розміщені в окремих стовпцях, тому одна змінна "місяць" фактично розподілена між багатьма стовпцями. Після `melt()` усі місяці перетворюються на значення одного стовпця "місяць", а значення хмарності — на стовпець "хмарність". Кожен рядок після цього описує одне спостереження.

### 3. Чим CSV відрізняється від JSON?

CSV зберігає таблицю у вигляді рядків і стовпців, розділених спеціальним символом. JSON зберігає дані у структурованому форматі об'єктів і масивів. Для цієї таблиці CSV є простішим табличним представленням, а JSON — більш структурованим форматом, який зручно використовувати для обміну даними між програмами.
