```python
orders.isna().sum()
```

```python
median_quantity = orders["кількість"].median()

orders["кількість"] = orders["кількість"].fillna(median_quantity)

orders
```

### Чому медіана доречніша за середнє?

Медіана є доречнішою для цього набору, оскільки маємо лише 8 спостережень і в даних може бути велике або нетипове значення. Медіана менш чутлива до викидів, ніж середнє, тому дозволяє стабільніше заповнити пропущені значення кількості.

```python
orders.duplicated().sum()
```

```python
orders = orders.drop_duplicates()

print("Кількість рядків після видалення дубліката:", len(orders))
orders
```

```python
orders["ціна"] = (
    orders["ціна"]
    .astype(str)
    .str.replace(" грн", "", regex=False)
    .astype(float)
)

orders.dtypes
```

```python
orders["місто"] = (
    orders["місто"]
    .str.strip()
    .str.lower()
    .replace({"суми": "Суми"})
)

orders["місто"].unique()
```

```python
q1, q3 = orders["ціна"].quantile([0.25, 0.75])

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Нижня межа =", lower)
print("Верхня межа =", upper)
```

```python
outliers = orders[
    (orders["ціна"] < lower) |
    (orders["ціна"] > upper)
]

outliers
```

```python
orders.loc[orders["ціна"] == 7300, "ціна"] = 730

orders
```

```python
print("Пропуски:")
print(orders.isna().sum())

print("\nДублікати:")
print(orders.duplicated().sum())

print("\nТипи даних:")
print(orders.dtypes)

print("\nУнікальні міста:")
print(orders["місто"].unique())

print("\nКількість рядків:", len(orders))
```

## Підсумкові письмові відповіді

### 1. Чому заповнення пропуску середнім може спотворити стандартне відхилення?

Якщо замінювати пропуски середнім значенням, кілька спостережень отримують однакове центральне значення. Це може штучно зменшити розкид даних і стандартне відхилення. Медіана є стійкішою до нетипових значень і тому добре підходить для невеликого набору даних.

### 2. Чим відрізняється duplicated() без subset і з subset?

`duplicated()` без `subset` перевіряє повний збіг усіх значень рядка. `duplicated(subset=[...])` перевіряє дублікати лише за вказаними стовпцями. Тому результати можуть відрізнятися.

### 3. Чому не можна автоматично видаляти всі значення поза межами IQR?

Значення поза межами IQR не обов'язково є помилками. Вони можуть бути реальними рідкісними спостереженнями або великими замовленнями. Тому IQR використовується для виявлення підозрілих значень, а рішення про їх видалення або виправлення потрібно приймати після аналізу даних.
