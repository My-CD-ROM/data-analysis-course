## Завдання 1

### Відповідність показників `.describe()` мірам статистики

- `count` — кількість спостережень.
- `mean` — середнє арифметичне.
- `std` — стандартне відхилення.
- `min` — мінімальне значення; разом із `max` використовується для визначення розмаху.
- `25%` — перший квартиль Q1.
- `50%` — медіана, або другий квартиль Q2.
- `75%` — третій квартиль Q3.
- `max` — максимальне значення; разом із `min` використовується для визначення розмаху.

```python
s = humidity

rozmakh = s.max() - s.min()

variance_ddof1 = s.var(ddof=1)
variance_ddof0 = s.var(ddof=0)

std_ddof1 = s.std(ddof=1)
std_ddof0 = s.std(ddof=0)

q1 = s.quantile(0.25)
q3 = s.quantile(0.75)

iqr = q3 - q1

cv = std_ddof1 / s.mean() * 100

print(f"Розмах: {rozmakh:.2f}")
print(f"Дисперсія ddof=1: {variance_ddof1:.4f}")
print(f"Дисперсія ddof=0: {variance_ddof0:.4f}")
print(f"Стандартне відхилення ddof=1: {std_ddof1:.4f}")
print(f"Стандартне відхилення ddof=0: {std_ddof0:.4f}")
print(f"Q1: {q1:.2f}")
print(f"Q3: {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Коефіцієнт варіації: {cv:.2f}%")
```

```python
print("Q1 з ручного розрахунку:", q1)
print("Q1 з describe():", humidity.describe()["25%"])

print("Q3 з ручного розрахунку:", q3)
print("Q3 з describe():", humidity.describe()["75%"])
```

### Порівняння дисперсії

При `ddof=1` pandas обчислює вибіркову дисперсію, тобто ділить
суму квадратів відхилень на `n - 1`.

При `ddof=0` використовується дисперсія всієї сукупності, де
ділення виконується на `n`.

Тому значення дисперсії та стандартного відхилення при `ddof=1`
трохи більші, ніж при `ddof=0`.

Коефіцієнт варіації для Сум становить приблизно **14%** і показує
відносний рівень розсіювання вологості щодо її середнього значення.

### Порівняння регіонів

Для порівняння було використано середню річну вологість міст
регіонів Схід і Захід.

У регіоні **Схід** стандартне відхилення показує розкид середніх
річних значень вологості між Харковом, Дніпром і Сумами.

У регіоні **Захід** стандартне відхилення показує розкид між

Для коректного порівняння також використовується коефіцієнт
варіації (CV), оскільки він враховує різницю між середніми
значеннями груп. Більший CV означає більшу відносну варіативність.

За отриманими результатами більший відносний розкид має регіон,

```python
mean_humidity = humidity.mean()
median_humidity = humidity.median()

print(f"Середнє: {mean_humidity:.2f}%")
print(f"Медіана: {median_humidity:.2f}%")
print(f"Різниця: {abs(mean_humidity - median_humidity):.2f}%")
```

Середнє значення вологості становить **71,83%**, а медіана —
**69,00%**. Між ними є помітна різниця.

Значення вологості мають нерівномірний розподіл: у зимові місяці

Тому розподіл можна вважати **помірно скошеним**, а не повністю
симетричним.

Для опису центральної тенденції в цьому наборі **медіана є дещо
репрезентативнішою**, оскільки вона менше залежить від крайніх
значень, ніж середнє.

### Порівняння Сум і Львова

Для порівняння варіативності вологості Сум і Львова доцільно
використовувати коефіцієнт варіації, а не лише стандартне
відхилення.

`std` показує абсолютний розкид значень у відсоткових пунктах,
але не враховує рівень середньої вологості.

Коефіцієнт варіації визначається як:

CV = std / mean × 100%.

Тому CV показує розкид відносно середнього значення і дозволяє
коректніше порівнювати варіативність двох міст.

Навіть якщо середні значення вологості Сум і Львова близькі,
CV залишається корисним показником, оскільки враховує не тільки
абсолютний розкид, а й величину середнього значення.

## Підсумкові питання

### 1. Чи можете пояснити, яка міра розсіювання з лекції стоїть за кожним із рядків `std`, `25%`/`75%` і `min`/`max` у `.describe()`?

`std` — це стандартне відхилення, яке показує середній рівень
розсіювання значень відносно середнього.

`25%` і `75%` — це перший та третій квартилі (Q1 і Q3).
За ними можна визначити міжквартильний розмах:
IQR = Q3 − Q1.

`min` і `max` — мінімальне та максимальне значення. За ними
визначається розмах:
R = max − min.

### 2. Чому заповнення пропуску середнім штучно зменшує `std`?

Якщо пропуск замінити середнім значенням, додане значення
дорівнює центру розподілу і практично не збільшує розсіювання.
Тому після такого заповнення стандартне відхилення стає меншим.

Наприклад, для значень `[60, 70, 80]` середнє дорівнює 70.
Якщо додати пропуск як 70, отримаємо `[60, 70, 80, 70]`.
Нове значення знаходиться точно в центрі розподілу, тому воно
зменшує відносний розкид даних.

### 3. Чому порівняння абсолютних значень `std` двох груп може привести до хибного висновку?

Стандартне відхилення показує абсолютний розкид і не враховує
різницю між середніми значеннями груп.

Тому група з більшим `std` не обов'язково є більш варіативною
відносно власного середнього.

Для коректного порівняння використовується коефіцієнт варіації:

CV = std / mean × 100%.

Він нормує стандартне відхилення відносно середнього значення.

### 4. Судячи з власного розподілу, чи він симетричний і яка міра положення репрезентативніша?

Для Сум середнє значення вологості становить **71,83%**, а медіана —
**69,00%**. Їхня різниця становить приблизно **2,83 процентного
пункту**, тому розподіл не є повністю симетричним.

Вологість має вищі значення взимку та нижчі влітку, тому розподіл
можна вважати помірно скошеним.

Для цих даних **медіана є дещо репрезентативнішою**, оскільки вона
менше залежить від крайніх значень, ніж середнє.
