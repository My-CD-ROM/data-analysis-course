```python
# Практика 9. Перетворення даних
# Коць Артем, ІТ-42
# Варіант 9 — Суми

import numpy as np
import pandas as pd

customers = pd.DataFrame({
    "вік": [19, 24, 31, 38, 45, 52, 27, 61, 34, 68],
    "дохід": [18000, 24000, 32000, 41000, 50000, 58000, 27000, 65000, 36000, 72000],
    "кількість": [2, 3, 1, 4, 2, 3, 5, 2, 3, 1],
    "місто_доставки": [
        "Суми",
        "Київ",
        "Львів",
        "Одеса",
        "Суми",
        "Харків",
        "Полтава",
        "Київ",
        "Суми",
        "Дніпро"
    ]
})

customers
```

```python
base_price = 730

customers["сума_замовлення"] = (
    base_price * customers["кількість"]
)

customers["частка_доходу"] = (
    customers["сума_замовлення"] / customers["дохід"]
)

customers
```

### Завдання 2. Похідна змінна

Стовпець "сума_замовлення" обчислено як добуток базової ціни
730 грн на кількість товарів у замовленні.

Стовпець "частка_доходу" показує, яку частину місячного доходу
клієнта становить його замовлення.

Ця змінна може бути кориснішою для аналізу поведінки клієнтів,
оскільки враховує не тільки розмір замовлення, а й фінансові
можливості клієнта. Однакова сума замовлення може мати зовсім
різне значення для клієнтів із різним доходом.

```python
customers["вікова_група_cut"] = pd.cut(
    customers["вік"],
    bins=3
)

customers["вікова_група_cut"].value_counts().sort_index()
```

```python
customers["вікова_група_qcut"] = pd.qcut(
    customers["вік"],
    q=3
)

customers["вікова_група_qcut"].value_counts().sort_index()
```

### Завдання 3. Порівняння cut() і qcut()

Метод `pd.cut()` ділить діапазон значень на інтервали однакової
довжини. Тому кількість клієнтів у різних групах може сильно
відрізнятися, якщо значення віку розподілені нерівномірно.

Метод `pd.qcut()` ділить дані за квантилями та намагається створити
групи з приблизно однаковою кількістю спостережень. При цьому межі
вікових інтервалів можуть мати різну довжину.

Отже, `cut()` забезпечує однакову ширину інтервалів, а `qcut()`
забезпечує приблизно однакову кількість спостережень у групах.
Саме тому результати та розміри груп відрізняються, хоча вхідні
дані однакові.

```python
customers["вік_minmax"] = (
    (customers["вік"] - customers["вік"].min()) /
    (customers["вік"].max() - customers["вік"].min())
)

customers["дохід_minmax"] = (
    (customers["дохід"] - customers["дохід"].min()) /
    (customers["дохід"].max() - customers["дохід"].min())
)

customers[[
    "вік",
    "дохід",
    "вік_minmax",
    "дохід_minmax"
]].head()
```

```python
customers["вік_z"] = (
    (customers["вік"] - customers["вік"].mean()) /
    customers["вік"].std()
)

customers["дохід_z"] = (
    (customers["дохід"] - customers["дохід"].mean()) /
    customers["дохід"].std()
)

customers[[
    "вік",
    "дохід",
    "вік_minmax",
    "дохід_minmax",
    "вік_z",
    "дохід_z"
]].head()
```

### Завдання 4. Масштабування

Метод min-max нормалізації переводить значення віку та доходу
до спільного діапазону [0, 1]. Це дозволяє порівнювати змінні,
які спочатку мають різний масштаб.

Z-оцінка стандартизує змінні відносно їхнього середнього значення
та стандартного відхилення. Після стандартизації середнє значення
змінної дорівнює приблизно 0, а стандартне відхилення — приблизно 1.

До масштабування вік і дохід не можна було напряму порівнювати
або використовувати разом у методах, чутливих до масштабу, оскільки
дохід має значно більші числові значення. У такому випадку дохід
міг би непропорційно сильно впливати на результат.

Масштабування усуває цю проблему та дозволяє змінним мати
порівнюваний внесок у методах, чутливих до масштабу.

```python
city_dummies = pd.get_dummies(
    customers["місто_доставки"],
    prefix="місто"
)

customers = pd.concat(
    [customers, city_dummies],
    axis=1
)

city_dummies
```

```python
print("Унікальних міст:", customers["місто_доставки"].nunique())
print("Нових one-hot стовпців:", len(city_dummies.columns))
```

### Завдання 5. Індикаторне кодування

Для змінної "місто_доставки" використано one-hot кодування
за допомогою pd.get_dummies().

У наборі даних є 6 унікальних міст, тому було створено 6
індикаторних стовпців. Для кожного рядка відповідний стовпець
має значення True, якщо замовлення доставляється до цього міста,
і False в іншому випадку.

One-hot кодування є правильним для міста, оскільки міста є
номінальною категорією і не мають природного порядку. Якщо
замапити міста у числа 1, 2, 3 тощо, це створило б помилкове
враження, що одне місто "більше" або "вище" за інше.

Тому для номінальної змінної "місто_доставки" доцільно
використовувати саме one-hot кодування.

## Контрольні питання

### 1. У чому різниця між pd.cut() і pd.qcut()?

`pd.cut()` ділить діапазон числових значень на інтервали однакової
довжини. Кількість спостережень у групах при цьому може бути різною.

`pd.qcut()` ділить дані за квантилями та намагається створити групи
з приблизно однаковою кількістю спостережень. Межі інтервалів при
цьому можуть бути різної довжини.

### 2. Чому для міста застосовують one-hot кодування?

Місто є номінальною змінною і не має природного порядку. Кодування
міст числами 1, 2, 3 тощо створило б хибний порядок і відстань між
категоріями.

One-hot кодування створює окремий індикатор для кожного міста та
не вносить штучного порядку між категоріями.

### 3. Навіщо масштабувати змінні перед методами, чутливими до масштабу?

Змінні можуть мати дуже різні діапазони значень. У нашому випадку
вік вимірюється десятками років, а дохід — десятками тисяч гривень.

У дистанційних методах або моделях із регуляризацією змінна з більшим
числовим масштабом може отримати непропорційно великий вплив.
Масштабування приводить змінні до порівнюваної шкали.

### 4. Чому похідна змінна може бути кориснішою за вихідні складники?

Похідна змінна може показувати співвідношення між двома показниками,
яке має безпосередній практичний зміст.

Наприклад, "частка_доходу" показує, яку частину місячного доходу
клієнта становить замовлення. Тому вона дає змогу порівнювати
фінансове навантаження від замовлення для клієнтів із різним доходом,
чого не можна побачити лише за сумою замовлення або доходом окремо.
