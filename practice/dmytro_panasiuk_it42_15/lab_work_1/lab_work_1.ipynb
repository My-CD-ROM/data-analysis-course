## Завдання 1. Створення масивів

**Дніпро, варіант 5.** Температури взято з таблиці завдання, 12 місяців від січня до грудня. Кількості днів відповідають невисокосному року.

```python
city = "Дніпро"
variant = 5
months = np.array(["Січень", "Лютий", "Березень", "Квітень", "Травень", "Червень",
                   "Липень", "Серпень", "Вересень", "Жовтень", "Листопад", "Грудень"])

temps = np.array([-4, -3, 3, 11, 17, 21, 23, 22, 16, 9, 3, -2])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

for name, arr in [("temps", temps), ("days_in_month", days_in_month),
                  ("month_numbers", month_numbers)]:
    print(f"{name}: {arr}")
    print(f"  shape: {arr.shape}, dtype: {arr.dtype}")
```

## Завдання 2. Векторизовані обчислення

Формула переведення температури: **°F = °C × 9/5 + 32**. Для підрахунку теплих днів беруться всі дні тих місяців, де **середньомісячна температура строго вища за 15 °C**.

```python
temps_fahrenheit = temps * 9 / 5 + 32
warm_mask = temps > 15
warm_days = days_in_month[warm_mask].sum()
total_days = days_in_month.sum()

print("Температури у °F:", temps_fahrenheit)
print("Місяці з температурою > 15 °C:", months[warm_mask])
print("Кількість днів у цих місяцях:", days_in_month[warm_mask])
print("Загалом днів із середньомісячною температурою > 15 °C:", warm_days)
print("Кількість днів за рік:", total_days)
assert total_days == 365, "Очікується 365 днів для невисокосного року"
print("Перевірка: 365 днів — успішно.")
```

## Завдання 3. Індексація і фільтрація

`argmax()` і `argmin()` повертають **індекси, починаючи з нуля**. Номер місяця для користувача беремо з `month_numbers`, щоб уникнути плутанини.

```python
max_idx = temps.argmax()
min_idx = temps.argmin()
negative_month_numbers = month_numbers[temps < 0]
sorted_temps = np.sort(temps)

print(f"Найтепліший місяць: {months[max_idx]} (№{month_numbers[max_idx]}),"
      f" {temps[max_idx]} °C; індекс {max_idx}")
print(f"Найхолодніший місяць: {months[min_idx]} (№{month_numbers[min_idx]}),"
      f" {temps[min_idx]} °C; індекс {min_idx}")
print("Номери місяців із температурою нижче 0 °C:", negative_month_numbers)
print("Відсортовані температури:", sorted_temps)
print("Оригінальний масив temps (не змінено):", temps)
```

## Завдання 4. Двовимірний масив та `axis`

Додаємо **умовні** (вигадані лише для тренування NumPy) години сонячного сяйва для кожного місяця. Масив `data` має три рядки:

- рядок 0 — температура, °C;
- рядок 1 — дні в місяці;
- рядок 2 — умовні години сонячного сяйва, год.

**Зауваження:** механічне підсумовування цих величин із різними одиницями вимірювання коректне як вправа з NumPy, але **не має самостійного фізичного змісту**.

```python
sunshine_hours = np.array([55, 70, 120, 160, 210, 240, 260, 230, 170, 115, 65, 45])
data = np.array([temps, days_in_month, sunshine_hours])

row_sums = data.sum(axis=1)
row_means = data.mean(axis=1)
column_sums = data.sum(axis=0)

print("Масив data (3 × 12):")
print(data)
print("Форма масиву:", data.shape)
print("Сума по кожному рядку (axis=1):", row_sums)
print("Середнє по кожному рядку (axis=1):", row_means)
print("Суми по кожному стовпцю (axis=0):", column_sums)
print("Січень:", temps[0], "+", days_in_month[0], "+", sunshine_hours[0],
      "=", column_sums[0])
assert data.shape == (3, 12)
```

**Пояснення першого стовпця (січня):** результат дорівнює **82**: −4 (°C) + 31 (день) + 55 (умовних сонячних годин). Він показує **формальну суму трьох елементів першого стовпця**, а не змістовний кліматичний показник, бо величини мають різні одиниці.

`axis=1` агрегує елементи **вздовж стовпців усередині кожного рядка** (тому результат довжини 3); `axis=0` — **вздовж рядків для кожного стовпця** (результат довжини 12).

## Завдання 5. `reshape` і broadcasting

`rows.shape == (2, 12)`, а `correction.shape == (2,)`. NumPy порівнює розміри **справа наліво**: `12` та `2` не збігаються, і жоден із них не дорівнює `1`. Тому пряме додавання має викликати `ValueError`.

У наступній клітинці **справді виконуємо** неправильне додавання, але ловимо очікувану помилку, щоб виконання решти ноутбука не зупинилося.

```python
rows = np.array([temps, days_in_month])
correction = np.array([1, 0])
print("Форма rows:", rows.shape)
print("Форма correction:", correction.shape)

try:
    invalid_result = rows + correction
except ValueError as exc:
    print("Очікувана помилка ValueError:", exc)
else:
    raise AssertionError("Очікувалася помилка broadcasting")
```

```python
correction_2d = correction.reshape(2, 1)
corrected_rows = rows + correction_2d

print("Форма correction після reshape:", correction_2d.shape)
print("Форма результату:", corrected_rows.shape)
print("Масив після додавання поправки до кожного рядка:")
print(corrected_rows)

assert corrected_rows.shape == (2, 12)
assert np.array_equal(corrected_rows[0], temps + 1)
assert np.array_equal(corrected_rows[1], days_in_month)
print("Перевірка: +1 до температур, +0 до кількості днів — успішно.")
```

**Пояснення:** форми `(2, 12)` і `(2, 1)` сумісні: при порівнянні справа наліво останні розміри `12` та `1` дозволяють поширити (broadcast) одне значення на всі 12 стовпців, а передостанні `2` та `2` однакові. У результаті `1` додається до **кожної температури** в першому рядку, а `0` — до **кожного значення днів** у другому.

## Письмові відповіді на контрольні запитання

**1. Яка різниця між `shape == (12,)` та `shape == (1, 12)`?**  
`(12,)` — це одновимірний масив з **однією віссю** і 12 елементами. `(1, 12)` — двовимірний масив з **двома осями**: один рядок і 12 стовпців. Кількість елементів однакова, але кількість осей різна, що впливає на індексацію та broadcasting.

**2. Чому `temps[temps > 15]` повертає масив іншої довжини?**  
Умова `temps > 15` утворює булеву маску з 12 значень. Булева індексація залишає лише ті елементи, для яких маска має значення `True`. Для Дніпра це `[17, 21, 23, 22, 16]`, тобто **5 замість 12** температур.

**3. Що станеться, якщо додати масив форми `(3,)` до масиву форми `(3, 12)`?**  
Виникне **`ValueError` через несумісні форми**. NumPy порівнює осі справа наліво. В останній осі маємо `12` проти `3`: вони різні, і жоден не дорівнює `1`. Щоб додати окреме значення до кожного з трьох рядків, масив `(3,)` треба перетворити на `(3, 1)`.

**4. Чому `rows + correction` дає помилку, а `rows + correction.reshape(2, 1)` — ні?**  
У першому випадку форми `(2, 12)` і `(2,)` конфліктують по останній осі: `12 ≠ 2`, і жоден розмір не дорівнює `1`. Після `.reshape(2, 1)` розміри стають `(2, 12)` та `(2, 1)`: `2` збігається з `2`, а розмір `1` розтягується до `12`. Тому одна поправка застосовується до кожного елемента відповідного рядка.
