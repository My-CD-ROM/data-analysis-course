**Вихідні дані:** навчальні середньомісячні температури (°C) з Практики №1 та кількість опадів (мм) із таблиці Практики №2. Усі місяці розташовані в календарному порядку.

## Крок 1. Побудова DataFrame

Імпортуємо `pandas`, створюємо таблицю з трьома стовпцями: `month`, `avg_temp`, `precip`. Переглядаємо перші п'ять рядків, розмір і типи даних.

```python
import pandas as pd

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
avg_temp = [-4, -3, 3, 11, 17, 21, 23, 22, 16, 9, 3, -2]
precip = [36, 34, 30, 32, 42, 50, 54, 44, 32, 32, 38, 40]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip,
})

print("Місто: Дніпро | Варіант №5")
print("Перші 5 рядків:")
print(df.head().to_string())
print("\nФорма DataFrame:", df.shape)
print("\nТипи стовпців:")
print(df.dtypes.to_string())
```

## Завдання 1. Індекс

1. Показати початкову таблицю та індекс від 0 до 11.  
2. Зробити стовпець `month` індексом і дістати рядок за його назвою через `.loc`.  
3. Повернути звичайну нумерацію рядків.

```python
print("1. Початкова таблиця з індексом 0–11:")
print(df.to_string())
print("\nІндекс:", df.index)
assert df.index.tolist() == list(range(12))

# Створення таблиці з назвами місяців як підписами індексу
df_by_month = df.set_index("month")
print("\n2. Таблиця після set_index('month'):")
print(df_by_month.to_string())
print("\nРядок для місяця 'Лип' через .loc:")
print(df_by_month.loc["Лип"].to_string())

# Повернення стовпця month із індексу та відновлення нумерації 0–11
df = df_by_month.reset_index()
print("\n3. Індекс після reset_index():", df.index)
print(df.head().to_string())
assert df.index.tolist() == list(range(12))
```

## Завдання 2. Відбір рядків і стовпців

Призначаємо індекс `1, 2, ..., 12`, тобто номер календарного місяця.

- `.loc[1:6, "avg_temp"]` — вибір **за підписами**, включно з обома межами.
- `.iloc[0:6, 1]` — вибір **за позиціями**, позиція 6 не включається; `avg_temp` є другим стовпцем (позиція 1).
- Виведемо тільки `month` та `precip` для всіх 12 місяців.

```python
df.index = range(1, 13)
df.index.name = "month_number"

by_label = df.loc[1:6, "avg_temp"]
by_position = df.iloc[0:6, 1]

print("1. .loc: температури місяців з підписами 1–6:")
print(by_label.to_string())
print("\n2. .iloc: температури за позиціями 0–5:")
print(by_position.to_string())
print("\n3. Тільки month та precip:")
print(df.loc[:, ["month", "precip"]].to_string())

assert by_label.equals(by_position)
```

**Пояснення до пункту 2.** У нашій таблиці `.loc[1:6]` і `.iloc[0:6]` повертають ті самі шість місяців, оскільки індекс 1–12 відповідає порядку рядків. Проте `.loc` орієнтується на **значення індексу**, а `.iloc` — на **фізичну позицію**. Якщо індекс змінити (наприклад, на 0–11, довільні числа або підписи із пропусками), вибрані рядки та іноді їхня кількість можуть відрізнятися.

## Завдання 3. Фільтрація за умовою

1. Місяці з температурою понад **15 °C**.
2. Місяці, у яких **одночасно** температура понад 10 °C **і** опади понад 50 мм (оператор `&`).
3. Місяці, у яких температура нижча за 0 °C **або** опади понад 80 мм (оператор `|`).

```python
warm_months = df[df["avg_temp"] > 15]
warm_and_rainy = df[(df["avg_temp"] > 10) & (df["precip"] > 50)]
cold_or_very_rainy = df[(df["avg_temp"] < 0) | (df["precip"] > 80)]

print("1. Температура > 15 °C:")
print(warm_months.to_string())
print("\n2. Температура > 10 °C І опади > 50 мм:")
print(warm_and_rainy.to_string())
print("\n3. Температура < 0 °C АБО опади > 80 мм:")
print(cold_or_very_rainy.to_string())

assert warm_months["month"].tolist() == ["Тра", "Чер", "Лип", "Сер", "Вер"]
assert warm_and_rainy["month"].tolist() == ["Лип"]
assert cold_or_very_rainy["month"].tolist() == ["Січ", "Лют", "Гру"]
```

**Результат для Дніпра:** `avg_temp > 15` — травень–вересень (5 місяців); обидві умови `avg_temp > 10` і `precip > 50` задовольняє тільки липень (23 °C, 54 мм); умову `avg_temp < 0` або `precip > 80` задовольняють січень, лютий і грудень — через від'ємні температури. За вихідними даними опадів понад 80 мм немає.

## Завдання 4. Обчислювані стовпці

1. `avg_temp_f`: перетворення температури за формулою °F = °C × 9/5 + 32 без циклу.
2. `is_wet`: `True`, якщо кількість опадів більша за середню по всій таблиці.
3. `season`: назва пори року на основі **номера місяця** із застосуванням `.apply()` та власної функції.

```python
# 1. Векторизована формула
# 2. Векторизоване порівняння зі середньою кількістю опадів
df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32
mean_precip = df["precip"].mean()
df["is_wet"] = df["precip"] > mean_precip

# 3. Власна функція + .apply() до значень номерів місяців
def season_from_month(month_number):
    if month_number in (12, 1, 2):
        return "зима"
    elif month_number in (3, 4, 5):
        return "весна"
    elif month_number in (6, 7, 8):
        return "літо"
    else:
        return "осінь"

df["season"] = df.index.to_series().apply(season_from_month)

print(f"Середня кількість опадів: {mean_precip:.2f} мм")
print("\nПідсумкова таблиця з новими стовпцями:")
print(df.to_string())

assert len(df) == 12
assert df.shape == (12, 6)
assert abs(df.loc[7, "avg_temp_f"] - 73.4) < 1e-9
assert df["is_wet"].sum() == 5
assert df.loc[12, "season"] == "зима"
```

**Пояснення.** Середня кількість опадів становить приблизно **38,67 мм**. Стовпець `is_wet` має значення `True` тільки тоді, коли опадів **строго більше** за це число (у нашому варіанті: травень, червень, липень, серпень і грудень). Функція `season_from_month()` визначає сезони відповідно до календарних місяців, а `.apply()` застосовує її до всіх 12 номерів місяців.

## Контрольні питання — письмові відповіді

**1. Чому `.loc[1:6]` і `.iloc[0:6]` можуть повернути різну кількість рядків залежно від індексу?**

`.loc` обирає рядки за **мітками індексу** і зазвичай включає обидві межі (`1` та `6`). `.iloc` обирає за **порядковими позиціями** від 0 до 5; права межа (`6`) не входить. Наприклад, для індексу `0, 1, ..., 11` виклик `.loc[1:6]` поверне шість рядків (із мітками 1–6), а `.iloc[0:6]` — також шість, але це будуть **інші** рядки (із мітками 0–5). Якщо індекс має пропущені чи повторювані мітки, кількість рядків теж може відрізнятися. У нашій таблиці з індексом `1, ..., 12` обидва способи вибирають однакові шість рядків.

**2. Чому `df["avg_temp"] > 10 and df["precip"] > 50` викличе помилку, а `(df["avg_temp"] > 10) & (df["precip"] > 50)` — ні?**

Вираз `df["avg_temp"] > 10` повертає не одне логічне значення, а **цілу серію** `True`/`False` (для кожного місяця). Оператор `and` очікує одне логічне значення, тому виникає помилка про неоднозначність істинності серії (`ValueError: The truth value of a Series is ambiguous`). Оператор `&` порівнює значення **поелементно**, тобто залишає `True` лише там, де виконуються обидві умови. Кожну умову треба взяти в дужки через правила пріоритету операторів Python.

**3. Чим відрізняється обчислюваний стовпець, створений формулою, від створеного через `.apply()`?**

**Формула** виконує операцію одразу для всіх значень стовпця: наприклад, `df["avg_temp"] * 9/5 + 32` перетворює температуру на °F. Це лаконічний **векторизований** спосіб, зазвичай швидший для числових обчислень. **`.apply()`** послідовно застосовує задану функцію до значень серії: наприклад, `season_from_month()` на основі номера місяця повертає назву сезону. Він зручний для складнішої логіки з умовами, хоча часто повільніший за векторизовані операції.
