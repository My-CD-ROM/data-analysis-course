## Наскрізний прикладДемонстраційний набір — вологість повітря (%) та температура (°C) за 8 днів спостережень. На ньому показано ті самі прийоми, які далі застосовуються до даних вашого варіанта.

```python
humidity = np.array([42, 55, 61, 38, 47, 66, 59, 44])
print("humidity.shape:", humidity.shape)
print("humidity.dtype:", humidity.dtype)

# векторизована операція — переведення відсотків у частку, без циклу
humidity_frac = humidity / 100
print("humidity_frac:", humidity_frac)

# булева індексація — тільки дні з вологістю вище 50%
humid_days = humidity[humidity > 50]
print("humid_days:", humid_days)

# argmax/argmin — індекс найвологішого й найсухішого дня
print("argmax (найвологіший день):", humidity.argmax())
print("argmin (найсухіший день):", humidity.argmin())

# сортування без зміни оригінального масиву
print("np.sort(humidity):", np.sort(humidity))

# другий рядок вимірювань — температура за ті самі 8 днів
temperature = np.array([18, 20, 19, 22, 21, 17, 23, 20])
data = np.array([humidity, temperature])
print("data.shape:", data.shape)

print("mean(axis=1) — середнє по кожному рядку:", data.mean(axis=1))
print("sum(axis=0) — сума по кожному дню (стовпцю):", data.sum(axis=0))
```

## Завдання 1. Створення масивівСтворюємо:- `temps` — масив температур свого варіанта- `days_in_month` — кількість днів у кожному місяці (не високосний рік)- `month_numbers` — номери місяців від 1 до 12 через `np.arange`Виводимо `shape` і `dtype` для всіх трьох масивів.

```python
# Варіант 1 — Київ. Щоб перейти на інший варіант — замініть значення temps тут.
temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

print("temps:", temps)
print("  shape:", temps.shape, "  dtype:", temps.dtype)

print("days_in_month:", days_in_month)
print("  shape:", days_in_month.shape, "  dtype:", days_in_month.dtype)

print("month_numbers:", month_numbers)
print("  shape:", month_numbers.shape, "  dtype:", month_numbers.dtype)
```

## Завдання 2. Векторизовані обчислення- Переводимо `temps` із °C у °F однією векторизованою операцією: `°F = °C × 9/5 + 32`.- Рахуємо, скільки всього днів у році тепліше 15°C — через булеву індексацію по `temps` і підсумовування відповідних `days_in_month`.- Перевіряємо, що сума `days_in_month` дорівнює 365.

```python
# 1) Переведення у °F — одна векторизована операція, без циклу
temps_f = temps * 9/5 + 32
print("temps_f:", temps_f)

# 2) Скільки днів у році тепліше 15°C
warm_mask = temps > 15
print("Маска теплих місяців (T > 15°C):", warm_mask)
print("Дні цих місяців:", days_in_month[warm_mask])
warm_days_total = days_in_month[warm_mask].sum()
print("Усього днів тепліше 15°C:", warm_days_total)

# 3) Перевірка суми днів у році
total_days = days_in_month.sum()
print("Сума days_in_month:", total_days)
print("Чи дорівнює 365?", total_days == 365)
```

## Завдання 3. Індексація і фільтрація- Знаходимо значення й індекс найтеплішого та найхолоднішого місяця.- Виводимо номери місяців, де температура нижче нуля (булева індексація по `month_numbers`).- Виводимо температури, відсортовані за зростанням, **без зміни порядку** в оригінальному масиві `temps`.

```python
# Найтепліший і найхолодніший місяць
idx_max = temps.argmax()
idx_min = temps.argmin()
print(f"Найтепліший місяць: номер {month_numbers[idx_max]} (індекс {idx_max}), "
      f"температура {temps[idx_max]}°C")
print(f"Найхолодніший місяць: номер {month_numbers[idx_min]} (індекс {idx_min}), "
      f"температура {temps[idx_min]}°C")

# Місяці з температурою нижче нуля
cold_months = month_numbers[temps < 0]
print("Місяці з T < 0°C:", cold_months)

# Сортування без зміни оригінального масиву
temps_sorted = np.sort(temps)
print("Відсортовані температури:", temps_sorted)
print("Оригінальний temps:    ", temps)
print("Оригінал не змінився?", np.array_equal(temps, np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])))
```

## Завдання 4. Двовимірний масив і axis- Додаємо **третій** рядок даних на власний розсуд (наприклад, умовна кількість сонячних годин за місяць).- Збираємо `temps`, `days_in_month` і цей третій рядок у масив форми `(3, 12)`.- Рахуємо суму й середнє по кожному рядку (`axis=1`).- Рахуємо суму по кожному стовпцю (`axis=0`) і пояснюємо, що означає перше число (для січня).

```python
# Умовна кількість сонячних годин за місяць у Києві (навчальні дані)
sunny_hours = np.array([31, 60, 120, 180, 250, 270, 310, 280, 200, 130, 60, 30])

data3 = np.array([temps, days_in_month, sunny_hours])
print("data3.shape:", data3.shape)
print("data3:")
print(data3)

# Сума й середнє по кожному рядку (axis=1)
print("\nСума по рядках (axis=1):", data3.sum(axis=1))
print("Середнє по рядках (axis=1):", data3.mean(axis=1))

# Сума по кожному стовпцю (axis=0)
print("\nСума по стовпцях (axis=0):", data3.sum(axis=0))

# Пояснення для першого стовпця (axis=0, січень)
jan_sum = data3.sum(axis=0)[0]
print(f"\nПерший стовпець (axis=0, січень) = {temps[0]} (T) + {days_in_month[0]} (дні) + "
      f"{sunny_hours[0]} (сонячні години) = {jan_sum}")
print("Це просто сума трьох різних показників за січень — фізичного сенсу як")
print("одна величина не має (°C + дні + години — різні одиниці виміру).")
```

## Завдання 5. reshape і broadcasting- Збираємо `rows` форми `(2, 12)`: перший рядок — `temps`, другий — `days_in_month`.- Створюємо `correction = np.array([1, 0])`.- Пробуємо `rows + correction`, дивимось на помилку й порівнюємо форми.- Перетворюємо `correction` на форму `(2, 1)` через `.reshape(2, 1)` і додаємо ще раз.- Пояснюємо словами, чому тепер спрацювало.

```python
rows = np.array([temps, days_in_month])
print("rows.shape:", rows.shape)
print("rows:")
print(rows)

correction = np.array([1, 0])
print("\ncorrection.shape:", correction.shape)
print("correction:", correction)
```

```python
# Спроба додавання — має впасти з помилкою через несумісність форм
try:
    bad = rows + correction
    print(bad)
except ValueError as e:
    print("Помилка:", e)

# Аналіз форм
print("\nАналіз несумісності:")
print(f"  rows.shape = {rows.shape}  -> виміри (2, 12)")
print(f"  correction.shape = {correction.shape}  -> виміри (2,)")
print("  NumPy порівнює виміри справа наліво: 12 (у rows) vs 2 (у correction).")
print("  12 != 2, і жоден з них не дорівнює 1 -> broadcasting неможливий.")
```

```python
# Перетворюємо correction на форму (2, 1) і додаємо ще раз
correction_2d = correction.reshape(2, 1)
print("correction.reshape(2, 1).shape:", correction_2d.shape)
print("correction_2d:")
print(correction_2d)

result = rows + correction_2d
print("\nrows + correction_2d =")
print(result)

print("\nПояснення:")
print("  Тепер correction має форму (2, 1). NumPy порівнює виміри справа наліво:")
print("    12 vs 1 -> 1 'розтягується' до 12 (broadcast),")
print("     2 vs 2 -> збігаються.")
print("  По суті, вектор-стовпець [1]^T дублюється 12 разів по горизонталі,")
print("  а [0]^T — так само. До першого рядка (temps) додається 1,")
print("  до другого рядка (days_in_month) додається 0.")
```
