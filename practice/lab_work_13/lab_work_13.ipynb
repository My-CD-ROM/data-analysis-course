````python
import numpy as np
from scipy import stats

variant = 1
july_loc = 21.0   # параметр генерації вибірки (Київ)
mu0 = 20.0        # стара норма з довідника

np.random.seed(variant)
daily_temps = np.random.normal(loc=july_loc, scale=2.5, size=30)
np.random.seed()

print(daily_temps)
````

````python
n = len(daily_temps)
x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)

print("n =", n)
print("x̄ (точкова оцінка середнього) = {:.4f}".format(x_bar))
print("s  (точкова оцінка σ)         = {:.4f}".format(s))
print("Стара норма μ₀                = {:.1f}".format(mu0))
````

````python
se = s / np.sqrt(n)                       # стандартна похибка
z = (x_bar - mu0) / se                    # z-статистика
p_value = 2 * (1 - stats.norm.cdf(abs(z)))  # двобічна формула (як у прикладі)

print("se = {:.4f}".format(se))
print("z  = {:.4f}".format(z))
print("p-value (двобічний) = {:.4f}".format(p_value))
````

````python
alpha = 0.05

if p_value < alpha:
    decision = "ВІДХИЛЯЄМО H0"
else:
    decision = "НЕ відхиляємо H0"

print("α = {:.2f}, p-value = {:.4f}".format(alpha, p_value))
print("Рішення:", decision)
````

````python
# 95% довірчий інтервал для середнього (z-наближення, як просить завдання: x̄ ± 1.96·se)
ci_low  = x_bar - 1.96 * se
ci_high = x_bar + 1.96 * se

inside = ci_low <= mu0 <= ci_high

print("95% CI для середнього: ({:.4f}, {:.4f})".format(ci_low, ci_high))
print("μ₀ = {:.1f} всередині інтервалу? {}".format(mu0, inside))
print("Узгодженість: CI виключає μ₀ <=> тест відхиляє H0 при α=0.05")
````

## Контрольні питання

### 1. Що означає p-value і чому це НЕ «ймовірність правильності μ₀»

p-value — це ймовірність отримати вибіркове середнє, відхилене від μ₀ так само сильно або сильніше, **за умови, що H0 істинна** (стара норма правильна). Тобто P(такі або екстремальніші дані | H0). Це ймовірність даних, а не гіпотези. Твердження «p = 0.006 — це ймовірність, що μ₀ = 20 правильне» є помилковим: у частотній статистиці гіпотеза або істинна, або ні, вона не має ймовірності. p-value описує, наскільки несумісні дані з H0, а не ймовірність самої H0.

### 2. Чому тип тесту обирають ДО перегляду результату

Якщо обрати однобічний тест після того, як побачив, куди «схилилось» x̄, можна штучно подвоїти шанси відхилити H0 (бо однобічний p-value удвічі менший за двобічний у той самий бік). Це називається p-hacking і руйнує валідність: рівень α перестає контролюватись. Тип тесту визначається змістом питання дослідника, а не даними, інакше висновок стає упередженим.

### 3. Помилка I vs помилка II для цих даних

Це різні помилки, бо вони протилежні за напрямком і наслідками:
- **I рід:** сказали «клімат змінився», а він не змінився (хибна тривога, зайве оновлення довідника).
- **II рід:** сказали «клімат не змінився», а він змінився (проґавили реальне потепління, довідник лишився хибним).
Перша — «крик „пожежа“ без пожежі», друга — «мовчання під час пожежі». Вони конфліктують: зменшити одну зазвичай означає збільшити іншу.

### 4. Чи узгодився тест із довірчим інтервалом і чому так має бути завжди

Так, узгодився. Двобічний тест при α = 0.05 відхиляє H0 тоді й тільки тоді, коли відповідний 95% довірчий інтервал **не** містить μ₀. Це математична тотожність: CI будується з тієї самої нерівності |x̄ − μ₀| > крит_значення · se, яку використовує тест. Тому за однакових даних і одного α обидва методи дають однакове рішення — якщо ні, десь помилка в обчисленнях.
