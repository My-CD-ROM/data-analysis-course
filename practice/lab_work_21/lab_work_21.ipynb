```python
import numpy as np
import matplotlib.pyplot as plt

# Варіант 1 (Київ, тип: площа, чверть кола)
# Область: квадрат [0, 1] × [0, 1], площа квадрата = 1
# Умова потрапляння у фігуру: x² + y² <= 1
# Точне значення: np.pi / 4 ≈ 0.785398

np.random.seed(1)
exact_value = np.pi / 4

N = 1000
x = np.random.uniform(0, 1, N)
y = np.random.uniform(0, 1, N)

inside = (x**2 + y**2) <= 1
estimate = inside.sum() / N  # оскільки площа охоплюючого квадрата дорівнює 1

print(f"Точне значення: {exact_value:.5f}")
print(f"Оцінка методом Монте-Карло (N={N}): {estimate:.5f}")
```

```python
abs_error = abs(estimate - exact_value)
relative_error_pct = (abs_error / exact_value) * 100

print(f"Абсолютна похибка: {abs_error:.5f}")
print(f"Відносна похибка: {relative_error_pct:.2f}%")
```

## Контрольні питання

1. **Чому оцінка Монте-Карло — це вибіркове середнє, і який результат Лекції 12 це гарантує?**
   Оцінка Монте-Карло обчислюється як середнє арифметичне випадкових індикаторів або значень функції. Закон великих чисел (ЗВЧ) гарантує її збіжність до істинного значення.

2. **Чому вчетверо більша кількість випробувань дає лише вдвічі меншу похибку?**
   Через наявність квадратичного кореню в знаменнику формули стандартної похибки вибіркового середнього ($\frac{\sigma}{\sqrt{N}}$): $\frac{1}{\sqrt{4N}} = \frac{1}{2 \sqrt{N}}$.

3. **Для варіанта типу "площа": чому частку точок треба множити на площу прямокутника-охоплення?**
   Метод Монте-Карло для площі обчислює безрозмірну *частку* площі фігури відносно площі зовнішньої прямокуттної області. Щоб отримати абсолютне значення площі фігури, цю частку необхідно помножити на загальну площу охоплюючого прямокутника.

4. **Для варіанта типу "сподівання": чому саме `.min(axis=1)` чи `.max(axis=1)` дає масив результатів окремо для кожного випробування?**
   Використання параметра `axis=1` виконує операцію мінімуму чи максимуму горизонтально в межах кожного окремого рядка матриці (де кожен рядок відповідає одному незалежному експерименту/випробуванню з кількома кидками кубика), утворюючи одновимірний масив результатів для подальшого усереднення.
