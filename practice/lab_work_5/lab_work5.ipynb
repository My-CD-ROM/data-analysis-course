### Завдання 1. Класифікація змінних за шкалою вимірювання

1. **Обґрунтування шкал вимірювання:**
   - **Місто**: *Номінальна шкала*. Значення є текстовими мітками без будь-якого природного впорядкування чи математичного співвідношення.
   - **Номер місяця (1–12)**: *Дискусійна шкала (Порядкова / Інтервальна)*. 
     - *Аргумент за порядкову:* Номери місяців мають чіткий хронологічний порядок (1 передує 2), але у цієї шкали **немає природного нуля** ("0-й місяць" як "відсутність місяця" не існує). Крім того, місяці мають різну кількість днів (28–31), тому проміжки між ними не є строго однаковими.
     - *Аргумент за інтервальну:* Календарний рік є циклічним із рівномірними часовими проміжками (приблизно 30 діб), що дозволяє порівнювати інтервали (наприклад, різниця між 1 і 3 місяцем така ж, як між 6 і 8).
   - **Кількість опадів (мм)**: *Відносна шкала (шкала відношень)*. Значення є неперервними числами, мають **природний (абсолютний) нуль** (0 мм означає повну відсутність опадів) та дозволяють виконувати всі арифметичні операції, зокрема ділення (80 мм опадів — це удвічі більше, ніж 40 мм).

2. **Допустимі міри положення:**
   - **Місто**: тільки **мода** (найчастіше викликуване місто). Середнє та медіана не мають сенсу.
   - **Номер місяця**: **мода** та **медіана** (якщо розглядати як порядкову); **середнє** також можна обчислювати математично, якщо розглядаємо як інтервальну, але воно потребує обережної інтерпретації через циклічність року.
   - **Кількість опадів**: коректні **всі три міри** (середнє, медіана, мода), а також квантилі.

### Завдання 2. Варіаційний ряд і частотна таблиця

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Дані вашого варіанта (підставте свої значення):
precip = pd.Series([36, 39, 39, 46, 57, 82, 71, 60, 57, 41, 50, 47])

# 1. Варіаційний ряд (сортування за зростанням)
var_series = np.sort(precip.values)
print("Варіаційний ряд (опади за зростанням):")
print(var_series)

# 2. Розбиття на 4 інтервали
bins = [35, 47, 59, 71, 83]
categories = pd.cut(precip, bins=bins, right=True)

# 3. Частотна таблиця
freq_abs = categories.value_counts().sort_index()
freq_rel = (categories.value_counts(normalize=True).sort_index() * 100).round(1)
freq_cum = freq_abs.cumsum()

freq_table = pd.DataFrame({
    "Абсолютна частота": freq_abs,
    "Відносна частота (%)": freq_rel,
    "Накопичена частота": freq_cum
})

print("\n--- Частотна таблиця ---")
display(freq_table)
```

**Відповідь на питання з Завдання 2:**
Межа третього інтервалу становить **71 мм**. 
Накопичена частота для цього інтервалу дорівнює **11** (із 12 місяців), що становить **91.7%** від усіх місяців року. Тобто, у 91.7% місяців року кількість опадів не перевищує 71 мм.

### Завдання 3. Гістограма та оцінка кількості інтервалів (bins)

```python
n = len(precip)

# Розрахунок кількості інтервалів за правилами:
bins_sqrt = int(np.ceil(np.sqrt(n)))               # Правило sqrt(n)
bins_sturges = int(np.ceil(1 + np.log2(n)))        # Правило Стерджеса

print(f"Кількість спостережень (n): {n}")
print(f"Кількість bins за правилом √n: {bins_sqrt}")
print(f"Кількість bins за правилом Стерджеса: {bins_sturges}")

# Побудова трьох гістограм для порівняння
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. Мала кількість bins (bins = 3)
axes[0].hist(precip, bins=3, color='skyblue', edgecolor='black')
axes[0].set_title("bins = 3 (Занадто згладжена)")
axes[0].set_xlabel("Опади (мм)")
axes[0].set_ylabel("Частота")

# 2. Рекомендована кількість bins (bins = 4 або 5, за Стерджесом / √n)
axes[1].hist(precip, bins=bins_sturges, color='lightgreen', edgecolor='black')
axes[1].set_title(f"bins = {bins_sturges} (За Стерджесом — Оптимальна)")
axes[1].set_xlabel("Опади (мм)")

# 3. Велика кількість bins (bins = 10)
axes[2].hist(precip, bins=10, color='salmon', edgecolor='black')
axes[2].set_title("bins = 10 (Надмірно подрібнена)")
axes[2].set_xlabel("Опади (мм)")

plt.tight_layout()
plt.show()
```

**Аналіз гістограм:**
- Для $n = 12$ правило $\sqrt{n} = \sqrt{12} \approx 3.46 \rightarrow 4$, а правило Стерджеса $k = 1 + \log_2(12) \approx 4.58 \rightarrow 5$.
- Гістограма з **3-4 інтервалами (bins)** найкраще показує форму розподілу опадів протягом року: вона виявляє помірну правосторонню асиметрію (більшість місяців зосереджена в діапазоні 35–60 мм, і лише 1-2 літні місяці є піково вологими) без зайвого шуму чи надмірного згладжування.

### Завдання 4. Міри положення й квантилі

```python
mean_val = precip.mean()
median_val = precip.median()
mode_val = precip.mode().values  # може бути декілька мод

q1 = precip.quantile(0.25)
q3 = precip.quantile(0.75)

print(f"Середнє (Mean): {mean_val:.2f} мм")
print(f"Медіана (Median): {median_val:.2f} мм")
print(f"Мода (Mode): {mode_val} мм")
print(f"Перший квартиль (Q1 / 25%): {q1:.2f} мм")
print(f"Третій квартиль (Q3 / 75%): {q3:.2f} мм")
```

**Аналіз асиметрії розподілу:**
Середнє значення ($52.08$ мм) є більшим за медіану ($48.50$ мм). Ця різниця свідчить про наявність **правосторонньої (позитивної) асиметрії** у розподілі опадів. Це цілком відповідає кліматичній логіці: більшість місяців року є відносно помірно зволоженими (сухішими), але 1–2 літні місяці з рясними зливами (викиди/піки вгору) "затягують" середнє арифметичне у більший бік.

### Завдання 5. Зв'язок зі шкалою вимірювання

**Пояснення:**
Змінна "кількість опадів (мм)" належить до **відносної шкали**, оскільки має природний нуль (0 мм = відсутність опадів) та постійні одиниці вимірювання. Це дозволяє виконувати додавання, віднімання, множення та ділення. Завдяки цьому для неї коректні:
1. **Мода** (знаходження найпоширенішого рівня опадів);
2. **Медіана** (знаходження середини впорядкованого ряду, оскільки є відношення порядку);
3. **Середнє арифметичне** (сумування значень і ділення на $n$, оскільки існує абсолютна одиниця вимірювання та арифметика дійсних чисел).

На противагу цьому, змінна **"місто"** належить до **номінальної шкали**. Міста — це лише якісні текстові назви. Для них не існує операцій додавання чи впорядкування ("Київ" + "Львів" не має математичного сенсу). Тому для номінальної шкали коректно вираховувати **виключно моду** (назва міста, яке зустрічається найчастіше).

### Відповіді на підсумкові питання:

**1. Чи можете обґрунтувати, чому середнє арифметичне некоректне для номінальної змінної "місто", навіть якщо завжди можна закодувати міста числами?**
Числове кодування (наприклад, Київ = 1, Львів = 2, Одеса = 3) є штучними умовними позначеннями, а не кількісними вимірами. Математичні операції над ними позбавлені змісту: середня величина $\frac{1 + 3}{2} = 2$ не означає, що середовище між Києвом та Одесою — це Львів. Числа в номінальній шкалі слугують лише замінниками імен, тому обчислення середнього арифметичного дає безглуздий результат.

**2. Чи розумієте, чому "номер місяця" — приклад, де класифікація за шкалою вимірювання не однозначна, на відміну від явно порядкової чи явно відносної змінної?**
"Номер місяця" має характеристики порядкової шкали (місяці йдуть у суворому порядку 1...12), але відрізняється від класичної відносної шкали тим, що **не має абсолютного нуля** (не існує 0-го місяця як відсутності часу). Водночас проміжки між місяцями є майже однаковими (календарні дні), що наближає її до інтервальної шкали. Додаткова складність полягає в циклічності часу (грудень 12 межує з січнем 1), що робить звичайну лінійну арифметику не зовсім коректною.

**3. Чи можете пояснити, чому для "кількості опадів" (майже всі 12 значень різні) потрібна саме гістограма з розбиттям на інтервали, а не стовпчикова діаграма за окремими значеннями?**
Оскільки "кількість опадів" є неперервною кількісною змінною і майже всі 12 значень у вибірці є унікальними числами, стовпчикова діаграма за окремими значеннями створила б 12 однакових за висотою стовпчиків (частота = 1 для кожного) і не показала б жодної структури розподілу. Розбиття на інтервали (групування) дозволяє об'єднати близькі значення та побачити реальний розподіл (де спостерігається скупчення опадів, а де — сухі періоди).

**4. Чи можете пояснити змістовно, що означає накопичена частота на межі вашого третього інтервалу з Завдання 2?**
Накопичена (кумулятивна) частота показує загальну кількість спостережень, чиї значення **не перевищують верхню межу** даного інтервалу. На межі 3-го інтервалу (71 мм) накопичена частота показує, скільки саме місяців у році (і яка їх частка у відсотках) мали кількість опадів $ \le 71$ мм. Це дає змогу швидко оцінити, яка більшість року характеризується помірним рівнем опадів.
