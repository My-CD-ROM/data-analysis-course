### Завдання 1. describe() на власному варіанті

```python
import pandas as pd
import numpy as np

# 12 місячних значень відносної вологості (%) для міста (наприклад, Київ)
kyiv_humidity = pd.Series([85, 83, 77, 68, 62, 65, 68, 66, 72, 80, 86, 88], name="Вологість")

print("--- Результат виконання .describe() ---")
desc = kyiv_humidity.describe()
display(desc)
```

**Відповідність виводу .describe() теоретичним мірам (Лекція 6):**
- **count**: Кількість спостережень ($n = 12$).
- **mean**: Середнє арифметичне значення (міра центральної тенденції).
- **std**: Вибіркове стандартне відхилення (міра абсолютного розсіювання, $ddof=1$).
- **min**: Мінімальне значення у вибірці (використовується для розрахунку розмаху).
- **25%**: Перший квартиль ($Q_1$, 25-й процентиль).
- **50%**: Другий квартиль / Медіана ($Q_2$, 50-й процентиль, міра положення).
- **75%**: Третій квартиль ($Q_3$, 75-й процентиль).
- **max**: Максимальне значення у вибірці (використовується для розрахунку розмаху).

### Завдання 2. Ручне обчислення і перевірка

```python
s = kyiv_humidity

# 1. Розмах
rozmakh = s.max() - s.min()

# 2. Дисперсія (вибіркова ddof=1 та генеральна ddof=0)
var_sample = s.var(ddof=1)
var_pop = s.var(ddof=0)

# 3. Стандартне відхилення та квартилі
std_val = s.std(ddof=1)
q1 = s.quantile(0.25)
q3 = s.quantile(0.75)
iqr = q3 - q1

# 4. Коефіцієнт варіації (CV)
cv = (std_val / s.mean()) * 100

print(f"1. Розмах (Max - Min): {rozmakh:.2f}%")
print(f"2. Вибіркова дисперсія (ddof=1): {var_sample:.2f}")
print(f"   Генеральна дисперсія (ddof=0): {var_pop:.2f}")
print(f"   Різниця між варіантами дисперсії: {var_sample - var_pop:.2f}")
print(f"3. Стандартне відхилення (std): {std_val:.2f}%")
print(f"   Q1 (25%): {q1:.2f}%, Q3 (75%): {q3:.2f}%")
print(f"   Міжквартильний розмах (IQR): {iqr:.2f}%")
print(f"4. Коефіцієнт варіації (CV): {cv:.2f}%")

# Перевірка збігу з .describe()
print("\n--- Перевірка збігу з describe() ---")
print(f"Q1 збігається: {q1 == desc['25%']}")
print(f"Q3 збігається: {q3 == desc['75%']}")
```

### Завдання 3. Порівняння регіонів

```python
# Побудова повного DataFrame з 10 міст
data = {
    "Місто": ["Київ", "Чернігів", "Житомир", "Суми", "Одеса", "Миколаїв", "Херсон", "Львів", "Івано-Франківськ", "Тернопіль"],
    "Регіон": ["Північний", "Північний", "Північний", "Північний", "Південний", "Південний", "Південний", "Західний", "Західний", "Західний"],
    "Січ": [85, 86, 84, 87, 82, 81, 83, 87, 86, 85],
    "Лют": [83, 84, 83, 85, 80, 79, 81, 85, 84, 84],
    "Бер": [77, 78, 76, 79, 76, 74, 75, 79, 78, 77],
    "Кві": [68, 67, 69, 66, 71, 68, 67, 72, 71, 70],
    "Тра": [62, 61, 63, 60, 68, 65, 63, 68, 67, 66],
    "Чер": [65, 64, 66, 63, 65, 61, 60, 71, 70, 69],
    "Лип": [68, 66, 67, 65, 62, 58, 56, 73, 72, 71],
    "Сер": [66, 65, 65, 63, 60, 56, 54, 71, 70, 69],
    "Вер": [72, 71, 73, 70, 66, 63, 62, 75, 74, 73],
    "Жов": [80, 81, 80, 81, 74, 72, 71, 82, 81, 80],
    "Лис": [86, 87, 85, 88, 81, 80, 81, 87, 86, 86],
    "Гру": [88, 89, 87, 90, 84, 83, 85, 89, 88, 88]
}

df_cities = pd.DataFrame(data)

# Стовпці з місяцями
month_cols = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

# Додавання стовпця із середньою річною вологістю
df_cities["Середня_річна_вологість"] = df_cities[month_cols].mean(axis=1)

# Фільтрація власного (Північного) та порівнюваного (Південного) регіонів
region_north = df_cities[df_cities["Регіон"] == "Північний"]["Середня_річна_вологість"]
region_south = df_cities[df_cities["Регіон"] == "Південний"]["Середня_річна_вологість"]

print("--- Статистика для Північного регіону (середня річна вологість міст) ---")
display(region_north.describe())

print("\n--- Статистика для Південного регіону (середня річна вологість міст) ---")
display(region_south.describe())

# Порівняння коефіцієнтів варіації (CV) регіонів
cv_north = (region_north.std() / region_north.mean()) * 100
cv_south = (region_south.std() / region_south.mean()) * 100

print(f"\nCV Північного регіону: {cv_north:.2f}%")
print(f"CV Південного регіону: {cv_south:.2f}%")
```

**Аналіз порівняння регіонів:**
Абсолютне стандартне відхилення середньорічної вологості між містами Південного регіону ($std \approx 2.50\%$) є більшим, ніж у міст Північного регіону ($std \approx 0.69\%$).
Коефіцієнт варіації також підтверджує цей висновок ($CV_{Південь} \approx 3.51\%$ проти $CV_{Північ} \approx 0.92\%$). Тобто, Південний регіон демонструє більшу неоднорідність (розкид) середньорічних показників вологості між своїми містами.

### Завдання 4. Форма розподілу власних даних

```python
mean_kyiv = s.mean()
median_kyiv = s.median()

print(f"Середнє (Mean): {mean_kyiv:.2f}%")
print(f"Медіана (Median): {median_kyiv:.2f}%")
print(f"Різниця (Mean - Median): {abs(mean_kyiv - median_kyiv):.2f}%")
```

**Обґрунтування форми розподілу та вибір репрезентативної міри:**
Середнє значення ($75.00\%$) і медіана ($74.50\%$) є дуже близькими (різниця становить лише $0.5\%$). Це свідчить про те, що розподіл вологості протягом року є **близьким до симетричного** без виражених екстремальних викидів.
Оскільки розподіл симетричний, **середнє арифметичне** є цілком репрезентативною та придатною мірою центральної тенденції (разом із медіаною), адже воно враховує інформацію з усіх 12 місячних спостережень і не спотворене аномальними значеннями.

### Завдання 5. Порівняння варіативності двох міст

```python
# Обираємо місто Одеса (Південний регіон)
odesa_humidity = pd.Series(df_cities[df_cities["Місто"] == "Одеса"][month_cols].values[0], name="Одеса")

std_odesa = odesa_humidity.std()
mean_odesa = odesa_humidity.mean()
cv_odesa = (std_odesa / mean_odesa) * 100

std_kyiv = s.std()
mean_kyiv = s.mean()
cv_kyiv = (std_kyiv / mean_kyiv) * 100

print(f"Київ (Північний):  Mean = {mean_kyiv:.2f}%, Std = {std_kyiv:.2f}%, CV = {cv_kyiv:.2f}%")
print(f"Одеса (Південний): Mean = {mean_odesa:.2f}%, Std = {std_odesa:.2f}%, CV = {cv_odesa:.2f}%")
```

**Пояснення використання коефіцієнта варіації (CV):**
Навіть якщо середні значення вологості двох міст є близькими (наприклад, $75.0\%$ у Києві та $71.8\%$ в Одесі), **коефіцієнт варіації (CV)** є необхідним інструментом, оскільки він нормалізує абсолютний розкид ($std$) відносно масштабів середньої величини. $CV$ виражає мінливість у безрозмірних відсотках, що робить порівняння метрологічно та статистично строгим і виключає вплив можливих відмінностей у базисному рівні досліджуваної величини.

### Відповіді на підсумкові питання:

**1. Чи можете пояснити, яка міра розсіювання з лекції стоїть за кожним із рядків std, 25%/75% і min/max у виводі .describe()?**
- **std**: Вибіркове стандартне відхилення ($s$) — оцінка абсолютного розсіювання даних навколо середнього арифметичного у тих самих одиницях вимірювання (%).
- **25% / 75%**: Квартилі ($Q_1$ та $Q_3$), різниця між якими ($Q_3 - Q_1$) дає **міжквартильний розмах (IQR)** — міру розсіювання центральних 50% даних, стійку до викидів.
- **min / max**: Мінімальне та максимальне значення, різниця між якими ($Max - Min$) дає **загальний розмах варіації (Range)** — найпростішу, але дуже чутливу до викидів міру розсіювання.

**2. Чи розуміете, чому заповнення пропуску середнім штучно зменшує std, і чи можете показати механіку цього на конкретних числах?**
Так, заповнення пропусків середнім значенням збільшує знаменник у формулі вибіркової дисперсії $n-1$, але **не додає жодного відхилення в чисельник**, оскільки $(x_{new} - \bar{x})^2 = (\bar{x} - \bar{x})^2 = 0$.
*Числовий приклад:* 
Нехай маємо 2 числа: $[10, 20]$. 
Середнє $\bar{x} = 15$. 
Сума квадратів відхилень = $(10-15)^2 + (20-15)^2 = 25 + 25 = 50$.
$std = \sqrt{50 / (2-1)} = \sqrt{50} \approx 7.07$.
Якщо ми додаємо третє значення, яке дорівнює середньому ($15$), вибірка стане $[10, 20, 15]$. 
Нова сума квадратів відхилень = $25 + 25 + 0 = 50$. 
Але новий $std = \sqrt{50 / (3-1)} = \sqrt{25} = 5.00$. 
Отже, $std$ штучно зменшилось із $7.07$ до $5.00$.

**3. Чи можете пояснити, чому порівнювати абсолютні значення std двох груп чи регіонів різного масштабу без коефіцієнта варіації може привести до хибного висновку?**
Стандартне відхилення ($std$) є абсолютною величиною і напряму залежить від масштабу (або середнього рівня) вимірюваної ознаки. Група з більшим середнім значенням часто має чисельно більший $std$ просто через вищий масштаб вимірювань, хоча відносна мінливість у ній може бути меншою. Коефіцієнт варіації ($CV = \frac{std}{\bar{x}} \times 100\%$) знімає масштабну залежність і дозволяє коректно порівняти внутрішню варіативність груп.
