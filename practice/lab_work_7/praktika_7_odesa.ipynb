### 4. Що таке chartjunk і чому він шкодить сприйняттю графіка?

Chartjunk — це зайві декоративні елементи графіка, які не додають корисної інформації та можуть заважати правильному сприйняттю даних.

Наприклад, для звичайної стовпчикової діаграми непотрібний 3D-ефект може створювати додаткові перспективні спотворення. Передня частина стовпчика може здаватися більшою або меншою, ніж вона є насправді.

Тому 3D-ефекти, надмірна кількість кольорів, декоративні елементи та зайві тіні можуть ускладнювати порівняння значень.

Для статистичних графіків краще використовувати просте оформлення, яке допомагає побачити дані, а не відволікає від них.


### 3. Чому підписи осей, одиниці вимірювання і заголовок — не оздоблення?

Підписи осей, одиниці вимірювання та заголовок визначають, що саме показує графік.

Наприклад, число 55 без контексту може означати 55 продажів, 55 °C, 55 людей або 55 гривень. Підпис «Продажі, шт./день» визначає зміст цього числа.

Заголовок додатково пояснює, які саме дані представлені та за який період. Тому підписи й заголовок є частиною інформації, яку передає графік, а не просто декоративними елементами.


### 2. Коли лінійний графік буде методологічною помилкою для категоріальних даних?

Лінійний графік може бути методологічною помилкою, коли точки належать до номінальних категорій, між якими немає природного порядку.

Наприклад, якщо порівнювати кількість продажів у містах Одеса, Львів, Київ і Харків, не варто просто з'єднувати ці точки лінією. Порядок міст на осі X є довільним, а лінія створює враження безперервної зміни від одного міста до іншого.

Міста є номінальною змінною, тому між ними немає числової відстані або природного порядку. Для такого порівняння краще використати стовпчикову діаграму.

Лінійний графік доречніший тоді, коли по осі X є впорядкована кількісна або часова змінна, де з'єднання сусідніх точок має зміст.


### 1. Чому обрізана вісь Y перебільшує різницю між категоріями, хоча підписані числа правильні?

Обрізана вісь Y змінює візуальний масштаб графіка. Якщо почати вісь не з нуля, то відносно невелика різниця між значеннями може займати більшу частину висоти рисунка.

Наприклад, різниця між 4 і 5 становить лише 1 одиницю. Якщо вісь починається з 0, ця різниця виглядає невеликою. Якщо ж вісь починається з 3, стовпчики висотою 4 і 5 візуально відрізняються набагато сильніше.

Тому числа на осі можуть залишатися правильними, але саме графічне представлення створює перебільшене враження про величину різниці.


### Пояснення спотворення

У правій частині рисунка я навмисно обрізав вісь Y: замість початку від 0 вона починається зі значення 3.

Це є спотворенням масштабу графіка. Реальні значення частот при цьому не змінюються, але візуальні відмінності між стовпцями стають перебільшеними.

Наприклад, якщо один інтервал має частоту 4, а інший — 5, реальна різниця становить лише 1 день. Але якщо вісь починається з 3, різниця між висотами стовпців візуально здається значно більшою.

Отже, обрізання осі може створити у глядача неправильне враження про величину відмінностей, хоча всі підписані числові значення залишаються правильними.

Коректна гістограма повинна мати такий масштаб, який не перебільшує реальні відмінності між значеннями.


```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].hist(
    kiosk["продажі"],
    bins=6,
    edgecolor="black"
)

axes[0].set_xlabel("Продажі, шт./день")
axes[0].set_ylabel("Кількість днів")
axes[0].set_title("Коректна гістограма")

axes[0].set_ylim(bottom=0)


axes[1].hist(
    kiosk["продажі"],
    bins=6,
    edgecolor="black"
)

axes[1].set_xlabel("Продажі, шт./день")
axes[1].set_ylabel("Кількість днів")
axes[1].set_title("Спотворена гістограма")

axes[1].set_ylim(3, None)


plt.tight_layout()
plt.show()
```

# Завдання 5. Навмисно спотворений графік

Для демонстрації спотворення побудуємо дві однакові гістограми поруч.

Ліворуч буде коректна версія з віссю Y, яка починається від 0.

Праворуч буде навмисно спотворена версія, де вісь Y обрізана. Через це невеликі відмінності між частотами інтервалів візуально здаватимуться значно більшими.


### Висновок за діаграмою розсіювання

На діаграмі розсіювання не спостерігається чіткої сильної лінійної залежності між температурою та кількістю продажів.

Це очікувано, оскільки під час генерації даних продажі залежали від базового рівня, випадкової варіації та збільшення у вихідні дні, але безпосередня залежність продажів від температури не задавалася.

Тому за цим набором не можна стверджувати, що вища температура однозначно призводить до більшої або меншої кількості продажів.


```python
fig, ax = plt.subplots(figsize=(7, 5))

ax.scatter(
    kiosk["температура"],
    kiosk["продажі"]
)

ax.set_xlabel("Температура, °C")
ax.set_ylabel("Продажі, шт./день")
ax.set_title("Залежність продажів від температури в Одесі")

plt.show()
```

# Завдання 4. Діаграма розсіювання

Побудуємо діаграму розсіювання, де по осі X буде температура повітря, а по осі Y — кількість продажів за день.


### Висновок за boxplot

На boxplot видно, що медіана продажів у вихідні дні вища, ніж у будні. Також розташування всього інтерквартильного діапазону вихідних зміщене в бік більших значень.

Це узгоджується зі способом генерації даних у Завданні 1: для вихідних днів ми спеціально збільшили базові продажі на 15–25%.

Отже, графік візуально підтверджує задану нами закономірність — у вихідні дні кіоск продає більше напоїв.


```python
fig, ax = plt.subplots(figsize=(7, 5))

weekday_sales = kiosk.loc[
    ~kiosk["вихідний"],
    "продажі"
]

weekend_sales = kiosk.loc[
    kiosk["вихідний"],
    "продажі"
]

groups = [
    weekday_sales,
    weekend_sales
]

ax.boxplot(
    groups,
    tick_labels=["Будні", "Вихідні"]
)

ax.set_xlabel("Тип дня")
ax.set_ylabel("Продажі, шт./день")
ax.set_title("Порівняння продажів у будні та вихідні")

plt.show()
```
