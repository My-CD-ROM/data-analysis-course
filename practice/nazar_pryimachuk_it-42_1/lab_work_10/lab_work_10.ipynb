# Практика 10. Неперервні розподіли в scipy.stats

**Варіант:** 7  
**Місто:** Івано-Франківськ  
**Липнева середньомісячна температура:** 17 °C

```python
import numpy as np
from scipy import stats 
import matplotlib.pyplot as plt
```

## Завдання 1. Побудова вибірки денних температур

Варіант 7 — Івано-Франківськ. Липнева середньомісячна температура становить 17 °C.
На її основі генеруємо вибірку з 30 умовних денних температур.

```python
np.random.seed(7)

daily_temps = np.random.normal(
    loc=17,
    scale=2.5,
    size=30
)

print("Вибірка денних температур:")
print(daily_temps)

print("\nСереднє значення:", daily_temps.mean())
print("Стандартне відхилення:", daily_temps.std())
```

## Завдання 2. Параметри нормального розподілу

Параметри нормального розподілу визначаємо безпосередньо з отриманої вибірки. Середнє значення є параметром μ, а стандартне відхилення — параметром σ.

```python
mean = daily_temps.mean()
std = daily_temps.std()

dist = stats.norm(loc=mean, scale=std)

print("Середнє значення (mu):", mean)
print("Стандартне відхилення (sigma):", std)
print("Нормальний розподіл створено")
```

## Завдання 3. Гістограма і теоретична pdf-крива Будуємо гістограму денних температур з параметром density=True та накладаємо на неї теоретичну криву нормального розподілу.

```python
x = np.linspace(
    daily_temps.min() - 2,
    daily_temps.max() + 2,
    200
)

plt.figure(figsize=(10, 6))

plt.hist(
    daily_temps,
    bins=8,
    density=True,
    edgecolor="white",
    alpha=0.7,
    label="Daily temperatures"
)

plt.plot(
    x,
    dist.pdf(x),
    linewidth=2,
    label="Normal PDF"
)

plt.xlabel("Temperature, °C")
plt.ylabel("Density")
plt.title("Daily temperatures and normal distribution")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

## Завдання 4. Функція розподілу CDF

Обчислюємо значення функції розподілу (CDF) для отриманої вибірки та будуємо графік теоретичної функції розподілу нормального розподілу.

x_sorted = np.sort(daily_temps)

cdf_values = dist.cdf(x_sorted)

plt.figure(figsize=(10, 6))

plt.plot(
    x_sorted,
    cdf_values,
    marker="o",
    label="Normal CDF"
)

plt.xlabel("Temperature, °C")
plt.ylabel("Cumulative probability")
plt.title("Cumulative distribution function")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
