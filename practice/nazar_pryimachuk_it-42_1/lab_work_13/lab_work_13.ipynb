# Практика 13. Точкові оцінки та перевірка гіпотез

**Варіант 7 — Івано-Франківськ**

Поточне липневе значення для генерації вибірки: **17 °C**

Стара норма середньої липневої температури:
**μ₀ = 16.4 °C**

У роботі перевіряється, чи узгоджуються поточні дані зі старою кліматичною нормою.

```python
import numpy as np
from scipy import stats
```

```python
np.random.seed(7)

july_temperature = 17
mu0 = 16.4

daily_temps = np.random.normal(
    loc=july_temperature,
    scale=2.5,
    size=30
)

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)
n = len(daily_temps)

print("Вибірка денних температур:")
print(daily_temps)

print("\nСереднє x_bar:", x_bar)
print("Стандартне відхилення s:", s)
print("Розмір вибірки n:", n)
print("Стара норма mu0:", mu0)
```

### Висновок до Завдання 1

Було згенеровано 30 поточних денних температур для Івано-Франківська навколо липневого значення 17 °C. Вибіркове середнє x̄ та вибіркове стандартне відхилення s є точковими оцінками відповідних параметрів генеральної сукупності.

## Завдання 2. Формулювання H0 і H1

Нульова гіпотеза:

**H0: μ = 16.4 °C**

Альтернативна гіпотеза:

**H1: μ ≠ 16.4 °C**

Обираємо двобічний тест, оскільки твердження "стара норма більше не відповідає дійсності" допускає відхилення середньої температури як у бік потепління, так і в бік похолодання. Тип тесту обрано до перегляду результату та p-value, щоб не підлаштовувати статистичний аналіз під отримане вибіркове середнє.

```python
se = s / np.sqrt(n)

z = (x_bar - mu0) / se

p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("Стандартна похибка:", se)
print("z-статистика:", z)
print("p-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    decision = "Відхиляємо H0"
else:
    decision = "Не відхиляємо H0"

print("Рівень значущості alpha:", alpha)
print("p-value:", p_value)
print("Рішення:", decision)
```

### Висновок до Завдання 4

При рівні значущості α = 0.05 рішення приймається шляхом порівняння p-value з α. Якщо p-value < 0.05, нульову гіпотезу відхиляємо; якщо p-value ≥ 0.05, немає достатніх статистичних підстав її відхилити.

Для Івано-Франківська отримане значення p-value та відповідне рішення показують, чи є статистично значущі підстави вважати, що середня липнева температура відрізняється від старої норми 16.4 °C.

```python
ci_95 = (
    x_bar - 1.96 * se,
    x_bar + 1.96 * se
)

print("95% довірчий інтервал:", ci_95)
print("Стара норма mu0:", mu0)

if ci_95[0] <= mu0 <= ci_95[1]:
    print("mu0 потрапляє у довірчий інтервал.")
else:
    print("mu0 НЕ потрапляє у довірчий інтервал.") 
```

### Порівняння з перевіркою гіпотези

95% довірчий інтервал для середнього потрібно порівняти зі старою нормою μ₀ = 16.4 °C.

Якщо μ₀ не потрапляє до 95% довірчого інтервалу, це узгоджується з відхиленням H0 при α = 0.05 для двобічного тесту. Якщо μ₀ потрапляє до інтервалу, це узгоджується з відсутністю достатніх підстав для відхилення H0.

### Помилка I роду

Для Івано-Франківська помилка I роду означала б, що ми відхилили H0 та зробили висновок, що стара норма μ₀ = 16.4 °C більше не відповідає дійсності, хоча насправді стара норма все ще правильна.

### Помилка II роду

Помилка II роду означала б, що ми не відхилили H0 та залишили висновок про узгодженість зі старою нормою, хоча насправді середня температура Івано-Франківська вже змінилася.

Для кліматологів, які готують офіційний оновлений довідник, помилка II роду може бути особливо важливою, оскільки вона може призвести до того, що реальну зміну кліматичних умов не буде вчасно виявлено. Водночас помилка I роду також має наслідки, адже передчасне оновлення норми на основі випадкової вибірки може створити неправильне уявлення про довгострокову зміну клімату.

# Контрольні питання

### 1. Що означає отримане p-value?

p-value показує, наскільки незвичним був би отриманий результат або ще більш екстремальний результат за умови, що нульова гіпотеза H0 є істинною. У цій роботі це означає, наскільки ймовірно отримати таке або ще сильніше відхилення вибіркового середнього від 16.4 °C за умови, що стара норма справді правильна.

p-value не є ймовірністю того, що стара норма μ₀ правильна. Воно не показує ймовірність істинності H0.

### 2. Чому тип тесту потрібно обирати до перегляду результату?

Тип тесту потрібно визначати до аналізу результату, оскільки він залежить від поставленого дослідницького питання, а не від отриманого вибіркового середнього. Якщо вибирати однобічний або двобічний тест після отримання результату, можна штучно отримати менше p-value та збільшити ризик помилкового відхилення H0.

У цій роботі заздалегідь обрано двобічний тест, оскільки нас цікавить будь-яка зміна старої норми — як підвищення, так і зниження температури.

### 3. Що означають помилки I та II роду?

Помилка I роду для наших даних означала б відхилення старої норми 16.4 °C, хоча вона насправді правильна.

Помилка II роду означала б невідхилення старої норми, хоча середня температура Івано-Франківська насправді вже змінилася.

Ці помилки різні: у першому випадку ми виявляємо зміну, якої насправді немає, а в другому — не помічаємо реальної зміни.

### 4. Чи узгоджується тест із довірчим інтервалом?

Висновок перевірки гіпотези має узгоджуватися з 95% довірчим інтервалом при α = 0.05.

Для двобічного тесту, якщо значення μ₀ = 16.4 °C не входить до 95% довірчого інтервалу, H0 відхиляється. Якщо μ₀ входить до інтервалу, H0 не відхиляється. Це пов'язано з відповідністю двобічного тесту рівня α = 0.05 та 95% довірчого інтервалу.
