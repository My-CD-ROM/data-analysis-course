```python
import pandas as pd
```

```python
wide = pd.DataFrame({
    "місто": ["Івано-Франківськ"],
    "Січ": [84],
    "Лют": [80],
    "Бер": [74],
    "Кві": [64],
    "Тра": [57],
    "Чер": [50],
    "Лип": [46],
    "Сер": [48],
    "Вер": [58],
    "Жов": [72],
    "Лис": [82],
    "Гру": [86]
})

print(wide)
```

```python
tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність"
)

print(tidy)
print("Форма таблиці:", tidy.shape)
```

```python
print(tidy.dtypes)
```

### Завдання 3. Класифікація змінних

- **місто — номінальна шкала.** Значення є назвами міст і використовуються лише для їх ідентифікації, без числового порядку між ними.
- **місяць — порядкова шкала.** Місяці мають природний порядок протягом року: січень, лютий, березень і так далі.
- **хмарність — шкала відношень.** Хмарність вимірюється у відсотках, має нуль як змістовну точку, а співвідношення значень має сенс.

```python
back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

back = back.reset_index()

print(back)
```

```python
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

back_sorted = back[["місто"] + months]

print("Дані збігаються:", wide.equals(back_sorted))
```

```python
tidy.to_csv("variant.csv", index=False)
```

```python
tidy.to_json(
    "variant.json",
    orient="records",
    force_ascii=False
)
```

```python
reloaded_csv = pd.read_csv("variant.csv")

print(reloaded_csv)
print("CSV збігається:", tidy.equals(reloaded_csv))
```

### Порівняння CSV та JSON

CSV зберігає таблицю у вигляді рядків і стовпців, де значення розділені спеціальним символом, у нашому випадку комою. Формат CSV простий і зручний для табличних даних.

JSON зберігає дані у структурованому вигляді як набір об'єктів і пар «ключ — значення». У нашому випадку кожен рядок таблиці представлений окремим JSON-об'єктом.

Отже, CSV орієнтований безпосередньо на табличне представлення, а JSON — на структуроване представлення даних.

## Підсумкові питання

### 1. Чому «місто» — номінальна шкала, а «хмарність» — шкала відношень?

Стовпець «місто» належить до номінальної шкали, оскільки назви міст є категоріями та не мають числового порядку або кількісного значення. Стовпець «хмарність» належить до шкали відношень, оскільки значення задані у відсотках, мають змістовний нуль і між ними можна коректно порівнювати співвідношення.

### 2. Чому wide-таблиця з місяцями у стовпцях не є tidy?

Wide-таблиця не є tidy, тому що значення місяців використовуються як назви окремих стовпців. Після застосування `melt()` назви місяців перетворюються на значення одного стовпця «місяць», а значення хмарності — на значення стовпця «хмарність». Таким чином, кожен рядок описує одне спостереження для конкретного міста і місяця.

### 3. Чим CSV принципово відрізняється від JSON?

CSV є табличним форматом, у якому дані представлені рядками та стовпцями, розділеними спеціальним символом. JSON є структурованим форматом, у якому дані представлені об'єктами, масивами та парами «ключ — значення». Тому CSV простіший для табличних даних, а JSON краще підходить для структурованого обміну даними.
