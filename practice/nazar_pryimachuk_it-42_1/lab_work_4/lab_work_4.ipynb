### Різниця між duplicated() без subset і з subset

Метод duplicated() без параметра subset перевіряє повний збіг усіх значень рядка. Якщо всі значення двох рядків однакові, другий рядок визначається як дублікат.

Метод duplicated() з subset перевіряє тільки вказані стовпці. У цій роботі для визначення унікального замовлення використано поле "id". Тому два рядки з однаковим id вважаються дубльованими.

```python
print(orders.dtypes)
```

```python
orders["ціна"] = (
    orders["ціна"]
    .astype(str)
    .str.replace(" грн", "", regex=False)
    .astype(float)
)

print(orders.dtypes)
```

```python
print("Варіанти назви міста:")
print(orders["місто"].unique())
```

```python
city_mapping = {
    "Івано-Франківськ": "Івано-Франківськ",
    " Івано-Франківськ ": "Івано-Франківськ",
    "ІВАНО-ФРАНКІВСЬК": "Івано-Франківськ"
}

orders["місто"] = orders["місто"].replace(city_mapping)

print("Унікальні значення після узгодження:")
print(orders["місто"].unique())
```

```python
print("Тип ціни:", orders["ціна"].dtype)
print("Унікальні міста:", orders["місто"].unique())

orders
```

```python
q1 = orders["ціна"].quantile(0.25)
q3 = orders["ціна"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)
print("Нижня межа =", lower)
print("Верхня межа =", upper)
```

```python
outliers = orders[
    (orders["ціна"] < lower) |
    (orders["ціна"] > upper)
]

print("Викиди:")
print(outliers)
```

### Рішення щодо викиду

Значення 7600 грн було визначено як викид за правилом IQR. Для цього набору я вважаю це значення помилкою введення, оскільки базова ціна товару становить 760 грн, а значення 7600 грн рівно у 10 разів перевищує базову ціну. Найімовірніше, під час введення було випадково додано зайвий нуль. Тому це значення доцільно виправити до 760 грн, а не залишати як реальне велике замовлення.

```python
orders.loc[orders["ціна"] == 7600, "ціна"] = 760

print("Ціни після виправлення:")
print(orders["ціна"].tolist())
```

## Письмові відповіді

### 1. Чому середнє може спотворити стандартне відхилення?

Заповнення пропусків середнім може зменшити розкид даних, тому стандартне відхилення може вийти штучно меншим. Медіана краще підходить для невеликого набору даних і менше залежить від крайніх значень.

### 2. У чому різниця між duplicated() без subset і з subset?

Без subset метод перевіряє повний збіг усього рядка. З subset перевіряються тільки вказані стовпці. Тому кількість знайдених дублікатів може відрізнятися.

### 3. Чому не можна автоматично видаляти всі значення поза IQR?

Значення поза межами IQR не завжди є помилками. Це може бути реальне незвичайне значення. Тому його потрібно перевірити перед видаленням.
