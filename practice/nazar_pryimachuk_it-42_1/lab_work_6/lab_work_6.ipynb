```python
import pandas as pd
```

```python
humidity = pd.Series(
    [86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87],
    name="Вологість"
)

humidity.describe()
```

### Пояснення результатів `.describe()`

- `count` — кількість спостережень, у нашому випадку 12 місяців.
- `mean` — середнє арифметичне значення вологості.
- `std` — стандартне відхилення, міра розсіювання даних відносно середнього.
- `min` — мінімальне значення.
- `25%` — перший квартиль Q1.
- `50%` — медіана, або другий квартиль Q2.
- `75%` — третій квартиль Q3.
- `max` — максимальне значення.

Розмах визначається як `max - min`, а міжквартильний розмах IQR — як `Q3 - Q1`.

```python
minimum = humidity.min()
maximum = humidity.max()

rozmakh = maximum - minimum

variance_ddof1 = humidity.var(ddof=1)
variance_ddof0 = humidity.var(ddof=0)

std = humidity.std(ddof=1)

q1 = humidity.quantile(0.25)
q3 = humidity.quantile(0.75)

iqr = q3 - q1

mean = humidity.mean()
cv = std / mean * 100

print("Мінімум:", minimum)
print("Максимум:", maximum)
print("Розмах:", rozmakh)
print("Дисперсія ddof=1:", variance_ddof1)
print("Дисперсія ddof=0:", variance_ddof0)
print("Стандартне відхилення:", std)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Середнє:", mean)
print("Коефіцієнт варіації:", cv, "%")
```

```python
description = humidity.describe()

print("Q1 з describe():", description["25%"])
print("Q1 вручну:", q1)

print("Q3 з describe():", description["75%"])
print("Q3 вручну:", q3)

print("Q1 збігається:", description["25%"] == q1)
print("Q3 збігається:", description["75%"] == q3)
```

## Відповіді на контрольні питання

### 1. Які міри розсіювання стоять за рядками `.describe()`?

`std` — стандартне відхилення, яке характеризує розсіювання значень відносно середнього.

`25%` і `75%` — перший та третій квартилі. За ними визначається міжквартильний розмах IQR:

`IQR = Q3 - Q1`.

`min` і `max` — мінімальне та максимальне значення. За ними можна визначити загальний розмах:

`Range = max - min`.

### 2. Чому заповнення пропуску середнім штучно зменшує std?

Якщо пропущене значення замінити середнім, воно буде розташоване точно в центрі розподілу та не матиме відхилення від середнього. Наприклад, для значень `10, 20, 30` середнє дорівнює `20`. Якщо замінити пропуск середнім, отримуємо додаткове значення `20`. Воно не збільшує суму квадратів відхилень, тому відносне розсіювання набору зменшується. Таким чином, таке заповнення може штучно зменшувати `std`.

### 3. Чому не можна безпосередньо порівнювати std груп різного масштабу?

Стандартне відхилення є абсолютною мірою розсіювання. Якщо дві групи мають різні середні значення або різний масштаб, однаковий або більший `std` не обов'язково означає більшу відносну варіативність. Для коректного порівняння використовують коефіцієнт варіації:

`CV = std / mean * 100%`.

Він показує розсіювання відносно середнього значення.

### 4. Чи симетричний розподіл власних даних?

Для Івано-Франківська середнє значення становить 76.25%, а медіана — 76.50%. Вони дуже близькі, тому розподіл є близьким до симетричного.

Через відсутність значної різниці між середнім і медіаною середнє арифметичне є репрезентативною мірою положення для цих даних. Медіана також добре характеризує центр розподілу.
