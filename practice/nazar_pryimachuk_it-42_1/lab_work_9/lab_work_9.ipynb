## Завдання 1. Побудова набору клієнтів

Створимо DataFrame із 10 замовленнями інтернет-магазину. Для кожного клієнта вкажемо вік, місячний дохід, кількість товару та місто доставки.

```python
import pandas as pd

base_price = 760

df = pd.DataFrame({
    "вік": [19, 23, 28, 34, 39, 45, 51, 57, 63, 69],
    "дохід": [18000, 24000, 32000, 41000, 47000, 55000, 62000, 70000, 82000, 95000],
    "кількість": [2, 3, 4, 3, 5, 2, 4, 3, 6, 2],
    "місто_доставки": [
        "Івано-Франківськ",
        "Львів",
        "Київ",
        "Одеса",
        "Івано-Франківськ",
        "Дніпро",
        "Львів",
        "Київ",
        "Одеса",
        "Івано-Франківськ"
    ]
})

df
```

```python
df["сума_замовлення"] = base_price * df["кількість"]

df["частка_доходу"] = df["сума_замовлення"] / df["дохід"]

df
```

## Завдання 3. Дискретизація

Розіб'ємо вік клієнтів на 3 групи двома способами: за допомогою `pd.cut()` та `pd.qcut()`.

```python
df["вік_група_cut"] = pd.cut(df["вік"], bins=3)

cut_counts = df["вік_група_cut"].value_counts().sort_index()

print("Розподіл клієнтів за pd.cut():")
print(cut_counts)
```

```python
df["вік_група_qcut"] = pd.qcut(df["вік"], q=3)

qcut_counts = df["вік_група_qcut"].value_counts().sort_index()

print("Розподіл клієнтів за pd.qcut():")
print(qcut_counts)
```

```python
comparison = pd.DataFrame({
    "cut": cut_counts.reset_index(drop=True),
    "qcut": qcut_counts.reset_index(drop=True)
})

comparison
```

### Порівняння `cut()` і `qcut()`

`pd.cut()` ділить діапазон значень на інтервали однакової ширини, тому кількість клієнтів у групах може відрізнятися.

`pd.qcut()` ділить дані за квантилями та намагається отримати приблизно однакову кількість спостережень у кожній групі.

Тому при однакових вхідних даних результати `cut()` і `qcut()` можуть відрізнятися: `cut()` орієнтується на ширину інтервалів, а `qcut()` — на кількість спостережень.

## Завдання 4. Масштабування

Виконаємо min-max нормалізацію та стандартизацію методом z-оцінки для змінних `вік` і `дохід`.

```python
# Min-Max нормалізація

df["вік_minmax"] = (
    (df["вік"] - df["вік"].min()) /
    (df["вік"].max() - df["вік"].min())
)

df["дохід_minmax"] = (
    (df["дохід"] - df["дохід"].min()) /
    (df["дохід"].max() - df["дохід"].min())
)

df[["вік", "дохід", "вік_minmax", "дохід_minmax"]].head()
```

```python
df["вік_z"] = (
    (df["вік"] - df["вік"].mean()) /
    df["вік"].std()
)

df["дохід_z"] = (
    (df["дохід"] - df["дохід"].mean()) /
    df["дохід"].std()
)
```

```python
df[[
    "вік",
    "дохід",
    "вік_minmax",
    "дохід_minmax",
    "вік_z",
    "дохід_z"
]].head()
```

### Пояснення

До масштабування `вік` і `дохід` мають різний порядок величини та різні одиниці вимірювання. Вік має значення від 19 до 69 років, а дохід — від 18000 до 95000 гривень.

Через це в методах, чутливих до масштабу, дохід міг би мати значно більший вплив на результат лише через великі числові значення.

Min-max нормалізація переводить значення у діапазон від 0 до 1.

Z-стандартизація перетворює значення так, щоб вони характеризували відхилення від середнього у стандартних відхиленнях.

Тому після масштабування `вік` і `дохід` можна коректніше використовувати разом.

```python
city_dummies = pd.get_dummies(
    df["місто_доставки"],
    prefix="місто",
    dtype=int
)

city_dummies
```

```python
unique_cities = df["місто_доставки"].nunique()
dummy_columns = city_dummies.shape[1]

print("Кількість унікальних міст:", unique_cities)
print("Кількість dummy-стовпців:", dummy_columns)
```

```python
df_encoded = pd.concat([df, city_dummies], axis=1)

df_encoded
```

```python
assert unique_cities == dummy_columns

print("Перевірка успішна!")
print("Кількість dummy-стовпців дорівнює кількості унікальних міст.")
```

### Пояснення one-hot кодування

`місто_доставки` є номінальною змінною, оскільки міста не мають природного порядку.

Якщо присвоїти містам числа 1, 2, 3, 4, алгоритм може помилково сприйняти ці числа як порядок або відстань між містами.

One-hot кодування створює окремий бінарний стовпець для кожного міста. Значення 1 означає, що замовлення доставляється до цього міста, а 0 — що ні.

Тому для змінної `місто_доставки` one-hot кодування є правильнішим за порядкове або просте числове кодування.

## Відповіді на контрольні питання

### 1. Яка різниця між `pd.cut()` і `pd.qcut()`?

`pd.cut()` розбиває діапазон числових значень на інтервали однакової ширини. Кількість спостережень у групах при цьому може бути різною.

`pd.qcut()` використовує квантилі та намагається сформувати групи з приблизно однаковою кількістю спостережень. При цьому ширина інтервалів може відрізнятися.

### 2. Чому для міста використовують one-hot кодування?

Місто є номінальною змінною і не має природного порядку. Присвоєння містам чисел 1, 2, 3 тощо створило б штучний порядок.

One-hot кодування створює окрему ознаку для кожного міста без створення такого порядку.

### 3. Навіщо масштабувати змінні?

Масштабування потрібне, щоб змінні з різними одиницями вимірювання та різним порядком величини мали порівнянний вплив на алгоритм.

Це особливо важливо для дистанційних методів та регресії з регуляризацією.

### 4. Навіщо створювати похідні змінні?

Похідна змінна може показувати співвідношення між декількома характеристиками.

Наприклад, `частка_доходу` показує, яку частину місячного доходу клієнт витратив на замовлення. Це дозволяє краще порівнювати клієнтів із різним рівнем доходу.
