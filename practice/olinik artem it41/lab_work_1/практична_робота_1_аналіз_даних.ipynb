```python
# Варіант 2 — Львів

temps = np.array([-3, -2, 2, 8, 13, 16, 18, 17, 13, 8, 3, -1])

days_in_month = np.array([
    31, 28, 31, 30, 31, 30,
    31, 31, 30, 31, 30, 31
])

month_numbers = np.arange(1, 13)

print("Місто: Львів")
print()

print("temps:")
print(temps)
print("shape:", temps.shape)
print("dtype:", temps.dtype)

print()

print("days_in_month:")
print(days_in_month)
print("shape:", days_in_month.shape)
print("dtype:", days_in_month.dtype)

print()

print("month_numbers:")
print(month_numbers)
print("shape:", month_numbers.shape)
print("dtype:", month_numbers.dtype)
```

```python
temps_f = temps * 9 / 5 + 32

print("Температура в °C:")
print(temps)

print()

print("Температура в °F:")
print(temps_f)
```

```python
warm_mask = temps > 15

print("Температура > 15°C:")
print(warm_mask)

warm_days = days_in_month[warm_mask]

print()
print("Кількість днів у теплих місяцях:")
print(warm_days)

print()
print("Всього днів тепліше 15°C:", warm_days.sum())
```

```python
total_days = days_in_month.sum()

print("Всього днів у році:", total_days)
print("Чи дорівнює 365:", total_days == 365)

```

```python
warmest_index = temps.argmax()
coldest_index = temps.argmin()

warmest_temp = temps[warmest_index]
coldest_temp = temps[coldest_index]

warmest_month = month_numbers[warmest_index]
coldest_month = month_numbers[coldest_index]

print("Найтепліший місяць:")
print("Номер місяця:", warmest_month)
print("Температура:", warmest_temp, "°C")
print("Індекс:", warmest_index)

print()

print("Найхолодніший місяць:")
print("Номер місяця:", coldest_month)
print("Температура:", coldest_temp, "°C")
print("Індекс:", coldest_index)
```

```python
negative_mask = temps < 0

negative_months = month_numbers[negative_mask]

print("Місяці з температурою нижче 0°C:")
print(negative_months)
```

```python
sorted_temps = np.sort(temps)

print("Температури за зростанням:")
print(sorted_temps)

print()

print("Оригінальний масив temps:")
print(temps)
```

```python
rows = np.array([
    temps,
    days_in_month
])

print("rows:")
print(rows)

print()
print("shape:", rows.shape)
```

```python
correction = np.array([1, 0])

print("correction:")
print(correction)

print("shape:", correction.shape)
```

```python
rows + correction
```

```python
correction_2d = correction.reshape(2, 1)

print("correction після reshape:")
print(correction_2d)

print("shape:", correction_2d.shape)
```

```python
result = rows + correction_2d

print("Результат:")
print(result)

print()
print("shape:", result.shape)
```

Письмові відповіді
1. Різниця між shape == (12,) і shape == (1, 12)
shape == (12,) означає одновимірний масив із 12 елементів.

shape == (1, 12) означає двовимірний масив з 1 рядком і 12 стовпцями.

Кількість елементів однакова, але структура масивів різна.

2. Чому temps[temps > 15] повертає масив іншої довжини?
temps > 15 створює булеву маску. Для кожного елемента вона визначає, чи температура більша за 15°C.

NumPy вибирає тільки ті елементи, де значення маски дорівнює True.

Для Львова результат:

[16 18 17]

Тому результат має 3 елементи замість 12.

3. Що станеться, якщо додати масив форми (3,) до масиву форми (3, 12)?
Виникне помилка broadcasting.

Форми (3, 12) і (3,) є несумісними, тому що при порівнянні справа наліво розміри 12 і 3 не збігаються і жоден з них не дорівнює 1.

4. Чому rows + correction спочатку дає помилку, а rows + correction.reshape(2, 1) — ні?
rows має форму (2, 12), а correction має форму (2,).

При broadcasting форми порівнюються справа наліво. Числа 12 і 2 несумісні, тому виникає помилка.

Після correction.reshape(2, 1) форма стає (2, 1).

Тепер форми (2, 12) і (2, 1) сумісні: 2 збігається з 2, а 1 може бути розширена до 12.

Тому додавання працює. До всіх елементів першого рядка додається 1, а до всіх елементів другого рядка додається 0.
