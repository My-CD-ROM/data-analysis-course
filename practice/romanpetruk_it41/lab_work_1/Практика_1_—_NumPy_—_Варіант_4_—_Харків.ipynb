```python
# Завдання 1. Створення масивів

# Температури Харкова за місяцями
temps = np.array([-6, -5, 1, 10, 17, 20, 22, 21, 15, 8, 1, -4])

# Кількість днів у кожному місяці невисокосного року
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])

# Номери місяців від 1 до 12
month_numbers = np.arange(1, 13)

print("temps:")
print(temps)
print("shape:", temps.shape)
print("dtype:", temps.dtype)

print("\ndays_in_month:")
print(days_in_month)
print("shape:", days_in_month.shape)
print("dtype:", days_in_month.dtype)

print("\nmonth_numbers:")
print(month_numbers)
print("shape:", month_numbers.shape)
print("dtype:", month_numbers.dtype)
```

```python
# Завдання 2. Векторизовані обчислення

# Переведення температури з °C у °F
temps_f = temps * 9 / 5 + 32

print("Температура у °F:")
print(temps_f)

# Місяці, де температура вище 15°C
warm_months = temps > 15

print("\nБулева маска для температур > 15°C:")
print(warm_months)

# Кількість днів у таких місяцях
warm_days = days_in_month[warm_months]

print("\nКількість днів у теплих місяцях:")
print(warm_days)

print("Всього днів із температурою вище 15°C:", warm_days.sum())

# Перевірка кількості днів у році
print("\nСума днів у році:", days_in_month.sum())
print("Чи дорівнює 365:", days_in_month.sum() == 365)
```

```python
# Завдання 3. Індексація і фільтрація

# Найвища температура
max_temp = temps.max()
max_index = temps.argmax()

print("Найтепліший місяць:")
print("Температура:", max_temp, "°C")
print("Індекс:", max_index)
print("Номер місяця:", month_numbers[max_index])

# Найнижча температура
min_temp = temps.min()
min_index = temps.argmin()

print("\nНайхолодніший місяць:")
print("Температура:", min_temp, "°C")
print("Індекс:", min_index)
print("Номер місяця:", month_numbers[min_index])

# Місяці з температурою нижче 0°C
below_zero_months = month_numbers[temps < 0]

print("\nНомери місяців з температурою нижче 0°C:")
print(below_zero_months)

# Сортування температур
sorted_temps = np.sort(temps)

print("\nТемператури за зростанням:")
print(sorted_temps)

print("\nОригінальний масив temps:")
print(temps)
```

```python
# Завдання 4. Двовимірний масив і axis

# Умовна кількість сонячних годин за місяць
sun_hours = np.array([45, 65, 110, 160, 220, 240, 250, 230, 170, 120, 60, 40])

# Об'єднуємо три рядки в один двовимірний масив
data = np.array([temps, days_in_month, sun_hours])

print("Двовимірний масив:")
print(data)

print("\nФорма data:")
print(data.shape)

# Сума по кожному рядку
print("\nСума по кожному рядку (axis=1):")
print(data.sum(axis=1))

# Середнє по кожному рядку
print("\nСереднє по кожному рядку (axis=1):")
print(data.mean(axis=1))

# Сума по кожному стовпцю
print("\nСума по кожному стовпцю (axis=0):")
print(data.sum(axis=0))
```

```python
# Завдання 5. reshape і broadcasting

# Масив із двох рядків
rows = np.array([temps, days_in_month])

print("rows:")
print(rows)

print("\nФорма rows:")
print(rows.shape)

# Масив поправок
correction = np.array([1, 0])

print("\ncorrection:")
print(correction)

print("Форма correction:")
print(correction.shape)

# Спроба додавання
try:
    result = rows + correction
    print(result)
except ValueError as e:
    print("\nВиникла помилка:")
    print(e)
```

```python
# Змінюємо форму correction
correction_2d = correction.reshape(2, 1)

print("Нова форма correction:")
print(correction_2d.shape)

# Додаємо поправки
result = rows + correction_2d

print("\nРезультат:")
print(result)

print("\nФорма результату:")
print(result.shape)
```

Відповіді на контрольні питання
1. Різниця між shape == (12,) і shape == (1, 12)

shape == (12,) означає одновимірний масив із 12 елементів. Він має одну вісь.

shape == (1, 12) означає двовимірний масив із 1 рядком і 12 стовпцями. Кількість елементів однакова, але структура масиву різна.

2. Чому temps[temps > 15] має іншу довжину?

Вираз temps > 15 створює булеву маску. Для кожної температури отримуємо True або False.

Під час булевої індексації NumPy залишає тільки елементи, для яких отримано True. Тому результат містить лише температури, що перевищують 15°C, і має меншу довжину, ніж початковий масив temps.

3. Що станеться при додаванні масиву (3,) до (3, 12)?

При broadcasting NumPy порівнює форми справа наліво. Форми (3, 12) та (3,) можна розглядати як (3, 12) та (1, 3).

Числа 12 і 3 несумісні, тому виникне помилка broadcasting.

Щоб додати по одному значенню до кожного з трьох рядків, потрібно використати масив форми (3, 1).

4. Чому rows + correction дає помилку, а rows + correction.reshape(2, 1) працює?

rows має форму (2, 12), а correction — (2,). При порівнянні останніх вимірів отримуємо 12 і 2, які несумісні.

Після correction.reshape(2, 1) форма стає (2, 1). Тепер NumPy може розширити вимір 1 до 12, тому форми (2, 12) та (2, 1) стають сумісними.
