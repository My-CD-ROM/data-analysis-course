```python
# ============================================================
# Практика 12. Довірчі інтервали для середнього і частки
# Варіант 4 — Харків
# ============================================================

import numpy as np
from scipy import stats
import pandas as pd


# ============================================================
# ЗАВДАННЯ 1. Вибірка денних температур
# ============================================================

np.random.seed(4)

july_temp = 22
n = 30

daily_temps = np.random.normal(
    loc=july_temp,
    scale=2.5,
    size=n
)

print("ЗАВДАННЯ 1. Вибірка денних температур")
print("-" * 50)

print("Вибірка:")
print(daily_temps)

mean = daily_temps.mean()
s = daily_temps.std(ddof=1)

print("\nСереднє:", mean)
print("Вибіркове стандартне відхилення:", s)


# ============================================================
# ЗАВДАННЯ 2. Довірчий інтервал для середньої температури
# ============================================================

print("\n")
print("ЗАВДАННЯ 2. Довірчий інтервал для середньої")
print("-" * 50)

se = s / np.sqrt(n)

ci_mean = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean,
    scale=se
)

print("Стандартна похибка:", se)
print("95% довірчий інтервал:", ci_mean)


# ============================================================
# ЗАВДАННЯ 3. Довірчий інтервал для частки
# ============================================================

print("\n")
print("ЗАВДАННЯ 3. Довірчий інтервал для частки")
print("-" * 50)

p_hat = (daily_temps > july_temp).mean()

n_p = n * p_hat
n_1p = n * (1 - p_hat)

print("p_hat:", p_hat)
print("n * p_hat:", n_p)
print("n * (1 - p_hat):", n_1p)

print("\nПеревірка умови нормального наближення:")
print("n * p_hat >= 5:", n_p >= 5)
print("n * (1 - p_hat) >= 5:", n_1p >= 5)

se_p = np.sqrt(
    p_hat * (1 - p_hat) / n
)

ci_prop = stats.norm.interval(
    0.95,
    loc=p_hat,
    scale=se_p
)

print("\nСтандартна похибка частки:", se_p)
print("95% довірчий інтервал для частки:", ci_prop)

print(
    "Чи входить 0.5 до інтервалу:",
    ci_prop[0] <= 0.5 <= ci_prop[1]
)


# ============================================================
# ЗАВДАННЯ 4. Ширина інтервалу при різних рівнях довіри
# ============================================================

print("\n")
print("ЗАВДАННЯ 4. Ширина інтервалу при різних рівнях довіри")
print("-" * 50)

ci_90 = stats.t.interval(
    0.90,
    df=n - 1,
    loc=mean,
    scale=se
)

ci_95 = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean,
    scale=se
)

ci_99 = stats.t.interval(
    0.99,
    df=n - 1,
    loc=mean,
    scale=se
)

width_90 = ci_90[1] - ci_90[0]
width_95 = ci_95[1] - ci_95[0]
width_99 = ci_99[1] - ci_99[0]

print("90% ДІ:", ci_90)
print("Ширина 90%:", width_90)

print("\n95% ДІ:", ci_95)
print("Ширина 95%:", width_95)

print("\n99% ДІ:", ci_99)
print("Ширина 99%:", width_99)


# Таблиця результатів

results = pd.DataFrame({
    "Рівень довіри": ["90%", "95%", "99%"],
    "Нижня межа": [
        ci_90[0],
        ci_95[0],
        ci_99[0]
    ],
    "Верхня межа": [
        ci_90[1],
        ci_95[1],
        ci_99[1]
    ],
    "Ширина": [
        width_90,
        width_95,
        width_99
    ]
})

print("\nТаблиця:")
print(results)


# ============================================================
# ЗАВДАННЯ 5. Інтерпретація результату
# ============================================================

print("\n")
print("ЗАВДАННЯ 5. Інтерпретація 95% довірчого інтервалу")
print("-" * 50)

print(
    "Якби ми багато разів повторювали процедуру вибірки "
    "по 30 спостережень і щоразу будували 95% довірчий "
    "інтервал таким самим способом, приблизно 95% "
    "отриманих інтервалів накривали б істинну середню "
    "температуру."
)
```

Контрольні питання
1. Яка різниця між точковою оцінкою і довірчим інтервалом?
Точкова оцінка — це одне число, яке оцінює невідомий параметр генеральної сукупності. Наприклад, вибіркове середнє є точковою оцінкою генерального середнього. Довірчий інтервал, крім оцінки, показує діапазон значень і враховує невизначеність оцінювання.

2. Чому використовують t-розподіл, а не нормальний?
Оскільки справжнє стандартне відхилення σ невідоме, ми використовуємо його вибіркову оцінку s. Через додаткову невизначеність застосовується t-розподіл із n − 1 ступенями вільності. Для великих вибірок t-розподіл наближається до нормального.

3. Що насправді означає «95% довіри»?
95% довіри означає, що при багаторазовому повторенні процедури формування вибірки та побудови інтервалу приблизно 95% отриманих інтервалів накриватимуть істинне значення параметра. Формулювання «ймовірність 95%, що істинне значення знаходиться в цьому конкретному інтервалі» є некоректним у частотній інтерпретації, оскільки після побудови інтервалу його межі вже визначені, а параметр вважається фіксованим.

4. Які три фактори визначають ширину довірчого інтервалу?
Ширину інтервалу визначають рівень довіри, варіативність даних та обсяг вибірки. Вищий рівень довіри збільшує ширину інтервалу, більша варіативність також збільшує її, а збільшення обсягу вибірки зменшує ширину інтервалу.
