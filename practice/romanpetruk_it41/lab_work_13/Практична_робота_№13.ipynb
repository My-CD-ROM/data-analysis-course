```python
# ============================================================
# ПРАКТИКА 13
# Точкові оцінки та перевірка гіпотез: p-value
# Варіант 4 — Харків
# ============================================================

import numpy as np
from scipy import stats


# ============================================================
# ЗАВДАННЯ 1. Вибірка і точкові оцінки
# ============================================================

np.random.seed(4)

# Параметри для варіанта 4 — Харків
july_temp = 22       # параметр генерації вибірки
mu0 = 20.9           # стара норма
n = 30

# Генеруємо вибірку поточних денних температур
daily_temps = np.random.normal(
    loc=july_temp,
    scale=2.5,
    size=n
)

# Точкові оцінки
x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)

print("ЗАВДАННЯ 1. Вибірка і точкові оцінки")
print("-" * 60)

print("Місто: Харків")
print("Поточна липнева температура для генерації:", july_temp, "°C")
print("Стара норма μ0:", mu0, "°C")
print("\nВибірка daily_temps:")
print(daily_temps)

print("\nВибіркове середнє x̄:", x_bar)
print("Вибіркове стандартне відхилення s:", s)


# ============================================================
# ЗАВДАННЯ 2. Формулювання H0 і H1
# ============================================================

print("\n")
print("ЗАВДАННЯ 2. Формулювання H0 і H1")
print("-" * 60)

# Використовуємо двобічний тест:
# нас цікавить будь-яке відхилення від старої норми

print("H0: μ =", mu0)
print("H1: μ ≠", mu0)
print("Тип тесту: двобічний")


# ============================================================
# ЗАВДАННЯ 3. z-статистика і p-value
# ============================================================

print("\n")
print("ЗАВДАННЯ 3. z-статистика і p-value")
print("-" * 60)

# Стандартна похибка
se = s / np.sqrt(n)

# z-статистика
z = (x_bar - mu0) / se

# Двобічний p-value
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print("Стандартна похибка se:", se)
print("z-статистика:", z)
print("p-value:", p_value)


# ============================================================
# ЗАВДАННЯ 4. Рішення при α = 0.05 і висновок
# ============================================================

print("\n")
print("ЗАВДАННЯ 4. Рішення при α = 0.05")
print("-" * 60)

alpha = 0.05

if p_value < alpha:
    decision = "Відхиляємо H0"
else:
    decision = "Не відхиляємо H0"

print("α =", alpha)
print("p-value =", p_value)
print("Рішення:", decision)

if p_value < alpha:
    print(
        "\nВисновок: отримані дані дають статистично значущі "
        "підстави вважати, що середня липнева температура "
        "Харкова відрізняється від старої норми 20.9 °C."
    )
else:
    print(
        "\nВисновок: отриманих даних недостатньо, щоб "
        "стверджувати, що середня липнева температура "
        "Харкова відрізняється від старої норми 20.9 °C."
    )


# ============================================================
# ЗАВДАННЯ 5. Перевірка через довірчий інтервал
# ============================================================

print("\n")
print("ЗАВДАННЯ 5. 95% довірчий інтервал і помилки I/II роду")
print("-" * 60)

# 95% ДІ за формулою x̄ ± 1.96 * se
ci_lower = x_bar - 1.96 * se
ci_upper = x_bar + 1.96 * se

print("95% довірчий інтервал:")
print("Нижня межа:", ci_lower)
print("Верхня межа:", ci_upper)

# Перевіряємо, чи потрапляє стара норма μ0 в інтервал
mu0_in_ci = ci_lower <= mu0 <= ci_upper

print("\nЧи потрапляє μ0 =", mu0, "до інтервалу?", mu0_in_ci)

if mu0_in_ci:
    print("Стара норма входить до 95% ДІ.")
else:
    print("Стара норма НЕ входить до 95% ДІ.")


# ============================================================
# ПІДСУМКОВИЙ РЕЗУЛЬТАТ
# ============================================================

print("\n")
print("=" * 60)
print("ПІДСУМОК")
print("=" * 60)

print("Місто: Харків")
print("x̄ =", x_bar)
print("s =", s)
print("μ0 =", mu0)
print("se =", se)
print("z =", z)
print("p-value =", p_value)
print("α =", alpha)
print("Рішення:", decision)
print("95% ДІ:", (ci_lower, ci_upper))
```

1. Що означає отримане p-value і чому це не «ймовірність того, що стара норма μ₀ правильна»?
p-value показує ймовірність отримати таке саме або ще більш екстремальне відхилення від старої норми за умови, що H₀: μ = 20.9 °C є істинною. Це не є ймовірністю правильності H₀, оскільки p-value розраховується за припущення, що H₀ вже є істинною.

2. Чому тип тесту потрібно обирати до того, як побачено результат?
Тип тесту потрібно визначити заздалегідь, виходячи з дослідницького питання. У нашому випадку ми перевіряємо будь-яке відхилення від старої норми 20.9 °C, тому використовуємо двобічний тест: H₁: μ ≠ 20.9. Якщо вибирати тип тесту після перегляду результату, можна штучно отримати менше p-value та збільшити ймовірність помилки I роду.

3. Що в контексті наших даних означали б помилки I та II роду?
Помилка I роду — відхилити H₀ і зробити висновок, що стара норма 20.9 °C більше не відповідає Харкову, хоча насправді вона правильна.

Помилка II роду — не відхилити H₀ і залишити стару норму 20.9 °C, хоча насправді середня липнева температура Харкова вже змінилася.

Ці помилки відрізняються тим, що в першому випадку ми помилково виявляємо зміну, а в другому — не помічаємо реальну зміну.

4. Чи узгодився висновок тесту з довірчим інтервалом і чому?
Так, висновки мають узгоджуватися, якщо використовується той самий рівень значущості α = 0.05 і відповідний 95% довірчий інтервал. Якщо стара норма 20.9 °C не входить до 95% довірчого інтервалу, H₀ відхиляється. Якщо 20.9 °C входить до інтервалу, H₀ не відхиляється. Це є наслідком відповідності двобічного тесту при α = 0.05 та 95% довірчого інтервалу.
