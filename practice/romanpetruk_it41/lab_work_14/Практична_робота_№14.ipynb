```python
# ============================================================
# ПРАКТИКА 14
# Параметричні критерії: t-критерій та порівняння дисперсій
# Варіант 4 — Харків
# ============================================================

import numpy as np
from scipy import stats


# ============================================================
# ЗАВДАННЯ 1. Одновибірковий t-критерій проти норми
# ============================================================

# Дані Харкова, вимірювання вологості о 8:00
humidity = np.array([
    83, 80, 75, 68, 62, 60,
    58, 60, 66, 73, 79, 84
])

# Довідкова норма
norm = 71

# H0: середня вологість дорівнює нормі
# H1: середня вологість відрізняється від норми

t_stat, p_value = stats.ttest_1samp(
    humidity,
    popmean=norm
)

print("ЗАВДАННЯ 1. Одновибірковий t-критерій")
print("-" * 60)

print("Місто: Харків")
print("Норма:", norm, "%")
print("Середня вологість:", humidity.mean(), "%")
print("t-статистика:", t_stat)
print("p-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("\nРішення: відхиляємо H0.")
    print(
        "Висновок: середня вологість Харкова "
        "статистично значуще відрізняється від норми 71%."
    )
else:
    print("\nРішення: не відхиляємо H0.")
    print(
        "Висновок: недостатньо статистичних підстав "
        "стверджувати, що середня вологість відрізняється від норми 71%."
    )


# ============================================================
# ЗАВДАННЯ 2. Ручна перевірка t-статистики
# ============================================================

print("\n")
print("ЗАВДАННЯ 2. Ручна перевірка t-статистики")
print("-" * 60)

x_bar = humidity.mean()
s = humidity.std(ddof=1)
n = len(humidity)

t_manual = (x_bar - norm) / (s / np.sqrt(n))

print("Середнє x̄:", x_bar)
print("Стандартне відхилення s:", s)
print("Кількість спостережень n:", n)
print("t-статистика вручну:", t_manual)
print("t-статистика scipy:", t_stat)
print("Збіг:", np.isclose(t_manual, t_stat))


# ============================================================
# ЗАВДАННЯ 3. Незалежний двовибірковий критерій
# Харків проти міста іншого регіону
# ============================================================

# Обираємо Львів — регіон Захід
lviv = np.array([
    85, 83, 80, 73, 68, 65,
    63, 65, 70, 77, 82, 86
])

print("\n")
print("ЗАВДАННЯ 3. Незалежний двовибірковий критерій")
print("-" * 60)

print("Порівнюємо: Харків і Львів")
print("Регіони: Схід і Захід")

# Перевірка рівності дисперсій
levene_stat_1, levene_p_1 = stats.levene(
    humidity,
    lviv
)

print("\nКритерій Лівена:")
print("Статистика:", levene_stat_1)
print("p-value:", levene_p_1)

if levene_p_1 >= alpha:
    equal_var_1 = True
    print("p >= 0.05 → немає підстав вважати дисперсії різними.")
    print("Використовуємо стандартний t-критерій: equal_var=True")
else:
    equal_var_1 = False
    print("p < 0.05 → дисперсії статистично значуще відрізняються.")
    print("Використовуємо Welch t-критерій: equal_var=False")

t_stat_1, p_value_1 = stats.ttest_ind(
    humidity,
    lviv,
    equal_var=equal_var_1
)

print("\nДвовибірковий t-критерій:")
print("t-статистика:", t_stat_1)
print("p-value:", p_value_1)

if p_value_1 < alpha:
    print(
        "Висновок: середня вологість Харкова і Львова "
        "статистично значуще відрізняється."
    )
else:
    print(
        "Висновок: немає достатніх статистичних підстав "
        "стверджувати, що середня вологість Харкова і Львова відрізняється."
    )


# ============================================================
# ЗАВДАННЯ 4. Порівняння з протилежним висновком Лівена
# ============================================================

# Обираємо Одесу — регіон Південь
odesa = np.array([
    82, 80, 78, 74, 71, 69,
    67, 68, 72, 76, 80, 83
])

print("\n")
print("ЗАВДАННЯ 4. Порівняння з протилежним висновком Лівена")
print("-" * 60)

print("Порівнюємо: Харків і Одеса")
print("Регіони: Схід і Південь")

levene_stat_2, levene_p_2 = stats.levene(
    humidity,
    odesa
)

print("\nКритерій Лівена:")
print("Статистика:", levene_stat_2)
print("p-value:", levene_p_2)

if levene_p_2 >= alpha:
    equal_var_2 = True
    print("p >= 0.05 → використовуємо equal_var=True")
else:
    equal_var_2 = False
    print("p < 0.05 → використовуємо Welch, equal_var=False")

t_stat_2, p_value_2 = stats.ttest_ind(
    humidity,
    odesa,
    equal_var=equal_var_2
)

print("\nДвовибірковий t-критерій:")
print("t-статистика:", t_stat_2)
print("p-value:", p_value_2)

if p_value_2 < alpha:
    print(
        "Висновок: середня вологість Харкова і Одеси "
        "статистично значуще відрізняється."
    )
else:
    print(
        "Висновок: немає достатніх статистичних підстав "
        "стверджувати, що середня вологість Харкова і Одеси відрізняється."
    )


# ============================================================
# ЗАВДАННЯ 5. Парний t-критерій: ранок проти вечора
# ============================================================

# Для кожного місяця від ранкового значення віднімаємо
# відповідну добову амплітуду:
# 3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3

amplitude = np.array([
    3, 3, 4, 5, 6, 7,
    7, 7, 6, 5, 4, 3
])

evening = humidity - amplitude

print("\n")
print("ЗАВДАННЯ 5. Парний t-критерій: ранок проти вечора")
print("-" * 60)

print("Ранкові значення:")
print(humidity)

print("\nВечірні значення:")
print(evening)

# Парний t-критерій
t_stat_rel, p_value_rel = stats.ttest_rel(
    humidity,
    evening
)

print("\nПарний t-критерій:")
print("t-статистика:", t_stat_rel)
print("p-value:", p_value_rel)

if p_value_rel < alpha:
    print(
        "\nВисновок: середня вологість о 8:00 "
        "статистично значуще відрізняється від "
        "середньої вологості о 20:00."
    )
else:
    print(
        "\nВисновок: немає достатніх статистичних підстав "
        "стверджувати, що середня вологість о 8:00 "
        "відрізняється від середньої вологості о 20:00."
    )
```

Контрольні питання
1. Чому для однієї вибірки використовується t-, а не z-розподіл?
Тому що справжнє стандартне відхилення генеральної сукупності σ невідоме. Ми оцінюємо його за вибірковим стандартним відхиленням s, тому використовуємо t-розподіл. У нашому випадку вибірка містить лише 12 місячних значень.

2. Чому потрібно перевіряти equal_var через критерій Лівена?
equal_var визначає, чи припускаємо ми однакові дисперсії двох вибірок. Якщо дисперсії статистично відрізняються, потрібно використовувати Welch-варіант (equal_var=False). Неправильний вибір може змінити розрахунок стандартної похибки та ступенів вільності, а отже призвести до неправильного p-value.

3. Чому парний критерій кращий для даних «ранок — вечір»?
Ранкове та вечірнє значення належать одному й тому самому місяцю, тому ці спостереження пов'язані між собою. Парний критерій спочатку розглядає різницю для кожної пари «ранок − вечір», завдяки чому усувається вплив індивідуальних особливостей кожного місяця та перевіряється саме середня різниця між двома вимірюваннями.

4. Чи узгоджується критерій Лівена з порівнянням .std()?
Критерій Лівена дає формальну статистичну перевірку рівності дисперсій, тоді як просте порівняння .std() показує лише величину вибіркового розкиду. Тому навіть якщо стандартні відхилення здаються різними, остаточний вибір між equal_var=True та equal_var=False потрібно робити за результатом критерію Лівена при α = 0.05.
