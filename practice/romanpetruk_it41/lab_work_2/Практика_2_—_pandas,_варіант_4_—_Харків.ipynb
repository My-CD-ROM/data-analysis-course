Робота з DataFrame, індексами, .loc, .iloc, фільтрацією та обчислюваними стовпцями.

```python
import pandas as pd

print("Версія pandas:", pd.__version__)
```

```python
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

avg_temp = [-6, -5, 1, 10, 17, 20, 22, 21, 15, 8, 1, -4]

precip = [42, 40, 36, 38, 48, 58, 62, 50, 38, 38, 46, 46]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip
})

print(df)
```

```python
print("Перші рядки:")
print(df.head())

print("\nФорма:")
print(df.shape)

print("\nТипи даних:")
print(df.dtypes)
```

```python
# Завдання 1. Індекс

print("DataFrame з індексом за замовчуванням:")
print(df)

print("\nІндекс:")
print(df.index)
```

```python
# Робимо стовпець month індексом
df_by_month = df.set_index("month")

print(df_by_month)
```

```python
print(df_by_month.loc["Лип"])
```

```python
df = df_by_month.reset_index()

print(df)
```

```python
# Завдання 2

df.index = range(1, 13)

print("Через .loc:")
print(df.loc[1:6, "avg_temp"])

print("\nЧерез .iloc:")
print(df.iloc[0:6, 1])

print("\nСтовпці month і precip:")
print(df[["month", "precip"]])
```

```python
# Завдання 3

print("avg_temp > 15:")
print(df[df["avg_temp"] > 15][["month", "avg_temp"]])

print("\navg_temp > 10 і precip > 50:")
result = df[
    (df["avg_temp"] > 10) &
    (df["precip"] > 50)
]
print(result[["month", "avg_temp", "precip"]])

print("\navg_temp < 0 або precip > 80:")
result = df[
    (df["avg_temp"] < 0) |
    (df["precip"] > 80)
]
print(result[["month", "avg_temp", "precip"]])
```

Відповіді на контрольні питання
1. Чому .loc[1:6] і .iloc[0:6] можуть повернути різну кількість рядків?

.loc працює за мітками індексу, а .iloc — за позиціями. .loc[1:6] включає мітку 6, а .iloc[0:6] вибирає позиції від 0 до 5. Якщо індекс не збігається з позицією, результати можуть відрізнятися.

2. Чому df["avg_temp"] > 10 and df["precip"] > 50 викличе помилку?

and не призначений для поелементної роботи з Series pandas. Для поелементного порівняння потрібно використовувати &, а кожну умову брати в дужки:

(df["avg_temp"] > 10) & (df["precip"] > 50)

3. У чому різниця між обчислюваним стовпцем через формулу і через .apply()?

Формула виконує векторизовану операцію одразу над усім стовпцем. .apply() застосовує власну функцію до окремих елементів і зручний для складнішої логіки, наприклад визначення пори року.
