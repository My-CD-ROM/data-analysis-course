```python
import pandas as pd
```

```python
# Завдання 1

wide = pd.DataFrame({
    "місто": ["Харків"],
    "Січ": [76],
    "Лют": [72],
    "Бер": [65],
    "Кві": [55],
    "Тра": [46],
    "Чер": [40],
    "Лип": [36],
    "Сер": [38],
    "Вер": [48],
    "Жов": [62],
    "Лис": [74],
    "Гру": [80]
})

print(wide)
print("\nShape:", wide.shape)
```

```python
# Завдання 2

tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність"
)

print(tidy)
print("\nShape:", tidy.shape)
```

```python
# Завдання 3

print("місто — номінальна шкала")
print("місяць — порядкова шкала")
print("хмарність — шкала відношень")
```

```python
# Завдання 4

back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

print(back)
```

```python


back = back.reset_index()

back = back[[
    "місто", "Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"
]]

print("Початкова таблиця:")
print(wide)

print("\nТаблиця після pivot:")
print(back)

print("\nЧи однакові таблиці:")
print(wide.equals(back))
```

```python
# Завдання 5

tidy.to_csv("variant.csv", index=False)

tidy.to_json(
    "variant.json",
    orient="records",
    force_ascii=False
)

print("Файли variant.csv та variant.json створено.")
```

```python


reloaded_csv = pd.read_csv("variant.csv")
reloaded_json = pd.read_json("variant.json")

print("CSV:")
print(reloaded_csv)

print("\nJSON:")
print(reloaded_json)
```

```python


print("CSV збігається з оригіналом:")
print(tidy.equals(reloaded_csv))

print("\nJSON збігається з оригіналом:")
print(tidy.equals(reloaded_json))
```

Відповіді на контрольні питання
1. Чому «місто» — номінальна шкала, а «хмарність» — шкала відношень?

«Місто» є номінальною шкалою, тому що назви міст є категоріями, між якими немає числового порядку або відстані. «Хмарність» є шкалою відношень, оскільки вимірюється у відсотках, має природне нульове значення та дозволяє порівнювати значення у відношеннях.

2. Чому wide-таблиця з місяцями в стовпцях не є tidy?

У wide-таблиці кожен місяць представлений окремим стовпцем, тому одна змінна «місяць» фактично розподілена між багатьма стовпцями. Після melt() усі місяці стають значеннями одного стовпця місяць, а хмарність — значеннями одного стовпця хмарність. Таким чином, кожен рядок описує одне спостереження.

3. Чим CSV відрізняється від JSON?

CSV представляє дані як таблицю з рядками та стовпцями, значення в яких розділяються роздільником, наприклад комою. JSON має більш структурований формат і представляє дані через об'єкти, поля та масиви. JSON може зберігати складніші вкладені структури, тоді як CSV орієнтований на просту табличну структуру.
