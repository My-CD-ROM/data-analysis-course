```python
import numpy as np
import pandas as pd
```

```python
# Завдання 1

orders = pd.DataFrame([
    {"замовлення": 1, "місто": "Харків",       "товар": "Ноутбук", "ціна": "780 грн", "кількість": 2},
    {"замовлення": 2, "місто": " Харків ",     "товар": "Ноутбук", "ціна": 780,        "кількість": 3},
    {"замовлення": 3, "місто": "ХАРКІВ",       "товар": "Ноутбук", "ціна": "780 грн", "кількість": np.nan},
    {"замовлення": 4, "місто": "Харків",       "товар": "Ноутбук", "ціна": 780,        "кількість": 1},
    {"замовлення": 5, "місто": " Харків ",     "товар": "Ноутбук", "ціна": 780,        "кількість": 2},
    {"замовлення": 6, "місто": "ХАРКІВ",       "товар": "Ноутбук", "ціна": 7800,       "кількість": 4},
    {"замовлення": 7, "місто": "Харків",       "товар": "Ноутбук", "ціна": 780,        "кількість": np.nan},
    {"замовлення": 8, "місто": "Харків",       "товар": "Ноутбук", "ціна": 780,        "кількість": 3},
])

print(orders)
```

```python
# Завдання 2

print("Кількість пропусків:")
print(orders.isna().sum())

median_quantity = orders["кількість"].median()

orders["кількість"] = orders["кількість"].fillna(median_quantity)

print("\nМедіана кількості:", median_quantity)
print("\nПісля заповнення пропусків:")
print(orders)
```

```python
# Завдання 3

print("Дублікати без subset:")
print(orders.duplicated())

print("\nКількість повних дублікатів:")
print(orders.duplicated().sum())

print("\nДублікати за містом, товаром, ціною та кількістю:")
print(orders.duplicated(
    subset=["місто", "товар", "ціна", "кількість"]
))

orders = orders.drop_duplicates()

print("\nКількість рядків після видалення дубліката:")
print(len(orders))

print("\nОчищений DataFrame:")
print(orders)
```

```python
# Завдання 4

orders["ціна"] = (
    orders["ціна"]
    .astype(str)
    .str.replace(" грн", "", regex=False)
    .astype(float)
)

orders["місто"] = orders["місто"].str.strip().str.lower()

orders["місто"] = orders["місто"].replace({
    "харків": "Харків"
})

print(orders)

print("\nТип ціни:")
print(orders["ціна"].dtype)

print("\nУнікальні міста:")
print(orders["місто"].unique())
```

```python
# Завдання 5

q1 = orders["ціна"].quantile(0.25)
q3 = orders["ціна"].quantile(0.75)

iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Нижня межа:", lower)
print("Верхня межа:", upper)

outliers = orders[
    (orders["ціна"] < lower) |
    (orders["ціна"] > upper)
]

print("\nВикиди:")
print(outliers)
```

1. Заповнення пропуску середнім може вплинути на стандартне відхилення,
тому що середнє значення штучно додається до набору даних і не збільшує
розкид. Медіана також є способом заповнення, який менше залежить від
екстремальних значень.

2. duplicated() без subset шукає повністю однакові рядки за всіма стовпцями.
duplicated() з subset перевіряє дублікати тільки за вказаними стовпцями.
Тому результати можуть відрізнятися.

3. Автоматично видаляти всі значення поза межами IQR не завжди правильно,
тому що викид може бути не помилкою, а реальним рідкісним спостереженням.
Перед видаленням потрібно перевірити причину появи такого значення.
