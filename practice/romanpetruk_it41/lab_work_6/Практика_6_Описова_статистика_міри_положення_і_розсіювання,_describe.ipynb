```python
import numpy as np
import pandas as pd
```

```python
humidity = pd.Series([
    83, 80, 75, 68, 62, 60,
    58, 60, 66, 73, 79, 84
])

print(humidity)
```

```python
# Завдання 1

print(humidity.describe())
```

```python
# Завдання 2

range_value = humidity.max() - humidity.min()

variance_sample = humidity.var(ddof=1)
variance_population = humidity.var(ddof=0)

std_value = humidity.std(ddof=1)

q1 = humidity.quantile(0.25)
q3 = humidity.quantile(0.75)

iqr = q3 - q1

cv = std_value / humidity.mean() * 100

print("Розмах:", range_value)
print("Дисперсія ddof=1:", variance_sample)
print("Дисперсія ddof=0:", variance_population)
print("Стандартне відхилення:", std_value)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Коефіцієнт варіації:", cv, "%")
```

```python
print("Q1 з describe:", humidity.describe()["25%"])
print("Q1 вручну:", q1)

print("Q3 з describe:", humidity.describe()["75%"])
print("Q3 вручну:", q3)
```

```python
# Завдання 3

df = pd.DataFrame({
    "Місто": [
        "Київ", "Львів", "Одеса", "Харків", "Дніпро",
        "Чернігів", "Івано-Франківськ", "Полтава", "Суми", "Ужгород"
    ],

    "Регіон": [
        "Центр", "Захід", "Південь", "Схід", "Схід",
        "Центр", "Захід", "Центр", "Схід", "Захід"
    ],

    "Січ": [84, 85, 82, 83, 82, 85, 86, 84, 85, 84],
    "Лют": [82, 83, 80, 80, 79, 83, 84, 81, 82, 82],
    "Бер": [78, 80, 78, 75, 74, 79, 81, 76, 77, 79],
    "Кві": [70, 73, 74, 68, 67, 71, 75, 69, 70, 73],
    "Тра": [65, 68, 71, 62, 61, 66, 70, 63, 64, 68],
    "Чер": [63, 65, 69, 60, 59, 63, 67, 61, 62, 66],
    "Лип": [61, 63, 67, 58, 57, 61, 65, 59, 60, 64],
    "Сер": [63, 65, 68, 60, 59, 63, 67, 61, 62, 66],
    "Вер": [68, 70, 72, 66, 65, 69, 72, 67, 68, 71],
    "Жов": [75, 77, 76, 73, 72, 76, 78, 74, 75, 77],
    "Лис": [81, 82, 80, 79, 78, 82, 83, 80, 81, 82],
    "Гру": [85, 86, 83, 84, 83, 86, 87, 85, 86, 85]
})

months = [
    "Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"
]

df["Середня_річна_вологість"] = df[months].mean(axis=1)

print(df)
```

```python
east = df[df["Регіон"] == "Схід"]

print("Регіон Схід:")
print(east)

print("\nОписова статистика:")
print(east["Середня_річна_вологість"].describe())
```

```python
west = df[df["Регіон"] == "Захід"]

print("Регіон Захід:")
print(west)

print("\nОписова статистика:")
print(west["Середня_річна_вологість"].describe())
```

```python
east_mean = east["Середня_річна_вологість"].mean()
east_std = east["Середня_річна_вологість"].std()
east_cv = east_std / east_mean * 100

west_mean = west["Середня_річна_вологість"].mean()
west_std = west["Середня_річна_вологість"].std()
west_cv = west_std / west_mean * 100

print("CV Сходу:", east_cv, "%")
print("CV Заходу:", west_cv, "%")
```

```python
# Завдання 4

mean_value = humidity.mean()
median_value = humidity.median()

print("Середнє:", mean_value)
print("Медіана:", median_value)
print("Різниця:", abs(mean_value - median_value))
```

```python
# Завдання 5

kharkiv_cv = humidity.std() / humidity.mean() * 100

lviv = df[df["Місто"] == "Львів"][months].iloc[0]

lviv_mean = lviv.mean()
lviv_std = lviv.std()
lviv_cv = lviv_std / lviv_mean * 100

print("Харків:")
print("Середнє:", humidity.mean())
print("std:", humidity.std())
print("CV:", kharkiv_cv, "%")

print("\nЛьвів:")
print("Середнє:", lviv_mean)
print("std:", lviv_std)
print("CV:", lviv_cv, "%")
```

1. У .describe() рядок std відповідає стандартному відхиленню.
Різниця між 75% і 25% дає IQR — міжквартильний розмах.
Різниця між max і min дає розмах.

2. Якщо пропуски заповнювати середнім, нове значення знаходиться
в центрі розподілу. Воно не додає такого розкиду, як реальне
спостереження, тому стандартне відхилення може штучно зменшитися.

3. Сирий std показує абсолютний розкид і залежить від масштабу
значень. Коефіцієнт варіації ділить std на середнє і тому показує
відносний розкид. Це робить його зручнішим для порівняння груп
із різними середніми значеннями.

4. Для Харкова середнє приблизно 69.83%, а медіана 70.5%.
Вони досить близькі, тому розподіл близький до симетричного.
Середнє можна вважати репрезентативною мірою положення для цих даних,
оскільки немає значної різниці між середнім і медіаною.
