```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Завдання 1


np.random.seed(4)

days = np.arange(1, 31)
weekend = (days % 7 == 6) | (days % 7 == 0)

temperature = np.round(-2 + np.random.uniform(-3, 3, size=30), 1)

sales = np.round(
    30
    + np.random.normal(0, 4, size=30)
    + weekend * (30 * 0.20)
).astype(int)

data = pd.DataFrame({
    "день": days,
    "вихідний": weekend,
    "температура": temperature,
    "продажі": sales
})

data
```

```python
# Завдання 2

fig, ax = plt.subplots(figsize=(6, 4))

ax.hist(data["продажі"], bins=6, edgecolor="white")

ax.set_xlabel("Продажі, шт./день")
ax.set_ylabel("Кількість днів")
ax.set_title("Розподіл продажів за 30 днів — Харків")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 4))

ax.hist(data["продажі"], bins=10, edgecolor="white")

ax.set_xlabel("Продажі, шт./день")
ax.set_ylabel("Кількість днів")
ax.set_title("Розподіл продажів за 30 днів — Харків")

plt.show()
```

```python
# Завдання 3

fig, ax = plt.subplots(figsize=(5, 4))

groups = [
    data.loc[~data["вихідний"], "продажі"],
    data.loc[data["вихідний"], "продажі"]
]

ax.boxplot(groups, labels=["Будні", "Вихідні"])

ax.set_ylabel("Продажі, шт./день")
ax.set_title("Продажі у будні та вихідні — Харків")

plt.show()
```

```python
# Завдання 4

fig, ax = plt.subplots(figsize=(6, 4))

ax.scatter(data["температура"], data["продажі"])

ax.set_xlabel("Температура, °C")
ax.set_ylabel("Продажі, шт./день")
ax.set_title("Продажі залежно від температури — Харків")

plt.show()
```

```python
# Завдання 5
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].hist(data["продажі"], bins=6, edgecolor="white")
ax[0].set_xlabel("Продажі, шт./день")
ax[0].set_ylabel("Кількість днів")
ax[0].set_title("Коректний графік")

ax[1].hist(data["продажі"], bins=6, edgecolor="white")
ax[1].set_ylim(0, 3)
ax[1].set_xlabel("Продажі, шт./день")
ax[1].set_ylabel("Кількість днів")
ax[1].set_title("Спотворений графік")

plt.show()
```

1. Обрізана вісь Y може перебільшувати різницю між категоріями, тому що графічна висота стовпців починається не від нуля. Через це невелика числова різниця може виглядати значною.

2. Лінійний графік може бути методологічною помилкою для категоріальних даних, наприклад для порівняння міст. З'єднання міст лінією створює враження природного числового порядку та зміни між категоріями, хоча такого зв'язку може не існувати.

3. Підписи осей, одиниці вимірювання та заголовок пояснюють, що саме показує графік. Без них значення можуть бути неправильно інтерпретовані.

4. Chartjunk — це зайві графічні елементи, які не передають інформації, наприклад 3D-ефекти, зайві кольори або декоративні елементи. Вони ускладнюють порівняння даних і можуть спотворювати їхнє сприйняття.
