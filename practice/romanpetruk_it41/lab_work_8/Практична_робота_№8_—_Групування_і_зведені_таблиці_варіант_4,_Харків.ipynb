```python
# Завдання 1

import numpy as np
import pandas as pd

np.random.seed(42)

base_temp = 8.8
amplitude = 15
city = "Харків"

rows = []

for year in [2021, 2022, 2023, 2024]:
    for month in range(1, 13):
        seasonal = amplitude * np.cos((month - 7) / 12 * 2 * np.pi)
        noise = np.random.normal(0, 1.0)

        rows.append({
            "місто": city,
            "рік": year,
            "місяць": month,
            "температура": round(base_temp + seasonal + noise, 1)
        })

climate = pd.DataFrame(rows)

climate
```

```python
# Завдання 2

year_stats = climate.groupby("рік")["температура"].agg(
    ["mean", "min", "max"]
)

year_stats
```

```python

month_stats = climate.groupby("місяць")["температура"].agg(
    ["mean", "std"]
)

month_stats
```

```python


month_stats["std"].idxmax()
```

```python
# Завдання 3

pivot_temp = climate.pivot_table(
    index="місяць",
    columns="рік",
    values="температура",
    aggfunc="mean"
)

pivot_temp
```

```python
# Завдання 4

def get_season(month):
    if month in [12, 1, 2]:
        return "зима"
    elif month in [3, 4, 5]:
        return "весна"
    elif month in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"


climate["сезон"] = climate["місяць"].apply(get_season)

climate["тепліше_за_середнє"] = climate["температура"] > base_temp

climate
```

```python


pd.crosstab(
    climate["сезон"],
    climate["тепліше_за_середнє"]
)
```

```python
# Завдання 5

pivot_result = climate.pivot(
    index="місяць",
    columns="рік",
    values="температура"
)

pivot_result
```

1. groupby() працює за принципом split-apply-combine: спочатку дані розділяються на групи за певною ознакою, потім до кожної групи застосовується потрібна операція, а після цього результати об'єднуються в одну таблицю.

2. pivot() працює лише тоді, коли кожній комбінації індексу та стовпця відповідає одне значення. pivot_table() може працювати з дубльованими комбінаціями, тому що додатково агрегує їх за допомогою mean, sum або іншої функції.

3. crosstab() показує частоти комбінацій категоріальних змінних у вигляді таблиці. groupby(...).size() також може рахувати кількість спостережень у групах, але crosstab() зручніше використовувати саме для побудови перехресної таблиці категорій.

4. Довгий tidy-формат зручний для groupby() і pivot_table(), тому що кожне спостереження знаходиться в окремому рядку, а змінні — у стовпцях. Це дозволяє легко групувати, фільтрувати та агрегувати дані без залежності від назв стовпців.
