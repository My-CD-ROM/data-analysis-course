```python
# Завдання 1

import pandas as pd

df = pd.DataFrame({
    "вік": [19, 24, 31, 37, 42, 48, 53, 27, 61, 35],
    "дохід": [18000, 25000, 32000, 41000, 50000, 58000, 67000, 29000, 75000, 38000],
    "кількість": [1, 2, 3, 2, 4, 1, 3, 2, 5, 2],
    "місто_доставки": [
        "Харків", "Київ", "Львів", "Одеса", "Харків",
        "Дніпро", "Київ", "Полтава", "Львів", "Харків"
    ]
})

df
```

```python
# Завдання 2

base_price = 780

df["сума_замовлення"] = base_price * df["кількість"]
df["частка_доходу"] = df["сума_замовлення"] / df["дохід"]

df
```

```python
# Завдання 3

df["вікова_група_cut"] = pd.cut(
    df["вік"],
    bins=3
)

df["вікова_група_cut"].value_counts().sort_index()
```

```python

df["вікова_група_qcut"] = pd.qcut(
    df["вік"],
    q=3
)

df["вікова_група_qcut"].value_counts().sort_index()
```

```python
# Завдання 4

df["вік_minmax"] = (
    (df["вік"] - df["вік"].min()) /
    (df["вік"].max() - df["вік"].min())
)

df["дохід_minmax"] = (
    (df["дохід"] - df["дохід"].min()) /
    (df["дохід"].max() - df["дохід"].min())
)

df["вік_z"] = (
    (df["вік"] - df["вік"].mean()) /
    df["вік"].std()
)

df["дохід_z"] = (
    (df["дохід"] - df["дохід"].mean()) /
    df["дохід"].std()
)

df[
    [
        "вік",
        "дохід",
        "вік_minmax",
        "дохід_minmax",
        "вік_z",
        "дохід_z"
    ]
].head()
```

```python
# Завдання 5

city_dummies = pd.get_dummies(
    df["місто_доставки"],
    prefix="місто"
)

df = pd.concat([df, city_dummies], axis=1)

df
```

```python

print("Кількість унікальних міст:", df["місто_доставки"].nunique())
print("Кількість dummy-стовпців:", len(city_dummies.columns))
```

1. pd.cut() ділить діапазон значень на інтервали приблизно однакової довжини, тому кількість спостережень у групах може бути різною. pd.qcut() формує групи так, щоб у кожній було приблизно однакове число спостережень.

2. Для змінної "місто" використовується one-hot кодування, тому що міста є номінальними категоріями і не мають природного порядку. Кодування міст числами 1, 2, 3 створило б помилкове припущення, що одне місто більше або менше за інше.

3. Масштабування потрібне для того, щоб змінні з різними одиницями та масштабами мали зіставний вплив на методи, чутливі до масштабу. Наприклад, дохід у тисячах гривень має значно більші числові значення, ніж вік.

4. Похідна змінна "частка доходу" показує, яку частину доходу клієнт витратив на замовлення. Вона дозволяє порівнювати клієнтів із різним рівнем доходу, тоді як сама сума замовлення без урахування доходу дає менше інформації про відносне фінансове навантаження.
