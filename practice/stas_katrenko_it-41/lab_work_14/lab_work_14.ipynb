```python
import numpy as np
import pandas as pd
from scipy import stats

pd.set_option('display.width', 200)
pd.set_option('display.max_columns', 30)
np.set_printoptions(precision=4, suppress=True)
```

```python
months = ['Січ','Лют','Бер','Кві','Тра','Чер','Лип','Сер','Вер','Жов','Лис','Гру']

# Добова амплітуда вологості (п.п.), яку віднімаємо від ранкового значення => вечірнє (20:00)
amplitude = np.array([3, 3, 4, 5, 6, 7, 7, 7, 6, 5, 4, 3], dtype=float)

rows = [
    ('Київ',                'Центр', 74, [84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85]),
    ('Львів',               'Захід',  76, [85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86]),
    ('Одеса',               'Південь',73, [82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83]),
    ('Харків',              'Схід',   71, [83, 80, 75, 68, 62, 60, 58, 60, 66, 73, 79, 84]),
    ('Дніпро',              'Схід',   70, [82, 79, 74, 67, 61, 59, 57, 59, 65, 72, 78, 83]),
    ('Чернігів',            'Центр', 75, [85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86]),
    ('Івано-Франківськ',    'Захід',  77, [86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87]),
    ('Полтава',             'Центр', 72, [84, 81, 76, 69, 63, 61, 59, 61, 67, 74, 80, 85]),
    ('Суми',                'Схід',   73, [85, 82, 77, 70, 64, 62, 60, 62, 68, 75, 81, 86]),
    ('Ужгород',             'Захід',  76, [84, 82, 79, 73, 68, 66, 64, 66, 71, 77, 82, 85]),
]

df = pd.DataFrame(
    [(city, region, norm, *vals) for city, region, norm, vals in rows],
    columns=['Місто', 'Регіон', 'Норма'] + months
)
df
```

```python
# Мій варіант: Одеса. Регіональні змінні для подальших завдань.
MY_CITY = 'Одеса'
MY_NORM = float(df.loc[df['Місто'] == MY_CITY, 'Норма'].iloc[0])
my_region = df.loc[df['Місто'] == MY_CITY, 'Регіон'].iloc[0]
morning = df.loc[df['Місто'] == MY_CITY, months].to_numpy(dtype=float)[0]

def humidity(city):
    return df.loc[df['Місто'] == city, months].to_numpy(dtype=float)[0]

print('Місто:', MY_CITY, '| регіон:', my_region, '| норма:', MY_NORM, '%')
print('Ранкові значення:', morning)
print('n =', morning.size, '| x̄ =', round(morning.mean(), 4), '| s =', round(morning.std(ddof=1), 4))
```

```python
t_stat, p_value = stats.ttest_1samp(morning, popmean=MY_NORM)

print('t_stat =', round(t_stat, 6))
print('p_value =', round(p_value, 6))
print('x̄ =', round(morning.mean(), 4), '| μ0 =', MY_NORM, '| різниця =', round(morning.mean() - MY_NORM, 4))

alpha = 0.05
print('H0 відхиляємо?' , p_value < alpha)
```

```python
n = morning.size
x_mean = morning.mean()
s = morning.std(ddof=1)
se = s / np.sqrt(n)
t_manual = (x_mean - MY_NORM) / se

print('x̄      =', x_mean)
print('s(ddof=1) =', s)
print('s/√n   =', se)
print('t_manual =', t_manual)
print('t_stat  (ttest_1samp) =', t_stat)
print('різниця =', abs(t_manual - t_stat))
print('збігається після округлення?', round(t_manual, 4) == round(t_stat, 4))
```

```python
CITY_3 = 'Львів'
city_3 = humidity(CITY_3)

print('Одеса  (Центр/Південь): mean =', round(morning.mean(), 4), '| std =', round(morning.std(ddof=1), 4))
print(f'{CITY_3}: mean =', round(city_3.mean(), 4), '| std =', round(city_3.std(ddof=1), 4))

stat_lev, p_lev = stats.levene(morning, city_3)
print('Лівен: W =', round(stat_lev, 6), '| p_lev =', round(p_lev, 6))
print('Дисперсії рівні (p_lev >= 0.05)?', p_lev >= alpha)
print('Відношення дисперсій s2_%s / s2_%s =' % (CITY_3, MY_CITY), round(city_3.var(ddof=1) / morning.var(ddof=1), 3))
```

```python
# Крок 2 — Welch-варіант, бо p_lev < 0.05 (дисперсії НЕ рівні)
res_welch = stats.ttest_ind(morning, city_3, equal_var=False)
print('Welch (equal_var=False): t =', round(res_welch.statistic, 6),
      '| p =', round(res_welch.pvalue, 6), '| df =', round(res_welch.df, 4))

print('різниця середніх (Одеса − %s) =' % CITY_3, round(morning.mean() - city_3.mean(), 4), 'п.п.')
```

```python
print('%-20s %-9s %8s %8s %10s' % ('Місто', 'Регіон', 's_Одеса', 's_міста', 'p_lev'))
print('-' * 62)
candidates = {}
for _, r in df.iterrows():
    if r['Регіон'] == my_region or r['Місто'] == CITY_3:
        continue
    v = r[months].to_numpy(dtype=float)
    w, p = stats.levene(morning, v)
    candidates[r['Місто']] = p
    print('%-20s %-9s %8.4f %8.4f %10.5f  %s' % (
        r['Місто'], r['Регіон'], morning.std(ddof=1), v.std(ddof=1), p,
        'рівні' if p >= alpha else 'не рівні'))
```

```python
# Обираю місто з протилежним результатом Лівена (p_lev >= 0,05), інший регіон
CITY_4 = 'Івано-Франківськ'
city_4 = humidity(CITY_4)
print('Обране місто:', CITY_4, '| регіон:', df.loc[df['Місто'] == CITY_4, 'Регіон'].iloc[0])
print('Std Одеси =', round(morning.std(ddof=1), 4), '| Std %s =' % CITY_4, round(city_4.std(ddof=1), 4))

stat_lev4, p_lev4 = stats.levene(morning, city_4)
print('Лівен: W =', round(stat_lev4, 6), '| p_lev =', round(p_lev4, 6))
print('Дисперсії рівні (p_lev >= 0.05)?', p_lev4 >= alpha)
```

```python
# Крок 2 — тепер p_lev >= 0.05, тож коректний СТАНДАРТНИЙ варіант (equal_var=True)
res_std = stats.ttest_ind(morning, city_4, equal_var=True)
print('Стандартний (equal_var=True): t =', round(res_std.statistic, 6),
      '| p =', round(res_std.pvalue, 6), '| df =', round(res_std.df, 4))
print('різниця середніх (Одеса − %s) =' % CITY_4, round(morning.mean() - city_4.mean(), 4), 'п.п.')
```

```python
evening = morning - amplitude
pairs = pd.DataFrame({'Місяць': months, 'Ранок (8:00)': morning,
                      'Вечір (20:00)': evening, 'Різниця': morning - evening})
pairs
```

```python
res_rel = stats.ttest_rel(morning, evening)
print('t_rel =', round(res_rel.statistic, 6), '| p =', res_rel.pvalue, '| df =', res_rel.df)
print('середня добова різниця =', round((morning - evening).mean(), 4),
      '| std різниць =', round((morning - evening).std(ddof=1), 4))
print('p_value < 0.05?', res_rel.pvalue < alpha)
```

```python
# Для порівняння потужності: незалежний критерій на тих самих даних (коректно не застосовується)
res_ind_wrong = stats.ttest_ind(morning, evening, equal_var=False)
print('ttest_ind (некоректний тут): t =', round(res_ind_wrong.statistic, 4), '| p = %.4f' % res_ind_wrong.pvalue)
print('ttest_rel (коректний):      t =', round(res_rel.statistic, 4), '| p = %.3e' % res_rel.pvalue)
print('Std ранку =', round(morning.std(ddof=1), 4), '| Std вечора =', round(evening.std(ddof=1), 4),
      '| Std різниць =', round((morning - evening).std(ddof=1), 4))
```

```python
summary = pd.DataFrame([
    ['1. ttest_1samp (Одеса vs 73%)', 't = %.4f' % t_stat, 'p = %.4f' % p_value,
     'H0 не відхиляємо: значущої відмінності від норми нет'],
    ['2. Ручний t = (x̄−μ₀)/(s/√n)', 't = %.4f' % t_manual, '—',
     'збігається з ttest_1samp'],
    ['3. levene (Одеса, Львів)', 'W = %.4f' % stat_lev, 'p_lev = %.4f' % p_lev,
     'дисперсії НЕ рівні → equal_var=False (Welch)'],
    ['3. ttest_ind (Одеса, Львів)', 't = %.4f' % res_welch.statistic, 'p = %.4f' % res_welch.pvalue,
     'значущої різниці середніх немає'],
    ['4. levene (Одеса, Івано-Франківськ)', 'W = %.4f' % stat_lev4, 'p_lev = %.4f' % p_lev4,
     'дисперсії рівні → equal_var=True (стандартний)'],
    ['4. ttest_ind (Одеса, Івано-Франківськ)', 't = %.4f' % res_std.statistic, 'p = %.4f' % res_std.pvalue,
     'значущої різниці середніх немає'],
    ['5. ttest_rel (ранок vs вечір)', 't = %.4f' % res_rel.statistic, 'p = %.2e' % res_rel.pvalue,
     'добова різниця значуща (5,0 п.п.)'],
], columns=['Крок', 'Статистика', 'p-значення', 'Висновок'])
summary
```

## Письмові відповіді на контрольні питання

### 1. Чому для перевірки середнього однієї вибірки використовується саме t-, а не z-розподіл?

Тому що z-критерій (нормальний) вимагає, щоб **дисперсія генеральної сукупності σ² була відома** і щоб вибірка була нормально розподілена. На практиці σ майже ніколи не відома, а 12 місячних значень — це мала вибірка, для якої нормальність розподілу перевірити ненадійно, а навіть за нормальної форми середнє з n = 12 не є нормальним (за CLT воно наближається до нормального лише при великих n). t-критерій замінює відоме σ на **оцінку** s = σ-hat (з ddof = 1), через що сам статистичний розподіл t = (x̄ − μ₀)/(s/√n) вже не є нормальним, а має важкі хвости (закон Стьюдента з df = n − 1 = 11). Саме ця «плата» за невідоме σ і робить t-критерій коректним для малих вибірок: він дає ширший, але саме правильний інтервал і p-значення. При n → ∞ t-розподіл автоматично наближається до нормального, і t-критерій перетворюється на z-критерій.

### 2. Чому неправильний вибір `equal_var` у `ttest_ind` дає систематично неправильний p-value, а не просто «менш точний»?

Стандартний варіант використовує **спільну** оцінку дисперсії s²_pooled і, відповідно, стандартну помилку SE_pooled = √(s²_pooled·(1/n₁ + 1/n₂)); Welch-варіант використовує SE_Welch = √(s₁²/n₁ + s₂²/n₂) і менше число степенів свободи. Якщо дисперсії насправді різні, але застосовано `equal_var=True`, то s²_pooled усереднює дуже різні дисперсії й **систематично переоцінює** SE для групи з меншою дисперсією (або недооцінює для іншої) — це зміщений знаменник, а t-статистика виявляється зміщеною відносно його правилього нульового розподілу. Наслідок — p-value обчислюється не з того розподілу, якому реально підпорядковане спостережене t, тобто воно **некоректне за рівнем** (p-хиба): реальний рівень significance може виявитись істотно більшим за nominal 0,05 (класично — завищена ймовірність хибного відхилення H0). Це не «недостатня точність» (яка зменшила б потужність, але не змінила правильність висновку), а порушення правила першого рівня. Симетрично, якщо дисперсії рівні, а застосовано Welch, ми платимо втратою ефективних степенів свободи ((df ≈ 19 замість 22) — тобто втратою потужності, але не коректністю p-value. Саме тому Лівен треба застосовувати **до** вибору варіанта, а в цьому ноутбуці вибір `equal_var` у Завданнях 3 і 4 ґрунтується саме на його результаті.

### 3. Чому парний критерій потужніший за незалежний на тих самих даних «до/після» і що саме усуває різниця для кожної пари?

При обчисленні dᵢ = xᵢ − yᵢ з пари зникає **спільна (міжумова) варіація** — усе, що однаково впливало на обидва вимірювання: сезонний фон, тип погоди, тривалість дня, властивості конкретного дня. Математично Var(d) = Var(x) + Var(y) − 2Cov(x, y): за повної додатковості (Cov = 0) дисперсії **додавалися б**, а при сильному додатковому зв'язку (Cov ≈ Var) різниця має майже нульову дисперсію, бо xᵢ ≈ yᵢ. Тобто різниця «очищає» сигнал від індивідуального зсуву об'єкта, і стандартна помилка зменшується пропорційно корелюванні пар. Чим сильніший зв'язок у парах, тим більше виграш у потужності. Емпірично в Завданні 5: Std різниць = 1,595 п.п. проти Std окремих вимірювань ≈ 5,6 і 7,2 п.п.; t_rel = 10,86 (p = 3,2·10⁻⁷) проти t_ind = 1,90 (p ≈ 0,072) на тих самих числах. Незалежний критерій фактично «викидає» інформацію про те, що місяці попарно узгоджені.

### 4. Чи узгоджується висновок формального критерію Лівена з порівнянням `.std()` «на око» (за власними результатами Завдань 3–4)?

**У Завданні 3 — так, повністю.** Std(Одеса) = 5,62, Std(Львів) = 8,44: різниця помітна «неозброєним оком», дисперсія Львова у ≈ 2,25 раза більша, і Лівен це підтвердив (W = 4,57, p = 0,0438 < 0,05 → «не рівні»).

**У Завданні 4 — висновок Лівена збігається з напрямом порівняння, але magnitude «на око» виглядає інакше.** Std(Одеса) = 5,62, Std(Івано-Франківськ) = 7,97 — різниця std досить помітна, проте Лівен дає p = 0,0986 ≥ 0,05, тобто **статистично значущої** відмінності дисперсій при n = 12 у кожній групі не виявлено.

Пояснення: критерій Лівена (центральна версія, median-centered) перевіряє гіпотезу не про рівність самих дисперсій «у точку», а перевіряє, чи відрізняються **масштаби розкиду** так, щоб це не пояснювалося випадковим розкидом і доведено до рівня значущості. З n₁ = n₂ = 12 потужність критерію мала: щоб відхилити H0, різниця розкиду має бути досить великою **і** відповідати формі розподілів. Тому «на око» ми бачимо різницю стандартних відхилень, але формальний критерій при α = 0,05 не вважає її статистично значущою. Практичний висновок: око варте для **вибору кандидатів** і для sanity-check, але рішення про `equal_var` має спиратися на `levene()` — інакше ми знову підмінятимемо формальну перевірку інтуїцією. (Обидва погляди тут збігаються напрямом: у Завданні 3 Лівен каже «не рівні» — і видно різницю; у Завданні 4 Лівен каже «немає підстав відкидати рівність» — і це теж не суперечить поміркованому висновку про близькість розкидів. Просто рівність дисперсій у статистичному сенсі ніколи не означає їхню повну ідентичність.)

4. **Парний** `ttest_rel` (ранок 8:00 vs вечір 20:00, 12 пар): t(11) = 10,8562, p = 3,23·10⁻⁷ → добова різниця 5,0 п.п. **статистично значуща**; парний критерій коректний, бо різниця для кожної пари усуває спільний сезонний/погодний складник (Std різниць 1,60 п.п. проти ≈ 5,6 і 7,2 п.п. для окремих вимірювань), що підвищує потужність порівняно з незалежним критерієм (t = 1,896, p ≈ 0,072 на тих самих даних).
