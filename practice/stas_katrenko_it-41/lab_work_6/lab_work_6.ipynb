```python
import pandas as pd
import numpy as np

# Створюємо Series для Варіанта 3 (Одеса)
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
odessa_data = [82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83]

odessa_s = pd.Series(odessa_data, name="Одеса")
```

**1**

```python
# Отримання основних мір положення й розсіювання
desc = odessa_s.describe()

print("=== ЗАВДАННЯ 1: describe() для м. Одеса ===")
print(desc)
```

**2**

```python
rozmakh = odessa_s.max() - odessa_s.min()
var_sample = odessa_s.var(ddof=1)
var_pop = odessa_s.var(ddof=0)
std_val = odessa_s.std(ddof=1)
q1 = odessa_s.quantile(0.25)
q3 = odessa_s.quantile(0.75)
iqr = q3 - q1
cv = std_val / odessa_s.mean() * 100

print("=== ЗАВДАННЯ 2: Ручні розрахунки ===")
print(f"Розмах (Range): {rozmakh}")
print(f"Вибіркова дисперсія (ddof=1): {var_sample:.4f}")
print(f"Генеральна дисперсія (ddof=0): {var_pop:.4f}")
print(f"Різниця між дисперсіями: {var_sample - var_pop:.4f}")
print(f"Стандартне відхилення (std): {std_val:.4f}")
print(f"Перший квартиль (Q1 / 25%): {q1:.2f}")
print(f"Третій квартиль (Q3 / 75%): {q3:.2f}")
print(f"Міжквартильний розмах (IQR): {iqr:.2f}")
print(f"Коефіцієнт варіації (CV): {cv:.2f}%")

assert np.isclose(q1, desc['25%']), "Q1 не збігається!"
assert np.isclose(q3, desc['75%']), "Q3 не збігається!"
print("\nУспіх: Значення 25% та 75% повністю збігаються з .describe()!")
```

**3**

```python
all_cities_data = {
    "Місто": ["Київ", "Львів", "Одеса", "Харків", "Дніпро", "Чернігів", "Івано-Франківськ", "Полтава", "Суми", "Ужгород"],
    "Регіон": ["Центр", "Захід", "Південь", "Схід", "Схід", "Центр", "Захід", "Центр", "Схід", "Захід"],
    "Січ": [84, 85, 82, 83, 82, 85, 86, 84, 85, 84],
    "Лют": [82, 83, 80, 80, 79, 83, 84, 81, 82, 82],
    "Бер": [78, 80, 78, 75, 74, 79, 81, 76, 77, 79],
    "Кві": [70, 73, 74, 68, 67, 71, 75, 69, 70, 73],
    "Тра": [65, 68, 71, 62, 61, 66, 70, 63, 64, 68],
    "Чер": [63, 65, 69, 60, 59, 63, 67, 61, 62, 66],
    "Лип": [61, 63, 67, 58, 57, 61, 65, 59, 60, 64],
    "Сер": [63, 65, 68, 60, 59, 63, 67, 61, 62, 66],
    "Вер": [68, 70, 72, 66, 65, 69, 72, 67, 68, 71],
    "Жов": [75, 77, 76, 73, 72, 76, 78, 74, 75, 77],
    "Лис": [81, 82, 80, 79, 78, 82, 83, 80, 81, 82],
    "Гру": [85, 86, 83, 84, 83, 86, 87, 85, 86, 85]
}

df = pd.DataFrame(all_cities_data)

df["Середня_річна_вологість"] = df[months].mean(axis=1)

south_region = df[df["Регіон"] == "Південь"]["Середня_річна_вологість"]
east_region = df[df["Регіон"] == "Схід"]["Середня_річна_вологість"]

print("=== ЗАВДАННЯ 3: describe() для Півдня (1 місто) ===")
print(south_region.describe())

print("\n=== ЗАВДАННЯ 3: describe() для Сходу (3 міста) ===")
print(east_region.describe())

cv_east = east_region.std() / east_region.mean() * 100
print(f"\nКоефіцієнт варіації регіону Схід: {cv_east:.2f}%")
```

**4**

```python
mean_val = odessa_s.mean()
median_val = odessa_s.median()

print("=== ЗАВДАННЯ 4: Форма розподілу ===")
print(f"Середнє (mean): {mean_val:.2f}%")
print(f"Медіана (50%):  {median_val:.2f}%")
print(f"Різниця:        {abs(mean_val - median_val):.2f}%")
```

**5**

```python
lviv_s = pd.Series([85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86], name="Львів")

cv_odessa = odessa_s.std() / odessa_s.mean() * 100
cv_lviv = lviv_s.std() / lviv_s.mean() * 100

print("=== ЗАВДАННЯ 5: Порівняння CV двох міст ===")
print(f"Одеса: Mean = {odessa_s.mean():.2f}%, std = {odessa_s.std():.2f}%, CV = {cv_odessa:.2f}%")
print(f"Львів: Mean = {lviv_s.mean():.2f}%, std = {lviv_s.std():.2f}%, CV = {cv_lviv:.2f}%")
```

## Головні практичні висновки, які ви закріпили в цій роботі: ##

1. **`.describe()` як базовий інструмент**: Один виклик надає повний стат-паспорт вибірки — від центрів групування (`mean`, `50%`) до оцінки варіативності й меж (`std`, `IQR`, `min`/`max`).
2. **Булева фільтрація vs `.groupby()**`: Фільтрація вигляду `df[df["Регіон"] == "..."]` дає повний контроль над окремою групою, коли потрібно швидко порівняти 2–3 підмножини без перебудови всієї структури таблиці.
3. **Нормування через CV**: Сирий `std` показує відхилення в початкових одиницях (в даному випадку в %), але лише **коефіцієнт варіації ($CV = \frac{std}{mean} \cdot 100\%$)** дозволяє об'єктивно порівняти рівень нестабільності/розсіювання між різними містами та регіонами.
4. **Форма розподілу має значення**: Збіг чи розходження середнього та медіани одразу вказує на симетрію або наявність перекосів/викидів у даних, підказуючи, який показник репрезентативніший.
