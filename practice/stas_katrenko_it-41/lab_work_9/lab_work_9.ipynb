```python
import pandas as pd
import numpy as np

np.random.seed(42)

BASE_PRICE = 640
```

**1**

```python
data = {
    "вік": [21, 28, 35, 42, 19, 58, 63, 31, 47, 52],
    "дохід": [18000, 32000, 45000, 28000, 15000, 60000, 22000, 38000, 50000, 41000],
    "кількість": [4, 5, 6, 3, 5, 8, 2, 5, 7, 4],
    "місто_доставки": ["Одеса", "Київ", "Львів", "Харків", "Одеса", "Дніпро", "Одеса", "Київ", "Полтава", "Чернігів"]
}

df = pd.DataFrame(data)

print("--- Завдання 1: Початковий DataFrame ---")
print(df)
print("\n" + "="*50 + "\n")
```

**2**

```python
df["сума_замовлення"] = BASE_PRICE * df["кількість"]
df["частка_доходу"] = df["сума_замовлення"] / df["дохід"]

print("--- Завдання 2: З урахуванням похідних змінних ---")
print(df[["вік", "дохід", "кількість", "сума_замовлення", "частка_доходу"]])
print("\n" + "="*50 + "\n")
```

**3**

```python
df["вік_група_cut"] = pd.cut(df["вік"], bins=3)
df["вік_група_qcut"] = pd.qcut(df["вік"], q=3)

print("--- Завдання 3: pd.cut() (однакова ширина інтервалів) ---")
print(df["вік_група_cut"].value_counts().sort_index())

print("\n--- Завдання 3: pd.qcut() (однакова кількість клієнтів) ---")
print(df["вік_група_qcut"].value_counts().sort_index())
print("\n" + "="*50 + "\n")
```

**4**

```python
df["вік_minmax"] = (df["вік"] - df["вік"].min()) / (df["вік"].max() - df["вік"].min())
df["дохід_minmax"] = (df["дохід"] - df["дохід"].min()) / (df["дохід"].max() - df["дохід"].min())

# Z-score Стандартизація
df["вік_zscore"] = (df["вік"] - df["вік"].mean()) / df["вік"].std()
df["дохід_zscore"] = (df["дохід"] - df["дохід"].mean()) / df["дохід"].std()

print("--- Завдання 4: Порівняння оригінальних та масштабованих змінних ---")
print(df[["вік", "вік_minmax", "вік_zscore", "дохід", "дохід_minmax", "дохід_zscore"]].head())
print("\n" + "="*50 + "\n")
```

**5**

```python
city_dummies = pd.get_dummies(df["місто_доставки"], prefix="місто", dtype=int)
df_final = pd.concat([df, city_dummies], axis=1)

print("--- Завдання 5: Фінальна таблиця з One-Hot кодуванням міста ---")
print(df_final.head())
print(f"\nКількість унікальних міст у даних: {df['місто_доставки'].nunique()}")
print(f"Кількість створених індикаторних стовпчиків: {city_dummies.shape[1]}")
```

## Відповіді на підсумкові запитання оцінювання

1. **Різниця між `pd.cut()` та `pd.qcut()`:**
* `pd.cut()` фіксує **довжину інтервалів** ($\Delta = \frac{\text{max} - \text{min}}{k}$), але кількість об'єктів у кожному інтервалі може бути різною.
* `pd.qcut()` фіксує **кількість об'єктів** у кожній групі ($N_{\text{group}} \approx \frac{N_{\text{total}}}{q}$), через що межі інтервалів мають різну довжину.


2. **Чому One-Hot, а не 1, 2, 3 для міст:**
Порядкове кодування використовується лише для **ординальних** змінних, де є логічний порядок (наприклад, *"низький" < "середній" < "високий"*). Для **номінальних** змінних (кольори, міста, бренди) числове кодування вносить хибні математичні відносини порядку та відстані, тому застосовують One-Hot (Dummy) кодування.
3. **Навіщо масштабувати змінні:**
Більшість метричних алгоритмів ($k$-NN, $k$-means, PCA) та регресій з регуляризацією (Lasso, Ridge) обчислюють відстані або градієнти. Якщо ознаки мають різний масштаб, ознака з більшим числовим діапазоном буде повністю домінувати над ознакою з меншим діапазоном, викривляючи результат.
4. **Чому похідна змінна несе більше сенсу:**
Похідні змінні агрегують контекст декількох показників у єдиний коефіцієнт (відношення, щільність, частка). Вона виражає відносну характеристику ($\%$, відносне навантаження, ефективність), яка краще відображає приховану поведінку клієнта, ніж сирі абсолютні числа окремо.
