```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
print(a)
print("shape:", a.shape)
print("dtype:", a.dtype)
```

```python
temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])

days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])

month_numbers = np.arange(1, 13)

print("temps:", temps)
print("shape:", temps.shape, "dtype:", temps.dtype)

print("days_in_month:", days_in_month)
print("shape:", days_in_month.shape, "dtype:", days_in_month.dtype)

print("month_numbers:", month_numbers)
print("shape:", month_numbers.shape, "dtype:", month_numbers.dtype)
```

```python
temps_f = temps * 9 / 5 + 32

print("Температура у °F:")
print(temps_f)

warm_days = days_in_month[temps > 15]
total_warm_days = warm_days.sum()

print("Кількість днів із температурою вище 15°C:", total_warm_days)

print("Кількість днів у році:", days_in_month.sum())
print("Чи дорівнює 365:", days_in_month.sum() == 365)
```

```python
warmest_index = temps.argmax()
coldest_index = temps.argmin()

print("Найтепліший місяць:")
print("індекс:", warmest_index)
print("номер місяця:", month_numbers[warmest_index])
print("температура:", temps[warmest_index], "°C")

print("\nНайхолодніший місяць:")
print("індекс:", coldest_index)
print("номер місяця:", month_numbers[coldest_index])
print("температура:", temps[coldest_index], "°C")

below_zero_months = month_numbers[temps < 0]
print("\nМісяці з температурою нижче нуля:", below_zero_months)

sorted_temps = np.sort(temps)
print("\nТемператури за зростанням:", sorted_temps)

print("Оригінальний temps:", temps)
```

```python
sun_hours = np.array([
    55, 70, 110, 150, 220, 240,
    260, 245, 180, 120, 65, 50
])

data = np.array([temps, days_in_month, sun_hours])

print("data:")
print(data)

print("shape:", data.shape)

print("\nСума кожного рядка:")
print(data.sum(axis=1))

print("\nСереднє кожного рядка:")
print(data.mean(axis=1))

print("\nСума кожного стовпця:")
print(data.sum(axis=0))
```

```python
rows = np.array([temps, days_in_month])

print("rows:")
print(rows)
print("shape:", rows.shape)

correction = np.array([1, 0])

print("\ncorrection:")
print(correction)
print("shape:", correction.shape)

# Спробуйте:
rows + correction
```

```python
correction_2d = correction.reshape(2, 1)

print("correction_2d:")
print(correction_2d)
print("shape:", correction_2d.shape)

result = rows + correction_2d

print("\nРезультат:")
print(result)
```
