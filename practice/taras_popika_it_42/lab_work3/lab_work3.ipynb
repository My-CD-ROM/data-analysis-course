```python
# Варіант 6
# Завдання 1
import pandas as pd

wide = pd.DataFrame({
    "місто": ["Чернігів"],
    "Січ": [80],
    "Лют": [76],
    "Бер": [70],
    "Кві": [60],
    "Тра": [52],
    "Чер": [46],
    "Лип": [42],
    "Сер": [44],
    "Вер": [54],
    "Жов": [68],
    "Лис": [78],
    "Гру": [84]
})

print(wide)
```

```python
# Завдання 2 
tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність"
)

print(tidy)
print("Форма таблиці:", tidy.shape)
```

```python
# Завдання 4
back = tidy.pivot(
    index="місто",
    columns="місяць",
    values="хмарність"
)

print(back)
```

```python
months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

back = back[months]

print(back)
```

```python
back = back.reset_index()

print(wide.equals(back))
```

```python
# Завдання 5
tidy.to_csv("variant.csv", index=False)

tidy.to_json(
    "variant.json",
    orient="records",
    force_ascii=False
)
```

```python
reloaded_csv = pd.read_csv("variant.csv")
reloaded_json = pd.read_json("variant.json")

print("CSV:")
print(reloaded_csv)

print("\nJSON:")
print(reloaded_json)
```

```python
print("CSV збігається:", tidy.equals(reloaded_csv))
print("JSON збігається:", tidy.equals(reloaded_json))
```

```python
# Завдання 3. Класифікація змінних
# місто — номінальна шкала, тому що назва міста є категорією без природного порядку між значеннями.
# місяць — порядкова шкала, тому що місяці мають природний порядок протягом року: січень, лютий, березень тощо.
# хмарність — шкала відношень, тому що це числова величина у відсотках із природним нульовим значенням, а співвідношення значень має зміст.
```

```python
# 1. Чому «місто» — номінальна шкала, а «хмарність» — шкала відношень?
# Місто є номінальною шкалою, оскільки назви міст лише позначають різні категорії, і між ними немає числового порядку. Хмарність є шкалою відношень, оскільки вимірюється у відсотках, має нульове значення (0% хмарності) і можна порівнювати відношення між значеннями, наприклад 80% — це вдвічі більше за 40%.

# 2. Чому wide-таблиця з місяцями у стовпцях не є tidy?
# Wide-таблиця не є tidy, тому що змінна «місяць» фактично закодована в назвах стовпців (Січ, Лют, Бер тощо), а не представлена окремим стовпцем. Після melt() назви місяців перетворюються на значення стовпця місяць, а відповідні показники — на значення стовпця хмарність, тому кожен рядок описує одне спостереження.

# 3. Чим CSV принципово відрізняється від JSON?
# CSV зберігає дані як таблицю з рядками та стовпцями, де значення зазвичай розділені комами. JSON зберігає дані у вигляді структурованих об'єктів і масивів, тому він краще підтримує вкладені та складні структури даних. У цій роботі orient="records" представляє кожен рядок таблиці як окремий JSON-об'єкт.

```
