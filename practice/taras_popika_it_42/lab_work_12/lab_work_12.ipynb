```python
# Практика 12
# Варіант 6 — Чернігів


import numpy as np
from scipy import stats


# ============================================================
# ЗАВДАННЯ 1. Вибірка денних температур
# ============================================================

np.random.seed(6)

july_temp = 20
daily_temps = np.random.normal(
    loc=july_temp,
    scale=2.5,
    size=30
)

mean = daily_temps.mean()
s = daily_temps.std(ddof=1)
n = len(daily_temps)

print("ЗАВДАННЯ 1")
print("-" * 50)

print("Місто: Чернігів")
print(f"Липнева температура: {july_temp} °C")
print("\nВибірка денних температур:")
print(daily_temps)

print(f"\nСереднє значення: {mean:.4f} °C")
print(f"Вибіркове стандартне відхилення: {s:.4f} °C")
print(f"Розмір вибірки: {n}")


# ============================================================
# ЗАВДАННЯ 2. Довірчий інтервал для середньої температури
# ============================================================

se = s / np.sqrt(n)

ci_mean = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean,
    scale=se
)

print("\n\nЗАВДАННЯ 2")
print("-" * 50)

print(f"Стандартна похибка: {se:.4f}")
print(f"95% довірчий інтервал для середньої температури: "
      f"({ci_mean[0]:.4f}, {ci_mean[1]:.4f}) °C")


# Письмова відповідь:
#
# Для побудови довірчого інтервалу використовується t-розподіл,
# оскільки справжнє стандартне відхилення генеральної сукупності
# (σ) невідоме. Ми маємо лише вибіркове стандартне відхилення s,
# обчислене за 30 спостереженнями.
#
# Z-розподіл можна було б використовувати, якби справжнє σ було
# відоме. У нашому випадку ця умова не виконана, тому для оцінки
# середньої температури при невідомому σ використовується
# t-розподіл із n - 1 = 29 ступенями вільності.


# ============================================================
# ЗАВДАННЯ 3. Довірчий інтервал для частки
# ============================================================

above_july_temp = daily_temps > july_temp

p_hat = above_july_temp.mean()

n_p = n * p_hat
n_1p = n * (1 - p_hat)

se_p = np.sqrt(p_hat * (1 - p_hat) / n)

ci_prop = stats.norm.interval(
    0.95,
    loc=p_hat,
    scale=se_p
)

print("\n\nЗАВДАННЯ 3")
print("-" * 50)

print(f"Кількість днів із температурою вище {july_temp} °C: "
      f"{above_july_temp.sum()}")

print(f"Оцінка частки p_hat: {p_hat:.4f}")
print(f"n * p_hat = {n_p:.4f}")
print(f"n * (1 - p_hat) = {n_1p:.4f}")

print(f"95% довірчий інтервал для частки: "
      f"({ci_prop[0]:.4f}, {ci_prop[1]:.4f})")


# Письмова відповідь:
#
# Для нормального наближення необхідно, щоб n * p_hat та
# n * (1 - p_hat) були достатньо великими. У нашій вибірці
# обидва значення потрібно перевірити перед використанням
# формули для довірчого інтервалу.
#
# Якщо обидва значення не менші за 5, умова нормального
# наближення виконується, тому побудований довірчий інтервал
# для частки є прийнятним.
#
# Вибірка генерується симетрично навколо 20 °C, тому теоретично
# частка днів із температурою вище 20 °C повинна бути близькою
# до 0.5.
#
# Якщо 0.5 потрапляє в отриманий інтервал, це означає, що
# результат вибірки узгоджується з теоретичним значенням 0.5.
#
# Якщо 0.5 не потрапляє в інтервал, це не означає, що теоретичне
# значення неправильне. Вибірка містить лише 30 випадкових
# спостережень, тому через випадковість конкретна вибірка може
# дати інтервал, який не накриває теоретичне значення. При
# повторенні процедури приблизно 95% побудованих інтервалів
# повинні містити справжнє значення, а близько 5% — ні.


# ============================================================
# ЗАВДАННЯ 4. Ширина інтервалу при різних рівнях довіри
# ============================================================

ci_90 = stats.t.interval(
    0.90,
    df=n - 1,
    loc=mean,
    scale=se
)

ci_95 = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean,
    scale=se
)

ci_99 = stats.t.interval(
    0.99,
    df=n - 1,
    loc=mean,
    scale=se
)

width_90 = ci_90[1] - ci_90[0]
width_95 = ci_95[1] - ci_95[0]
width_99 = ci_99[1] - ci_99[0]

print("\n\nЗАВДАННЯ 4")
print("-" * 50)

print("Рівень довіри | Нижня межа | Верхня межа | Ширина")
print("-" * 55)

print(f"90%           | {ci_90[0]:10.4f} | "
      f"{ci_90[1]:10.4f} | {width_90:.4f}")

print(f"95%           | {ci_95[0]:10.4f} | "
      f"{ci_95[1]:10.4f} | {width_95:.4f}")

print(f"99%           | {ci_99[0]:10.4f} | "
      f"{ci_99[1]:10.4f} | {width_99:.4f}")


# Письмова відповідь:
#
# Ширина довірчого інтервалу збільшується при переході від
# 90% до 95% і далі до 99% рівня довіри.
#
# Це відбувається тому, що для більшої впевненості в накритті
# істинного значення потрібен ширший діапазон можливих значень.
# Тому 99% довірчий інтервал ширший за 95%, а 95% — ширший
# за 90%.
#
# Щоб отримати вужчий інтервал при тому самому рівні довіри,
# потрібно збільшити розмір вибірки n. При збільшенні n
# стандартна похибка s / sqrt(n) зменшується, тому інтервал
# стає вужчим.


# ============================================================
# ЗАВДАННЯ 5. Інтерпретація результату словами
# ============================================================

print("\n\nЗАВДАННЯ 5")
print("-" * 50)

print(f"95% довірчий інтервал: "
      f"({ci_mean[0]:.4f}, {ci_mean[1]:.4f}) °C")


# Письмова відповідь:
#
# Отриманий 95% довірчий інтервал показує діапазон значень,
# отриманий за заданою процедурою оцінювання середньої
# температури Чернігова на основі нашої вибірки.
#
# Якщо багато разів генерувати нові вибірки по 30 днів і щоразу
# будувати 95% довірчий інтервал таким самим способом, то
# приблизно 95% таких інтервалів накриватимуть істинне середнє
# значення генеральної сукупності.
#
# Це не означає, що для вже побудованого конкретного інтервалу
# існує 95% ймовірність містити істинне значення. Істинне
# значення є фіксованим, а змінюється сам інтервал від вибірки
# до вибірки.


# ============================================================
# КОНТРОЛЬНІ ПИТАННЯ
# ============================================================


# 1. Яка різниця між точковою оцінкою і довірчим інтервалом?
#
# Точкова оцінка — це одне число, яке використовується для
# оцінювання невідомого параметра генеральної сукупності.
# Наприклад, вибіркове середнє x̄ є точковою оцінкою
# генерального середнього.
#
# Але одне число не показує, наскільки точно ми оцінили
# параметр. Довірчий інтервал дає діапазон можливих значень
# параметра та враховує випадкову похибку вибірки.


# 2. Чому використовують t-розподіл, а не нормальний?
#
# Якщо σ генеральної сукупності невідоме, ми замінюємо його
# вибірковим стандартним відхиленням s. Через цю додаткову
# невизначеність використовується t-розподіл.
#
# Особливо це важливо при невеликих вибірках. У нашому випадку
# n = 30, а справжнє σ невідоме, тому використовується
# t-розподіл із n - 1 = 29 ступенями вільності.


# 3. Що насправді означає "95% довіри"?
#
# 95% довіри означає властивість процедури побудови інтервалу.
# Якщо багато разів брати незалежні вибірки одного розміру та
# для кожної будувати інтервал однаковим способом, приблизно
# 95% таких інтервалів накриватимуть істинний параметр.
#
# Не можна говорити, що для вже отриманого конкретного
# інтервалу існує 95% ймовірність того, що істинне значення
# знаходиться всередині. Після побудови інтервал уже визначений,
# а істинний параметр вважається фіксованим.


# 4. Які три фактори визначають ширину довірчого інтервалу?
#
# Ширина інтервалу залежить від:
#
# 1) Рівня довіри:
#    чим вищий рівень довіри, тим ширший інтервал.
#
# 2) Розміру вибірки n:
#    чим більша вибірка, тим менша стандартна похибка
#    і тим вужчий інтервал.
#
# 3) Варіативності даних (стандартного відхилення s):
#    чим більша мінливість даних, тим більша стандартна
#    похибка і тим ширший довірчий інтервал.
#
# Отже, щоб отримати вужчий інтервал без зменшення рівня
# довіри, потрібно збільшити розмір вибірки.
```
