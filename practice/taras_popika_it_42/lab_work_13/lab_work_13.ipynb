```python
# ПРАКТИКА 13
# Варіант 6 — Чернігів

import numpy as np
from scipy import stats


# ============================================================
# ВИХІДНІ ДАНІ
# ============================================================

np.random.seed(6)

city = "Чернігів"
july_temp = 20
mu0 = 19.4
n = 30
alpha = 0.05

daily_temps = np.random.normal(
    loc=july_temp,
    scale=2.5,
    size=n
)

x_bar = daily_temps.mean()
s = daily_temps.std(ddof=1)


print("ПРАКТИКА 13")
print("=" * 60)
print(f"Варіант: 6")
print(f"Місто: {city}")
print(f"Стара норма μ0: {mu0} °C")
print(f"Кількість вимірювань n: {n}")
print()
print("Згенерована вибірка daily_temps:")
print(daily_temps)
print()
print(f"Вибіркове середнє x̄ = {x_bar:.4f} °C")
print(f"Вибіркове стандартне відхилення s = {s:.4f} °C")


# ============================================================
# ЗАВДАННЯ 1. ВИБІРКА І ТОЧКОВІ ОЦІНКИ
# ============================================================

print("\n")
print("ЗАВДАННЯ 1")
print("=" * 60)

print(f"Вибіркове середнє x̄ = {x_bar:.4f} °C")
print(f"Стандартне відхилення s = {s:.4f} °C")

# Письмова відповідь:
#
# Вибіркове середнє x̄ є точковою оцінкою теоретичного
# середнього значення температури μ.
#
# Вибіркове стандартне відхилення s є точковою оцінкою
# стандартного відхилення генеральної сукупності σ.
#
# Для міста Чернігів за згенерованими 30 поточними
# вимірюваннями отримано вибіркове середнє x̄ та
# стандартне відхилення s, значення яких наведені вище.


# ============================================================
# ЗАВДАННЯ 2. ФОРМУЛЮВАННЯ H0 І H1
# ============================================================

print("\n")
print("ЗАВДАННЯ 2")
print("=" * 60)

print("H0: μ = 19.4 °C")
print("H1: μ ≠ 19.4 °C")
print("Тип тесту: двобічний")

# Письмова відповідь:
#
# Нульова гіпотеза:
# H0: μ = 19.4 °C.
#
# Це означає, що стара кліматична норма середньої липневої
# температури для Чернігова залишається правильною.
#
# Альтернативна гіпотеза:
# H1: μ ≠ 19.4 °C.
#
# Вона означає, що середня липнева температура Чернігова
# відрізняється від старої норми 19.4 °C.
#
# Обрано ДВОБІЧНИЙ тест, оскільки за умовою потрібно
# перевірити, чи стара норма більше не відповідає дійсності.
# Нас цікавить будь-яке відхилення від 19.4 °C:
# як підвищення температури, так і її зниження.
#
# Тип тесту визначено ДО обчислення p-value, щоб не
# підлаштовувати статистичний тест під отриманий результат.


# ============================================================
# ЗАВДАННЯ 3. z-СТАТИСТИКА І p-value
# ============================================================

print("\n")
print("ЗАВДАННЯ 3")
print("=" * 60)

se = s / np.sqrt(n)

z = (x_bar - mu0) / se

# Двобічний тест
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

print(f"Стандартна похибка se = {se:.4f}")
print(f"z-статистика = {z:.4f}")
print(f"p-value = {p_value:.4f}")

# Письмова відповідь:
#
# Стандартна похибка обчислюється за формулою:
#
# se = s / sqrt(n)
#
# Вона показує типовий розмір випадкового відхилення
# вибіркового середнього від істинного середнього.
#
# z-статистика показує, на скільки стандартних похибок
# вибіркове середнє x̄ відрізняється від старої норми μ0.
#
# Для двобічного тесту p-value обчислюється за формулою:
#
# p-value = 2 * (1 - Φ(|z|))
#
# де Φ — функція розподілу стандартного нормального
# розподілу.


# ============================================================
# ЗАВДАННЯ 4. РІШЕННЯ ПРИ α = 0.05
# ============================================================

print("\n")
print("ЗАВДАННЯ 4")
print("=" * 60)

print(f"p-value = {p_value:.4f}")
print(f"alpha = {alpha}")

if p_value < alpha:
    decision = "Відхиляємо H0"
    print(decision)
    
    print(
        f"Висновок: для міста {city} отримано статистично "
        f"значущі підстави вважати, що середня липнева "
        f"температура відрізняється від старої норми "
        f"{mu0} °C (p = {p_value:.4f}, двобічний тест, "
        f"α = 0.05)."
    )
else:
    decision = "Не відхиляємо H0"
    print(decision)
    
    print(
        f"Висновок: для міста {city} на рівні значущості "
        f"α = 0.05 недостатньо статистичних підстав "
        f"стверджувати, що середня липнева температура "
        f"відрізняється від старої норми {mu0} °C "
        f"(p = {p_value:.4f}, двобічний тест)."
    )


# Письмова відповідь:
#
# Правило прийняття рішення:
#
# якщо p-value < α, то H0 відхиляємо;
# якщо p-value >= α, то H0 не відхиляємо.
#
# У нашому випадку рішення автоматично визначається
# отриманим вище p-value.
#
# Важливо: "не відхиляємо H0" не означає, що ми довели
# абсолютну правильність старої норми. Це означає лише,
# що наявних даних недостатньо для статистично значущого
# відхилення від старої норми при α = 0.05.


# ============================================================
# ЗАВДАННЯ 5. ДОВІРЧИЙ ІНТЕРВАЛ ТА ПОМИЛКИ I/II РОДУ
# ============================================================

print("\n")
print("ЗАВДАННЯ 5")
print("=" * 60)

# 95%-й довірчий інтервал за формулою x̄ ± 1.96 * se

ci_low = x_bar - 1.96 * se
ci_high = x_bar + 1.96 * se

print(f"95%-й довірчий інтервал: ({ci_low:.4f}; {ci_high:.4f}) °C")
print(f"Стара норма μ0 = {mu0:.4f} °C")

if ci_low <= mu0 <= ci_high:
    print("Стара норма μ0 ПОТРАПЛЯЄ у 95%-й довірчий інтервал.")
else:
    print("Стара норма μ0 НЕ ПОТРАПЛЯЄ у 95%-й довірчий інтервал.")


# Письмова відповідь:
#
# 95%-й довірчий інтервал для середньої липневої
# температури Чернігова побудовано за формулою:
#
# x̄ ± 1.96 * se
#
# Якщо μ0 потрапляє в цей інтервал, то результат узгоджується
# з нульовою гіпотезою H0 на рівні значущості 0.05.
#
# Якщо μ0 не потрапляє в інтервал, то H0 відхиляється
# на рівні значущості 0.05.
#
# Тому висновок через 95%-й довірчий інтервал має узгоджуватися
# з рішенням у Завданні 4.


# ============================================================
# ПОМИЛКА I РОДУ
# ============================================================

print("\n")
print("ПОМИЛКА I РОДУ")
print("-" * 60)

# Письмова відповідь:
#
# Помилка I роду означала б, що ми відхилили H0 і заявили,
# що стара норма 19.4 °C більше не відповідає дійсності,
# хоча насправді середня липнева температура Чернігова
# все ще дорівнює 19.4 °C.
#
# Тобто ми помилково оголосили про зміну кліматичної норми,
# хоча реальної зміни не було.
#
# Імовірність такої помилки контролюється рівнем значущості
# α = 0.05.


# ============================================================
# ПОМИЛКА II РОДУ
# ============================================================

print("\n")
print("ПОМИЛКА II РОДУ")
print("-" * 60)

# Письмова відповідь:
#
# Помилка II роду означала б, що ми не відхилили H0
# і залишили стару норму 19.4 °C, хоча насправді
# середня липнева температура Чернігова вже змінилася.
#
# Тобто реальна зміна клімату відбулася, але за результатами
# нашої вибірки ми не змогли виявити її статистично.
#
# Ця помилка відрізняється від помилки I роду:
# при помилці I роду ми заявляємо про зміну, якої насправді
# немає, а при помилці II роду ми не помічаємо реальної зміни.


# ============================================================
# ЯКА ПОМИЛКА ДОРОЖЧА?
# ============================================================

print("\n")
print("ПОРІВНЯННЯ ПОМИЛОК I ТА II РОДУ")
print("-" * 60)

# Письмова відповідь:
#
# Для кліматологів, які готують офіційний оновлений довідник,
# більш дорогою можна вважати помилку II роду.
#
# Якщо реальна зміна середньої температури вже відбулася,
# але її не виявити, офіційний довідник продовжить містити
# застарілу норму. Це може впливати на подальший аналіз
# клімату, планування та прийняття рішень.
#
# Водночас помилка I роду також має наслідки, оскільки може
# призвести до передчасного оновлення офіційної норми.
#
# Тому вибір того, яка помилка є дорожчою, залежить від
# практичної мети дослідження. У випадку моніторингу клімату
# пропуск реальної довгострокової зміни може бути особливо
# небезпечним.


# ============================================================
# КОНТРОЛЬНІ ПИТАННЯ
# ============================================================

print("\n")
print("КОНТРОЛЬНІ ПИТАННЯ")
print("=" * 60)

# 1. Що означає p-value?
#
# p-value — це ймовірність отримати результат, такий самий
# або ще більш екстремальний за спостережений, за умови,
# що нульова гіпотеза H0 є істинною.
#
# У нашій задачі це означає: якщо середня липнева температура
# Чернігова насправді дорівнює 19.4 °C, p-value показує,
# наскільки ймовірно отримати таке або ще сильніше відхилення
# вибіркового середнього від 19.4 °C у випадковій вибірці
# такого самого розміру.
#
# p-value НЕ є ймовірністю того, що H0 правильна.
# Ми не можемо трактувати його як "ймовірність того,
# що стара норма правильна".


# 2. Чому тест обирається до перегляду результату?
#
# Тип тесту потрібно визначити до перегляду x̄ та p-value,
# щоб уникнути підлаштування статистичного аналізу під
# отриманий результат.
#
# Якщо спочатку подивитися на дані, а потім вирішити,
# використовувати однобічний чи двобічний тест, це може
# збільшити ймовірність отримання помилково значущого
# результату.
#
# У нашій задачі заздалегідь обрано двобічний тест, оскільки
# нас цікавить будь-яке відхилення від старої норми 19.4 °C.


# 3. Що означають помилки I та II роду?
#
# Помилка I роду:
# ми відхилили H0 та заявили, що стара норма Чернігова
# 19.4 °C більше не відповідає дійсності, хоча насправді
# вона все ще правильна.
#
# Помилка II роду:
# ми не відхилили H0 і залишили стару норму 19.4 °C,
# хоча насправді середня температура Чернігова вже змінилася.


# 4. Чи узгоджується тест із довірчим інтервалом?
#
# Так. При двобічному тесті з α = 0.05 рішення узгоджується
# з 95%-м довірчим інтервалом.
#
# Якщо μ0 НЕ входить у 95%-й довірчий інтервал, H0
# відхиляється при α = 0.05.
#
# Якщо μ0 входить у 95%-й довірчий інтервал, H0
# не відхиляється при α = 0.05.
#
# Це пов'язано з тим, що 95%-й довірчий інтервал відповідає
# двобічному тесту з рівнем значущості α = 0.05.
```
