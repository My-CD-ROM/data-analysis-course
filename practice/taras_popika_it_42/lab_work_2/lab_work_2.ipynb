```python
import pandas as pd
import numpy as np

# Варіант 1 — Київ

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
          "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]

avg_temp = [-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2]

precip = [38, 36, 38, 41, 53, 74, 84, 69, 47, 36, 44, 44]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip
})

print(df)

print("\nПерші 5 рядків:")
print(df.head())

print("\nРозмір:")
print(df.shape)

print("\nТипи даних:")
print(df.dtypes)

```

```python
# Завдання 1. 
print("Індекс за замовчуванням:")
print(df.index)

df_by_month = df.set_index("month")

print("\nDataFrame з індексом month:")
print(df_by_month)

print("\nРядок для липня:")
print(df_by_month.loc["Лип"])

df = df_by_month.reset_index()

print("\nПовернули звичайний індекс:")
print(df)
```

```python

df.index = range(1, 13)

print("DataFrame:")
print(df)

print("\n.avg_temp через .loc:")
print(df.loc[1:6, "avg_temp"])

print("\navg_temp через .iloc:")
print(df.iloc[0:6, 1])

print("\nmonth і precip:")
print(df[["month", "precip"]])
```

```python
# Завдання 3

print("\nТемпература > 15:")
print(df[df["avg_temp"] > 15][["month", "avg_temp"]])

print("\nТемпература > 10 і precip > 50:")
print(
    df[
        (df["avg_temp"] > 10) &
        (df["precip"] > 50)
    ][["month", "avg_temp", "precip"]]
)

print("\nТемпература < 0 або precip > 80:")
print(
    df[
        (df["avg_temp"] < 0) |
        (df["precip"] > 80)
    ][["month", "avg_temp", "precip"]]
)
```

```python
# Завдання 4. 

df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32


mean_precip = df["precip"].mean()
df["is_wet"] = df["precip"] > mean_precip


def get_season(month_number):
    if month_number in [12, 1, 2]:
        return "зима"
    elif month_number in [3, 4, 5]:
        return "весна"
    elif month_number in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"

df["season"] = df.index.to_series().apply(get_season)

print("\nФінальний DataFrame:")
print(df)


df = df.reset_index(drop=True)

print("\nФінальний DataFrame з індексом 0–11:")
print(df)
```
