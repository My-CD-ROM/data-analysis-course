```python

# Варіант 6 — ЧЕРНІГІВ

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# ВИХІДНІ ДАНІ
# ============================================================

city = "Чернігів"

months = [
    "Січень", "Лютий", "Березень", "Квітень",
    "Травень", "Червень", "Липень", "Серпень",
    "Вересень", "Жовтень", "Листопад", "Грудень"
]

rain_values = [42, 38, 35, 40, 55, 72, 80, 65, 45, 40, 44, 46]

rain = pd.Series(rain_values, index=months, name="Опади, мм")

print("=" * 70)
print("ПРАКТИКА 5. ШКАЛИ ВИМІРЮВАННЯ І ЧАСТОТНИЙ РОЗПОДІЛ")
print("Варіант 6 — Чернігів")
print("=" * 70)

print("\nВихідні дані:")
print(rain)


# ============================================================
# ЗАВДАННЯ 1. КЛАСИФІКАЦІЯ ЗМІННИХ ЗА ШКАЛОЮ ВИМІРЮВАННЯ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 1. КЛАСИФІКАЦІЯ ЗМІННИХ ЗА ШКАЛОЮ ВИМІРЮВАННЯ")
print("=" * 70)

print("""
1. Місто — номінальна шкала.
   Місто є якісною категоріальною змінною. Назви міст лише
   розрізняють категорії і не мають природного порядку.
   Для номінальної шкали коректною мірою положення є мода.

2. Номер місяця — порядкова шкала.
   Місяці мають природний порядок: 1, 2, 3, ..., 12.
   Проте число 1 не є природним нулем, а "нульового місяця"
   як відсутності місяця не існує. Тому в цій роботі
   номер місяця класифікуємо як порядкову шкалу.
   Для неї можна використовувати медіану і моду.

3. Кількість опадів (мм) — відносна шкала.
   Вона має природний нуль: 0 мм означає відсутність опадів.
   Можна порівнювати не тільки різницю, а й відношення значень.
   Наприклад, 80 мм — удвічі більше, ніж 40 мм.
   Для відносної шкали коректні середнє, медіана і мода.
""")

scale_table = pd.DataFrame({
    "Змінна": ["Місто", "Номер місяця", "Кількість опадів (мм)"],
    "Шкала": ["Номінальна", "Порядкова", "Відносна"],
    "Коректні міри положення": [
        "Мода",
        "Медіана, мода",
        "Середнє, медіана, мода"
    ]
})

print("Підсумкова класифікація:")
print(scale_table.to_string(index=False))


# ============================================================
# ЗАВДАННЯ 2. ВАРІАЦІЙНИЙ РЯД І ЧАСТОТНА ТАБЛИЦЯ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 2. ВАРІАЦІЙНИЙ РЯД І ЧАСТОТНА ТАБЛИЦЯ")
print("=" * 70)

# Варіаційний ряд
variation_row = np.sort(rain_values)

print("\nВаріаційний ряд:")
print(", ".join(map(str, variation_row)))

# ------------------------------------------------------------
# Вибираємо 4 інтервали:
# [35; 45)
# [45; 55)
# [55; 65)
# [65; 80]
#
# Для технічної реалізації останню межу робимо 81,
# щоб значення 80 потрапило в останній інтервал.
# ------------------------------------------------------------

bins_4 = [35, 45, 55, 65, 81]

interval_labels = [
    "[35; 45)",
    "[45; 55)",
    "[55; 65)",
    "[65; 81)"
]

intervals = pd.cut(
    rain_values,
    bins=bins_4,
    right=False,
    labels=interval_labels,
    include_lowest=True
)

frequency = pd.Series(intervals).value_counts().sort_index()

relative_frequency = frequency / len(rain_values) * 100

cumulative_frequency = frequency.cumsum()

frequency_table = pd.DataFrame({
    "Інтервал, мм": interval_labels,
    "Абсолютна частота": frequency.values,
    "Відносна частота, %": relative_frequency.values.round(1),
    "Накопичена частота": cumulative_frequency.values
})

print("\nЧастотна таблиця:")
print(frequency_table.to_string(index=False))

# Перевірка
print("\nСума абсолютних частот:", frequency.sum())
print("Сума відносних частот:", round(relative_frequency.sum(), 1), "%")

# Частка до межі третього інтервалу
third_cumulative = cumulative_frequency.iloc[2]
third_share = third_cumulative / len(rain_values) * 100

print("\nНакопичена частота після третього інтервалу:",
      third_cumulative)

print(
    f"Частка місяців із кількістю опадів не більшою за "
    f"65 мм: {third_cumulative} із {len(rain_values)} = "
    f"{third_share:.1f}%"
)

# Відповідь:
# Накопичена частота після третього інтервалу дорівнює 8.
# Отже, 8 із 12 місяців, тобто 66,7% місяців року,
# мають кількість опадів не більшу за 65 мм.


# ============================================================
# ДОДАТКОВО: СТОВПЧИКОВА ДІАГРАМА ЗА МІСЯЦЯМИ
# ============================================================

plt.figure(figsize=(12, 5))

plt.bar(months, rain_values, color="steelblue", edgecolor="black")

plt.title("Середньомісячна кількість опадів у Чернігові")
plt.xlabel("Місяць")
plt.ylabel("Опади, мм")

plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()


# ============================================================
# ЗАВДАННЯ 3. ГІСТОГРАМИ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 3. ГІСТОГРАМИ")
print("=" * 70)

# ------------------------------------------------------------
# 1. Гістограма з кількістю bins за замовчуванням
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    rain_values,
    edgecolor="black",
    color="skyblue"
)

plt.title("Гістограма опадів у Чернігові — bins за замовчуванням")
plt.xlabel("Кількість опадів, мм")
plt.ylabel("Частота")
plt.grid(axis="y", alpha=0.3)

plt.show()


# ------------------------------------------------------------
# 2. Гістограма з 3 bins
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    rain_values,
    bins=3,
    edgecolor="black",
    color="lightgreen"
)

plt.title("Гістограма опадів у Чернігові — 3 bins")
plt.xlabel("Кількість опадів, мм")
plt.ylabel("Частота")
plt.grid(axis="y", alpha=0.3)

plt.show()


# ------------------------------------------------------------
# 3. Гістограма з 10 bins
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.hist(
    rain_values,
    bins=10,
    edgecolor="black",
    color="salmon"
)

plt.title("Гістограма опадів у Чернігові — 10 bins")
plt.xlabel("Кількість опадів, мм")
plt.ylabel("Частота")
plt.grid(axis="y", alpha=0.3)

plt.show()


# ------------------------------------------------------------
# Правило sqrt(n)
# ------------------------------------------------------------

n = len(rain_values)

sqrt_rule = np.sqrt(n)

# Правило Стерджеса
sturges_rule = 1 + np.log2(n)

print(f"\nКількість спостережень n = {n}")
print(f"Правило √n: √{n} = {sqrt_rule:.2f}")
print(f"Орієнтовна кількість інтервалів за √n: {round(sqrt_rule)}")

print(f"\nПравило Стерджеса:")
print(f"k = 1 + log2({n}) = {sturges_rule:.2f}")
print(f"Орієнтовна кількість інтервалів за Стерджесом: "
      f"{round(sturges_rule)}")

# Письмове пояснення:
# За правилом √n для n=12 отримуємо приблизно 3–4 інтервали.
# За правилом Стерджеса отримуємо приблизно 5 інтервалів.
#
# Гістограма з 10 bins є надто деталізованою для лише 12 спостережень:
# багато інтервалів мають дуже малу або нульову частоту.
#
# Гістограма з 3 bins є більш простою і добре показує загальну форму
# розподілу, не створюючи надмірної кількості порожніх інтервалів.
#
# Для цього набору найбільш наочною є гістограма з 3 bins
# або гістограма з кількістю bins за замовчуванням.


# ============================================================
# ЗАВДАННЯ 4. МІРИ ПОЛОЖЕННЯ Й КВАНТИЛІ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 4. МІРИ ПОЛОЖЕННЯ Й КВАНТИЛІ")
print("=" * 70)

mean_value = rain.mean()
median_value = rain.median()
mode_values = rain.mode()

q1 = rain.quantile(0.25)
q3 = rain.quantile(0.75)

print(f"\nСереднє арифметичне: {mean_value:.2f} мм")
print(f"Медіана: {median_value:.2f} мм")
print(
    "Мода:",
    ", ".join(f"{x:.0f} мм" for x in mode_values)
)
print(f"Q1 (25-й процентиль): {q1:.2f} мм")
print(f"Q3 (75-й процентиль): {q3:.2f} мм")

position_table = pd.DataFrame({
    "Міра": [
        "Середнє",
        "Медіана",
        "Мода",
        "Q1",
        "Q3"
    ],
    "Значення, мм": [
        mean_value,
        median_value,
        mode_values.iloc[0],
        q1,
        q3
    ]
})

print("\nПідсумкова таблиця:")
print(position_table.to_string(index=False))

# ------------------------------------------------------------
# Письмове пояснення:
# ------------------------------------------------------------

difference = mean_value - median_value

print("\nАналіз середнього та медіани:")
print(
    f"Середнє ({mean_value:.2f} мм) більше за медіану "
    f"({median_value:.2f} мм) на {difference:.2f} мм."
)

print("""
Ця різниця є помітною. Причиною є кілька великих літніх значень
опадів — 65, 72 та 80 мм, які збільшують середнє арифметичне.

Отже, розподіл має ознаки правосторонньої асиметрії:
більшість значень знаходиться приблизно в діапазоні 35–55 мм,
а кілька більших значень улітку зміщують середнє вгору.
""")


# ============================================================
# ЗАВДАННЯ 5. ЗВ'ЯЗОК ЗІ ШКАЛОЮ ВИМІРЮВАННЯ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 5. ЗВ'ЯЗОК ЗІ ШКАЛОЮ ВИМІРЮВАННЯ")
print("=" * 70)

print("""
Кількість опадів належить до відносної шкали, оскільки має
природний нуль: 0 мм означає повну відсутність опадів.

Для цієї шкали можна коректно використовувати всі три міри
положення:

- середнє — характеризує середню кількість опадів;
- медіана — визначає центральне значення впорядкованого ряду;
- мода — показує найчастіше значення.

Це принципово відрізняється від змінної "місто".
Місто належить до номінальної шкали. Якщо закодувати міста
числами, ці числа залишаються лише умовними кодами категорій.
Їх не можна змістовно додавати, ділити або усереднювати.

Тому для змінної "місто" з трьох розглянутих мір положення
коректною є лише мода.
""")


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 1
# ============================================================

print("\n" + "=" * 70)
print("КОНТРОЛЬНЕ ПИТАННЯ 1")
print("=" * 70)

# Чи можете обґрунтувати, чому середнє арифметичне некоректне
# для номінальної змінної "місто", навіть якщо завжди можна
# закодувати міста числами?
#
# Відповідь:
# Кодування міст числами не робить номінальну змінну кількісною.
# Наприклад, якщо Київ = 1, Львів = 2, Чернігів = 3,
# ці числа є лише кодами категорій.
# Число 3 не означає, що Чернігів у три рази "більший" за Київ.
# Тому середнє арифметичне таких кодів не має змістовної
# статистичної інтерпретації.
# Для номінальної змінної коректною мірою положення є мода.


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 2
# ============================================================

print("\n" + "=" * 70)
print("КОНТРОЛЬНЕ ПИТАННЯ 2")
print("=" * 70)

# Чому "номер місяця" — приклад, де класифікація за шкалою
# вимірювання не однозначна?
#
# Відповідь:
# Місяці мають природний порядок, тому номер місяця можна
# розглядати як порядкову змінну.
# Водночас календарні інтервали між сусідніми місяцями
# мають однаковий порядок часу, тому можна аргументувати
# й інтервальну інтерпретацію.
#
# Однак у номера місяця немає природного нуля:
# "нульовий місяць" не означає відсутність місяця.
# Крім того, календар є циклічним: після грудня (12) йде
# січень (1). Тому класифікація може бути неоднозначною.
# У цій роботі номер місяця класифіковано як порядкову шкалу.


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 3
# ============================================================

print("\n" + "=" * 70)
print("КОНТРОЛЬНЕ ПИТАННЯ 3")
print("=" * 70)

# Чому для "кількості опадів" потрібна гістограма з інтервалами,
# а не стовпчикова діаграма за окремими значеннями?
#
# Відповідь:
# Кількість опадів є кількісною змінною, а при малому наборі
# даних багато значень можуть бути різними.
# Якщо створити окремий стовпчик для кожного значення,
# наприклад 35, 38, 40, 42, 44, 45, 46 тощо,
# діаграма буде погано показувати загальну форму розподілу.
#
# Гістограма об'єднує близькі значення в інтервали.
# Це дозволяє побачити, у яких діапазонах опадів
# спостережень найбільше або найменше.


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 4
# ============================================================

print("\n" + "=" * 70)
print("КОНТРОЛЬНЕ ПИТАННЯ 4")
print("=" * 70)

# Що означає накопичена частота на межі третього інтервалу?
#
# Відповідь:
# Накопичена частота після третього інтервалу дорівнює 8.
# Це означає, що 8 із 12 місяців мають кількість опадів
# не більшу за 65 мм.
#
# У відсотках:
# 8 / 12 * 100% = 66,7%.
#
# Отже, 66,7% місяців року мають кількість опадів
# не більшу за 65 мм.


# ============================================================
# ПІДСУМКОВИЙ ВИВІД
# ============================================================

print("\n" + "=" * 70)
print("ПІДСУМОК")
print("=" * 70)

print(f"""
Місто: {city}
Кількість спостережень: {n}

Варіаційний ряд:
{", ".join(map(str, variation_row))}

Середнє: {mean_value:.2f} мм
Медіана: {median_value:.2f} мм
Мода: {mode_values.iloc[0]:.0f} мм
Q1: {q1:.2f} мм
Q3: {q3:.2f} мм

Правило √n: приблизно {sqrt_rule:.2f} інтервали
Правило Стерджеса: приблизно {sturges_rule:.2f} інтервали

Накопичена частота після 3-го інтервалу:
{third_cumulative} із {n} = {third_share:.1f}%

Висновок:
Розподіл опадів у Чернігові має ознаки правосторонньої
асиметрії, оскільки середнє значення ({mean_value:.2f} мм)
помітно перевищує медіану ({median_value:.2f} мм).
""")

```
