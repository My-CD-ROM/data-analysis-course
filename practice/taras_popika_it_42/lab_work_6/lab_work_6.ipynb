```python

# ПРАКТИКА 6. 
# Варіант 6 — Чернігів

import pandas as pd
import numpy as np

months = [
    "Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"
]

chernihiv_values = [85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86]

humidity = pd.Series(
    chernihiv_values,
    index=months,
    name="Вологість (%)"
)

print("=" * 70)
print("ПРАКТИКА 6. ОПИСОВА СТАТИСТИКА")
print("Варіант 6 — Чернігів")
print("=" * 70)

print("\nСередньомісячна відносна вологість повітря:")
print(humidity)


# ============================================================
# ЗАВДАННЯ 1. describe() НА ВЛАСНОМУ ВАРІАНТІ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 1. describe() НА ВЛАСНОМУ ВАРІАНТІ")
print("=" * 70)

description = humidity.describe()

print("\nРезультат .describe():")
print(description)

# Відповідність рядків .describe() мірам статистики:
#
# count — кількість спостережень, у нашому випадку 12.
# mean — середнє арифметичне.
# std — стандартне відхилення вибірки (ddof=1).
# min — мінімальне значення.
# 25% — перший квартиль Q1.
# 50% — медіана (Q2).
# 75% — третій квартиль Q3.
# max — максимальне значення.
#
# Отже:
# mean = середнє
# std = стандартне відхилення
# min = мінімум
# 25% = Q1
# 50% = медіана
# 75% = Q3
# max = максимум


# ============================================================
# ЗАВДАННЯ 2. РУЧНЕ ОБЧИСЛЕННЯ І ПЕРЕВІРКА
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 2. РУЧНЕ ОБЧИСЛЕННЯ І ПЕРЕВІРКА")
print("=" * 70)

# Розмах
rozmakh = humidity.max() - humidity.min()

# Дисперсія для вибірки: ddof=1
variance_ddof1 = humidity.var(ddof=1)

# Дисперсія для генеральної сукупності: ddof=0
variance_ddof0 = humidity.var(ddof=0)

# Стандартне відхилення вибірки
std = humidity.std(ddof=1)

# Квартилі
q1 = humidity.quantile(0.25)
q3 = humidity.quantile(0.75)

# Міжквартильний розмах
iqr = q3 - q1

# Коефіцієнт варіації
mean = humidity.mean()
cv = std / mean * 100


print("\nРучні обчислення:")
print(f"Розмах: {rozmakh:.2f}")
print(f"Дисперсія (ddof=1): {variance_ddof1:.4f}")
print(f"Дисперсія (ddof=0): {variance_ddof0:.4f}")
print(f"Стандартне відхилення (ddof=1): {std:.4f}")
print(f"Q1: {q1:.2f}")
print(f"Q3: {q3:.2f}")
print(f"IQR: {iqr:.2f}")
print(f"Коефіцієнт варіації: {cv:.2f}%")


# Перевірка Q1 і Q3 з describe()
print("\nПеревірка квартилів:")
print(f"Q1 з describe(): {description['25%']:.2f}")
print(f"Q1 ручний:        {q1:.2f}")

print(f"Q3 з describe(): {description['75%']:.2f}")
print(f"Q3 ручний:        {q3:.2f}")

print("\nЧи збігаються Q1:", np.isclose(description["25%"], q1))
print("Чи збігаються Q3:", np.isclose(description["75%"], q3))


# Порівняння ddof=1 та ddof=0
difference_variance = variance_ddof1 - variance_ddof0

print("\nПорівняння дисперсій:")
print(f"Різниця: {difference_variance:.4f}")

# Пояснення:
#
# Дисперсія з ddof=1 є вибірковою дисперсією.
# Саме її використовує pandas для .std() за замовчуванням.
#
# Дисперсія з ddof=0 розглядає ці 12 значень як всю генеральну
# сукупність і ділить суму квадратів відхилень на n.
#
# Дисперсія з ddof=1 дещо більша, оскільки використовується
# поправка Бесселя: сума квадратів відхилень ділиться на n-1.


# ============================================================
# ЗАВДАННЯ 3. ПОРІВНЯННЯ РЕГІОНІВ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 3. ПОРІВНЯННЯ РЕГІОНІВ")
print("=" * 70)

# Повний DataFrame усіх 10 міст

data = {
    "Місто": [
        "Київ",
        "Львів",
        "Одеса",
        "Харків",
        "Дніпро",
        "Чернігів",
        "Івано-Франківськ",
        "Полтава",
        "Суми",
        "Ужгород"
    ],

    "Регіон": [
        "Центр",
        "Захід",
        "Південь",
        "Схід",
        "Схід",
        "Центр",
        "Захід",
        "Центр",
        "Схід",
        "Захід"
    ],

    "Січ": [84, 85, 82, 83, 82, 85, 86, 84, 85, 84],
    "Лют": [82, 83, 80, 80, 79, 83, 84, 81, 82, 82],
    "Бер": [78, 80, 78, 75, 74, 79, 81, 76, 77, 79],
    "Кві": [70, 73, 74, 68, 67, 71, 75, 69, 70, 73],
    "Тра": [65, 68, 71, 62, 61, 66, 70, 63, 64, 68],
    "Чер": [63, 65, 69, 60, 59, 63, 67, 61, 62, 66],
    "Лип": [61, 63, 67, 58, 57, 61, 65, 59, 60, 64],
    "Сер": [63, 65, 68, 60, 59, 63, 67, 61, 62, 66],
    "Вер": [68, 70, 72, 66, 65, 69, 72, 67, 68, 71],
    "Жов": [75, 77, 76, 73, 72, 76, 78, 74, 75, 77],
    "Лис": [81, 82, 80, 79, 78, 82, 83, 80, 81, 82],
    "Гру": [85, 86, 83, 84, 83, 86, 87, 85, 86, 85]
}

df = pd.DataFrame(data)

print("\nПовний DataFrame:")
print(df.to_string(index=False))


# Список місячних стовпців
month_columns = [
    "Січ", "Лют", "Бер", "Кві", "Тра", "Чер",
    "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"
]

# Середня річна вологість для кожного міста
df["Середня річна вологість"] = df[month_columns].mean(axis=1)

print("\nDataFrame із середньою річною вологістю:")
print(
    df[["Місто", "Регіон", "Середня річна вологість"]]
    .to_string(index=False)
)


# ------------------------------------------------------------
# Порівнюємо Центр і Захід.
# Чернігів належить до регіону Центр.
# Інший регіон для порівняння — Захід.
# ------------------------------------------------------------

center = df[df["Регіон"] == "Центр"]["Середня річна вологість"]

west = df[df["Регіон"] == "Захід"]["Середня річна вологість"]


print("\nОписова статистика регіону Центр:")
print(center.describe())

print("\nОписова статистика регіону Захід:")
print(west.describe())


# Стандартні відхилення регіонів
std_center = center.std()
std_west = west.std()

# Середні значення регіонів
mean_center = center.mean()
mean_west = west.mean()

# Коефіцієнти варіації
cv_center = std_center / mean_center * 100
cv_west = std_west / mean_west * 100

print("\nПорівняння регіонів:")
print(f"Центр: std = {std_center:.4f}")
print(f"Захід: std = {std_west:.4f}")

print(f"Центр: CV = {cv_center:.4f}%")
print(f"Захід: CV = {cv_west:.4f}%")


# Письмовий висновок:
#
# Для порівняння обираємо Центр і Захід.
# У регіоні Центр три міста: Київ, Чернігів і Полтава.
# У регіоні Захід три міста: Львів, Івано-Франківськ та Ужгород.
#
# Порівнюємо стандартні відхилення середньої річної вологості.
# Більший std означає більший абсолютний розкид між містами.
#
# Для коректного порівняння також розглядаємо коефіцієнт варіації,
# оскільки він нормує стандартне відхилення на середнє значення.
#
# У нашому наборі середні значення вологості регіонів близькі,
# тому висновок за std і CV буде дуже подібним.
# Проте CV є зручнішим показником для порівняння варіативності,
# оскільки враховує масштаб середнього значення.


# ============================================================
# ЗАВДАННЯ 4. ФОРМА РОЗПОДІЛУ ВЛАСНИХ ДАНИХ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 4. ФОРМА РОЗПОДІЛУ ВЛАСНИХ ДАНИХ")
print("=" * 70)

median = humidity.median()

print(f"\nСереднє: {mean:.2f}%")
print(f"Медіана: {median:.2f}%")
print(f"Різниця між середнім і медіаною: {abs(mean - median):.2f}")


# Письмове пояснення:
#
# Для Чернігова середнє і медіана близькі за значенням.
# Це свідчить про те, що розподіл 12 місячних значень
# відносної вологості повітря є ближчим до симетричного,
# а не сильно скошеним.
#
# Значення вологості зменшуються від зимових місяців до літа,
# а потім знову збільшуються восени та взимку.
# Тобто ряд має приблизно симетричну сезонну форму.
#
# Оскільки середнє і медіана близькі, середнє арифметичне
# є репрезентативною мірою положення для цих даних.
# Медіана також є коректною, але суттєвої переваги над середнім
# у цьому випадку немає.


# ============================================================
# ЗАВДАННЯ 5. ПОРІВНЯННЯ ВАРІАТИВНОСТІ ДВОХ МІСТ
# ============================================================

print("\n" + "=" * 70)
print("ЗАВДАННЯ 5. ПОРІВНЯННЯ ВАРІАТИВНОСТІ ДВОХ МІСТ")
print("=" * 70)

# Обираємо Львів — інший регіон (Захід)
lviv = df[df["Місто"] == "Львів"][month_columns].iloc[0]

lviv_mean = lviv.mean()
lviv_std = lviv.std()
lviv_cv = lviv_std / lviv_mean * 100

print("\nЧернігів:")
print(f"Середнє: {mean:.2f}%")
print(f"Std: {std:.4f}")
print(f"CV: {cv:.4f}%")

print("\nЛьвів:")
print(f"Середнє: {lviv_mean:.2f}%")
print(f"Std: {lviv_std:.4f}")
print(f"CV: {lviv_cv:.4f}%")


# Письмовий висновок:
#
# Для порівняння обираємо Чернігів і Львів.
# Вони належать до різних регіонів:
# Чернігів — Центр, Львів — Захід.
#
# Сирий std показує абсолютний розкид значень у відсоткових пунктах.
# Але для порівняння варіативності між наборами даних важливо
# враховувати рівень їхніх середніх значень.
#
# Коефіцієнт варіації:
# CV = std / mean * 100%
#
# Він показує величину стандартного відхилення у відсотках
# відносно середнього значення.
#
# Тому CV є коректнішим інструментом для порівняння відносної
# варіативності двох міст.
#
# Навіть якщо середні вологості Чернігова і Львова досить близькі,
# використання CV все одно є коректним, оскільки воно дозволяє
# порівнювати розкид з урахуванням масштабу кожного набору.


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 1
# ============================================================

# Чи можете пояснити, яка міра розсіювання з лекції стоїть
# за кожним із рядків std, 25%/75% і min/max у .describe()?
#
# Відповідь:
# std — стандартне відхилення, яке характеризує типовий розкид
# значень відносно середнього.
#
# 25% і 75% — перший і третій квартилі.
# Їх різниця утворює міжквартильний розмах (IQR):
# IQR = Q3 - Q1.
#
# min і max — мінімальне та максимальне значення.
# Їх різниця утворює розмах:
# Range = max - min.


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 2
# ============================================================

# Чи розумієте, чому заповнення пропуску середнім штучно
# зменшує std, і чи можете показати механіку цього на конкретних числах?
#
# Відповідь:
# Якщо пропущене значення замінити середнім, воно точно дорівнює
# середньому і тому має нульове відхилення від нього.
#
# Наприклад, маємо значення:
# 10, 20, 30.
#
# Середнє = 20.
#
# Якщо замість пропущеного значення поставити середнє 20,
# отримуємо:
# 10, 20, 20, 30.
#
# Нове значення 20 не додає жодного відхилення від середнього.
# Через це загальний розкид даних стає меншим порівняно
# з ситуацією, коли реальне невідоме значення могло бути,
# наприклад, 50 або 5.
#
# Отже, заповнення пропусків середнім може штучно зменшувати std.


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 3
# ============================================================

# Чому порівнювати абсолютні std двох груп або регіонів
# різного масштабу без коефіцієнта варіації може привести
# до хибного висновку?
#
# Відповідь:
# Стандартне відхилення показує абсолютний розкид і залежить
# від масштабу значень.
#
# Наприклад, група A має середнє 100 і std 10,
# а група B має середнє 1000 і std 20.
#
# За сирим std група B є більш розкиданою:
# 20 > 10.
#
# Але у відносному вираженні:
# CV(A) = 10 / 100 * 100% = 10%
# CV(B) = 20 / 1000 * 100% = 2%
#
# Отже, відносно свого середнього група A значно варіативніша.
#
# Саме тому для порівняння відносного розкиду використовують CV.


# ============================================================
# КОНТРОЛЬНЕ ПИТАННЯ 4
# ============================================================

# Судячи з власного розподілу (Завдання 4), чи він симетричний,
# і яка міра положення репрезентативніша саме для ваших даних — і чому?
#
# Відповідь:
# Для Чернігова середнє і медіана дуже близькі.
# Це свідчить про відсутність сильної асиметрії.
#
# Розподіл місячної вологості ближчий до симетричного:
# взимку значення високі, влітку низькі, а восени та навесні
# вони поступово переходять між цими рівнями.
#
# Тому середнє арифметичне є репрезентативною мірою положення.
# Медіана також коректна, але через відсутність сильної асиметрії
# немає необхідності надавати їй перевагу над середнім.


# ============================================================
# ПІДСУМКОВИЙ ВИВІД
# ============================================================

print("\n" + "=" * 70)
print("ПІДСУМКОВІ РЕЗУЛЬТАТИ — ВАРІАНТ 6, ЧЕРНІГІВ")
print("=" * 70)

print(f"""
Кількість спостережень: {len(humidity)}

Середнє: {mean:.2f}%
Медіана: {median:.2f}%
Мінімум: {humidity.min():.2f}%
Максимум: {humidity.max():.2f}%
Розмах: {rozmakh:.2f}%

Q1: {q1:.2f}%
Q3: {q3:.2f}%
IQR: {iqr:.2f}%

Дисперсія (ddof=1): {variance_ddof1:.4f}
Дисперсія (ddof=0): {variance_ddof0:.4f}

Стандартне відхилення: {std:.4f}
Коефіцієнт варіації: {cv:.2f}%

Регіон Чернігова: Центр
Регіон для порівняння: Захід
Місто для порівняння: Львів

CV Чернігова: {cv:.2f}%
CV Львова: {lviv_cv:.2f}%

Висновок щодо форми розподілу:
середнє та медіана близькі, тому розподіл ближчий
до симетричного, а середнє є репрезентативною
мірою положення.
""")

print("=" * 70)
print("ПРАКТИКУ 6 ЗАВЕРШЕНО")
print("=" * 70)
```
