## Завдання 1. Створення масивів

Середньомісячні температури Києва задані в градусах Цельсія. Рік невисокосний.

```python
temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

print('temps:', temps)
print('temps shape:', temps.shape, 'dtype:', temps.dtype)
print('days_in_month:', days_in_month)
print('days_in_month shape:', days_in_month.shape, 'dtype:', days_in_month.dtype)
print('month_numbers:', month_numbers)
print('month_numbers shape:', month_numbers.shape, 'dtype:', month_numbers.dtype)
```

## Завдання 2. Векторизовані обчислення

Перетворення температури та підрахунок кількості днів виконуються без циклів.

```python
temps_fahrenheit = temps * 9 / 5 + 32
warm_days = days_in_month[temps > 15]

print('Температури у °F:', temps_fahrenheit)
print('Температури вище 15 °C:', temps[temps > 15])
print('Днів із температурою вище 15 °C:', warm_days.sum())
print('Сума днів у році:', days_in_month.sum())
print('Рік невисокосний:', days_in_month.sum() == 365)
```

## Завдання 3. Індексація і фільтрація

```python
warmest_index = temps.argmax()
coldest_index = temps.argmin()
negative_months = month_numbers[temps < 0]
sorted_temps = np.sort(temps)

print('Найтепліший місяць:', month_numbers[warmest_index], 'значення:', temps[warmest_index], '°C, індекс:', warmest_index)
print('Найхолодніший місяць:', month_numbers[coldest_index], 'значення:', temps[coldest_index], '°C, індекс:', coldest_index)
print('Місяці з температурою нижче нуля:', negative_months)
print('Відсортовані температури:', sorted_temps)
print('Оригінальний temps не змінено:', temps)
```

## Завдання 4. Двовимірний масив і `axis`

Третій рядок містить умовну кількість сонячних годин за кожен місяць.

```python
sunny_hours = np.array([55, 70, 95, 140, 190, 220, 240, 225, 170, 120, 75, 50])
data = np.vstack([temps, days_in_month, sunny_hours])

print('data:\n', data)
print('Форма data:', data.shape)
print('Сума кожного рядка:', data.sum(axis=1))
print('Середнє кожного рядка:', data.mean(axis=1))
column_sums = data.sum(axis=0)
print('Сума кожного стовпця:', column_sums)
```

Сума першого стовпця відповідає січню: вона додає температуру січня, кількість днів у січні та умовні сонячні години січня. Оскільки це різні фізичні величини, число має сенс лише як результат демонстрації `axis=0`, а не як фізична характеристика.

## Завдання 5. `reshape` і broadcasting

```python
rows = np.vstack([temps, days_in_month])
correction = np.array([1, 0])

print('Форма rows:', rows.shape)
print('Форма correction:', correction.shape)
try:
    incompatible_result = rows + correction
except ValueError as error:
    print('Очікувана помилка broadcasting:', error)

correction_column = correction.reshape(2, 1)
corrected_rows = rows + correction_column
print('Форма correction після reshape:', correction_column.shape)
print('Результат rows + correction.reshape(2, 1):')
print(corrected_rows)
```

`rows` має форму `(2, 12)`, а `correction` форму `(2,)`. Під час broadcasting останній вимір порівнюється з останнім: `12` і `2` несумісні. Після `reshape(2, 1)` форми стають `(2, 12)` і `(2, 1)`: одиничний другий вимір розширюється до 12, тому поправка застосовується окремо до кожного рядка.

## Письмові відповіді

1. `shape == (12,)` означає одновимірний масив із 12 елементів без окремого поняття рядка чи стовпця. `shape == (1, 12)` означає двовимірний масив: один рядок і 12 стовпців.

2. Вираз `temps[temps > 15]` використовує булеву маску й залишає тільки елементи, для яких умова істинна. Тому результат має довжину, що дорівнює кількості місяців із температурою вище 15 °C, а не обов'язково 12.

3. Додавання масиву форми `(3,)` до масиву форми `(3, 12)` дає помилку: останні виміри `12` і `3` несумісні. Щоб додати по одному значенню до кожного рядка, потрібна форма `(3, 1)`.

4. `rows + correction` дає помилку, бо `(2,)` порівнюється з останнім виміром `(2, 12)`, тобто `2` порівнюється з `12`. Після `correction.reshape(2, 1)` форми `(2, 12)` і `(2, 1)` сумісні: одиниця розширюється до 12 стовпців.
