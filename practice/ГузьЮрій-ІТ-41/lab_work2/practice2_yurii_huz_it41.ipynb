Мета: побудувати `DataFrame`, попрацювати з індексом, відбором через `.loc` і `.iloc`, фільтрацією та обчислюваними стовпцями.

```python
import pandas as pd

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
avg_temp = [-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2]
precip = [38, 36, 38, 41, 53, 74, 84, 69, 47, 36, 44, 44]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip,
})

print(df)
print()
print("Форма:", df.shape)
print()
print("Типи стовпців:")
print(df.dtypes)
```

## Завдання 1. Індекс

Спочатку використовується індекс за замовчуванням від 0 до 11. Потім індексом стає назва місяця, а після вибірки повертаємо числову нумерацію рядків.

```python
print("Індекс за замовчуванням:", df.index)

df_by_month = df.set_index("month")
print()
print("Рядок для травня через .loc:")
print(df_by_month.loc["Тра"])

df = df_by_month.reset_index()
print()
print("Після reset_index():")
print(df)
```

## Завдання 2. Відбір рядків і стовпців

```python
df.index = range(1, 13)

print("avg_temp для індексів 1..6 через .loc:")
print(df.loc[1:6, "avg_temp"])

print("avg_temp для позицій 0..5 через .iloc:")
print(df.iloc[0:6, 1])

print("Стовпці month і precip:")
print(df[["month", "precip"]])
```

Оскільки в цій таблиці індекси `1..12` відповідають позиціям рядків зі зміщенням на 1, `.loc[1:6]` і `.iloc[0:6]` вибирають однакові шість місяців. За іншого індексу або за відсутності потрібних підписів результати могли б відрізнятися.

## Завдання 3. Фільтрація за умовою

```python
print("Місяці, де avg_temp > 15:")
print(df.loc[df["avg_temp"] > 15, ["month", "avg_temp"]])

warm_and_wet = (df["avg_temp"] > 10) & (df["precip"] > 50)
print()
print("Місяці, де avg_temp > 10 і precip > 50:")
print(df.loc[warm_and_wet, ["month", "avg_temp", "precip"]])

cold_or_rainy = (df["avg_temp"] < 0) | (df["precip"] > 80)
print()
print("Місяці, де avg_temp < 0 або precip > 80:")
print(df.loc[cold_or_rainy, ["month", "avg_temp", "precip"]])
```

## Завдання 4. Обчислювані стовпці

```python
df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32
df["is_wet"] = df["precip"] > df["precip"].mean()

def season_by_month(month_number):
    if month_number in [12, 1, 2]:
        return "зима"
    if month_number in [3, 4, 5]:
        return "весна"
    if month_number in [6, 7, 8]:
        return "літо"
    return "осінь"

df["season"] = df.index.to_series().apply(season_by_month)
print("Середнє значення precip:", df["precip"].mean())
print(df)
```

## Письмові відповіді

1. `.loc` працює за підписами індексу, а `.iloc` — за позиціями рядків, що починаються з нуля. Тому `.loc[1:6]` включає обидва підписи 1 і 6, а `.iloc[0:6]` використовує правило зрізу Python і не включає позицію 6. Якщо індекс не збігається з позицією або містить інші підписи, кількість і склад рядків можуть відрізнятися.

2. `and` працює з одиничними логічними значеннями, а порівняння стовпця pandas повертає цілу булеву `Series`. Через це `df["avg_temp"] > 10 and df["precip"] > 50` викликає помилку про неоднозначність істинності Series. Оператор `&` виконує поелементне логічне «і», тому з дужками `(df["avg_temp"] > 10) & (df["precip"] > 50)` працює правильно.

3. Обчислюваний стовпець, створений формулою, використовує векторизовану операцію над усім стовпцем одразу. Це зазвичай коротше й швидше, наприклад формула переведення температури у Фаренгейти. `.apply()` викликає власну функцію для кожного елемента і зручний для складнішої умовної логіки, як визначення пори року.
