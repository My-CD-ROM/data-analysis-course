Мета: перетворити wide-таблицю хмарності у tidy-формат, виконати зворотне перетворення, класифікувати змінні та попрацювати з CSV і JSON.

```python
import pandas as pd

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
cloudiness = [78, 74, 68, 58, 50, 44, 40, 42, 52, 66, 76, 82]
```

## Завдання 1. Побудова wide-таблиці

Вихідна таблиця має один рядок для міста, а назви місяців є окремими стовпцями.

```python
wide = pd.DataFrame({
    "місто": ["Київ"],
    "Січ": [78],
    "Лют": [74],
    "Бер": [68],
    "Кві": [58],
    "Тра": [50],
    "Чер": [44],
    "Лип": [40],
    "Сер": [42],
    "Вер": [52],
    "Жов": [66],
    "Лис": [76],
    "Гру": [82],
})

print(wide)
print("Форма wide-таблиці:", wide.shape)
```

## Завдання 2. Приведення до tidy через `melt()`

```python
tidy = wide.melt(
    id_vars="місто",
    var_name="місяць",
    value_name="хмарність",
)

print(tidy)
print("Форма tidy-таблиці:", tidy.shape)
```

У wide-форматі 12 місяців зберігалися в 12 окремих стовпцях одного рядка. `melt()` перетворює назви цих стовпців на значення стовпця `місяць`, а відповідні числа — на значення стовпця `хмарність`. Тому один рядок розгортається у 12 рядків: по одному спостереженню для кожного місяця.

## Завдання 3. Класифікація змінних

```python
measurement_scales = pd.DataFrame({
    "змінна": ["місто", "місяць", "хмарність"],
    "шкала": ["номінальна", "порядкова", "відносна"],
    "пояснення": [
        "Назви міст є категоріями без природного порядку.",
        "Місяці мають природну послідовність у межах року.",
        "Відсоток хмарності має нуль і дає змогу порівнювати відношення значень.",
    ],
})
print(measurement_scales)
```

## Завдання 4. Зворотне перетворення й перевірка

```python
back = tidy.pivot(index="місто", columns="місяць", values="хмарність").reset_index()
back = back[["місто"] + months]

same_values = wide.equals(back)
print(back)
print("Таблиці мають однакові значення:", same_values)
```

## Завдання 5. Читання й запис CSV та JSON

```python
tidy.to_csv("variant.csv", index=False)
tidy.to_json("variant.json", orient="records", force_ascii=False)

reloaded_csv = pd.read_csv("variant.csv")
reloaded_json = pd.read_json("variant.json")

print("CSV після читання:")
print(reloaded_csv)
print()
print("JSON після читання:")
print(reloaded_json)
print()
print("CSV і JSON містять однакові дані:", reloaded_csv.equals(reloaded_json))
print()
print("Вміст variant.csv:")
print(open("variant.csv", encoding="utf-8").read())
print("Вміст variant.json:")
print(open("variant.json", encoding="utf-8").read())
```

CSV зберігає таблицю як текст із рядками та роздільниками між полями. Перший рядок зазвичай містить заголовки стовпців. JSON зберігає ті самі дані як масив об'єктів: кожен рядок є окремим об'єктом із парами «ключ — значення». JSON має більш явну структуровану форму, а CSV простіший і компактніший для табличного обміну.

## Письмові відповіді

1. `місто` — номінальна шкала, бо це назва категорії без числового значення та природного порядку. `хмарність` — шкала відношень: 0% означає повну відсутність хмарності, а значення можна порівнювати як відношення.

2. Wide-таблиця не є tidy, тому що одна змінна «місяць» розподілена між назвами 12 стовпців. Після `melt()` кожен рядок містить одне спостереження, а місяць стає значенням окремого стовпця.

3. CSV є плоским текстовим форматом із роздільниками, який природно описує рядки та стовпці. JSON є ієрархічним текстовим форматом: таблиця подається як масив об'єктів із назвами полів.
