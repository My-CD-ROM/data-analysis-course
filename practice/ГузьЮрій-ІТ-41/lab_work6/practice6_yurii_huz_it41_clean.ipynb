Описова статистика вологості за 12 місяців.

```python
import numpy as np
import pandas as pd

months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
kyiv = pd.Series([84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85], index=months, name="вологість")
print(kyiv)
```

## Завдання 1. `describe()`

`count` — кількість спостережень; `mean` — середнє; `std` — вибіркове стандартне відхилення; `min` і `max` — мінімум і максимум; `25%` і `75%` — квартилі; `50%` — медіана.

```python
description = kyiv.describe()
print(description)
```

## Завдання 2. Ручне обчислення і перевірка

```python
range_value = kyiv.max() - kyiv.min()
variance_ddof1 = kyiv.var(ddof=1)
variance_ddof0 = kyiv.var(ddof=0)
std_value = kyiv.std(ddof=1)
q1, q3 = kyiv.quantile([0.25, 0.75])
iqr = q3 - q1
cv_kyiv = std_value / kyiv.mean() * 100
print("Розмах:", range_value)
print("Дисперсія ddof=1:", variance_ddof1)
print("Дисперсія ddof=0:", variance_ddof0)
print("Стандартне відхилення:", std_value)
print("Q1, Q3, IQR:", q1, q3, iqr)
print("CV (%):", cv_kyiv)
print("Q1 збігається:", q1 == description.loc["25%"])
print("Q3 збігається:", q3 == description.loc["75%"]) 
```

## Завдання 3. Порівняння регіонів

Повна таблиця міст використовується для порівняння Центру і Сходу через булеву фільтрацію, без `groupby()`. 

```python
rows = [
    ["Київ", "Центр", 84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85],
    ["Львів", "Захід", 85, 83, 80, 73, 68, 65, 63, 65, 70, 77, 82, 86],
    ["Одеса", "Південь", 82, 80, 78, 74, 71, 69, 67, 68, 72, 76, 80, 83],
    ["Харків", "Схід", 83, 80, 75, 68, 62, 60, 58, 60, 66, 73, 79, 84],
    ["Дніпро", "Схід", 82, 79, 74, 67, 61, 59, 57, 59, 65, 72, 78, 83],
    ["Чернігів", "Центр", 85, 83, 79, 71, 66, 63, 61, 63, 69, 76, 82, 86],
    ["Івано-Франківськ", "Захід", 86, 84, 81, 75, 70, 67, 65, 67, 72, 78, 83, 87],
    ["Полтава", "Центр", 84, 81, 76, 69, 63, 61, 59, 61, 67, 74, 80, 85],
    ["Суми", "Схід", 85, 82, 77, 70, 64, 62, 60, 62, 68, 75, 81, 86],
    ["Ужгород", "Захід", 84, 82, 79, 73, 68, 66, 64, 66, 71, 77, 82, 85]
]
columns = ["Місто", "Регіон"] + months
regions = pd.DataFrame(rows, columns=columns)
regions["Середня_вологість"] = regions[months].mean(axis=1)
center = regions[regions["Регіон"] == "Центр"]["Середня_вологість"]
east = regions[regions["Регіон"] == "Схід"]["Середня_вологість"]
print(regions[["Місто", "Регіон", "Середня_вологість"]])
print()
print("Центр:")
print(center.describe())
print("Схід:")
print(east.describe())
print("CV Центр (%):", center.std() / center.mean() * 100)
print("CV Схід (%):", east.std() / east.mean() * 100)
```

## Завдання 4. Форма розподілу

Для Києва середнє і медіана близькі, тому розподіл ближчий до симетричного. Середнє є репрезентативним, а медіана дає стійку перевірку впливу крайніх місяців.

```python
lviv = regions[regions["Місто"] == "Львів"][months].iloc[0]
lviv_cv = lviv.std() / lviv.mean() * 100
print("Київ: середнє =", kyiv.mean(), "std =", kyiv.std(), "CV =", cv_kyiv)
print("Львів: середнє =", lviv.mean(), "std =", lviv.std(), "CV =", lviv_cv)
```

## Завдання 5 і письмові відповіді

Львів належить до іншого регіону — Заходу. CV коректніший за сирий `std`, бо показує розкид у відсотках від середнього і дає масштабно зіставне порівняння.

Середнє для міста некоректне, якщо місто закодувати числами: коди є умовними мітками, а їх заміна змінила б середнє.

Заповнення пропуску середнім додає значення без відхилення від середнього й тому штучно зменшує стандартне відхилення.

У `.describe()` `std` — стандартне відхилення, `25%` і `75%` — квартилі, `min` і `max` задають розмах.

Номер місяця можна трактувати як порядкову шкалу через природний порядок і відсутність природного нуля; водночас рівні календарні кроки дають аргумент на користь інтервального трактування.

Для Києва середнє і медіана близькі, отже розподіл вологості майже симетричний, і середнє добре описує типовий рівень.
