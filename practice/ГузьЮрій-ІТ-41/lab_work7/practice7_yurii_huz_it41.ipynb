Мета: побудувати гістограму, boxplot і діаграму розсіювання через об'єктний інтерфейс Matplotlib та показати приклад спотвореної візуалізації.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(1)
days = np.arange(1, 31)
base_sales = 40
base_temperature = 3
weekend = (days % 7 == 6) | (days % 7 == 0)
temperature = np.round(base_temperature + np.random.uniform(-3, 3, size=30), 1)
random_variation = np.random.randint(-5, 6, size=30)
weekend_factor = np.where(weekend, np.random.uniform(1.15, 1.25, size=30), 1)
sales = np.round((base_sales + random_variation - 0.8 * (temperature - base_temperature)) * weekend_factor).astype(int)

kiosk = pd.DataFrame({"день": days, "вихідний": weekend, "температура": temperature, "продажі": sales})
print(kiosk.head())
print("Кількість будніх днів:", (~kiosk["вихідний"]).sum())
print("Кількість вихідних днів:", kiosk["вихідний"].sum())
```

## Завдання 1. Побудова власного набору

Стовпець `вихідний` позначає кожен 6-й і 7-й день тижня. На вихідні продажі збільшуються приблизно на 15–25%, а невелика температурна поправка моделює дещо більший попит на гарячі напої в холодніші дні.

## Завдання 2. Гістограма

Для 30 спостережень побудовано гістограми з 6 і 10 інтервалами. Шість інтервалів зазвичай дають читабельнішу картину, а десять показують більше деталей, але можуть надмірно подрібнити невелику вибірку.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(kiosk["продажі"], bins=6, edgecolor="black")
axes[0].set_xlabel("Продажі, шт./день")
axes[0].set_ylabel("Кількість днів")
axes[0].set_title("Розподіл продажів: 6 інтервалів")
axes[1].hist(kiosk["продажі"], bins=10, edgecolor="black")
axes[1].set_xlabel("Продажі, шт./день")
axes[1].set_ylabel("Кількість днів")
axes[1].set_title("Розподіл продажів: 10 інтервалів")
fig.suptitle("Порівняння кількості інтервалів гістограми")
fig.tight_layout()
plt.show()
```

## Завдання 3. Boxplot

Оскільки у вихідних даних для вихідних днів закладено множник 1,15–1,25, медіана продажів у вихідні очікується вищою, ніж у будні. Розкид може бути також більшим через випадковий множник.

```python
weekday_sales = kiosk.loc[~kiosk["вихідний"], "продажі"]
weekend_sales = kiosk.loc[kiosk["вихідний"], "продажі"]
fig, ax = plt.subplots(figsize=(6, 4))
ax.boxplot([weekday_sales, weekend_sales], labels=["Будні", "Вихідні"])
ax.set_xlabel("Тип дня")
ax.set_ylabel("Продажі, шт./день")
ax.set_title("Продажі в будні та вихідні")
fig.tight_layout()
plt.show()
print("Медіана буднів:", weekday_sales.median())
print("Медіана вихідних:", weekend_sales.median())
```

## Завдання 4. Діаграма розсіювання

На графіку очікується слабкий негативний зв'язок: за нижчої температури поправка для гарячих напоїв трохи збільшує продажі. Водночас вихідний множник і випадкова варіація додають шум, тому зв'язок не має бути сильним.

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(kiosk["температура"], kiosk["продажі"], alpha=0.8)
ax.set_xlabel("Температура, °C")
ax.set_ylabel("Продажі, шт./день")
ax.set_title("Продажі напоїв залежно від температури")
fig.tight_layout()
plt.show()
print("Кореляція температура-продажі:", kiosk["температура"].corr(kiosk["продажі"]))
```

## Завдання 5. Навмисно спотворений графік

Поруч показано коректну гістограму і спотворену версію. У правій версії навмисно прибрано підписи осей і заголовок. Самі числа не змінилися, але без контексту глядачеві важче зрозуміти, що вимірюється, у яких одиницях і яку саме вибірку показано.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(kiosk["продажі"], bins=6, edgecolor="black")
axes[0].set_xlabel("Продажі, шт./день")
axes[0].set_ylabel("Кількість днів")
axes[0].set_title("Коректна гістограма")
axes[1].hist(kiosk["продажі"], bins=6, edgecolor="black")
axes[1].set_title("Спотворена версія без підписів")
axes[1].set_xticks([])
axes[1].set_yticks([])
fig.suptitle("Вплив відсутності контексту на сприйняття")
fig.tight_layout()
plt.show()
```

## Письмові відповіді

1. Обрізана вісь Y на стовпчиковій діаграмі перебільшує різницю між категоріями, бо око порівнює висоти стовпців відносно видимого нуля. Навіть правильні підписані числа можуть здаватися набагато різнішими.

2. Лінійний графік буде методологічною помилкою для номінальних категорій, наприклад для порівняння міст, бо з'єднання точок створює штучний порядок і натяк на неперервний перехід між категоріями. Для номінальної шкали коректніші стовпчикова діаграма або точки без лінії.

3. Підписи осей, одиниці вимірювання і заголовок є частиною твердження графіка: вони визначають змінні, масштаб і контекст. Без них одна й та сама форма може бути неправильно інтерпретована.

4. Chartjunk — зайві декоративні елементи, наприклад 3D-ефект, які не додають інформації. Він ускладнює порівняння висот і може створювати оптичні спотворення, тому шкодить точності сприйняття.
