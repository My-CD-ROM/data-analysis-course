```python
import numpy as np

temps = np.array([-5, -4, 1, 9, 15, 18, 20, 19, 13, 7, 1, -3])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

print("temps shape:", temps.shape, "dtype:", temps.dtype)
print("days_in_month shape:", days_in_month.shape, "dtype:", days_in_month.dtype)
print("month_numbers shape:", month_numbers.shape, "dtype:", month_numbers.dtype)
```

```python
print("temps shape:", temps.shape, "dtype:", temps.dtype)
print("days_in_month shape:", days_in_month.shape, "dtype:", days_in_month.dtype)
print("month_numbers shape:", month_numbers.shape, "dtype:", month_numbers.dtype)
```

```python
temps_f = temps * 9 / 5 + 32
warm_months = temps > 15
days_warm = days_in_month[warm_months].sum()

print("Temps in Fahrenheit:", temps_f)
print("Days warmer than 15°C:", days_warm)
print("Days in year:", days_in_month.sum())
```

```python
max_temp = temps.max()
max_idx = temps.argmax()
min_temp = temps.min()
min_idx = temps.argmin()
cold_months = month_numbers[temps < 0]
sorted_temps = np.sort(temps)

print(f"Hottest month: {month_numbers[max_idx]} with {max_temp}°C")
print(f"Coldest month: {month_numbers[min_idx]} with {min_temp}°C")
print("Months with temperatures below 0°C:", cold_months)
print("Sorted temperatures:", sorted_temps)
```

```python
sun_hours = np.array([45, 50, 90, 140, 190, 210, 230, 220, 170, 110, 60, 40])
data = np.array([temps, days_in_month, sun_hours])

row_sums = data.sum(axis=1)
row_means = data.mean(axis=1)
col_sums = data.sum(axis=0)

print("Row sums:", row_sums)
print("Row means:", row_means)
print("Column sums:", col_sums)
print("Sum for January (first column):", col_sums[0])
print("This is the sum of temperature, days in January, and sun hours.")
```

```python
rows = np.array([temps, days_in_month])
correction = np.array([1, 0])

print("rows shape:", rows.shape)
print("correction shape:", correction.shape)

try:
    result = rows + correction
except ValueError as e:
    print("Error:", e)

correction_reshaped = correction.reshape(2, 1)
result = rows + correction_reshaped
print("Result after reshape and addition:\n", result)
```
