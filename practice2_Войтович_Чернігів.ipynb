```python
import pandas as pd

months = ["Січ","Лют","Бер","Кві","Тра","Чер","Лип","Сер","Вер","Жов","Лис","Гру"]

avg_temp = [-5, -4, 1, 9, 15, 18, 20, 19, 13, 7, 1, -3]
precip = [40, 38, 40, 44, 56, 78, 86, 70, 50, 40, 48, 46]

df = pd.DataFrame({
    "month": months,
    "avg_temp": avg_temp,
    "precip": precip
})

df.head()
```

```python
# Перевірка початкового індексу
print(df.index)

# Встановлення індексу на місяць
df.set_index("month", inplace=True)
print(df.loc["Сер"])  # приклад однієї вибірки по місяцю "Серпань"

# Повернення індексу назад (звичайна числова нумерація)
df.reset_index(inplace=True)
print(df.head())
```

```python
# Встановлюємо числовий індекс від 1 до 12
df.index = range(1, 13)

# Вивід avg_temp для місяців з індексом 1 до 6 за .loc
print(df.loc[1:6, "avg_temp"])

# Вивід avg_temp за позиціями 0 до 5 за .iloc
print(df.iloc[0:6]["avg_temp"])

# Вивід лише двох колонок month і precip
print(df[["month", "precip"]])
```

```python
# Вибрати місяці, де температура > 15°C
print(df[df["avg_temp"] > 15])

# Вибрати місяці, де temp > 10 і precip > 50
print(df[(df["avg_temp"] > 10) & (df["precip"] > 50)])

# Вибрати місяці, де temp < 0 або precip > 80
print(df[(df["avg_temp"] < 0) | (df["precip"] > 80)])
```

```python
# Температура у Фаренгейтах
df["avg_temp_f"] = df["avg_temp"] * 9 / 5 + 32

# Булевий стовпець, якщо опади більші за середнє
df["is_wet"] = df["precip"] > df["precip"].mean()

# Функція для визначення сезону за номером місяця
def get_season(month_num):
    if month_num in [12, 1, 2]:
        return "зима"
    elif month_num in [3, 4, 5]:
        return "весна"
    elif month_num in [6, 7, 8]:
        return "літо"
    else:
        return "осінь"

# Додаємо номер місяця в DataFrame
df["month_num"] = range(1, 13)

# Застосовуємо функцію
df["season"] = df["month_num"].apply(get_season)

print(df)
```

### Відповіді на питання

1. `.loc[1:6]` включає рядки з індексами 1 до 6 включно, а `.iloc[0:6]` — позиції з 0 по 5, тому може бути різниця, якщо індекси не співпадають з позиціями.

2. Вираз `df["avg_temp"] > 10 and df["precip"] > 50` викликає помилку, бо `and` не працює з масивами pandas, потрібно використовувати побітовий оператор `&` з дужками: `(df["avg_temp"] > 10) & (df["precip"] > 50)`.

3. Обчислюваний стовпець за формулою — це векторна операція над колонками, а `.apply()` виконує функцію по рядку або стовпцю, корисно для складніших або умовних обчислень.
