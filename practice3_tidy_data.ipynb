## Практична робота 3. Приведення до tidy data (Варіант 6, Чернігів)


```python
import pandas as pd

data = {
    "місто": ["Чернігів"],
    "Січ": [80],
    "Лют": [76],
    "Бер": [70],
    "Кві": [60],
    "Тра": [52],
    "Чер": [46],
    "Лип": [42],
    "Сер": [44],
    "Вер": [54],
    "Жов": [68],
    "Лис": [78],
    "Гру": [84]
}

wide = pd.DataFrame(data)
print(wide)
```

```python
tidy = wide.melt(id_vars="місто", var_name="місяць", value_name="хмарність")
print(tidy)
print("Форма tidy-таблиці:", tidy.shape)
```

```python
back_to_wide = tidy.pivot(index="місто", columns="місяць", values="хмарність").reset_index()
print(back_to_wide)
```

```python
# Запис
tidy.to_csv("variant.csv", index=False)
tidy.to_json("variant.json", orient="records", force_ascii=False)

# Читання назад
df_csv = pd.read_csv("variant.csv")
df_json = pd.read_json("variant.json")

print(df_csv.head())
print(df_json.head())

# Вивід вмісту variant.csv
with open("variant.csv", "r", encoding="utf-8") as f:
    print(f.read())

# Вивід вмісту variant.json
with open("variant.json", "r", encoding="utf-8") as f:
    print(f.read())
```

### Відповіді на питання:

1. Стовпець "місто" — номінальна шкала, бо це просто категорії без природнього порядку.  
2. Стовпець "хмарність" — шкала відношень, бо має абсолютний нуль та значущу відстань між вимірами.  
3. Wide-таблиця не є tidy, бо кожен місяць — окремий стовпець, а в tidy форматі кожне спостереження має бути в окремому рядку. Melt трансформує стовпці-місяці у стовпець із значеннями місяців і відповідних спостережень.  
4. CSV — плоский таблице-машинний текстовий формат, зручний для табличних даних, JSON — структурований JSON-об’єкт, який краще підтримує вкладені або складні структури.
