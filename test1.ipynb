```python
import numpy as np

print("Версія NumPy:", np.__version__)

a = np.array([1, 2, 3, 4, 5])
print("Масив a:", a)
print("shape:", a.shape)
print("dtype:", a.dtype)
```

```python
temps = np.array([-4, -3, 1, 7, 12, 15, 17, 16, 12, 7, 2, -2])

days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])

month_numbers = np.arange(1, 13)

print("temps: shape =", temps.shape, "| dtype =", temps.dtype)
print(
    "days_in_month: shape =",
    days_in_month.shape,
    "| dtype =",
    days_in_month.dtype,
)
print(
    "month_numbers: shape =",
    month_numbers.shape,
    "| dtype =",
    month_numbers.dtype,
)
```

```python
temps_fahrenheit = temps * 9 / 5 + 32
print("Temperatures in °F:", temps_fahrenheit)

warm_days_total = days_in_month[temps > 15].sum()
print("Всього днів у році з temp > 15°C:", warm_days_total)

total_days = days_in_month.sum()
print("Загальна сума днів у році:", total_days, "(Перевірка пройшла!)")
```

```python
max_idx = temps.argmax()
min_idx = temps.argmin()

print(
    f"Найтепліший місяць: №{month_numbers[max_idx]} зі значенням {temps[max_idx]}°C"
)
print(
    f"Найхолодніший місяць: №{month_numbers[min_idx]} зі значенням {temps[min_idx]}°C"
)

cold_months = month_numbers[temps < 0]
print("Номери місяців з температурою < 0°C:", cold_months)

sorted_temps = np.sort(temps)
print("Відсортовані температури:", sorted_temps)
print("Оригінальний масив temps (не змінився):", temps)
```

```python
sun_hours = np.array([60, 80, 120, 160, 210, 240, 260, 230, 180, 130, 70, 50])
matrix_3x12 = np.array([temps, days_in_month, sun_hours])
print("Форма матриці:", matrix_3x12.shape)

row_sums = matrix_3x12.sum(axis=1)
row_means = matrix_3x12.mean(axis=1)

print("\n--- Результати по рядках (axis=1) ---")
print("Суми по рядках:", row_sums)
print("Середні по рядках:", row_means)

col_sums = matrix_3x12.sum(axis=0)
print("\n--- Результати по стовпцях (axis=0) ---")
print("Суми по стовпцях (для кожного місяця):", col_sums)
```

```python
rows = np.array([temps, days_in_month])
print("Форма rows:", rows.shape)

correction = np.array([1, 0])
print("Форма correction:", correction.shape)

try:
    result = rows + correction
except ValueError as e:
    print("\n[ПОМИЛКА BROADCASTING]:", e)
```
